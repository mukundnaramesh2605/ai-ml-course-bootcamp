```python
# CHeck for missing values
df.isna().sum()
```

```python
# visualizing missing values using heatmap
plt.figure(figsize=(10,5))
missing_values = df.isnull().astype(int)
plt.imshow(missing_values,cmap='viridis',aspect="auto")
plt.title("Missing Values")
plt.ylabel("Rows")
plt.xlabel("Columns")
plt.yticks([]) # hide y axis
plt.colorbar(label = "Missing Data 1 = missing, 0 = present")
plt.show()
```

```python
#Histogramp : distribution of Engagement Metrics
fig,ax = plt.subplots(2,2,figsize = (12,10))
ax[0,0].hist(df['comments_count'].dropna(),bins=30,color='blue',alpha = 0.7)
ax[0,0].set_title("comments_count")
ax[0,1].hist(df['engagement_rate'].dropna(),bins=30,color='green',alpha = 0.7)
ax[0,1].set_title("engagement_rate")
ax[1,0].hist(df['likes_count'].dropna(),bins=30,color='red',alpha = 0.7)
ax[1,0].set_title("likes_count")
ax[1,1].hist(df['shares_count'].dropna(),bins=30,color='purple',alpha = 0.7)
ax[1,1].set_title("shares_count")
plt.tight_layout()
plt.show()
```
