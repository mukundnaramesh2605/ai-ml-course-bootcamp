```python
# ============================================================
# EXERCISE 3
# Goal: Understand the distribution of customer ages.
# ------------------------------------------------------------
import numpy as np
np.random.seed(10)
ages = np.concatenate([
    np.random.normal(28, 5, 400),
    np.random.normal(45, 8, 300),
    np.random.normal(60, 6, 150)
])
mean_age = np.mean(ages)
median_age = np.median(ages)
fig, ax = plt.subplots(figsize=(10, 6))
counts, bins, patches = ax.hist(
    ages, 
    bins=20, 
    color='#1f77b4', 
    edgecolor='white', 
    alpha=0.85
)
ax.bar_label(
    patches, 
    fmt='%d',                  # Formats the bin count as an integer
    padding=3,                 # Places text just above the bar
    fontsize=9, 
    color='#333333'
)
ax.axvline(mean_age, color='red', linestyle='--', linewidth=1.5, label=f'Mean: {mean_age:.1f}')
ax.text(
    mean_age + 1,              # X position (slightly to the right of the line)
    max(counts) * 0.9,         # Y position (near the top of the chart)
    f'Average Age:\n{mean_age:.1f} years', 
    color='red', 
    fontweight='bold'
)

# 6. Clean up and format the chart
ax.set_title('Customer Age Distribution', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Age (Years)', fontsize=11)
ax.set_ylabel('Customer Count (Frequency)', fontsize=11)
ax.set_xlim(10, 80)
ax.set_ylim(0, max(counts) + 15) # Add vertical padding for the bar labels

# Remove top and right borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate the exercise data
np.random.seed(11)
hours_studied = np.random.uniform(0, 12, 80)
exam_score = np.clip(30 + 5.5 * hours_studied + np.random.randn(80) * 8, 0, 100)

# 2. Setup the figure
fig, ax = plt.subplots(figsize=(10, 6))

# 3. Create the scatter plot
ax.scatter(hours_studied, exam_score, color='#69b3a2', edgecolors='black', alpha=0.8, s=50, label='Student Data')

# 4. Calculate and plot a linear trend line (Best Fit Line)
m, c = np.polyfit(hours_studied, exam_score, 1) # Linear regression formula (y = mx + c)
x_trend = np.linspace(0, 12, 100)
y_trend = m * x_trend + c
ax.plot(x_trend, y_trend, color='#e67e22', linestyle='-', linewidth=2, label=f'Trend (Slope: {m:.2f} pts/hr)')

# 5. Annotate specific data points of interest
# Find the student who studied the most
idx_max_hours = np.argmax(hours_studied)
ax.annotate(
    'Max Effort',
    xy=(hours_studied[idx_max_hours], exam_score[idx_max_hours]),
    xytext=(hours_studied[idx_max_hours] - 2.5, exam_score[idx_max_hours] - 5),
    arrowprops=dict(arrowstyle="->", color="black", lw=1.2),
    fontsize=10, fontweight='bold'
)

# Find an outlier (e.g., high study hours but lower score than expected)
ax.annotate(
    'High Study / Lower Score Outlier',
    xy=(9.7, 72.5), # Points near an explicit visual outlier in this seed
    xytext=(5, 85),
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=-0.1", color="red", lw=1.2),
    fontsize=10, color='red'
)

# 6. Clean up layout and style definitions
ax.set_title('Relationship Between Study Hours and Exam Scores', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Hours Studied (Hours)', fontsize=11)
ax.set_ylabel('Exam Score (%)', fontsize=11)
ax.set_xlim(-0.5, 13)
ax.set_ylim(0, 105)

# Remove top and right chart borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower right')

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXERCISE 5
# Goal: Compare test score distributions across 4 schools.
# ------------------------------------------------------------
import numpy as np
np.random.seed(12)
school_1 = np.random.normal(72, 10, 120)
school_2 = np.random.normal(68, 14, 120)
school_3 = np.random.normal(75, 7, 120)
school_4 = np.random.normal(70, 12, 120)
data = [school_1, school_2, school_3, school_4]
labels = ['School 1', 'School 2', 'School 3', 'School 4']
fig, ax = plt.subplots(figsize=(10, 6))
box_plot = ax.boxplot(
    data, 
    tick_labels=labels, 
    patch_artist=True,      # Enables fill color
    notch=False, 
    medianprops=dict(color='black', linewidth=1.5)
)
colors = ['#1f77b4', '#aec7e8', '#ff7f0e', '#ffbb78']
for patch, color in zip(box_plot['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
for i, line in enumerate(box_plot['medians']):
    x_coords, y_coords = line.get_data()
    median_val = y_coords[0]  # Grab the numerical value from the coordinate array
    x_pos = (x_coords[0] + x_coords[1]) / 2
    
    ax.text(
        x_pos, 
        median_val + 1, 
        f'{median_val:.1f}', 
        ha='center', 
        va='bottom', 
        fontsize=10, 
        fontweight='bold',
        color='#222222'
    )
```

```python
# ============================================================
# EXERCISE 12
# Goal: Visualize temperature by month and hour of day (which hours/months
#       are hottest).
# ------------------------------------------------------------
import numpy as np
np.random.seed(15)
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
hours = list(range(0, 24))
# rows = months, cols = hours
base = np.array([15,18,22,26,30,34,36,33,29,25,20,16]).reshape(-1, 1)
daily = np.sin((np.array(hours) - 6) / 24 * 2 * np.pi) * 6
temp_grid = base + daily + np.random.randn(12, 24) * 1.5
plt.figure(figsize=(12, 6))

# Create heatmap using Seaborn
sns.heatmap(
    temp_grid,
    xticklabels=hours,
    yticklabels=months,
    cmap='YlOrRd',  # Warm colormap (Yellow -> Orange -> Red)
    cbar_kws={'label': 'Temperature (°C)'},
)

# Formatting title & labels
plt.title(
    'Temperature Distribution by Month and Hour of Day',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('Hour of Day (0–23)', fontsize=12)
plt.ylabel('Month', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXERCISE 8
# Goal: Explore how strongly 6 health metrics relate to each other.
# ------------------------------------------------------------
import numpy as np
import pandas as pd
np.random.seed(14)
n = 200
weight = np.random.normal(70, 12, n)
height = np.random.normal(170, 10, n)
bmi = weight / ((height/100) ** 2)
bp = 90 + 0.5 * bmi + np.random.randn(n) * 8
sugar = 80 + 0.3 * bmi + np.random.randn(n) * 10
cholesterol = 150 + 0.6 * bmi + np.random.randn(n) * 15
df = pd.DataFrame({
    'weight': weight, 'height': height, 'bmi': bmi,
    'blood_pressure': bp, 'sugar': sugar, 'cholesterol': cholesterol
})
```

```python
corr_matrix = df.corr()
```

```python
sns.heatmap(
    corr_matrix,
    annot=True,  # Display correlation values inside cells
    fmt='.2f',  # Format floating-point numbers to 2 decimal places
    cmap='coolwarm',  # Diverging color palette (Blue = negative, Red = positive)
    vmin=-1,
    vmax=1,  # Standardize colorbar range between -1 and +1
    linewidths=0.5,
    cbar_kws={'label': 'Pearson Correlation Coefficient'},
)

plt.title(
    'Correlation Heatmap of Health Metrics',
    fontsize=14,
    fontweight='bold',
    pad=15,
)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXERCISE 1
# Show quarterly sales split across 3 product lines, stacked per quarter.
# ------------------------------------------------------------
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
electronics = np.array([45, 52, 48, 60])
clothing    = np.array([30, 35, 40, 38])
groceries   = np.array([25, 28, 33, 42])

fig, ax = plt.subplots(1,1,figsize=(10,6))
electronics_bar = ax.bar(x=quarters,height=electronics,edgecolor='black',alpha=0.6,color='#4e79a7',label='electronics')
clothing_bar = ax.bar(x=quarters,height=clothing,bottom=electronics,edgecolor='black',alpha=0.6,color='#f28e2b',label='clothing')
groceries_bar = ax.bar(x=quarters,height=groceries,bottom=electronics+clothing,edgecolor='black',alpha=0.6,color='#e15759',label='groceries')
plt.bar_label(electronics_bar,label_type='center', color='black', weight='bold', fontsize=9)
plt.bar_label(clothing_bar,label_type='center', color='black', weight='bold', fontsize=9)
plt.bar_label(groceries_bar,label_type='center', color='black', weight='bold', fontsize=9)
totals = electronics + clothing + groceries
for idx, total in enumerate(totals):
    plt.text(idx, total + 1.5, str(total), ha='center', weight='bold', fontsize=10)
plt.legend(loc='upper left', ncol=3)
plt.show()
```

```python
# ============================================================
# EXERCISE 5
# Show each region's total revenue split into 3 quarters, as proportions
# (100% stacked) so shares are comparable across regions.
# ------------------------------------------------------------
regions = ['North', 'South', 'East', 'West']
q1 = np.array([120, 90, 150, 80])
q2 = np.array([140, 110, 130, 95])
q3 = np.array([160, 100, 170, 120])

totals = q1 + q2 + q3
q1_pct = (q1 / totals) * 100
q2_pct = (q2 / totals) * 100
q3_pct = (q3 / totals) * 100

# Create plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plot bars stacked on top of each other
ax.bar(regions, q1_pct, label='Q1', color='#4c72b0')
ax.bar(regions, q2_pct, bottom=q1_pct, label='Q2', color='#55a868')
ax.bar(regions, q3_pct, bottom=q1_pct + q2_pct, label='Q3', color='#c44e52')

# Labels and Styling
plt.title("Quarterly Revenue Distribution by Region (100% Stacked)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Region", fontsize=12)
plt.ylabel("Percentage of Total Revenue (%)", fontsize=12)
plt.ylim(0, 100)

# Add Legend and Grid
plt.legend(title="Quarter", bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
