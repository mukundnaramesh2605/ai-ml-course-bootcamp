```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ============================================================
# EXERCISE 1
# Goal: Show how monthly revenue changed over the year.
# ------------------------------------------------------------
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
revenue = [120, 135, 128, 160, 175, 190, 185, 205, 220, 210, 240, 265]
plt.plot(months,revenue,marker='o',color='blue')
plt.title("Yearly Sales revenue")
plt.xlabel("Month")
plt.ylabel("Total revenue ($)")
plt.grid(True)
plt.show()
sns.lineplot(x=months,y=revenue,errorbar=None, marker='o')
plt.grid(True)
plt.title("Yearly Sales revenue using Seaborn")
plt.xlabel("Month")
plt.ylabel("Total revenue ($)")
plt.show()
```

```python
# ============================================================
# EXERCISE 2
# Goal: Compare market share across 5 competitors.
# ------------------------------------------------------------
companies = ['Alpha', 'Beta', 'Gamma', 'Delta', 'Epsilon']
market_share = [34, 22, 18, 15, 11]
colors = ['#4e79a7', '#f28e2b', '#e15759', '#76b7b2', '#59a14f']
bar = plt.bar(companies,market_share,color=colors)
plt.title('Market Share')
plt.xlabel('Companies')
plt.ylabel('Market Share')
plt.ylim(plt.ylim(0, max(market_share) * 1.12))
plt.bar_label(bar,fmt='%d%%')
plt.show()

ax = sns.barplot(x=companies,y=market_share,palette="viridis")
plt.title('Market Share')
plt.xlabel('Companies')
plt.ylabel('Market Share')
for container in ax.containers:
    ax.bar_label(
        container, 
        fmt='%d%%',               # Formats the number as an integer percentage (e.g., 34%)
        padding=4,                # Sets distance between top of bar and text
        fontsize=11, 
        fontweight='bold',
        color='black'
    )
plt.ylim(plt.ylim(0, max(market_share) * 1.12))
plt.show()
```

```python
# ============================================================
# EXERCISE 3
# Goal: Understand the distribution of customer ages.
# ------------------------------------------------------------
import numpy as np
np.random.seed(10)
ages = np.concatenate([
    np.random.normal(28, 5, 400),
    np.random.normal(45, 8, 300),
    np.random.normal(60, 6, 150)
])
mean_age = np.mean(ages)
median_age = np.median(ages)
fig, ax = plt.subplots(figsize=(10, 6))
counts, bins, patches = ax.hist(
    ages, 
    bins=20, 
    color='#1f77b4', 
    edgecolor='white', 
    alpha=0.85
)
ax.bar_label(
    patches, 
    fmt='%d',                  # Formats the bin count as an integer
    padding=3,                 # Places text just above the bar
    fontsize=9, 
    color='#333333'
)
ax.axvline(mean_age, color='red', linestyle='--', linewidth=1.5, label=f'Mean: {mean_age:.1f}')
ax.text(
    mean_age + 1,              # X position (slightly to the right of the line)
    max(counts) * 0.9,         # Y position (near the top of the chart)
    f'Average Age:\n{mean_age:.1f} years', 
    color='red', 
    fontweight='bold'
)

# 6. Clean up and format the chart
ax.set_title('Customer Age Distribution', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Age (Years)', fontsize=11)
ax.set_ylabel('Customer Count (Frequency)', fontsize=11)
ax.set_xlim(10, 80)
ax.set_ylim(0, max(counts) + 15) # Add vertical padding for the bar labels

# Remove top and right borders for a cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXERCISE 4
# Goal: See if there's a relationship between hours studied and exam score.
# ------------------------------------------------------------
import numpy as np
np.random.seed(11)
hours_studied = np.random.uniform(0, 12, 80)
exam_score = np.clip(30 + 5.5 * hours_studied + np.random.randn(80) * 8, 0, 100)
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(hours_studied, exam_score, color='#69b3a2', edgecolors='black', alpha=0.8, s=50, label='Student Data')
ax.set_title('Relationship Between Study Hours and Exam Scores', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Hours Studied (Hours)', fontsize=11)
ax.set_ylabel('Exam Score (%)', fontsize=11)
ax.set_xlim(-0.5, 13)
ax.set_ylim(0, 105)

# Remove top and right chart bo
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate the exercise data
np.random.seed(11)
hours_studied = np.random.uniform(0, 12, 80)
exam_score = np.clip(30 + 5.5 * hours_studied + np.random.randn(80) * 8, 0, 100)

# 2. Setup the figure
fig, ax = plt.subplots(figsize=(10, 6))

# 3. Create the scatter plot
ax.scatter(hours_studied, exam_score, color='#69b3a2', edgecolors='black', alpha=0.8, s=50, label='Student Data')

# 4. Calculate and plot a linear trend line (Best Fit Line)
m, c = np.polyfit(hours_studied, exam_score, 1) # Linear regression formula (y = mx + c)
x_trend = np.linspace(0, 12, 100)
y_trend = m * x_trend + c
ax.plot(x_trend, y_trend, color='#e67e22', linestyle='-', linewidth=2, label=f'Trend (Slope: {m:.2f} pts/hr)')

# 5. Annotate specific data points of interest
# Find the student who studied the most
idx_max_hours = np.argmax(hours_studied)
ax.annotate(
    'Max Effort',
    xy=(hours_studied[idx_max_hours], exam_score[idx_max_hours]),
    xytext=(hours_studied[idx_max_hours] - 2.5, exam_score[idx_max_hours] - 5),
    arrowprops=dict(arrowstyle="->", color="black", lw=1.2),
    fontsize=10, fontweight='bold'
)

# Find an outlier (e.g., high study hours but lower score than expected)
ax.annotate(
    'High Study / Lower Score Outlier',
    xy=(9.7, 72.5), # Points near an explicit visual outlier in this seed
    xytext=(5, 85),
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=-0.1", color="red", lw=1.2),
    fontsize=10, color='red'
)

# 6. Clean up layout and style definitions
ax.set_title('Relationship Between Study Hours and Exam Scores', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Hours Studied (Hours)', fontsize=11)
ax.set_ylabel('Exam Score (%)', fontsize=11)
ax.set_xlim(-0.5, 13)
ax.set_ylim(0, 105)

# Remove top and right chart borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower right')

plt.tight_layout()
plt.show()
```
