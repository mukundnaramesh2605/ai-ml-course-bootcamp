```python
import numpy as np
x = np.linspace(0,2,50)
plt.plot(x,x,label="Linear")
plt.plot(x,x**2,label="Quadratic")
plt.plot(x,x**3,label="Cubic")
plt.title("Simple Plot")
plt.xlabel("X label")
plt.ylabel("Y label")
plt.legend()
plt.show()
```

```python
x = [5,2,9,4,7]
y = [10,5,8,4,2]
z = [15,8,7,5,6]
plt.plot(x,y,marker="v",color='red',label = "XY")
plt.plot(x,z,marker="o",color='blue',label = "XZ")
plt.xlabel("X Value")
plt.ylabel("y & Z values")
plt.legend()
plt.show()
```
