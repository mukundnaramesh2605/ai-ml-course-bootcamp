```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
d1 = np.random.normal(0,1,1000)
plt.hist(d1,bins=20,color="#101249",alpha=0.7)
plt.title("Histogram")
plt.xlabel('Values')
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
# KDE plot - kernel desnsity plot
sns.kdeplot(d1,color="#101249")
plt.title('KDE plot for dummy values')
plt.xlabel('Values')
plt.grid()
plt.show()
```

```python
# KDE plot with histogram
d1 = np.random.normal(0,1,1000)
plt.hist(d1,bins=20,color="#101249",alpha=0.7,density=True)
sns.kdeplot(d1,color="#003366")
plt.title("Histogram")
plt.xlabel('Values')
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
# Box Plot
data = [np.random.normal(0,1,1000) for i in range(5)]

data[0] += 5
data[3] -= 6

plt.boxplot(data,tick_labels = ["G1","G2","G3","G4","G5"])
plt.title("Box Plot of the Dummy Data")
plt.ylabel("Values")
plt.grid(True)
plt.show()
```

```python
import kagglehub
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("jayapriyadatascience/wineqt-dataset")

print("Path to dataset files:", path)
```

```python
import os
file_path = os.path.join(path,"WineQT.csv")
print(file_path)
```

```python
wine_df = pd.read_csv(file_path)
wine_df.info()
```

```python
wine_df.drop('Id',axis=1,inplace=True)
wine_df.head()
```

```python
plt.figure(figsize=(12,10))

#wine_df = wine_df.drop(columns="Id")

sns.violinplot(data = wine_df)
plt.show()
```

```python
#pie chart
sizes = [40, 30, 20, 10]
labels = ['Apples', 'Bananas', 'Cherries', 'Dates']
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
explode = (0.1, 0, 0, 0)  # "explode" the 1st slice (Apples)

# Create the pie chart
plt.pie(sizes, 
        explode=explode, 
        labels=labels, 
        colors=colors, 
        autopct='%1.1f%%', 
        shadow=True, 
        startangle=140)

# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')  

# Add a title and display
plt.title("Fruit Distribution")
plt.show()
```
