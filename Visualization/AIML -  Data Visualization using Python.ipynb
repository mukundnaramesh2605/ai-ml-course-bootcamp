```python
import matplotlib.pyplot as plt
import numpy as np

# Your Data
item_label = ["A", "B", "C", "D", "E"]
item_sold1 = np.array([10, 20, 30, 40, 50])
item_sold2 = np.array([5, 60, 50, 40, 70])
item_sold3 = np.array([50, 10, 30, 40, 15])
item_sold4 = np.array([50, 100, 24, 36, 1])

# Calculate bottom baselines for stacking
stack1 = item_sold1
stack2 = stack1 + item_sold2
stack3 = stack2 + item_sold3
totals = stack3 + item_sold4  # Total height of each bar

fig, ax = plt.subplots(figsize=(10, 6))

# Plot each layer
b1 = ax.bar(item_label, item_sold1, label="Sold 1")
b2 = ax.bar(item_label, item_sold2, bottom=stack1, label="Sold 2")
b3 = ax.bar(item_label, item_sold3, bottom=stack2, label="Sold 3")
b4 = ax.bar(item_label, item_sold4, bottom=stack3, label="Sold 4")

# 1. Annotate individual segment values inside the bars
for bar_container in [b1, b2, b3, b4]:
    ax.bar_label(
        bar_container, 
        label_type="center", 
        color="white", 
        fontweight="bold", 
        fontsize=9
    )

# 2. Annotate the final cumulative total at the top edge of each bar
ax.bar_label(
    b4, 
    labels=[f"Total: {t}" for t in totals], 
    label_type="edge", 
    padding=5, 
    fontweight="bold", 
    color="black"
)

# Customise layout
ax.set_ylabel("Units Sold")
ax.set_title("Stacked Item Sales Breakdown")
ax.legend(loc="upper left")

# Expand y-axis limit slightly so top text does not get cut off
ax.set_ylim(0, max(totals) * 1.1)

plt.tight_layout()
plt.show()
```
