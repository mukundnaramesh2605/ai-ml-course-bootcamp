```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Create a line plot of y = x² for x from -10 to 10. Add axis labels and a title.
x = np.linspace(-10, 10, 100)
y=x**2
plt.plot(x,y,color='blue', linewidth=2, label='$y = x^2$')
plt.xlabel("x value")
plt.ylabel("y value ($x^2$)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.title("Line Plot of Quadratic Function $y = x^2$")
plt.show()
```

```python
#Plot two lines (sin(x) and cos(x)) on the same axes with a legend and different colors.
x = np.linspace(0, 4 * np.pi, 200)
y_sin = np.sin(x)
y_cos = np.cos(x)
plt.plot(x, y_sin, color='royalblue', linewidth=2, label='$\sin(x)$')
plt.plot(x, y_cos, color='darkorange', linewidth=2, linestyle='--', label='$\cos(x)$')
plt.title("Trigonometric Functions: $\sin(x)$ vs $\cos(x)$")
plt.xlabel("Angle (radians)")
plt.ylabel("Amplitude")
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
# ============================================================
# EXERCISE 3: Bar chart with value labels
# ------------------------------------------------------------
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 12, 67, 34]
# TODO: Bar chart with custom colors and value labels on top of each bar.
colors = ['#4e79a7', '#f28e2b', '#e15759', '#76b7b2', '#59a14f']
bars = plt.bar(x=categories,height=values,color=colors)
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title("Categorical Distribution Analysis")
plt.bar_label(bars, padding=3, fontsize=10, weight='bold')
plt.ylim(0, max(values) * 1.12)
plt.show()
```

```python
# ============================================================
# EXERCISE 4: Horizontal bar chart, sorted
# ------------------------------------------------------------
products = ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard']
sales = [340, 890, 220, 410, 150]
# TODO: Horizontal bar chart sorted from highest to lowest.
colors = ['#4e79a7', '#f28e2b', '#e15759', '#76b7b2', '#59a14f']
sorted_data = sorted(zip(sales, categories))
sorted_sales, sorted_categories = zip(*sorted_data)
bars = plt.barh(sorted_categories, sorted_sales, color=colors,edgecolor='black')
plt.ylabel('products')
plt.title("Product Sales Analysis")
plt.bar_label(bars, padding=3, fontsize=10, weight='bold')
plt.xlim(0, max(sales) * 1.12)
plt.show()
```

```python
# ============================================================
# EXERCISE 5: Histogram with mean line
# ------------------------------------------------------------
import numpy as np
np.random.seed(0)
data = np.random.normal(50, 15, 1000)
# TODO: Histogram with 30 bins and a vertical line at the mean.
data_mean = np.mean(data)
plt.hist(data,bins=30)
plt.axvline(data_mean, color='red', linestyle='--', linewidth=2, 
            label=f'Mean: {data_mean:.2f}')
plt.title("Data Distribution Summary with Mean Line Indicator")
plt.xlabel("Measured Values")
plt.ylabel("Frequency Counts")
plt.legend()
plt.show()
```

```python
# ============================================================
# EXERCISE 6: Scatter colored by category
# ------------------------------------------------------------
import numpy as np
np.random.seed(1)
x = np.random.rand(100)
y = np.random.rand(100)
category = np.random.choice(['red', 'green', 'blue'], 100)
# TODO: Scatter plot colored by the categorical variable.
plt.scatter(x,y,c=category,edgecolors='black')
plt.title("Categorical Distribution Analysis (Scatter)", fontsize=13, pad=12)
plt.xlabel("X Coordinate Dimension")
plt.ylabel("Y Coordinate Dimension")
```

```python
# ============================================================
# EXERCISE 7: Bubble chart (size = 4th variable)
# ------------------------------------------------------------
import numpy as np
np.random.seed(2)
x = np.random.rand(50)
y = np.random.rand(50)
size = np.random.rand(50) * 1000
# TODO: Scatter where marker size encodes 'size'.
plt.scatter(x, y, s=size, color='teal', edgecolor='black')
plt.title("4D Bubble Chart Analysis (Marker Sizing Implementation)", fontsize=13, pad=12)
plt.xlabel("X Metric Dimension")
plt.ylabel("Y Metric Dimension")
plt.grid(True, linestyle='-', alpha=0.5)
plt.show()
```

```python
# ============================================================
# EXERCISE 8: 2x2 subplot grid
# ------------------------------------------------------------
import numpy as np
np.random.seed(3)
x = np.linspace(0, 10, 50)
line_y = np.sin(x)
bar_cats = ['P', 'Q', 'R', 'S']
bar_vals = [10, 25, 15, 30]
scatter_x = np.random.rand(50)
scatter_y = np.random.rand(50)
hist_data = np.random.normal(0, 1, 500)
# TODO: 2x2 grid: line, bar, scatter, histogram.
fig,axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
axes[0, 0].plot(x, line_y, color='royalblue', linewidth=2)
axes[0, 0].set_title("Sine Wave Evolution")
axes[0, 0].grid(True, linestyle=':', alpha=0.6)
axes[0, 1].bar(bar_cats, bar_vals, color='darkorange', edgecolor='black', alpha=0.8)
axes[0, 1].set_title("Categorical Metrics")
axes[1, 0].scatter(scatter_x, scatter_y, color='forestgreen', edgecolor='black', alpha=0.7)
axes[1, 0].set_title("Random Dispersion Pattern")
axes[1, 0].grid(True, linestyle=':', alpha=0.6)
axes[1, 1].hist(hist_data, bins=25, color='crimson', edgecolor='black', alpha=0.7)
axes[1, 1].set_title("Gaussian Probability Density")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# EXERCISE 9: Pie chart with exploded slice
# ------------------------------------------------------------
labels = ['Rent', 'Food', 'Transport', 'Savings', 'Other']
amounts = [1200, 600, 300, 500, 200]
# TODO: Pie chart with percentage labels and one slice exploded.
explode = (0.1, 0, 0, 0, 0)
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']
plt.pie(
    amounts, 
    explode=explode, 
    labels=labels, 
    colors=colors,
    autopct='%1.1f%%')
plt.title("Monthly Expense Allocation Budget", fontsize=14, pad=20)
plt.axis('equal') 
plt.show()
```

```python
# ============================================================
# EXERCISE 10: Box plot across 3 groups
# ------------------------------------------------------------
import numpy as np
np.random.seed(4)
group_a = np.random.normal(60, 10, 100)
group_b = np.random.normal(70, 15, 100)
group_c = np.random.normal(65, 8, 100)
# TODO: Box plot comparing the three groups.
data_to_plot = [group_a, group_b, group_c]
labels = ['Group A', 'Group B', 'Group C']
plt.boxplot(data_to_plot,label=labels,patch_artist=True,
            boxprops=dict(facecolor='#e2eff1', color='darkcyan', linewidth=1.5),
           medianprops=dict(color='crimson', linewidth=2),
           flierprops=dict(marker='o', markerfacecolor='orange', alpha=0.6, markersize=6))
plt.title("Statistical Distribution Comparison Across Groups", fontsize=13, pad=15)
plt.xlabel("Experimental Groups", fontsize=11)
plt.ylabel("Measured Response Range", fontsize=11)
plt.show()
```

```python
# ============================================================
# EXERCISE 11: Violin plot (same data as Ex 10)
# ------------------------------------------------------------
import numpy as np
np.random.seed(4)
group_a = np.random.normal(60, 10, 100)
group_b = np.random.normal(70, 15, 100)
group_c = np.random.normal(65, 8, 100)
# TODO: Violin plot for the three groups. Compare with the box plot.
data_to_plot = [group_a, group_b, group_c]
positions = [1, 2, 3]
labels = ['Group A', 'Group B', 'Group C']
violins = plt.violinplot(
    data_to_plot, 
    positions=positions,
    showmeans=False, 
    showmedians=True, 
    showextrema=True
)

for body in violins['bodies']:
    body.set_facecolor('indigo')
    body.set_edgecolor('black')
    body.set_alpha(0.4)
violins['cmedians'].set_color('crimson')
violins['cmedians'].set_linewidth(2)
violins['cmins'].set_color('dimgray')
violins['cmaxes'].set_color('dimgray')
plt.xticks(positions, labels)
plt.title("Violin Plot Distribution Density Across Groups", fontsize=13, pad=15)
plt.xlabel("Experimental Groups", fontsize=11)
plt.ylabel("Measured Response Range", fontsize=11)
plt.show()
```

```python
# ============================================================
# EXERCISE 12: Correlation heatmap
# ------------------------------------------------------------
import numpy as np
import pandas as pd
np.random.seed(5)
df = pd.DataFrame(np.random.rand(100, 5), columns=['V1', 'V2', 'V3', 'V4', 'V5'])
df['V2'] = df['V1'] * 0.8 + np.random.rand(100) * 0.2   # inject correlation
df['V4'] = df['V3'] * -0.6 + np.random.rand(100) * 0.4
corr = df.corr()
# TODO: Annotated heatmap of the correlation matrix.
sns.heatmap(corr, annot=True)
plt.show()
```

```python
# ============================================================
# EXERCISE 13: Stacked bar chart
# ------------------------------------------------------------
import numpy as np
categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_x = [20, 35, 30, 35]
product_y = [25, 32, 34, 20]
product_z = [15, 18, 22, 28]
# TODO: Stacked bar chart of the three products across quarters.
x = np.array(product_x)
y = np.array(product_y)
z = np.array(product_z)
plt.figure(figsize=(12,10))
bar1 = plt.bar(categories, x, color='#4e79a7', edgecolor='black', alpha=0.9, label='Product X')
bar2 = plt.bar(categories, y, bottom=x, color='#f28e2b', edgecolor='black', alpha=0.9, label='Product Y')
bar3 = plt.bar(categories, z, bottom=x+y, color='#e15759', edgecolor='black', alpha=0.9, label='Product Z')
plt.bar_label(bar1, label_type='center', color='white', weight='bold', fontsize=9)
plt.bar_label(bar2, label_type='center', color='white', weight='bold', fontsize=9)
plt.bar_label(bar3, label_type='center', color='white', weight='bold', fontsize=9)
totals = x + y + z
for idx, total in enumerate(totals):
    plt.text(idx, total + 1.5, str(total), ha='center', weight='bold', fontsize=10)
plt.title("Quarterly Sales Share by Product Category", fontsize=13, pad=18)
plt.xlabel("Fiscal Quarters", fontsize=11)
plt.ylabel("Units Transacted", fontsize=11)
plt.ylim(0, max(totals) * 1.12)
plt.legend(loc='upper left', ncol=3)
plt.show()
```
