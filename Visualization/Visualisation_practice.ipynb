```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Create a line plot of y = x² for x from -10 to 10. Add axis labels and a title.
x = np.linspace(-10, 10, 100)
y=x**2
plt.plot(x,y,color='blue', linewidth=2, label='$y = x^2$')
plt.xlabel("x value")
plt.ylabel("y value ($x^2$)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.title("Line Plot of Quadratic Function $y = x^2$")
plt.show()
```
