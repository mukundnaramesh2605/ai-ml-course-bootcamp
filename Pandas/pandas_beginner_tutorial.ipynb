## Merging and Joining

```python
df = pd.read_csv(r"datasets/Pandas - Q1 Sales.csv")
df.head()
```

```python
df['Date of Purchase'] = pd.to_datetime(df['Date of Purchase'])
```

```python
df.head()
```

```python
df2 = pd.DataFrame()
df2["Employee Name"] = df['Sales Person'].unique()
exp = [1,2,4]
random.shuffle(exp)
df2["Experience"] = exp
df2["Total Eperience"] = df2["Experience"] + random.randint(1,8)
df2.head()
```

```python
df.merge(df2,left_on=["Sales Person"],right_on=["Employee Name"])
```

```python
df.join(df2)
```

```python
df.set_index("Sales Person",inplace=True)
df2.set_index("Employee Name",inplace=True)
```

```python
df.join(df2)
```
