```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'/Users/mukundanramesh/projects/ai_ml_course_bootcamp_practice/ai-ml-course-bootcamp/Pandas/datasets/cumulative.csv',index_col=0)
df.head(5)
```

```python
#How many rows and columns does the dataset have, and what are the dtypes?
shape = df.shape
print(f"THe dataset has {shape[0]} rows and {shape[1]} columns")
print(df.describe())
```

```python
#Which columns have missing values, and what percentage is missing in each?
missing = (df.isna().mean() * 100).round(2)
missing[missing > 0].sort_values(ascending=False)
```

```python
#Are there duplicate rows, or duplicate kepid/kepoi_name entries?
df.duplicated(subset=['kepid','kepoi_name'])

# Fully duplicate rows (every column identical)
df.duplicated().sum()

# Duplicate kepoi_name alone (should really be unique — it's the KOI identifier)
df.duplicated(subset=['kepoi_name']).sum()

# Duplicate kepid alone (NOT expected to be unique — one star can host multiple planet candidates)
df.duplicated(subset=['kepid']).sum()
```

```python
#Which columns are essentially "ID" columns (won't be useful as features)?
# For each column, what fraction of values are unique?
df.nunique() / len(df)

unique = pd.DataFrame({
    'dtype': df.dtypes,
    'n_unique': df.nunique(),
    'pct_unique': (df.nunique() / len(df) * 100).round(1)
}).sort_values('pct_unique', ascending=False)
unique.iloc[0]
```

```python
#What are the unique values of koi_disposition? What's the distribution (counts and percentages)?
unique_koi_disposition = df.koi_disposition.nunique()
pct_unique_koi_disposition = (df.koi_disposition.nunique() / len(df) * 100)
unique_koi_disposition,pct_unique_koi_disposition
```

```python
#How does koi_disposition relate to koi_pdisposition? Do they ever disagree?
pd.crosstab(df['koi_disposition'], df['koi_pdisposition'])
```

```python
#What does koi_score look like for each disposition class? (try a groupby + describe, or boxplot)
df.groupby("koi_disposition")["koi_score"].describe()
```
