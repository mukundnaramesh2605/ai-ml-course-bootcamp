```python
import pandas as pd

# ==========================================================
# Problem 1: Department Salary Report
#
# Business Requirement:
#
# Find the average salary of employees
# in each department.
#
# Expected Output:
# Department | Average Salary
# ==========================================================

df = pd.DataFrame({
    "Employee":["Alice","Bob","Charlie","David","Eva","Frank"],
    "Department":["HR","IT","IT","HR","Finance","Finance"],
    "Salary":[50000,70000,80000,55000,90000,85000]
})

df.groupby('Department').agg(Average_Salary=('Salary', 'mean'))
```
