```python
import pandas as pd

orders = pd.DataFrame({
    "customer": ["A","A","B","B","C","A","C","D"],
    "product": ["Laptop","Mouse","Laptop","Keyboard","Mouse","Keyboard","Laptop","Monitor"],
    "amount": [800,20,900,50,25,70,850,300]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Find the total amount spent by each customer.
display(orders.groupby('customer')['amount'].sum())
# 2. Find the average purchase amount.
display(orders.groupby('customer')['amount'].mean())
# 3. Find the highest purchase amount.
display(orders.groupby('customer')['amount'].max())
# 4. Find the number of purchases.
display(orders.groupby('customer')['amount'].count())
# 5. Use agg() to calculate all the above in one statement.
display(orders.groupby('customer').agg(total_purchase_amount=('amount','sum'),
                                                 avg_purchase_amount=('amount','mean'),
                                                highest_purchase_amount=('amount','max'),
                                                 number_of_purchases=('amount','count')))
```

```python
import pandas as pd

sales = pd.DataFrame({
    "employee":["John","John","Alice","Alice","Bob","Bob"],
    "sales":[100,150,200,250,300,100]
})

# ==============================
# QUESTIONS
# ==============================

# Without reducing the number of rows:

# 1. Create a column containing total sales for each employee.
sales['total_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('sum')
# 2. Create a column containing average sales.
sales['avg_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('mean')
# 3. Create a column containing maximum sales.
sales['max_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('max')
# 4. Create a column containing transaction count.
sales['no_of_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('count')
# 5. Sort by total sales.
display(sales.sort_values(by="total_sales_of_each_employee",ascending=False))
```

```python
import pandas as pd

employees = pd.DataFrame({
    "emp_id":[1,2,3,4],
    "name":["Alice","Bob","Charlie","David"],
    "dept_id":[10,20,30,20]
})

departments = pd.DataFrame({
    "dept_id":[10,20,30],
    "department":["IT","HR","Finance"]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Perform an inner merge.
inner_merge = pd.merge(employees,departments,how='inner',on='dept_id')
display(inner_merge)
# 2. Perform a left merge.
left_merge = pd.merge(employees,departments,how='left',on='dept_id')
display(left_merge)
# 3. Perform an outer merge.
outer_merge = pd.merge(employees,departments,how='outer',on='dept_id')
display(outer_merge)
# 4. Display only employee name and department.
outer_merge = pd.merge(employees,departments,how='outer',on='dept_id')
display(outer_merge[['name','department']])
# 5. Rename the merged department column to dept_name.
outer_merge = pd.merge(employees,departments,how='outer',on='dept_id').rename(columns={'department':'dept_name'})
display(outer_merge)
```

```python
import pandas as pd

customers = pd.DataFrame({
    "name":[" alice ","BOB","Charlie","david","Ankit"],
    "city":["Chennai","Delhi","Mumbai","Delhi","Chennai"]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Remove leading/trailing spaces.
customers['name'] = customers['name'].str.strip()
display(customers)
# 2. Convert names to title case.
customers['name'] = customers['name'].str.capitalize()
display(customers)
# 3. Find names containing the letter 'a'.
display(customers[customers['name'].str.contains('a')])
# 4. Find customers from Chennai.
display(customers[customers['city']=='Chennai'])
# 5. Sort names alphabetically.
display(customers.sort_values(by='name'))
```

```python
import pandas as pd

sales = pd.DataFrame({
    "city":["Chennai","Delhi","Delhi","Mumbai","Chennai","Mumbai"],
    "product":["Laptop","Laptop","Phone","Phone","Phone","Laptop"],
    "sales":[100,150,120,180,160,200]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Create a pivot table.
# 2. Show cities as rows.
# 3. Show products as columns.
# 4. Fill missing values with 0.
# 5. Add row and column totals.
pt = pd.pivot_table(sales,index='city',columns='product',values='sales',aggfunc='sum',margins=True,margins_name='Total')
pt
```

```python
import pandas as pd

students = pd.DataFrame({
    "id":[1,2,2,3,4,4,4],
    "name":["Alice","Bob","Bob","Charlie","David","David","David"],
    "marks":[90,80,80,95,85,85,85]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Find duplicate rows.
display(students[students.duplicated])
# 2. Count duplicate rows.
display(f"The number of duplicates is {students.duplicated().sum()}")
# 3. Remove duplicate rows.
students.drop_duplicates(inplace=True)
display(students)
# 4. Remove duplicates based only on "id".
students.drop_duplicates(subset=['id'],inplace=True)
display(students)
# 5. Keep only the last occurrence.
students.drop_duplicates(keep='last',inplace=True)
display(students)
```

```python
import pandas as pd

employees = pd.DataFrame({
    "department":["IT","IT","HR","HR","Finance","Finance"],
    "employee":["A","B","C","D","E","F"],
    "salary":[80000,90000,50000,60000,70000,75000]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Rank salaries within each department.
employees['rank'] = employees.groupby('department')['salary'].rank(method='dense')
display(employees)
# 2. Find the highest-paid employee in each department.
employees['rank'] = employees.groupby('department')['salary'].rank(method='dense')
display(employees)
# 3. Find the second highest-paid employee in each department.
display(employees[employees['rank'] == 2])
# 4. Sort by department and rank.
display(employees.sort_values(by=['department','rank']))
# 5. Display only rank 1 employees.
display(employees[employees['rank'] == 1])
```

```python
import pandas as pd

transactions = pd.DataFrame({
    "customer":["A","A","A","B","B","C","C","D"],
    "product":["Laptop","Mouse","Keyboard","Laptop","Mouse","Laptop","Monitor","Phone"],
    "amount":[800,20,50,900,25,850,300,700]
})

# ==============================
# QUESTIONS
# ==============================

# Without using loops:

# 1. Calculate total spending per customer.
display(transactions.groupby('customer')['amount'].sum())
# 2. Calculate average spending per customer.
display(transactions.groupby('customer')['amount'].mean())
# 3. Count purchases per customer.
display(transactions.groupby('customer')['amount'].count())
# 4. Create a column showing each customer's total spending.
transactions['Total_spends'] = transactions.groupby('customer')['amount'].transform('sum')
display(transactions)
# 5. Create a column showing the percentage contribution of each purchase
#    to that customer's total spending.
transactions['Pct_Of_Customer_Total'] = ((transactions['amount'] / transactions['Total_spends']) * 100).round(2)
display(transactions)
# 6. Sort customers by total spending.
transactions.sort_values(by='Total_spends',ascending=False)
```
