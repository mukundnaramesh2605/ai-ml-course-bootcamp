```python
import pandas as pd

orders = pd.DataFrame({
    "customer": ["A","A","B","B","C","A","C","D"],
    "product": ["Laptop","Mouse","Laptop","Keyboard","Mouse","Keyboard","Laptop","Monitor"],
    "amount": [800,20,900,50,25,70,850,300]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Find the total amount spent by each customer.
display(orders.groupby('customer')['amount'].sum())
# 2. Find the average purchase amount.
display(orders.groupby('customer')['amount'].mean())
# 3. Find the highest purchase amount.
display(orders.groupby('customer')['amount'].max())
# 4. Find the number of purchases.
display(orders.groupby('customer')['amount'].count())
# 5. Use agg() to calculate all the above in one statement.
display(orders.groupby('customer').agg(total_purchase_amount=('amount','sum'),
                                                 avg_purchase_amount=('amount','mean'),
                                                highest_purchase_amount=('amount','max'),
                                                 number_of_purchases=('amount','count')))
```

```python
import pandas as pd

sales = pd.DataFrame({
    "employee":["John","John","Alice","Alice","Bob","Bob"],
    "sales":[100,150,200,250,300,100]
})

# ==============================
# QUESTIONS
# ==============================

# Without reducing the number of rows:

# 1. Create a column containing total sales for each employee.
sales['total_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('sum')
# 2. Create a column containing average sales.
sales['avg_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('mean')
# 3. Create a column containing maximum sales.
sales['max_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('max')
# 4. Create a column containing transaction count.
sales['no_of_sales_of_each_employee'] = sales.groupby('employee')['sales'].transform('count')
# 5. Sort by total sales.
display(sales.sort_values(by="total_sales_of_each_employee",ascending=False))
```
