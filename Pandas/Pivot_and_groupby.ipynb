```python
import pandas as pd

# ==========================================================
# Problem 1: Regional Sales Dashboard
#
# Business Requirement:
#
# The CEO wants a dashboard showing:
#
# • Total Revenue
# • Average Revenue
#
# for every Region and Product combination.
#
# Missing combinations should display 0.
#
# Add Overall totals.
# ==========================================================

df = pd.DataFrame({
    "Region":["North","North","South","South","East","West","West","East"],
    "Product":["Laptop","Phone","Laptop","Tablet","Phone","Laptop","Phone","Tablet"],
    "Revenue":[120,80,140,100,90,150,70,110]
})

pt = pd.pivot_table(df,index='Region',columns='Product',values='Revenue',
                    aggfunc=['sum','mean'],fill_value=0).rename(columns={"sum":"Total Revenue","mean":"Average Revenue"})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 2: University Performance Report
#
# Business Requirement:
#
# Show:
#
# Average Marks
# Highest Marks
#
# for every Department and Semester.
#
# Display Semester as columns.
#
# ==========================================================

df = pd.DataFrame({
    "Department":["CSE","CSE","ECE","ECE","EEE","EEE","CSE","ECE"],
    "Semester":[1,2,1,2,1,2,2,1],
    "Marks":[82,91,76,88,80,94,95,84]
})

pt = pd.pivot_table(df,index='Department',columns='Semester',values='Marks',
                    aggfunc=['mean','max']).rename(columns={'mean':"Average Marks",'max':"Highest Mark"})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 3: Employee Salary Dashboard
#
# Business Requirement:
#
# Show:
#
# Average Salary
# Maximum Salary
#
# grouped by
#
# Department
# Gender
#
# Include Overall totals.
#
# Rename the total row to Company Total.
# ==========================================================

df = pd.DataFrame({
    "Department":["IT","IT","HR","HR","Finance","Finance","IT"],
    "Gender":["M","F","M","F","M","F","F"],
    "Salary":[70000,82000,55000,62000,90000,95000,78000]
})

pt = pd.pivot_table(df,index="Department",columns='Gender',values='Salary',
                    aggfunc=['mean','max'],fill_value=0,
                    margins=True,margins_name="Company Total").rename(columns={"mean":"Average Salary",'max':'Maximum Salary'})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 4: Bank Transactions
#
# Business Requirement:
#
# Create a dashboard showing
#
# Total Transaction Amount
#
# Branch × Month × Transaction Type
#
# Replace missing values with 0.
#
# ==========================================================

df = pd.DataFrame({
    "Branch":["A","A","A","B","B","C","C","C"],
    "Month":["Jan","Feb","Feb","Jan","Feb","Jan","Feb","Mar"],
    "Type":["Deposit","Deposit","Withdrawal","Deposit",
            "Withdrawal","Deposit","Deposit","Withdrawal"],
    "Amount":[5000,7000,3000,8000,2000,9000,6000,4000]
})

pt = pd.pivot_table(df,index=['Branch','Month'],columns='Type',aggfunc='sum',fill_value=0)
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 5: Retail Analytics
#
# Business Requirement:
#
# Build ONE pivot table showing:
#
# Rows:
# Store
#
# Columns:
# Product
# Year
#
# Values:
# Total Sales
#
# Include Overall totals.
#
# Replace missing values with 0.
# ==========================================================

df = pd.DataFrame({
    "Store":["A","A","A","B","B","C","C","A"],
    "Product":["Laptop","Phone","Tablet","Laptop","Phone","Laptop","Tablet","Laptop"],
    "Year":[2023,2023,2024,2024,2023,2023,2024,2024],
    "Sales":[120,80,90,150,100,140,110,170]
})

pt=pd.pivot_table(df,index='Store',columns=['Product','Year'],values ='Sales',aggfunc='sum',fill_value=0).rename(columns=
                                                                                                                 {'Laptop':'Total Laptop Sales',
                                                                                                                  'Phone':'Total Phone Sales',
                                                                                                                 'Tablet':'Total Tablet Sales'})
pt
```
