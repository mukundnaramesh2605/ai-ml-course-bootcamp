```python
import pandas as pd

orders = pd.DataFrame({
    "customer": ["A","A","B","C","B","A","C","D"],
    "amount": [100,200,150,300,250,50,100,500],
    "date": pd.to_datetime([
        "2025-01-01","2025-01-10","2025-01-02",
        "2025-01-05","2025-01-20","2025-02-01",
        "2025-02-05","2025-02-10"
    ])
})

# ==============================
# QUESTIONS
# ==============================
# 1. Find the total spending for each customer.
# 2. Find the average order value for each customer.
# 3. Find the customer who spent the most.
# 4. Find the first purchase date for every customer.
# 5. Sort customers by total spending in descending order.

# 1. Find the total spending for each customer.
total_speding_per_customer = orders.groupby('customer')['amount'].sum()
display(total_speding_per_customer)

# 2. Find the average order value for each customer.
average_speding_per_customer = orders.groupby('customer')['amount'].mean()
display(average_speding_per_customer)

# 3. Find the customer who spent the most.
print(f"The customer {orders.groupby('customer')['amount'].sum().idxmax()} has spent the most with the spend value of {orders.groupby('customer')['amount'].sum().max()}")
```

```python
# 4. Find the first purchase date for every customer.
first_purchases = orders.groupby("customer")["date"].min().reset_index()
display(first_purchases)
# 5. Sort customers by total spending in descending order.
orders.groupby("customer")['amount'].sum().sort_values(ascending=False)
```

```python
import pandas as pd

students = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "age": [20, 22, 19, 21, 23],
    "marks": [85, 70, 95, 60, 88]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Display only the "name" column.
# 2. Display the first 3 rows.
# 3. Display students with marks greater than 80.
# 4. Display students whose age is greater than 20.
# 5. Display only the "name" and "marks" columns.

# 1. Display only the "name" column.
display(students.name)

# 2. Display the first 3 rows.
display(students.head(3))

# 3. Display students with marks greater than 80.
display(students[students.marks > 80])

# 4. Display students whose age is greater than 20.
display(students[students.age > 20])

# 5. Display only the "name" and "marks" columns.
display(students[['name','marks']])
```

```python
import pandas as pd

employees = pd.DataFrame({
    "employee": ["A", "B", "C", "D"],
    "salary": [50000, 70000, 60000, 80000]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Sort salaries in ascending order.
# 2. Sort salaries in descending order.
# 3. Create a new column "bonus" equal to 10% of salary.
# 4. Create a new column "total_salary".
# 5. Rename "employee" to "Employee_Name".

# 1. Sort salaries in ascending order.
display(employees.sort_values(by='salary',ascending=True))

# 2. Sort salaries in descending order.
display(employees.sort_values(by='salary',ascending=False))

# 3. Create a new column "bonus" equal to 10% of salary.
employees['bonus'] = employees['salary'] * 0.10
display(employees)

# 4. Create a new column "total_salary".
employees['total_salary'] = employees['salary'] + employees['bonus']
display(employees)

# 5. Rename "employee" to "Employee_Name".
employees = employees.rename(columns={'employee':'Employee_Name'})
display(employees)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [20, None, 25, None],
    "city": ["NY", None, "LA", "Chicago"]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Count missing values.
# 2. Fill missing ages with the mean.
# 3. Fill missing city values with "Unknown".
# 4. Drop rows containing missing values.
# 5. Replace all missing values in the DataFrame with "N/A".

# 1. Count missing values.
display(df.isna().sum())

# 2. Fill missing ages with the mean.
df['age'].fillna(df['age'].mean(),inplace=True)
display(df)
# 3. Fill missing city values with "Unknown".
df['city'].fillna('Unknown',inplace=True)
display(df)
# 4. Drop rows containing missing values.
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [20, None, 25, None],
    "city": ["NY", None, "LA", "Chicago"]
})
df.dropna(inplace=True)
display(df)
# 5. Replace all missing values in the DataFrame with "N/A".
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [20, None, 25, None],
    "city": ["NY", None, "LA", "Chicago"]
})
df.fillna('N/A',inplace=True)
display(df)
```

```python
import pandas as pd

sales = pd.DataFrame({
    "city": ["Chennai", "Delhi", "Chennai", "Mumbai", "Delhi"],
    "sales": [100, 200, 150, 300, 250]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Find total sales for each city.
# 2. Find average sales for each city.
# 3. Find maximum sales for each city.
# 4. Find minimum sales for each city.
# 5. Count the number of records for each city.

# 1. Find total sales for each city.
display(sales.groupby('city')['sales'].sum())

# 2. Find average sales for each city.
display(sales.groupby('city')['sales'].mean())

# 3. Find maximum sales for each city.
display(sales.groupby('city')['sales'].max())

# 4. Find minimum sales for each city.
display(sales.groupby('city')['sales'].min())

# 4. Find minimum sales for each city.
display(sales.groupby('city')['sales'].count())
```

```python
import pandas as pd

orders = pd.DataFrame({
    "customer": ["A", "A", "B", "B", "C"],
    "amount": [100, 200, 300, 150, 250]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Find total amount spent by each customer.
# 2. Find average amount spent.
# 3. Find maximum purchase.
# 4. Find minimum purchase.
# 5. Use agg() to calculate sum, mean, max and min together.

# 1. Find total amount spent by each customer.
display(orders.groupby('customer').sum())

# 2. Find average amount spent.
display(orders.groupby('customer').mean())

# 3. Find maximum purchase.
display(orders.groupby('customer').max())

# 4. Find minimum purchase.
display(orders.groupby('customer').min())

# 5. Use agg() to calculate sum, mean, max and min together.
display(orders.groupby('customer').agg(total_amount=('amount','sum'),
                                      average_amount=('amount','mean'),
                                      maximum_purchase=('amount','max'),
                                      minimum_purchase=('amount','min')))
```

```python
import pandas as pd

customers = pd.DataFrame({
    "id": [1,2,3],
    "name": ["Alice","Bob","Charlie"]
})

orders = pd.DataFrame({
    "id": [1,2,2],
    "amount": [100,200,300]
})

# ==============================
# QUESTIONS
# ==============================

# 1. Perform an inner merge.
inner_merge = pd.merge(customers,orders,how='inner',on='id')
display(inner_merge)
# 2. Perform a left merge.
left_merge = pd.merge(customers,orders,how='left',on='id')
display(left_merge)
# 3. Perform a right merge.
right_merge = pd.merge(customers,orders,how='right',on='id')
display(right_merge)
# 4. Perform an outer merge.
outer_merge = pd.merge(customers,orders,how='outer',on='id')
display(outer_merge)
# 5. Find customers without orders.
merged = pd.merge(customers, orders, left_on='id', right_on='id', how='left')
no_orders = merged[merged['amount'].isna()][['id', 'name']]
display(no_orders)
```
