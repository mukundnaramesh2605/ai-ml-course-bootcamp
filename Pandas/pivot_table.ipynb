```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Department": ["HR", "IT", "IT", "HR", "Finance", "Finance"],
    "Salary": [50000, 70000, 80000, 55000, 90000, 85000]
})
"""
Task

Create a pivot table showing:

Department	Average Salary"""
pt1 = pd.pivot_table(df,index="Department",values='Salary',aggfunc='mean')
pt1
```

```python
df = pd.DataFrame({
    "Region": ["North","North","South","South","East","East"],
    "Product": ["Laptop","Phone","Laptop","Phone","Laptop","Phone"],
    "Sales": [100,80,120,90,150,110]
})
"""Task

Create a pivot table like

Region	Laptop	Phone
East	?	?
North	?	?
South	?	?

using the sum of sales."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc='sum')
pt
```

```python
#Exercise 3: Average Marks by Class and Subject (Medium)
df = pd.DataFrame({
    "Student":["A","B","C","D","E","F"],
    "Class":["10","10","10","9","9","9"],
    "Subject":["Math","Science","Math","Science","Math","Science"],
    "Marks":[88,75,92,81,95,78]
})
pt = pd.pivot_table(df,index='Class',columns='Subject',values='Marks',aggfunc='mean')
display(pt)
```

```python
#Exercise 4: Monthly Revenue Report (Medium)
df = pd.DataFrame({
    "Month":["Jan","Jan","Jan","Feb","Feb","Mar","Mar"],
    "Category":["Food","Electronics","Food","Food","Electronics","Food","Electronics"],
    "Revenue":[5000,12000,4000,6000,15000,7000,18000]
})
pt = pd.pivot_table(df,index='Month',columns='Category',values='Revenue',aggfunc='sum')
display(pt)
```

```python
#Exercise 5: Company Sales Dashboard (Medium)
df = pd.DataFrame({
    "Region":["North","North","North","South","South","East","East","East"],
    "Salesperson":["John","John","Mike","Mike","Sara","Sara","John","Mike"],
    "Product":["Laptop","Phone","Laptop","Phone","Laptop","Phone","Laptop","Phone"],
    "Sales":[100,150,120,170,200,180,110,140]
})
"""
Tasks
Part A

Create a pivot table showing

| Region | Total Sales |
"""
pt = pd.pivot_table(df,index='Region',values='Sales',aggfunc='sum')
pt = pt.rename(columns={'Sales':'Total Sales'})
pt
```

```python
"""Part B

Create a pivot table

| Region | Laptop | Phone |

using sum."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc='sum')
pt
```

```python
"""Part C

Create a pivot table

| Salesperson | Average Sales |"""
pt = pd.pivot_table(df,index='Salesperson',values='Sales',aggfunc="mean")
pt
```

```python
"""Part D (Challenge)

Create a pivot table

| Region | Laptop Sales | Phone Sales | Grand Total |

using margins."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc="sum",margins=True, 
    margins_name='Grand Total')
pt = pt.rename(columns={
    'Laptop': 'Laptop Sales',
    'Phone': 'Phone Sales'
})
pt.columns.name = None
pt
```
