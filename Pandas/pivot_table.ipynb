```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Department": ["HR", "IT", "IT", "HR", "Finance", "Finance"],
    "Salary": [50000, 70000, 80000, 55000, 90000, 85000]
})
"""
Task

Create a pivot table showing:

Department	Average Salary"""
pt1 = pd.pivot_table(df,index="Department",values='Salary',aggfunc='mean')
pt1
```

```python
df = pd.DataFrame({
    "Region": ["North","North","South","South","East","East"],
    "Product": ["Laptop","Phone","Laptop","Phone","Laptop","Phone"],
    "Sales": [100,80,120,90,150,110]
})
"""Task

Create a pivot table like

Region	Laptop	Phone
East	?	?
North	?	?
South	?	?

using the sum of sales."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc='sum')
pt
```

```python
#Exercise 4: Monthly Revenue Report (Medium)
df = pd.DataFrame({
    "Month":["Jan","Jan","Jan","Feb","Feb","Mar","Mar"],
    "Category":["Food","Electronics","Food","Food","Electronics","Food","Electronics"],
    "Revenue":[5000,12000,4000,6000,15000,7000,18000]
})
pt = pd.pivot_table(df,index='Month',columns='Category',values='Revenue',aggfunc='sum',fill_value=0)
display(pt)
```

```python
#Exercise 5: Company Sales Dashboard (Medium)
df = pd.DataFrame({
    "Region":["North","North","North","South","South","East","East","East"],
    "Salesperson":["John","John","Mike","Mike","Sara","Sara","John","Mike"],
    "Product":["Laptop","Phone","Laptop","Phone","Laptop","Phone","Laptop","Phone"],
    "Sales":[100,150,120,170,200,180,110,140]
})
"""
Tasks
Part A

Create a pivot table showing

| Region | Total Sales |
"""
pt = pd.pivot_table(df,index='Region',values='Sales',aggfunc='sum')
pt = pt.rename(columns={'Sales':'Total Sales'})
pt
```

```python
"""Part B

Create a pivot table

| Region | Laptop | Phone |

using sum."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc='sum')
pt
```

```python
"""Part C

Create a pivot table

| Salesperson | Average Sales |"""
pt = pd.pivot_table(df,index='Salesperson',values='Sales',aggfunc="mean")
pt
```

```python
"""Part D (Challenge)

Create a pivot table

| Region | Laptop Sales | Phone Sales | Grand Total |

using margins."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc="sum",margins=True, 
    margins_name='Grand Total')
pt = pt.rename(columns={
    'Laptop': 'Laptop Sales',
    'Phone': 'Phone Sales'
})
pt.columns.name = None
pt
```

```python
#Problem 1: Retail Store Sales Report ⭐ (Easy)
#Dataset
import pandas as pd

df = pd.DataFrame({
    "Store": ["A","A","A","B","B","B","C","C"],
    "Product": ["Laptop","Phone","Tablet","Laptop","Phone","Tablet","Laptop","Phone"],
    "Sales": [120000,50000,30000,90000,60000,25000,100000,55000]
})
"""Business Requirement

The regional manager wants to compare how much revenue each store generated from each product.

Expected Output

          Laptop   Phone   Tablet
Store
A
B
C"""

pt = pd.pivot_table(df,index='Store',columns='Product',values='Sales',aggfunc='sum',fill_value=0)
pt
```

```python
#Problem 2: Employee Performance Dashboard ⭐⭐
#Dataset
df = pd.DataFrame({
    "Employee":["Alice","Bob","Alice","Bob","Charlie","Charlie","Alice","Bob"],
    "Quarter":["Q1","Q1","Q2","Q2","Q1","Q2","Q3","Q3"],
    "Sales":[25,18,30,22,28,35,40,32]
})
"""Business Requirement

The HR team wants to know the average quarterly sales made by each employee.

Design a report where each employee is a row and each quarter is a separate column."""
pt = pd.pivot_table(df,index='Employee',columns='Quarter',values='Sales',aggfunc='mean',fill_value=0)
pt
```

```python
#Problem 3: Hospital Patient Analytics ⭐⭐⭐
#Dataset
df = pd.DataFrame({
    "Department":["Cardiology","Cardiology","Neurology","Neurology","Orthopedic","Orthopedic","Cardiology","Neurology"],
    "Gender":["Male","Female","Male","Female","Male","Female","Male","Female"],
    "Age":[62,58,45,50,38,41,70,55]
})
"""Business Requirement

The hospital administrator wants to compare the average age of patients by department and gender.

Bonus

Also display the oldest patient in every department-gender combination."""
pt = pd.pivot_table(df,index='Department',columns='Gender',values='Age',aggfunc=['mean','max'])
pt
```

```python
#Problem 4: E-commerce Analytics ⭐⭐⭐
#Dataset
df = pd.DataFrame({
    "Month":["Jan","Jan","Jan","Feb","Feb","Mar","Mar","Mar","Apr"],
    "Category":["Electronics","Furniture","Clothing",
                "Electronics","Furniture",
                "Electronics","Furniture","Clothing",
                "Electronics"],
    "Orders":[120,60,90,140,70,160,80,100,180]
})
"""Business Requirement

The CEO wants a dashboard showing the total number of orders placed for each product category every month.

Bonus

If a category has no orders in a month, show 0 instead of NaN."""
pt= pd.pivot_table(df,index='Month',columns='Category',values='Orders',aggfunc='sum',fill_value=0)
pt
```
