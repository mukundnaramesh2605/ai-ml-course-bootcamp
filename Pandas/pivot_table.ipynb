```python
#Exercise 5: Company Sales Dashboard (Medium)
df = pd.DataFrame({
    "Region":["North","North","North","South","South","East","East","East"],
    "Salesperson":["John","John","Mike","Mike","Sara","Sara","John","Mike"],
    "Product":["Laptop","Phone","Laptop","Phone","Laptop","Phone","Laptop","Phone"],
    "Sales":[100,150,120,170,200,180,110,140]
})
"""
Tasks
Part A

Create a pivot table showing

| Region | Total Sales |
"""
pt = pd.pivot_table(df,index='Region',values='Sales',aggfunc='sum')
pt = pt.rename(columns={'Sales':'Total Sales'})
pt
```

```python
"""Part B

Create a pivot table

| Region | Laptop | Phone |

using sum."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc='sum')
pt
```

```python
"""Part C

Create a pivot table

| Salesperson | Average Sales |"""
pt = pd.pivot_table(df,index='Salesperson',values='Sales',aggfunc="mean")
pt
```

```python
"""Part D (Challenge)

Create a pivot table

| Region | Laptop Sales | Phone Sales | Grand Total |

using margins."""
pt = pd.pivot_table(df,index='Region',columns='Product',values='Sales',aggfunc="sum",margins=True, 
    margins_name='Grand Total')
pt = pt.rename(columns={
    'Laptop': 'Laptop Sales',
    'Phone': 'Phone Sales'
})
pt.columns.name = None
pt
```

```python
#Problem 5: Company Salary Analytics ⭐⭐⭐⭐ (Interview-Level)
#Dataset
df = pd.DataFrame({
    "Department":["IT","IT","IT","HR","HR","Finance","Finance","Finance","Marketing"],
    "Gender":["M","F","M","F","M","F","M","F","M"],
    "Salary":[70000,75000,82000,50000,62000,90000,87000,95000,65000],
    "Experience":[2,3,5,1,4,8,6,10,3]
})
"""Business Requirement

The HR Director asks you to prepare a salary dashboard with the following requirements:

Show the average salary for each department broken down by gender.
Also show the maximum salary.
Add a Grand Total row and column.
Replace missing values with 0.
Rename the grand total label to Overall."""
pt =pd.pivot_table(df,index="Department",columns="Gender",values='Salary',aggfunc=['mean','max'],margins=True,margins_name='Overall',fill_value=0)
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 1: University Result Analysis
#
# Business Requirement:
#
# The Dean wants to compare the average marks of
# male and female students in every department.
#
# Bonus:
# Also display the highest marks.
# ==========================================================

df = pd.DataFrame({
    "Department":["CSE","CSE","ECE","ECE","EEE","EEE","CSE","ECE"],
    "Semester":[1,2,1,2,1,2,1,2],
    "Gender":["M","F","M","F","M","F","F","M"],
    "Marks":[82,91,75,88,79,94,95,84]
})

pt = pd.pivot_table(df,index='Department',columns="Gender",values='Marks',aggfunc=['mean','max'])
pt = pt.rename(columns={'mean':'Average marks','max': 'Highest Mark'})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 3: Hospital Operations
#
# Business Requirement:
#
# Find the average number of patients handled
# by every doctor in each department.
#
# Missing combinations should display 0.
# ==========================================================

df = pd.DataFrame({
    "Doctor":["A","A","B","B","C","C","A","B"],
    "Department":["Cardiology","Neurology","Cardiology",
                  "Orthopedic","Neurology","Orthopedic",
                  "Orthopedic","Neurology"],
    "Patients":[12,8,15,11,14,13,10,16]
})

pt = pd.pivot_table(df,index="Department",columns='Doctor',values='Patients',aggfunc='mean',fill_value=0)
pt = pt.rename(columns={'A': "Average Patients Handled by A",'B': "Average Patients Handled by B",'C': "Average Patients Handled by C"})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 4: Employee Salary Analytics
#
# Business Requirement:
#
# The HR team wants:
#
# 1. Average salary
# 2. Maximum salary
#
# for every department across every location.
#
# (Salary is in lakhs.)
# ==========================================================

df = pd.DataFrame({
    "Department":["IT","IT","IT","HR","HR",
                  "Finance","Finance","Marketing"],
    "Location":["Chennai","Bangalore","Chennai",
                "Mumbai","Mumbai","Delhi",
                "Delhi","Chennai"],
    "Salary":[7,8,9,5,6,10,11,7]
})
# index = Location
# columns = Department
# values = Salary
# aggfunc = ['mean','max']
# fill_value =
# margins =
# margins_name =
pt = pd.pivot_table(df,index='Location',columns='Department',values='Salary',aggfunc=['mean','max'],fill_value=0)
pt = pt.rename(columns={'mean':"Average Salary","max":'Highest Salary'})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 5: IPL Statistics
#
# Business Requirement:
#
# Create a report showing the total runs scored
# by each team in every season.
#
# Add grand totals.
# ==========================================================

df = pd.DataFrame({
    "Team":["CSK","CSK","RCB","RCB",
            "MI","MI","CSK","RCB"],
    "Season":[2023,2024,2023,2024,
              2023,2024,2023,2024],
    "Player":["Dhoni","Ruturaj","Kohli","Maxwell",
              "Rohit","Sky","Jadeja","Siraj"],
    "Runs":[180,620,650,420,500,540,350,80]
})

# index = Team
# columns = Season
# values = Runs
# aggfunc = sum
# fill_value = 0
# margins = True
# margins_name = grand totals
pt = pd.pivot_table(df,index='Team',columns='Season',values='Runs',aggfunc='sum',margins=True,margins_name='Grand total',fill_value=0)
pt = pt.rename(columns={2023:"Total run in 2023",2024:"Total run in 2024"})
pt
```

```python
import pandas as pd

# ==========================================================
# Problem 6: Startup Funding
#
# Business Requirement:
#
# The VC firm wants to know:
#
# 1. Total funding
# 2. Average funding
#
# received from every investor each year.
# ==========================================================

df = pd.DataFrame({
    "Startup":["A","A","B","B","C","C","A","B"],
    "Year":[2022,2023,2022,2023,2022,2023,2024,2024],
    "Investor":["Sequoia","Accel","Accel","Tiger",
                "Tiger","Sequoia","Tiger","Accel"],
    "Funding":[20,35,18,40,15,28,50,42]
})

# index = Investor
# columns = Year
# values = Funding 
# aggfunc = sum
# fill_value = 0
# margins =
# margins_name =
pt = pd.pivot_table(df,index='Investor',columns='Year',values='Funding',aggfunc=['sum','mean'],fill_value=0)
pt=pt.rename(columns={'sum':" Total Funding",'mean':"Average Funding"})
pt
```
