```python
n = int(input())
val=0
for i in range(n+1,1,-1):
    lst = [str(x) for x in range(1,i)]
    if val%2 != 0:
        lst.sort(reverse=True)
        print("".join(lst))
    else:
        print("".join(lst))
    val+=1
```

```python
n = int(input())
spaces_count = n-1
for i in range(1,n+1):
    stars = "*"*i
    spaces = " "*spaces_count
    spaces_count-=1
    print(spaces+stars)
```
