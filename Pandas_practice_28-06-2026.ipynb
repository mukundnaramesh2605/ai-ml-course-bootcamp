```python
import pandas as pd
import numpy as np
```

```python
# ============================================================================
# BLOCK 1: EASY PROBLEMS (1-10) - Fresh scenarios
# ============================================================================

# Problem 1: Basic info and data types
# DataFrame: id, name, email, age, salary
# Return data types of each column
# Expected: Series showing column dtypes
def easy_1():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'email': ['alice@test.com', 'bob@test.com', 'charlie@test.com'],
        'age': [25, 30, 28],
        'salary': [50000.50, 60000.75, 65000.00]
    })
    # Your code here
    return df.dtypes

# Test:
print("Easy Problem 1:")
print(easy_1())
print()
```

```python
# Problem 2: Get basic statistics
# DataFrame: id, score, attempts
# Return: min, max, mean of score column
# Expected: Series with min, max, mean values
def easy_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'score': [85, 92, 78, 88, 95],
        'attempts': [2, 1, 3, 2, 1]
    })
    # Your code here
    return df.agg(min_score =('score','min'),
                 max_score = ('score','max'),
                 mean_sore = ('score','mean'))

# Test:
print("Easy Problem 2:")
print(easy_2())
print()
```

```python
# Problem 3: Filter by string value
# DataFrame: id, status, amount
# Filter where status == 'active'
# Expected: 2 rows
def easy_3():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'status': ['active', 'inactive', 'active', 'pending'],
        'amount': [100, 200, 150, 300]
    })
    # Your code here
    return df[df.status == 'active']

# Test:
print("Easy Problem 3:")
print(easy_3())
print()
```

```python
# Problem 4: Drop a column
# DataFrame: id, name, email, phone
# Drop the 'phone' column
# Expected: DataFrame with 3 columns
def easy_4():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'email': ['a@test.com', 'b@test.com', 'c@test.com'],
        'phone': ['123-456', '234-567', '345-678']
    })
    # Your code here
    df= df.drop('phone',axis=1)
    return df

# Test:
print("Easy Problem 4:")
print(easy_4())
print()
```

```python
# Problem 5: Rename columns
# DataFrame: id, nm, sal
# Rename: nm -> name, sal -> salary
# Expected: DataFrame with renamed columns
def easy_5():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'nm': ['Alice', 'Bob', 'Charlie'],
        'sal': [50000, 60000, 65000]
    })
    # Your code here
    # Rename specific columns
    df = df.rename(columns={'nm': 'name', 'sal': 'salary'})
    return df

# Test:
print("Easy Problem 5:")
print(easy_5())
print()
```

```python
# Problem 6: Check for null values
# DataFrame: id, name, age (some NaN)
# Return count of null values per column
# Expected: Series with null counts
def easy_6():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'name': ['Alice', None, 'Charlie', 'David', 'Eve'],
        'age': [25, 30, None, 35, None]
    })
    # Your code here
    return df.isnull().sum()

# Test:
print("Easy Problem 6:")
print(easy_6())
print()
```

```python
# Problem 7: Get unique values
# DataFrame: id, city
# Return unique cities
# Expected: array with 3 unique values
def easy_7():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'city': ['NYC', 'LA', 'NYC', 'Chicago', 'LA']
    })
    # Your code here
    return df.city.unique
# Test:
print("Easy Problem 7:")
print(easy_7())
print()
```

```python
# Problem 8: Data type conversion
# DataFrame: id, date_str (string dates)
# Convert date_str to datetime
# Expected: DataFrame with datetime column
def easy_8():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'date_str': ['2023-01-15', '2023-02-20', '2023-03-10']
    })
    # Your code here
    df['date_str'] = pd.to_datetime(df['date_str'])
    return df

# Test:
print("Easy Problem 8:")
print(easy_8())
print()
```

```python
# Problem 9: String method - extract part
# DataFrame: id, full_name
# Create 'first_name' by extracting before space
# Expected: DataFrame with first_name column
def easy_9():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'full_name': ['Alice Smith', 'Bob Johnson', 'Charlie Brown']
    })
    # Your code here
    df['first_name'] = df['full_name'].str.split(' ').str[0]
    return df

# Test:
print("Easy Problem 9:")
print(easy_9())
print()
```

```python
# Problem 10: Get last N rows
# DataFrame: id, value (5 rows)
# Return last 2 rows
# Expected: 2 rows
def easy_10():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'value': [10, 20, 30, 40, 50]
    })
    # Your code here
    return df.iloc[-2:]
    #return df.tail(2)
    
# Test:
print("Easy Problem 10:")
print(easy_10())
print()
```
