```python
# ============================================================================
# BLOCK 1: EASY PROBLEMS (1-10) - Fresh scenarios
# ============================================================================

# Problem 1: Basic info and data types
# DataFrame: id, name, email, age, salary
# Return data types of each column
# Expected: Series showing column dtypes
def easy_1():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'email': ['alice@test.com', 'bob@test.com', 'charlie@test.com'],
        'age': [25, 30, 28],
        'salary': [50000.50, 60000.75, 65000.00]
    })
    # Your code here
    return df.dtypes

# Test:
print("Easy Problem 1:")
print(easy_1())
print()
```

```python
# Problem 2: Get basic statistics
# DataFrame: id, score, attempts
# Return: min, max, mean of score column
# Expected: Series with min, max, mean values
def easy_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'score': [85, 92, 78, 88, 95],
        'attempts': [2, 1, 3, 2, 1]
    })
    # Your code here
    return df.agg(min_score =('score','min'),
                 max_score = ('score','max'),
                 mean_sore = ('score','mean'))

# Test:
print("Easy Problem 2:")
print(easy_2())
print()
```

```python
# Problem 3: Filter by string value
# DataFrame: id, status, amount
# Filter where status == 'active'
# Expected: 2 rows
def easy_3():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'status': ['active', 'inactive', 'active', 'pending'],
        'amount': [100, 200, 150, 300]
    })
    # Your code here
    return df[df.status == 'active']

# Test:
print("Easy Problem 3:")
print(easy_3())
print()
```
