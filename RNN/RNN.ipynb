```python
import torch

batch_size = 4
sequence_length = 5
input_dim = 3

x = torch.randn(batch_size, sequence_length, input_dim)

print(x)

print(x.shape)
```

```python
import torch.nn as nn

rnn = nn.RNN(
    input_size = 3,
    hidden_size = 8,
    num_layers = 1,
    bidirectional = False,
    batch_first = True
)

output, hidden = rnn(x)

print("Input Shape ->",x.shape)
print("Output Shape -> y_hat ->",output.shape)
print("Hidden Shape -> a_n -> ",hidden.shape)
```

```python
import torch.nn as nn

rnn = nn.RNN(
    input_size = 3,
    hidden_size = 8,
    num_layers = 1,
    bidirectional = True,
    batch_first = True
)

output, hidden = rnn(x)

print("Input Shape ->",x.shape)
print("Output Shape -> y_hat ->",output.shape)
print("Hidden Shape -> a_n -> ",hidden.shape)
```

```python
lstm = nn.LSTM(
    input_size = 3,
    hidden_size = 8,
    num_layers = 1,
    batch_first = True
)

output, (hidden,cell) = lstm(x)

print("Input Shape ->",x.shape)
print("Output Shape -> y_hat ->",output.shape)
print("Hidden Shape -> a_n -> short term -> ",hidden.shape)
print("Cell Shape -> c_n -> Long term -> ",cell.shape)

```

```python
gru = nn.GRU(
    input_size = 3,
    hidden_size = 8,
    num_layers = 1,
    batch_first = True
)

output, hidden = gru(x)

print("Input Shape ->",x.shape)
print("Output Shape -> y_hat ->",output.shape)
print("Hidden Shape -> a_n -> ",hidden.shape)
```
