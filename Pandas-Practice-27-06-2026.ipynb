```python
import pandas as pd
import numpy as np
```

```python
# ============================================================================
# BLOCK 1: EASY PROBLEMS (1-10)
# ============================================================================

# Problem 1: Load and inspect
# Create DataFrame: id, name, age
# Return shape of DataFrame
# Expected: (3, 3)
def problem_easy_1():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'age': [25, 30, 28]
    })
    # Your code here
    return df.shape

# Test:
print("Easy Problem 1:")
print(problem_easy_1())
print()
```

```python
# Problem 2: Get first N rows
# DataFrame: id, name, age
# Return first 2 rows
def problem_easy_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'age': [25, 30, 28, 35, 22]
    })
    # Your code here
    return df.iloc[:2]

# Test:
print("Easy Problem 2:")
print(problem_easy_2())
print()
```

```python
# Problem 3: Get column as Series
# DataFrame: id, name, age
# Return the 'name' column
# Expected: Series with 3 values
def problem_easy_3():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'age': [25, 30, 28]
    })
    # Your code here
    return df.name

# Test:
print("Easy Problem 3:")
print(problem_easy_3())
print()
```

```python
# Problem 4: Simple filter - single condition
# DataFrame: id, name, age
# Filter: age > 25
# Expected: 2 rows (Bob and Charlie)
def problem_easy_4():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'age': [25, 30, 28]
    })
    # Your code here
    return df[df.age > 25]

# Test:
print("Easy Problem 4:")
print(problem_easy_4())
print()
```

```python
# Problem 5: Filter and select specific columns
# DataFrame: id, name, age, salary
# Filter: age >= 28
# Select: name, salary only
# Expected: 2 rows, 2 columns
def problem_easy_5():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [25, 30, 28, 22],
        'salary': [50000, 60000, 65000, 45000]
    })
    # Your code here
    return df[df.age >=28][['name','salary']]

# Test:
print("Easy Problem 5:")
print(problem_easy_5())
print()
```

```python
# Problem 6: Create new column - simple arithmetic
# DataFrame: id, price, quantity
# Create: 'total' = price * quantity
# Expected: new column with 3 values [100, 300, 200]
def problem_easy_6():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'price': [10, 15, 20],
        'quantity': [10, 20, 10]
    })
    # Your code here
    df['total'] = df.price * df.quantity
    return df.total

# Test:
print("Easy Problem 6:")
print(problem_easy_6())
print()
```

```python
# Problem 7: Create new column - conditional
# DataFrame: id, score
# Create: 'pass_fail'
#   - 'Pass' if score >= 60
#   - 'Fail' if score < 60
# Expected: ['Pass', 'Fail', 'Pass']
def problem_easy_7():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'score': [85, 45, 72]
    })
    # Your code here
    df['pass_fail'] = df.score.apply(lambda x : 'Pass' if x>=60 else 'Fail')
    return df['pass_fail']

# Test:
print("Easy Problem 7:")
print(problem_easy_7())
print()
```

```python
# Problem 8: Basic groupby - count
# DataFrame: id, category
# Group by category, count rows
# Expected: Series with category counts
def problem_easy_8():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'category': ['A', 'B', 'A', 'B', 'A']
    })
    # Your code here
    return df.groupby('category')['category'].value_counts()

# Test:
print("Easy Problem 8:")
print(problem_easy_8())
print()
```

```python
# Problem 9: Basic groupby - sum
# DataFrame: id, category, value
# Group by category, sum the values
# Expected: Series with category sums
def problem_easy_9():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'category': ['X', 'Y', 'X', 'Y', 'X'],
        'value': [10, 20, 15, 25, 30]
    })
    # Your code here
    return df.groupby('category')['value'].sum()

# Test:
print("Easy Problem 9:")
print(problem_easy_9())
print()
```

```python
# Problem 10: Sort by column
# DataFrame: id, name, salary
# Sort by salary ascending (lowest first)
# Expected: DataFrame sorted by salary column
def problem_easy_10():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [60000, 45000, 75000]
    })
    # Your code here
    sorted_salary = df.sort_values(by='salary')
    return sorted_salary

# Test:
print("Easy Problem 10:")
print(problem_easy_10())
print()
```

```python
# ============================================================================
# BLOCK 2: MEDIUM PROBLEMS (11-15)
# ============================================================================

# Problem 11: Multiple conditions filter
# DataFrame: id, name, age, department
# Filter: age > 25 AND department == 'Sales'
# Expected: 2 rows
def problem_medium_1():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'age': [28, 22, 30, 27, 25],
        'department': ['Sales', 'IT', 'Sales', 'HR', 'IT']
    })
    # Your code here
    filtered = df[(df.age > 25) & (df.department=='Sales')]
    return filtered

# Test:
print("Medium Problem 1:")
print(problem_medium_1())
print()
```

```python
# Problem 12: Groupby with multiple aggregations
# DataFrame: id, product, category, sales
# Group by category
# Aggregate: count products, sum sales, mean sales
# Expected: DataFrame with 3 columns
def problem_medium_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5, 6],
        'product': ['A', 'B', 'C', 'A', 'B', 'C'],
        'category': ['X', 'X', 'Y', 'Y', 'X', 'Y'],
        'sales': [100, 150, 200, 180, 120, 160]
    })
    # Your code here
    result = df.groupby('category').agg(
        count=('product', 'count'),
        total_sale=('sales', 'sum'),
        mean_sale=('sales', 'mean')
    ).reset_index()
    return result

# Test:
print("Medium Problem 2:")
print(problem_medium_2())
print()
```

```python
# Problem 13: Combine filter, transform, and groupby
# DataFrame: id, department, salary
# Step 1: Filter salary > 50000
# Step 2: Create 'bonus' = salary * 0.1
# Step 3: Group by department, sum bonuses
# Expected: Series with department totals
def problem_medium_3():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [60000, 55000, 75000, 65000, 48000]
    })
    # Your code here
    salary_filtered = df[df.salary > 50000]
    salary_filtered['bonus'] = salary_filtered['salary'] * 0.1
    grouped = salary_filtered.groupby('department')['bonus'].sum()
    return grouped 

# Test:
print("Medium Problem 3:")
print(problem_medium_3())
print()
```

```python
# Problem 14: Use .isin() for filtering
# DataFrame: id, city, sales
# Filter: city is either 'NYC' or 'LA'
# Expected: 3 rows
def problem_medium_4():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'city': ['NYC', 'LA', 'Chicago', 'NYC', 'Boston'],
        'sales': [100, 150, 200, 120, 80]
    })
    # Your code here
    return df[df.city.isin(['NYC','LA'])]

# Test:
print("Medium Problem 4:")
print(problem_medium_4())
print()
```

```python
# Problem 15: Sort by multiple columns
# DataFrame: id, department, salary
# Sort by: department (ascending), salary (descending)
# Expected: sorted DataFrame
def problem_medium_5():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'department': ['IT', 'Sales', 'IT', 'Sales', 'HR'],
        'salary': [65000, 60000, 55000, 75000, 48000]
    })
    # Your code here
    return df.sort_values(by=['department','salary'],ascending=[True,False])

# Test:
print("Medium Problem 5:")
print(problem_medium_5())
print()
```
