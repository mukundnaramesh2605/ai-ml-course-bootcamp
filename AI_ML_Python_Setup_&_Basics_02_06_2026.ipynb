```python
# Book
# attributes : title, author, price, discount, language
# methods (operation) : get to know the price, get the details of the book

# init method : Constructor method -> used to initialize the arrtibutes
# init method is the method that is called by default (if it exists) when we create the instance
```

```python
class Book:
  def __init__(self,title,author,price,discount,language):
    self.title = title
    self.author = author
    self.price = price
    self.discount = discount
    self.language = language

  def get_price(self):
    if self.discount is None:
      return self.price
    else :
      return self.price - (self.price * self.discount)

  def __repr__(self):
    return f"{self.title} is written by {self.author} in the language of {self.language} and the price of the book is {self.price}"


b1 = Book("Parva","S L Bhyrappa",750,0.1,"Kannada")

print(f"Price of {b1.title} is {b1.get_price()}")
```

```python
class Parrot: #class definition
  name = ""   #attributes
  age = 0

p1 = Parrot() #Instance
p1.name = "Blu"
p1.age = 10

p2 = Parrot()
p2.name = "Birdie"
p2.age = 5

print(f"{p1.name} is {p1.age} years old")
print(f"{p2.name} is {p2.age} years old")
```

```python
class Book:
  def __init__(self,title,author,price,discount,language):
    self.title = title
    self.author = author
    self.price = price
    self.discount = discount
    self.language = language

  def get_price(self):
    if self.discount is None:
      return self.price
    else:
      return self.price - (self.price * self.discount)

  def __repr__(self):
    return f"The book '{self.title}' by {self.author} is written in {self.language}."

title = "Hamlet"
author = "William Shakespeare"
price = 750
discount = 0.20
language = "English"
book = Book(title=title,author=author,price=price,discount=discount,language=language)

print(f"The book {book.title} is written by {book.author}")
print(f"The price of the book {book.title} is {book.get_price()}")
print(repr(book))
```
