```python

def positive_negative_list(l1):
  positive_list =[]
  negative_list = []
  for x in l1:
    if x >= 0:
      positive_list.append(x)
    else:
      negative_list.append(x)

  return positive_list,negative_list

var1 = [1,-3,-6,8,-9,12,-90,-45,-63,78]
p_list,n_list = positive_negative_list(var1)
print(p_list,n_list)
```

# **OOP Concepts**

```python
class Parrot: #class definition
  name = ""   #attributes
  age = 0

p1 = Parrot() #Instance
p1.name = "Blu"
p1.age = 10

p2 = Parrot()
p2.name = "Birdie"
p2.age = 5

print(f"{p1.name} is {p1.age} years old")
print(f"{p2.name} is {p2.age} years old")
```

```python
class Book:
  def __init__(self,title,author,price,discount,language):
    self.title = title
    self.author = author
    self.price = price
    self.discount = discount
    self.language = language

  def get_price(self):
    if self.discount is None:
      return self.price
    else:
      return self.price - (self.price * self.discount)

  def __repr__(self):
    return f"The book '{self.title}' by {self.author} is written in {self.language}."

title = "Hamlet"
author = "William Shakespeare"
price = 750
discount = 0.20
language = "English"
book = Book(title=title,author=author,price=price,discount=discount,language=language)

print(f"The book {book.title} is written by {book.author}")
print(f"The price of the book {book.title} is {book.get_price()}")
print(repr(book))
```

```python
print(book.__repr__())
```

```python
class Dog:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def get_name(self):
    return self.name

  def get_age(self):
    return self.age

  def make_sound(self):
    return f"{self.name} is making the sound Woof"

  def __repr__(self):
    return f"{self.name} is {self.age} years old and makes the sound Woof"

dog1 = Dog("Goldie",1)
print(dog1)
print(dog1.make_sound())
print(f"Age of the dog is {dog1.get_age()}")
```
