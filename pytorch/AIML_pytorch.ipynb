```python
# Creation of neural network

# input features : 4, output features : 3
# number of hidden layers :2
# number of nodes in each of the hidden layers : 4 and 5
# Activation function : ReLu

class Simple_MLP(nn.Module):
  def __init__(self):
    super().__init__()

    self.mlp_stack = nn.Sequential(
        nn.Linear(4,4),
        nn.ReLU(),
        nn.Linear(4,5),
        nn.ReLU(),
        nn.Linear(5,3)
    )

  def forward(self,x):
    return self.mlp_stack(x)

```

```python
sim = Simple_MLP()
x = torch.tensor([[1,2,3,4],[5,6,7,8]],dtype=torch.float32)
```

```python
# MLP
# 784 input features, number of classes 10
# 2 Hidden layers with 16 nodes each
# ReLU Activation

class MLP_MNIST(nn.Module):
  def __init__(self):
    super().__init__()

    self.mlp_stack = nn.Sequential(
        nn.Linear(784,16),
        nn.ReLU(),
        nn.Linear(16,16),
        nn.ReLU(),
        nn.Linear(16,10)
    )

  def forward(self,x):
    x = x.view(x.shape[0],-1)
    return self.mlp_stack(x)
```

```python
model = MLP_MNIST().to(device)
```

```python
import torchvision
import torchvision.transforms as transforms

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,),(0.5,))
])

train_dataset = torchvision.datasets.MNIST(
    root = "./data",
    train = True,
    download = True,
    transform = transform
)

test_dataset = torchvision.datasets.MNIST(
    root = "./data",
    train = False,
    download = True,
    transform = transform
)

train_loaders = torch.utils.data.DataLoader(
    train_dataset,
    batch_size = 128,
    shuffle = True
)

test_loaders = torch.utils.data.DataLoader(
    test_dataset,
    batch_size =128,
    shuffle = False
)
```

```python
# Visualize some images

import matplotlib.pyplot as plt

images,labels = next(iter(train_loaders))

print(images.shape)

plt.figure(figsize = (10,6))

for i in range(8):
  image = images[i]

  # CHW but for matplot HWC
  image = image.permute(1,2,0)

  label = labels[i]

  plt.subplot(2,4,i+1)
  plt.imshow(image,cmap = "grey")
  plt.title(f"Label : {label}")
  plt.axis("off")

plt.show()


```

```python
# hyper-parameters

num_epochs = 10
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(),lr = 0.001)
```

```python
def train_loop(model, train_loader, criterion, optimizer):
  model.train()

  running_loss = 0
  correct = 0
  total = 0

  for images,labels in train_loader:
    images = images.to(device)
    labels = labels.to(device)

    # Prediction
    logits = model(images)

    # Compute the loss
    loss = criterion(logits,labels)

    # Backpropagation
    optimizer.zero_grad()
    loss.backward()

    # Weight update
    optimizer.step()

    running_loss += loss.item()

    predictions = torch.argmax(logits,dim = 1)
    correct += (predictions == labels).sum().item()
    total += labels.size(0)

  epoch_loss = running_loss / len(train_loader)
  epoch_acc = correct / total

  print(f"Train Loss : {epoch_loss:.4f} | Train Acc : {epoch_acc:.4f}")



for epoch in range(num_epochs):
  print(f"Epoch : {epoch+1}")
  train_loop(model,train_loaders,criterion,optimizer)
```

```python
# Test Acc

def test_loop(model,test_loader):
  model.eval()

  correct = 0
  total = 0

  with torch.no_grad():
    for images,labels in test_loader:

      images = images.to(device)
      labels = labels.to(device)

      logits = model(images)

      predictions = torch.argmax(logits,dim = 1)

      correct += (predictions == labels).sum().item()
      total += labels.size(0)

  epoch_acc = correct / total

  print(f"Test Acc : {epoch_acc:.4f}")

test_loop(model,test_loaders)
```
