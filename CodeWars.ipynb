```python
def max_subarray_sum(arr):
    if not arr:
        return 0
        
    max_sum = 0
    current_sum = 0
    
    for num in arr:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
        
    return max_sum
```

```python
def sum_dig_pow(a, b):
    result = []
    
    # Loop through the inclusive range [a, b]
    for num in range(a, b + 1):
        # Convert number to string to easily iterate over digits
        digits = str(num)
        
        # Calculate the sum of digits raised to their consecutive positions
        total = sum(int(digit) ** (index + 1) for index, digit in enumerate(digits))
        
        # If the sum matches the original number, save it
        if total == num:
            result.append(num)
            
    return result
```

```python
import re

def increment_string(strng):
    match = re.match(r"(.*?)(\d*)$", strng)
    
    if match:
        prefix = match.group(1)
        number_str = match.group(2)
        if number_str:
            incremented_num = int(number_str) + 1
            new_number_str = str(incremented_num).zfill(len(number_str))
            return prefix + new_number_str
        else:
            return prefix + "1"
            
    return strng + "1"
```

```python
def declare_winner(fighter1, fighter2, first_attacker):
    # Set up our initial roles based on who attacks first
    if fighter1.name == first_attacker:
        attacker = fighter1
        defender = fighter2
    else:
        attacker = fighter2
        defender = fighter1

    # Keep fighting until someone's health drops to 0 or below
    while True:
        # Attacker strikes defender
        defender.health -= attacker.damage_per_attack
        
        # Check if the defender died
        if defender.health <= 0:
            return attacker.name
            
        # Swap roles for the next turn
        attacker, defender = defender, attacker
```
