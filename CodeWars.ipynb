```python
def max_subarray_sum(arr):
    if not arr:
        return 0
        
    max_sum = 0
    current_sum = 0
    
    for num in arr:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
        
    return max_sum
```

```python
def queue_time(customers, n):
    tills = [0] * n
    for time in customers:
        earliest_till_index = tills.index(min(tills))
        tills[earliest_till_index] += time
    return max(tills)
```

```python
import string
def high(x):
    alphabets = string.ascii_lowercase
    dicts = {}
    for i in x.split():
        sums = 0
        for j in i:
            sums+=(alphabets.index(j.lower()))+1
        dicts[i] = sums
    return [k for k,v in dicts.items() if v==max(dicts.values())][0]
high('man i need a taxi up to ubud')
```

```python
def sum_dig_pow(a, b):
    result = []
    
    # Loop through the inclusive range [a, b]
    for num in range(a, b + 1):
        # Convert number to string to easily iterate over digits
        digits = str(num)
        
        # Calculate the sum of digits raised to their consecutive positions
        total = sum(int(digit) ** (index + 1) for index, digit in enumerate(digits))
        
        # If the sum matches the original number, save it
        if total == num:
            result.append(num)
            
    return result
```

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i
two_sum([1 ,2, 3],4)
```

```python
def unique_in_order(sequence):
    unique = []
    for i in range(len(sequence)-1):
        if sequence[i] != sequence[i+1]:
            unique.append(sequence[i])
    unique.append(sequence[-1])
    return  unique
unique_in_order('AAAABBBCCDAABBB')
        
        
```

```python
from itertools import groupby

def unique_in_order(iterable):
    return [k for (k, _) in groupby(iterable)]
unique_in_order('AAAABBBCCDAABBB')
```

```python
def sort_array(source_array):
    odds = iter(sorted(x for x in source_array if x%2 != 0))
    return [next(odds) if x%2 !=0 else x for x in source_array]
sort_array([7,1,4,5,6,8,1])
```

```python
def nb_dig(n: int, d: int) -> int:
    # Convert digit to string once for efficient comparison
    digit_str = str(d)
    
    # Square each number, convert to string, and count occurrences of the digit
    return sum(str(k ** 2).count(digit_str) for k in range(n + 1))
```
