```python
def enough(cap, on, wait):
    return max(0, wait - (cap - on))
```

```python
def max_subarray_sum(arr):
    if not arr:
        return 0
        
    max_sum = 0
    current_sum = 0
    
    for num in arr:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
        
    return max_sum
```

```python
def grow(arr):
    product = arr[0]
    for i in arr[1:]:
        product *= i
    return product
```

```python
from functools import reduce

def grow(arr):
    return reduce(lambda x, y: x * y, arr)
```
