```python
def max_subarray_sum(arr):
    if not arr:
        return 0
        
    max_sum = 0
    current_sum = 0
    
    for num in arr:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
        
    return max_sum
```

```python
def queue_time(customers, n):
    tills = [0] * n
    for time in customers:
        earliest_till_index = tills.index(min(tills))
        tills[earliest_till_index] += time
    return max(tills)
```

```python
def sum_dig_pow(a, b):
    result = []
    
    # Loop through the inclusive range [a, b]
    for num in range(a, b + 1):
        # Convert number to string to easily iterate over digits
        digits = str(num)
        
        # Calculate the sum of digits raised to their consecutive positions
        total = sum(int(digit) ** (index + 1) for index, digit in enumerate(digits))
        
        # If the sum matches the original number, save it
        if total == num:
            result.append(num)
            
    return result
```

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i
two_sum([1 ,2, 3],4)
```

```python
def nb_dig(n: int, d: int) -> int:
    # Convert digit to string once for efficient comparison
    digit_str = str(d)
    
    # Square each number, convert to string, and count occurrences of the digit
    return sum(str(k ** 2).count(digit_str) for k in range(n + 1))
```

```python
def is_square(n: int) -> bool:
    # Negative numbers cannot be perfect squares
    if n < 0:
        return False
        
    # Calculate the square root and check if it's a whole number
    return (n ** 0.5) % 1 == 0
```

```python
import re

def increment_string(strng):
    match = re.match(r"(.*?)(\d*)$", strng)
    
    if match:
        prefix = match.group(1)
        number_str = match.group(2)
        if number_str:
            incremented_num = int(number_str) + 1
            new_number_str = str(incremented_num).zfill(len(number_str))
            return prefix + new_number_str
        else:
            return prefix + "1"
            
    return strng + "1"
```
