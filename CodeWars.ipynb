```python
def enough(cap, on, wait):
    return max(0, wait - (cap - on))
```

```python
def max_subarray_sum(arr):
    if not arr:
        return 0
        
    max_sum = 0
    current_sum = 0
    
    for num in arr:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
        
    return max_sum
```
