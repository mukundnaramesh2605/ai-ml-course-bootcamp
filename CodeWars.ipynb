```python
def enough(cap, on, wait):
    return max(0, wait - (cap - on))
```

```python
def max_subarray_sum(arr):
    if not arr:
        return 0
        
    max_sum = 0
    current_sum = 0
    
    for num in arr:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
        
    return max_sum
```

```python
def grow(arr):
    product = arr[0]
    for i in arr[1:]:
        product *= i
    return product
```

```python
from functools import reduce

def grow(arr):
    return reduce(lambda x, y: x * y, arr)
```

```python
def other_angle(a, b):
    return 180-(a+b)
```

```python
def queue_time(customers, n):
    tills = [0] * n
    for time in customers:
        earliest_till_index = tills.index(min(tills))
        tills[earliest_till_index] += time
    return max(tills)
```

```python
import string
def high(x):
    alphabets = string.ascii_lowercase
    dicts = {}
    for i in x.split():
        sums = 0
        for j in i:
            sums+=(alphabets.index(j.lower()))+1
        dicts[i] = sums
    return [k for k,v in dicts.items() if v==max(dicts.values())][0]
high('man i need a taxi up to ubud')
```

```python
def high(x):
    return max(x.split(), key=lambda k: sum(ord(c) - 96 for c in k))
high('man i need a taxi up to ubud')
```
