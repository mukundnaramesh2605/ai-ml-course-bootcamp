```python
n = int(input())
arr = list(map(int, input().split()))

# Pair each value with its original index, sort by value to find target positions
indexed = sorted(range(n), key=lambda i: arr[i])

visited = [False] * n
swaps = 0

for i in range(n):
    if visited[i] or indexed[i] == i:
        continue
    # walk the cycle starting at i
    cycle_size = 0
    j = i
    while not visited[j]:
        visited[j] = True
        j = indexed[j]
        cycle_size += 1
    swaps += cycle_size - 1

print(swaps)
```

```python
s = input().strip()
stack = []
mapping = {')': '(', '}': '{', ']': '['}
balanced = True

for char in s:
    if char in mapping:  
        if not stack or stack.pop() != mapping[char]:
            balanced = False
            break
    else:  
        stack.append(char)


if balanced and not stack:
    print("1")
else:
    print("0")
```

```python
def sum_proper_divisors(n):
    if n <= 1:
        return 0
    total = 1   # 1 is always a proper divisor for n > 1
    i = 2
    while i * i <= n:
        if n % i == 0:
            total += i
            if i != n // i:   # avoid counting sqrt twice for perfect squares
                total += n // i
        i += 1
    return total

n = int(input())
m = sum_proper_divisors(n)
if m != n and sum_proper_divisors(m) == n:
    print(m)
else:
    print(-1)
```

Write a program to find the Nth term in this series.


Input Description:
The value N in a positive integer that should be read from STDIN.


Output Description:
The Nth term, no other characters/ string or message should be written to STDOUT.


Explanation:
For example, if N=10, the 10th term in the series is to be derived from 9th term in the series. The 9th term is 8 so the 10th term is (8/2)=4. Only the value 4 should be printed to STDOUT. You can assume that N will not exceed 20,000.

10

```python
from collections import defaultdict, deque

n = int(input())
children = defaultdict(list)

for _ in range(n - 1):
    u, v = map(int, input().split())
    children[u].append(v)

if n == 1:
    print(1)
else:
    queue = deque([1])
    result = []
    while queue:
        node = queue.popleft()
        kids = children[node]
        # print edges in reversed order
        for child in reversed(kids):
            result.append(f"{node} {child}")
        # enqueue in original order
        for child in kids:
            queue.append(child)
    print('\n'.join(result))
```
