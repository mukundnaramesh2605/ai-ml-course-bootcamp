```python
"""399. Student Fee Calculation

Geekoin50
Medium
Topics
Problem Statement:
You are given an array of distinct roll numbers and an array of subjects. The student with the i th roll number , studies subject number i. (1 <= i <= N). Then you are given an array of subject-fee pairs. Subjects are unique in the array. Find the fee paid by all the N students with the N roll numbers to study that subject. If the fee for a subject is unavailable, print -1


Input Description:
First line contains an integer N. N is the size of the roll number array and subject name array
The next line contains N space separated integers denoting the roll numbers.
The next line contains N space separated strings denoting the subject names.
The ith student(roll no.) studies the ith subject
The next line contains an integer M denoting the number of subjects-fee pairs
Then M lines follow each containing a subject name and a fee separated by space


Output Description:
Print N lines, each containing a roll number and the fee paid by the student with that roll number separated by a single space.
Print it in increasing order of roll numbers. [Print -1 if fee for a subject is not available].


Explanation:
Output is self-explanatory


Sample Input:
5
99 81 8 82 32
math phy chem computers chem
3
phy 5000
chem 2000
computers 6000


Sample Output:
8 2000
32 2000
81 5000
82 6000
99 -1"""
n = int(input())
rolls = list(map(int, input().split()))
subjects = input().split()

m = int(input())
fee_map = {}
for _ in range(m):
    line = input().split()
    subj = line[0]
    fee = int(line[1])
    fee_map[subj] = fee

# pair each roll with its subject fee
students = []
for i in range(n):
    subj = subjects[i]
    fee = fee_map.get(subj, -1)
    students.append((rolls[i], fee))

students.sort()
for roll, fee in students:
    print(roll, fee)
```

```python
"""400. Counting Beautiful Indices

Geekoin40
Medium
Topics
Problem Statement:
J.S.Bhayana gave an integer array A of size N. If A[i] occurs A[i + 1] times in the array, i is called a beautiful index. Find the number of beautiful indices in the array. (1 <= i < N) (1 based indexing)


Input Description:
The first line contains an integer N, denoting the size of the array.
The next line contains N space separated integers.


Output Description:
Print a line containing the number of beautiful indices in the array.


Sample Input:
4
1 2 1 4 1 1


Sample Output:
3


Explanation:
A[2] occurs A[3] times
A[3] occurs A[4] times
A[4] occurs A[5] times
So 2, 1 and 4 are beautiful indices
So count is 3"""
n=int(input())
arr=list(map(int,input().split()))
count=0
for i in range(n-1):
    if arr.count(arr[i])==arr[i+1]:
        count+=1
print(count)
```

```python
"""410. Planet Gravity and Potential Energy

Geekoin50
Medium
Topics
Problem Statement:
The potential energy of a ball of mass m placed at height h is given by U = m.g.h. (m - mass, g - acceleration due to gravity). The acceleration due to gravity of K planets is given to you. Answer Q queries of the form,
U P h
U denotes the potential energy of a ball placed at height h on planet P.
Answer each query by printing the mass of the ball used.


Input Description:
The first line contains an integer K denoting the number of planets.
The next K lines contains the name of the planet and it's acceleration due to gravity, separated by a space.
The next line contains an integer Q denoting the number of queries.
The next Q lines contains a query as specified in the question.


Output Description:
Print Q lines denoting the mass of the ball used in each query.
Round off the mass to two decimal places.


Explanation:
earth has g = 20, so mass of the ball = 200 / (10 * 6) = 3.33


Sample Input:
2
earth 10
venus 20
1
200 earth 6


Sample Output:
3.33

"""
k = int(input())
planet_map = {}
for _ in range(k):
    line = input().split()
    planet = line[0]
    gravity = int(line[1])
    planet_map[planet] = gravity
query = int(input())
arr = list(input().split() for _ in range(query))
for i in arr:
  mass = round(int(i[0])/(planet_map[i[1]]*int(i[2])),2)
  print(f"{mass}")
```

```python
"""419. Palindrome Permutation with One Change

Geekoin50
Medium
Topics
Problem Statement:
Quinton is given a string, find if he can form a palindrome by permuting the characters of the string and changing at most one character of the string to another character. If he can form, print YES, else print NO.


Input Description:
The first and only line of input is a string S


Output Description:
Print a line containing YES or NO


Explanation:
'a' can be changed to 'b', we get "bbc". Permuting it we get "bcb"
, which is a palindrome.


Sample Input:
abc


Sample Output:
YES"""

from collections import Counter

s = input()
freq = Counter(s)
odd_count = sum(1 for v in freq.values() if v % 2 != 0)

if odd_count <= 3:
    print("YES")
else:
    print("NO")
```

```python
"""422. Maximize Single Element

Geekoin50
Medium
Topics
Problem Statement:
You are given an array A of N numbers. You need to perform the following operation on it until only one element is remaining in A. Take two numbers A[i] and A[j] for each valid i, j. Compute X = 2*(A[i] + A[j]). Delete A[i] and A[j] and add X to the array A. Find the maximum possible value of the single element left in the end.


Input Description:
The first line contains an integer N, denoting the number of array elements. The next line contains N space separated integers.


Output Description:
Print a single integer denoting the maximum possible value of the single element left in the end.


Explanation:
(4 + 3)^3 = 7^3 = 343


Sample Input:
2
4 3


Sample Output:
343

"""
import heapq

n = int(input())
arr = list(map(int, input().split()))

# max-heap using negation
heap = [-x for x in arr]
heapq.heapify(heap)

while len(heap) > 1:
    a = -heapq.heappop(heap)
    b = -heapq.heappop(heap)
    x = 2 * (a + b)
    heapq.heappush(heap, -x)

print(-heap[0])
```

```python
""" 445. Single Digit Sum

Geekoin40
Medium
Topics
Problem Statement:
Quinton was given an integer N. He was asked to find the digitized sum of N. The digitized sum is the summing up the digits until we obtain a single digit. Help Quinton solve this problem. For example: Digitized sum of 129 is 3, derived as,


1 + 2 + 9 = 12


1 + 2 = 3


Input Description:
The only line of input contains an integer N


Output Description:
Print one line, the digitized sum of N


Sample Input:
789


Sample Output:
6


Explanation:
7 + 8 + 9 = 24
2 + 4 = 6"""
n = input()
sum_s = sum(int(x) for x in n)
while len(str(sum_s)) != 1:
    sum_s = str(sum(int(x) for x in str(sum_s)))
print(sum_s)
```

```python
"""Problem Statement:
You will be given a list of N integers. Additionally you will be asked Q questions. For each question you will be given an integer X. You must report the number of integers in the given list that have X as a prefix. In other words you will have to find the count of numbers Z such that X+Y (+ denotes concatenation of two strings, not addition of two numbers) is equal to Z. As an example assume the list of integers is as follows: [ 1276,129,227,65 ]For X=12 count will be 2 since we have two numbers which have 12 as a prefix:1276 and 129 . (The prefix has been highlighted in bold)


Input Description:
The first line contains a single integer N denoting the number of integers that will be given to you
The next line will contain N space separated integers.
The next line will contain a single integer Q.
The next Q lines will contains a single integer in each line


Output Description:
Yout output must consist of Q lines :each line containing the required count.


Explanation:
Output is as explained in the problem statement


Sample Input:
4
1276 129 227 65
1
12


Sample Output:
2"""

n = int(input())
strings = list(input().split())
q = int(input())
for _ in range(q):
  count = 0
  prefix = input()
  for string in strings:
    if string.startswith(prefix):
      count += 1
  print(count)
```

```python
"""
455. Correcting Geometric Progression

Geekoin50
Medium
Topics
Problem Statement:
You are given a sequence containing N integers. Changing a maximum of one element of the sequence will result in a geometric progression with an integer common ratio. Your task is to print the correct geometric progression. ((Ref. https://en.wikipedia.org/wiki/Geometric_progression), N >= 4)


Input Description:
The first line contains the integer N.
The next line contains N space separated integers, A[0], A[1]...A[n-1]/


Output Description:
Print N space separated integers, which are in Geometric Progression"""
from fractions import Fraction

def is_gp(arr):
    r = Fraction(arr[1], arr[0])
    for i in range(2, len(arr)):
        if Fraction(arr[i], arr[i-1]) != r:
            return False
    return True

n = int(input())
arr = list(map(int, input().split()))

for i in range(n):
    test = arr[:]
    # derive test[i] from surrounding elements
    if i == 0:
        r = Fraction(arr[2], arr[1])
        val = Fraction(arr[1]) / r
    elif i == n-1:
        r = Fraction(arr[1], arr[0])
        val = Fraction(arr[-2]) * r
    else:
        # use ratio from the side that doesn't involve i
        if i >= 2:
            r = Fraction(arr[i-1], arr[i-2])
        else:
            r = Fraction(arr[i+2], arr[i+1])
        val = Fraction(arr[i-1]) * r

    if val.denominator == 1:
        test[i] = int(val)
        if is_gp(test):
            print(*test)
            break
```
