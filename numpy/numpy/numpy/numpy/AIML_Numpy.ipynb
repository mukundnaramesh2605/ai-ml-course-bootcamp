```python
# Manual implementaion of one pass through the layer
# in features = 3
# out features = 2

w = np.random.normal(0,1,(3,2))

b = np.random.uniform(0,1,(1,2))

x = np.array([1,2,3])

print(w)
print("----------------")

print(b)
print("----------------")

print(x @ w + b)
```

```python
# Implementing Activation Functions


# ReLu Activation

def relu(x):
  return np.maximum(0,x)

x = np.array([0,-2.12,3,3.8,-0.001])

print(relu(x))
```

```python
# Sigmoid Activation

def sigmoid(x):
  return 1/(1+np.exp(-x))

print(sigmoid(x))
```

```python
# Tanh activation

def tanh(x):
  return (np.exp(x) - np.exp(-x))/(np.exp(x) + np.exp(-x))

print(tanh(x))
```

```python
# Softmax function

def softmax(x):
  exp_x = np.exp(x-np.max(x))
  return exp_x / np.sum(exp_x)

print(softmax(x))
print(np.sum(softmax(x)))
```

```python
# MLP

# input -> Hidden -> ReLu -> Logits -> Prob vector

# input is 4 features

# output is 3 features

# hidden is 7 features

w1 = np.random.normal(1,2,(4,7))

w2 = np.random.normal(1,2,(7,3))

b1 = np.random.uniform(0,1,(1,7))

b2 = np.random.uniform(0,1,(1,3))

# batch size of 64

x = np.random.normal(0,1,(3,4))

z1 = x @ w1 + b1

a1 = relu(z1)

z2 = a1 @ w2 + b2

a2 = softmax(z2)

print(a2)

print(np.sum(a2))
```

```python
# MSE Error


def mse(y_true,y_pred):
  return np.mean((y_true-y_pred)**2)

y_true = np.array([1,2,3,4])

y_pred = np.array([1,2,3,4])

print(mse(y_true,y_pred))
```

```python
# CCE Loss

def cce_loss(y_true,y_pred):
  return -np.mean(y_true*np.log(y_pred))

y_true = np.array([1,0,0])

y_pred = np.array([0.9,0.05,0.05])

print(cce_loss(y_true,y_pred))
```
