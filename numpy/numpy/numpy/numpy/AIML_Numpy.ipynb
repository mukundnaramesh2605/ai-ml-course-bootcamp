```python
# Manual implementaion of one pass through the layer
# in features = 3
# out features = 2

w = np.random.normal(0,1,(3,2))

b = np.random.uniform(0,1,(1,2))

x = np.array([1,2,3])

print(w)
print("----------------")

print(b)
print("----------------")

print(x @ w + b)
```

```python
# for a batch of example

x = np.random.uniform(0,1,(4,3))

print(x)

print("----------------")

print(x @ w + b)
```

```python
# Implementing Activation Functions


# ReLu Activation

def relu(x):
  return np.maximum(0,x)

x = np.array([0,-2.12,3,3.8,-0.001])

print(relu(x))
```

```python
# Sigmoid Activation

def sigmoid(x):
  return 1/(1+np.exp(-x))

print(sigmoid(x))
```

```python
# Tanh activation

def tanh(x):
  return (np.exp(x) - np.exp(-x))/(np.exp(x) + np.exp(-x))

print(tanh(x))
```

```python
# Softmax function

def softmax(x):
  exp_x = np.exp(x-np.max(x))
  return exp_x / np.sum(exp_x)

print(softmax(x))
print(np.sum(softmax(x)))
```
