2.Highest and lowest per subject

3.Overall Class Topper

4.Paas Count per subject

5.Which subject is difficult

6.Ranking Students

```python
import numpy as np
from numpy.linalg import inv,det
```

```python
np.random.seed(25)
marks = np.random.randint(10,101,size=(20,5))
marks
```

```python
#1.Average Mark by subject and student
average_by_subject = np.mean(marks,axis=0)
average_by_studnet = np.mean(marks,axis=1)
print(f"The average mark per subject is: \n{average_by_subject}")
print(f"The average mark per student is: \n{average_by_studnet}")
```

```python
#2.Highest and lowest per subject
highest_mark_per_subject = np.max(marks,axis=0)
lowest_mark_per_subject = np.min(marks,axis=0)
print(f"The highest mark per subject is: \n{highest_mark_per_subject}")
print(f"The lowest mark per student is: \n{lowest_mark_per_subject}")
```

```python
#3.Overall Class Topper
total_marks_per_student = np.sum(marks,axis=1)
topper = np.argmax(total_marks_per_student)
class_topper_mark = np.max(total_marks_per_student)
print(f"The class topper {topper}'s mark is {class_topper_mark}")
```

```python
#4.Paas Count per subject
pass_count_per_subject = marks >=40
pass_count =np.sum(pass_count_per_subject,axis=0)
print(f"The number of pass count per subject is {pass_count}")
```
