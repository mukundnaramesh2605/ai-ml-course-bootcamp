```python
import numpy as np
```

```python
np.random.seed(25)
sales = np.random.randint(100,2000,size=(12,5))
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
products = ["Product A","Product B","Product C","Product D","Product E"]
sales
```

```python
#1.Total & average revenue by product and by month

total_revenune_by_product = np.sum(sales,axis=0)
total_revenune_by_month = np.sum(sales,axis=1)
avg_revenune_by_product = np.mean(sales,axis=0)
avg_revenune_by_month = np.mean(sales,axis=1)
print(f"The total revenue by product is {total_revenune_by_product}")
print(f"The total revenue by month is {total_revenune_by_month}")
print(f"The Average revenue by product is {avg_revenune_by_product}")
print(f"The Average revenue by month is {avg_revenune_by_month}")
```

```python
#2.Best/worst performing product
best_performing_product = products[np.argmax(total_revenune_by_product)]
worst_performing_product = products[np.argmin(total_revenune_by_product)]
print(f"The best performing product is {best_performing_product} with the sale of ₹{np.max(total_revenune_by_product)}")
print(f"The worst performing product is {worst_performing_product} with the sale of ₹{np.min(total_revenune_by_product)}")
#
```

```python
#3.Month with highest sales
month_with_the_highest_sales = np.argmax(total_revenune_by_month)
print(f"The month with the highest sales is {months[month_with_the_highest_sales]} with the sale of ₹{np.max(total_revenune_by_month)}")
```

```python
month_with_the_highest_sales
```

```python
#Products below target (filtering)
products_below_target = sales <=1000
count_of_products_below_target = np.sum(products_below_target,axis=0)
print(f"The number of products below target is {count_of_products_below_target}")
```

```python
#5.Revenue growth % month-over-month
# current = months 1-11, previous = months 0-10
growth = ((total_revenune_by_month[1:] - total_revenune_by_month[:-1]) / total_revenune_by_month[:-1]) * 100
growth = np.round(growth, 2)

for i in range(len(growth)):
    print(f"{months[i]} -> {months[i+1]}: {growth[i]}%")
print(f"\nThe month with highest sale is {months[np.argmax(growth)]} with a growth of {np.max(growth)}%")
```

```python
#Ranking
ranks = np.argsort(-total_revenune_by_product,)
print(f"First in sales is {products[ranks[0]]}")
print(f"First five sales are {ranks[0:5]}")
```
