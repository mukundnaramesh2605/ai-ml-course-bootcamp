```python
# Read matrix A
a_rows, a_cols = map(int, input().split())
A = [list(map(int, input().split())) for _ in range(a_rows)]

# Read matrix B
b_rows, b_cols = map(int, input().split())
B = [list(map(int, input().split())) for _ in range(b_rows)]

# Multiply: C[i][j] = sum of A[i][k] * B[k][j]
C = [[0] * b_cols for _ in range(a_rows)]
for i in range(a_rows):
    for j in range(b_cols):
        for k in range(a_cols):
            C[i][j] += A[i][k] * B[k][j]

# Print result
for row in C:
    print(*row)
```

```python
import math

n = int(input())
stiff = False
for a in range(2, n // 2 + 1):
    b = n - a
    if b > 1 and math.gcd(a, b) == 1:
        stiff = True
        break

print("Stiff" if stiff else "Not Stiff")
```

```python
z = int(input())
arr = list(map(int, input().split()))

def is_finest(n):
    t = 1
    while t**3 + (t+1)**3 <= n:
        if t**3 + (t+1)**3 == n:
            return True
        t += 1
    return False

finest = sorted([x for x in arr if is_finest(x)])
print(*finest if finest else [-1])
```
