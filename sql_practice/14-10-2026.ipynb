```python
import sqlite3, pandas as pd
con = sqlite3.connect("practice.db")
con.executescript(open("practice_setup.sql").read())
```

```python
"""-- 1. Rank Everyone by Salary  [AI screen / window]
-- Show name, salary, and each employee's rank by salary (highest = 1),
-- across the whole company. Keep all rows."""
query1 = """Select name,salary ,RANK() over(ORDER BY SALARY DESC) from employees"""
pd.read_sql(query1,con) 
```

```python
"""-- 2. Rank Within Department  [AI screen / window]
-- Show name, dept_id, salary, and each employee's salary rank WITHIN
-- their own department (highest = 1 per dept).
-- (PARTITION BY dept_id)"""
query2 = """Select name,dept_id,salary ,RANK() over(PARTITION BY dept_id ORDER BY SALARY DESC) from employees order by dept_id ASC"""
pd.read_sql(query2,con) 
```
