```python
users = {
    "Alice": ["Python", "ML", "AI"],
    "Bob": ["Python", "AI"],
    "Charlie": ["Java", "Spring"],
    "David": ["Python", "ML"]
}

def recommend(target_user):
    # Safe check if user doesn't exist
    if target_user not in users:
        return []
        
    target_interests = set(users[target_user])
    scores = {}
    
    # Calculate the intersection size for everyone else
    for user, interests in users.items():
        if user == target_user:
            continue
            
        shared_interests = target_interests.intersection(set(interests))
        if shared_interests:
            scores[user] = len(shared_interests)
            
    if not scores:
        return []
        
    # Find the highest intersection count
    max_score = max(scores.values())
    
    # Return all users who match that maximum score
    return [user for user, score in scores.items() if score == max_score]

# Testing the function
print(recommend("Bob"))
```

```python
rows = [
    "name,age,city",
    "Alice,25,Chennai",
    "Bob,30,Bangalore",
    "Charlie,22,Hyderabad"
]

# 1. Split the first string into a list of header keys
headers = rows[0].split(',')

# 2. Map headers to values for all remaining data rows
parsed_data = [dict(zip(headers, row.split(','))) for row in rows[1:]]

print(parsed_data)
```

```python
expenses = [
("Food",300),
("Travel",100),
("Food",200),
("Rent",1000),
("Travel",150)
]
expenses_sum = {}
for i,j in expenses:
    try:
        expenses_sum[i] = expenses_sum[i]+j
    except:
        expenses_sum[i] = j
expenses_sum
```
