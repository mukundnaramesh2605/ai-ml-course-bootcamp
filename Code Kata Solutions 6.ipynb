```python
def is_prime_simple(n: int) -> bool:
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
n = int(input())
print(is_prime_simple(n))
```

```python
rows, cols = map(int, input().split())
arr = [list(map(int, input().split())) for _ in range(rows)]
product = 1
seen = set()
for row in range(rows):
    for col in range(cols):
        if (row == col or row + col == rows - 1) and (row, col) not in seen:
            product *= arr[row][col]
            seen.add((row, col))
print(product)
```

```python
n = int(input())
arr = sorted(map(int, input().split()))
left, right = 0, n - 1
best_sum = float('inf')
best_pair = (arr[0], arr[1])

while left < right:
    s = arr[left] + arr[right]
    if abs(s) < abs(best_sum):
        best_sum = s
        best_pair = (arr[left], arr[right])
    if s < 0:
        left += 1
    elif s > 0:
        right -= 1
    else:
        break

print(best_pair[0], best_pair[1])
```
