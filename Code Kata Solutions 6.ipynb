```python
def is_prime_simple(n: int) -> bool:
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
n = int(input())
print(is_prime_simple(n))
```

```python
n, m = map(int, input().split())
list1 = input().split()
list2 = input().split()
result = []
for i in range(min(n, m)):
    result.append(list1[i] + list2[i])
# leftovers
if n > m:
    result.extend(list1[m:])
else:
    result.extend(list2[n:])
print(*result)
```
