```python
n = int(input())
grid = [input() for _ in range(n)]
plants = 0
weeds = 0
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            if grid[i][j] == '.':
                plants += 1
            elif grid[i][j] == '#':
                weeds += 1

print(plants, weeds)
```
