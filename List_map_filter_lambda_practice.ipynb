```python
words = ['pandas', 'numpy', 'scikit', 'matplotlib', 'seaborn', 'plotly', 'keras']
nums = [12, 7, 45, 3, 88, 21, 60, 9, 34, 17]
prices = {'laptop': 55000, 'mouse': 800, 'monitor': 12000, 'keyboard': 2500, 'cable': 300}
people = [('Arun', 28, 'Chennai'), ('Divya', 34, 'Bangalore'), ('Karthik', 22, 'Chennai'),
          ('Meera', 41, 'Mumbai'), ('Ravi', 29, 'Bangalore')]
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
```

```python
#1. From words, build a list of words longer than 6 characters, uppercased.
words_lst = list(filter(lambda x :  & (len(x) > 6),words))
words_lst
```

```python
#2. From nums, build a list where every even number is squared and every odd number is left as is.
nums_lst = [x**2 if x%2 == 0 else x for x in nums]
nums_lst
```

```python
#3. From prices, build a dict of items costing more than 1000, with values converted to a discounted price at 15 percent off, 
#rounded to the nearest integer.
prices_dict = {k:round(i*0.85) for k,i in prices.items() if i>1000}
prices_dict
```

```python
#4. Invert prices into a dict mapping price to item name.
inverse_prices_dict = {i:k for k,i in prices.items()}
inverse_prices_dict
```

```python
#5. Flatten matrix into a single list, then build a second list containing only the diagonal elements. 
#Do the diagonal with a comprehension, not indexing three times.
flattened_matrix = [item for items in matrix for item in items]
diagonal_matrix = [matrix[i][i] for i in range(len(matrix))]
flattened_matrix,diagonal_matrix
```

```python
#6. From people, build a dict grouping names by city. 
#Output should look like {'Chennai': ['Arun', 'Karthik'], ...}. Comprehension is allowed but a loop is fine here too.
grouped_by_city = {city: [name for name, age, c in people if c == city] for name, age, city in people}
grouped_by_city
```

```python
#7. Use filter and a lambda to get all numbers in nums divisible by 3. Then write the same thing as a list comprehension and note which reads better.
nums_filter_by_3 = list(filter(lambda x: x%3==0,nums))
print(nums_filter_by_3)
nums_list_by_3 = [x for x in nums if x%3 == 0]
nums_filter_by_3
```

```python
#8. Use map and a lambda to convert words into a list of (word, length) tuples. Then do it with a comprehension.
words_length_map = list(map(lambda x: (x,len(x)),words))
print(words_length_map)
words_length_list = [(x,len(x)) for x in words]
words_length_list
```

```python
#9. Sort people by age descending using sorted with a lambda key. Then sort by city first, age second.
sorted_people_age = sorted(people,reverse=True,key=lambda x:x[1])
sorted_people_age
```

```python
sorted_people_city = sorted(people,key=lambda x:(x[2],x[1]))
sorted_people_city
```

```python
#10. From nums, build a dict mapping each number to a label: 'low' under 20, 'mid' 20 to 50, 'high' above 50. Single dict comprehension with a nested conditional.
nums_mapping = {x: "high" if x > 50 else "mid" if x >= 20 else "low" for x in nums}
nums_mapping
```

```python
logs = ['INFO:login:ok', 'ERROR:db:timeout', 'WARN:cache:miss',
        'ERROR:api:500', 'INFO:logout:ok', 'ERROR:db:refused']
scores = {'Arun': [78, 92, 65], 'Divya': [88, 91, 95],
          'Karthik': [45, 58, 62], 'Meera': [90, 72, 84]}
items = [('pen', 10, 3), ('book', 250, 1), ('bag', 899, 2),
         ('mug', 150, 4), ('lamp', 1200, 1)]
sentence = 'the quick brown fox jumps over the lazy dog the end'
nested = {'a': {'x': 1, 'y': 2}, 'b': {'x': 3, 'y': 4}, 'c': {'x': 5, 'y': 6}}
```

```python
import re
#1. From logs, build a list of just the component names (middle field) for entries at ERROR level.
["".join(re.findall(r"ERROR:(\w+)",strng)) for strng in logs if re.findall(r"ERROR:(\w+:\w+)",strng)]
```

```python
#2. From logs, build a dict counting how many entries exist per level. No Counter.
{k:logs.count(k) for k in logs}
```

```python
#From scores, build a dict of name to average, rounded to 1 decimal.
{name:round(sum(score)/len(score),1) for name,score in scores.items()}
```

```python
#4. From scores, build a dict containing only people whose lowest score is above 60.
{name:score for name,score in scores.items() if min(score)>60}
```

```python
#5. From items, compute total spend as a single expression using sum and a comprehension. Each item is (name, price, qty).
sum(price*qty for name,price,qty in items)
```

```python
#6. From items, build a dict of name to total cost, sorted by total cost descending. Result must stay a dict.
dict(
    sorted(
        {name: price * qty for name, price, qty in items}.items(),
        key=lambda item: item[1],
        reverse=True
    )
)
```

```python
#7. From sentence, build a dict of word to frequency, then find the most frequent word using max with a lambda key.
sentence = 'the quick brown fox jumps over the lazy dog the end'
frequency_dict ={word:sentence.count(word) for word in sentence.split()}
max_frequency = max(frequency_dict.values())
"".join(k for k,i in frequency_dict.items() if i==max_frequency)
```
