```python
words = ['pandas', 'numpy', 'scikit', 'matplotlib', 'seaborn', 'plotly', 'keras']
nums = [12, 7, 45, 3, 88, 21, 60, 9, 34, 17]
prices = {'laptop': 55000, 'mouse': 800, 'monitor': 12000, 'keyboard': 2500, 'cable': 300}
people = [('Arun', 28, 'Chennai'), ('Divya', 34, 'Bangalore'), ('Karthik', 22, 'Chennai'),
          ('Meera', 41, 'Mumbai'), ('Ravi', 29, 'Bangalore')]
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
```

```python
#1. From words, build a list of words longer than 6 characters, uppercased.
words_lst = list(filter(lambda x : len(x) > 6,words))
words_lst
```

```python
#2. From nums, build a list where every even number is squared and every odd number is left as is.
nums_lst = [x**2 if x%2 == 0 else x for x in nums]
nums_lst
```

```python
#3. From prices, build a dict of items costing more than 1000, with values converted to a discounted price at 15 percent off, 
#rounded to the nearest integer.
prices_dict = {k:round(i*0.15) for k,i in prices.items()}
prices_dict
```

```python
#4. Invert prices into a dict mapping price to item name.
inverse_prices_dict = {i:k for k,i in prices.items()}
inverse_prices_dict
```
