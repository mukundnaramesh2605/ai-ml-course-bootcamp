```python
# ============================================================================
# 10 EASY Practice Problems: map(), filter(), list comprehensions
# Quick wins, fundamental patterns
# ============================================================================

# Problem 1: Map - add 5 to each number
# Input: [1, 2, 3, 4, 5]
# Expected: [6, 7, 8, 9, 10]
def easy_1():
    nums = [1, 2, 3, 4, 5]
    # Your code here
    return list(map(lambda x: x+5, nums))

# Test:
print("Easy Problem 1:")
print(easy_1())
print()
```

```python
# ============================================================================

# Problem 2: Filter - keep only numbers > 10
# Input: [5, 12, 8, 15, 3, 20]
# Expected: [12, 15, 20]
def easy_2():
    nums = [5, 12, 8, 15, 3, 20]
    # Your code here
    return list(filter(lambda x: x>10,nums))

# Test:
print("Easy Problem 2:")
print(easy_2())
print()
```

```python
# Problem 3: List comprehension - square all numbers
# Input: [1, 2, 3, 4, 5]
# Expected: [1, 4, 9, 16, 25]
def easy_3():
    nums = [1, 2, 3, 4, 5]
    # Your code here
    return [x**2 for x in nums]

# Test:
print("Easy Problem 3:")
print(easy_3())
print()
```

```python
# Problem 4: Map - uppercase all strings
# Input: ['apple', 'banana', 'cherry']
# Expected: ['APPLE', 'BANANA', 'CHERRY']
def easy_4():
    fruits = ['apple', 'banana', 'cherry']
    # Your code here
    return [x.upper() for x in fruits]

# Test:
print("Easy Problem 4:")
print(easy_4())
print()
```
