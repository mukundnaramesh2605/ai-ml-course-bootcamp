```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 📋 1. Create the Base Dataframe
data = {
    'Square_Feet': [1500, 2200, 1100],
    'Year_Built': [2005, 1998, 2015],
    'Sale_Price': [250000, 380000, 195000],
    'Days_on_Market': [14, 22, 8]
}
df = pd.DataFrame(data)

# =====================================================================
# 1️⃣ UNIVARIATE REGRESSION (1 Input -> 1 Output)
# =====================================================================
# Inputs must be a 2D Dataframe [[]], Outputs can be a 1D Series []
X_uni = df[['Square_Feet']]
y_uni = df['Sale_Price']

model_uni = LinearRegression()
model_uni.fit(X_uni, y_uni)

print("--- Univariate ---")
print("W (Weight):", model_uni.coef_)       # Output: 1 single scalar element inside an array
print("b (Bias):", model_uni.intercept_)   # Output: 1 single scalar float

# =====================================================================
# 2️⃣ MULTIPLE REGRESSION (Many Inputs -> 1 Output)
# =====================================================================
X_mult = df[['Square_Feet', 'Year_Built']]
y_mult = df['Sale_Price']

model_mult = LinearRegression()
model_mult.fit(X_mult, y_mult)

print("\n--- Multiple ---")
print("W (Weights):", model_mult.coef_)     # Output: 1D array with 2 weights [w1, w2]
print("b (Bias):", model_mult.intercept_)   # Output: 1 single scalar float

# =====================================================================
# 3️⃣ MULTIVARIATE REGRESSION (Many Inputs -> Many Outputs)
# =====================================================================
# Both inputs and outputs must be 2D Dataframes [[]]
X_mvar = df[['Square_Feet', 'Year_Built']]
Y_mvar = df[['Sale_Price', 'Days_on_Market']]

model_mvar = LinearRegression()
model_mvar.fit(X_mvar, Y_mvar)

print("\n--- Multivariate ---")
print("W (Weight Matrix):\n", model_mvar.coef_)      # Output: 2x2 Matrix grid 
print("b (Bias Vector):", model_mvar.intercept_)     # Output: 1D array with 2 biases [b1, b2]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load data
tips = sns.load_dataset("tips")

# 2. Create the figure and subplots
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# --- LEFT PLOT: Standard Boxplot ---
sns.boxplot(
    data=tips, 
    x="day", 
    y="total_bill", 
    ax=axs[0],
    medianprops={"color": "red", "linewidth": 2}
)
axs[0].set_title("Standard Boxplot")

# --- RIGHT PLOT: Boxplot with Median Annotations ---
sns.boxplot(
    data=tips, 
    x="day", 
    y="total_bill", 
    ax=axs[1],
    medianprops={"color": "red", "linewidth": 2}
)
axs[1].set_title("Boxplot with Median Labels")

# 3. Calculate and annotate medians on the second plot
# Group by the X-axis variable and find the median of the Y-axis variable
medians = tips.groupby("day", observed=False)["total_bill"].median()

# Loop through each group to place the text label
for i, day in enumerate(medians.index):
    median_val = medians[day]
    
    axs[1].text(
        x=i,                         # X-coordinate matches the box index (0, 1, 2...)
        y=median_val + 1.0,          # Y-coordinate slightly above the median line
        s=f"{median_val:.2f}",       # The formatted text string
        horizontalalignment="center", # Center text horizontally over the box
        size=10,                     # Font size
        color="black",               # Font color
        weight="semibold"            # Bold font style
    )

plt.tight_layout()
plt.show()
```
