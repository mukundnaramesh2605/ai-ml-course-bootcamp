```python
import math

deg, n = map(int, input().split())
x = math.radians(deg)

result = 0
for i in range(n):
    power = 2 * i
    term = ((-1)**i) * (x**power) / math.factorial(power)
    result += term

print(f"{result:.2f}")
```

```python
row, col = map(int, input().split())

full = "* " * col
full = full.rstrip()
hollow = "* " + "  " * (col - 2) + "*"

for i in range(row):
    if i == 0 or i == row - 1:
        print(full)
    else:
        print(hollow)
```

```python
n = int(input())
first_row = "1"
space_count = 0
last_row = "".join(str(x) for x in range(1,n+1))
for i in range(1,n+1):
    if i==1:
        print(first_row)
    elif i!=1 and i != n:
        arr = [x for x in range(1,i+1)]
        print(str(min(arr))+" "*space_count+str(max(arr)))
        space_count+=1
    elif i == n:
        print(last_row)
        
```

```python
n = int(input())
last_row = "1"
space_count = n-3
first_row = "".join(str(x) for x in range(1,n+1))
for i in range(n,0,-1):
    if i==n:
        print(first_row)
    elif i!=1 and i != n:
        arr = [x for x in range(1,i+1)]
        print(str(min(arr))+" "*space_count+str(max(arr)))
        space_count-=1
    elif i == 1:
        print(last_row)
        
```
