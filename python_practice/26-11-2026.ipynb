```python
# 1. Replace Each Word With Its Length  [easy / string]
# Return the string with every word replaced by its length. KEEP SPACES EXACTLY as given.
# "hi there" -> "2 5"
def word_lengths(s):
    return " ".join([str(len(i)) if i.isalpha() else i for i in s.split(" ")])

print(repr(word_lengths("hi there")))          # -> '2 5'
print(repr(word_lengths("I  love   python")))  # -> '1  4   6'   <- double/triple spaces must survive
print(repr(word_lengths("")))                  # -> ''
print(repr(word_lengths("a")))                 # -> '1'
print(repr(word_lengths("   ")))               # -> '   '        <- spaces only, no words
```

```python
# 2. Age Buckets  [easy / dict]
# Return a dict with EXACT keys "child", "adult", "senior" (all three ALWAYS present).
# child: age < 18, adult: 18 to 59 INCLUSIVE, senior: 60 and above.
# [5, 30, 70] -> {'child': 1, 'adult': 1, 'senior': 1}
def age_buckets(ages):
    grouped = {'child': 0, 'adult': 0, 'senior': 0}
    for i in ages:
        if i<18:
            grouped['child']+=1
        elif i>=18 and i<=59:
            grouped['adult']+=1
        else:
            grouped['senior']+=1
    return grouped 

print(age_buckets([5, 30, 70]))       # -> {'child': 1, 'adult': 1, 'senior': 1}
print(age_buckets([]))                # -> {'child': 0, 'adult': 0, 'senior': 0}   <- keys still present
print(age_buckets([17, 18, 59, 60]))  # -> {'child': 1, 'adult': 2, 'senior': 1}   <- boundaries
print(age_buckets([45]))              # -> {'child': 0, 'adult': 1, 'senior': 0}
print(age_buckets([60, 60, 10]))      # -> {'child': 1, 'adult': 0, 'senior': 2}
```

```python
# 3. Words Only In First Sentence  [medium / set]
# Return DISTINCT words of a that do NOT appear in b, in order of first appearance in a.
# Case-SENSITIVE. Words are separated by spaces.
# "the cat sat on the mat", "the dog sat" -> ['cat', 'on', 'mat']
def only_in_first(a, b):
    words_a = a.split()
    words_b = set(b.split())
    seen = set()
    result = []

    for word in words_a:
        if word not in words_b and word not in seen:
            result.append(word)
            seen.add(word)

    return result

print(only_in_first("the cat sat on the mat", "the dog sat"))  # -> ['cat', 'on', 'mat']
print(only_in_first("", "a b"))                                # -> []
print(only_in_first("a a b b", "c"))                           # -> ['a', 'b']    <- DISTINCT
print(only_in_first("Cat cat", "cat"))                         # -> ['Cat']       <- case-sensitive
print(only_in_first("x y", "x y"))                             # -> []            <- no match
```
