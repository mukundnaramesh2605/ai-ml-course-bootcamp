```python
# 1. Reverse Words, Keep Order of Letters  [Service / string]
# Reverse the order of words in a sentence.
# Example: "data science is fun" -> "fun is science data"
s = "AI will change the world"
print(" ".join(s.split()[::-1]))
```

```python
# 2. Count Each Word's Length  [Service / string]
# Return a dict mapping each word to its length.
# Example: "I love python" -> {"I": 1, "love": 4, "python": 6}
text = "machine learning is powerful"
{word:len(word) for word in text.split()}
```

```python
# 3. Find Common Elements in Two Lists  [Service / sets]
# Return the values present in both lists, no duplicates.
# Example: [1,2,3,4], [3,4,5,6] -> [3, 4]
a = [1, 2, 3, 4, 5]
b = [4, 5, 6, 7, 8]
print(set(a)&set(b))
```

```python
# 4. Merge Two Dicts, Keep Higher Value on Clash  [Service / dict]
# If a key is in both, keep the larger value.
# Example: {"a":1,"b":5}, {"b":3,"c":2} -> {"a":1,"b":5,"c":2}
d1 = {"x": 10, "y": 5}
d2 = {"y": 8, "z": 3}
result = d1.copy()
for k,v in d2.items():
    result[k] = max(result.get(k,0),v)
result
```

```python
# 5. Remove All Vowels  [Service / string]
# Return the string with every vowel removed.
# Example: "hello world" -> "hll wrld"
word = "artificial intelligence"
print("".join([i for i in word if i.lower() not in 'aeiou']))
```

```python
# 6. Group Numbers by Sign  [Service / dict + logic]
# Return a dict with "positive", "negative", "zero" lists.
# Example: [3, -1, 0, 5, -2] -> {"positive":[3,5], "negative":[-1,-2], "zero":[0]}
nums = [4, -3, 0, 7, -1, 0, 2]
grouped = {'positive':[],'negative':[],'zero':[]}
for i in nums:
    if i==0:
        grouped['zero'].append(i)
    elif i > 0:
        grouped['positive'].append(i)
    else:
        grouped['negative'].append(i)
grouped
```

```python
# 7. Check if All Characters Are Unique  [Service / sets]
# Return True if no character repeats in the string.
# Example: "python" -> True, "hello" -> False
check = "数据"  # just kidding, use: "learning"
len(set(check)) == len(check)
```

```python
# 8. Count Frequency of Each Word  [Service / dict]
# Return a dict of word -> how many times it appears, case-insensitive.
# Example: "the cat the dog" -> {"the":2, "cat":1, "dog":1}
from collections import Counter
sentence = "AI is AI and AI is here"
Counter(sentence.lower().split())
```

```python
# 1. Invert a Dictionary  [Service / dict]
# Swap keys and values. Assume values are unique.
# Example: {"a": 1, "b": 2} -> {1: "a", 2: "b"}
d = {"apple": 3, "banana": 5, "cherry": 7}
{v:k for k,v in d.items()}
```

```python
# 2. Find Elements in A but Not in B  [Service / sets]
# Return values that are in the first list but not the second.
# Example: [1,2,3,4], [3,4,5] -> [1, 2]
a = [1, 2, 3, 4, 5, 6]
b = [4, 5, 6, 7]
list(set(a)-set(b))
```

```python
# 3. Capitalize First Letter of Each Word  [Service / string]
# Without using .title() or .capitalize().
# Example: "hello world" -> "Hello World"
s = "machine learning rocks"
" ".join([i[0].upper()+i[1:] for i in s.split()])
```

```python
# 4. Sum Values in a Dictionary  [Service / dict]
# Return the total of all values.
# Example: {"a": 10, "b": 20, "c": 5} -> 35
scores = {"math": 80, "science": 90, "english": 75}
sum(scores.values())
```

```python
# 5. Count How Many Times Each Length Appears  [Service / dict + logic]
# Given words, return a dict of word-length -> count of words with that length.
# Example: ["a", "bb", "cc", "ddd"] -> {1: 1, 2: 2, 3: 1}
words = ["AI", "ML", "data", "code", "fun", "x"]
dict_len = {}
for word in words:
    dict_len[len(word)] = dict_len.get(len(word),0)+1
dict_len
```
