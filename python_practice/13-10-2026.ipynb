```python
# 3. Find Common Elements in Two Lists  [Service / sets]
# Return the values present in both lists, no duplicates.
# Example: [1,2,3,4], [3,4,5,6] -> [3, 4]
a = [1, 2, 3, 4, 5]
b = [4, 5, 6, 7, 8]
print(set(a)&set(b))
```

```python
# 4. Merge Two Dicts, Keep Higher Value on Clash  [Service / dict]
# If a key is in both, keep the larger value.
# Example: {"a":1,"b":5}, {"b":3,"c":2} -> {"a":1,"b":5,"c":2}
d1 = {"x": 10, "y": 5}
d2 = {"y": 8, "z": 3}
result = d1.copy()
for k,v in d2.items():
    result[k] = max(result.get(k,0),v)
result
```

```python
# 5. Remove All Vowels  [Service / string]
# Return the string with every vowel removed.
# Example: "hello world" -> "hll wrld"
word = "artificial intelligence"
print("".join([i for i in word if i.lower() not in 'aeiou']))
```

```python
# 6. Group Numbers by Sign  [Service / dict + logic]
# Return a dict with "positive", "negative", "zero" lists.
# Example: [3, -1, 0, 5, -2] -> {"positive":[3,5], "negative":[-1,-2], "zero":[0]}
nums = [4, -3, 0, 7, -1, 0, 2]
grouped = {'positive':[],'negative':[],'zero':[]}
for i in nums:
    if i==0:
        grouped['zero'].append(i)
    elif i > 0:
        grouped['positive'].append(i)
    else:
        grouped['negative'].append(i)
grouped
```

```python
# 5. Count How Many Times Each Length Appears  [Service / dict + logic]
# Given words, return a dict of word-length -> count of words with that length.
# Example: ["a", "bb", "cc", "ddd"] -> {1: 1, 2: 2, 3: 1}
words = ["AI", "ML", "data", "code", "fun", "x"]
dict_len = {}
for word in words:
    dict_len[len(word)] = dict_len.get(len(word),0)+1
dict_len
```

```python
# 1. Safe Division  [Service / exceptions]
# Write a function that divides a by b. If b is 0, catch the error and
# return the string "Cannot divide by zero" instead of crashing.
# Example: safe_divide(10, 2) -> 5.0 ;  safe_divide(5, 0) -> "Cannot divide by zero"
def safe_divide(a, b):
    try:
        return a/b
    except ZeroDivisionError as e:
        return f"Error:{e}"
safe_divide(1,0)
```

```python
# 2. Safe Int Conversion  [Service / exceptions]
# Convert a string to an int. If it is not a valid number, return None
# instead of crashing.
# Example: to_int("42") -> 42 ;  to_int("abc") -> None
def to_int(s):
    try:
        return int(s)
    except ValueError as e:
        return f'Error:{e}'
to_int('abc')
```

```python
# 3. Function With Default and *args  [Service / functions]
# Write a function that returns the sum of any number of arguments,
# plus a bonus that defaults to 0.
# Example: total(1, 2, 3) -> 6 ;  total(1, 2, bonus=10) -> 13
def total(*nums, bonus=0):
    return sum(nums)+bonus
total(1,2,3,4,bonus=5)
```

```python
# 4. Count Successes and Failures  [AI screen / exceptions]
# Given a list of strings, try to convert each to a float. Return a dict
# with how many converted successfully and how many failed.
# Example: ["1.5", "abc", "3", "x"] -> {"success": 2, "failed": 2}
def count_conversions(items):
    count = {}
    for i in items:
        try:
            float(i)
            count['success']=count.get('success',0)+1
        except:
            count['failed']=count.get('failed',0)+1
    return count
count_conversions(["1.5", "abc", "3", "x"])
```

```python
def count_up(n):
    for i in range(1, n+1):
        yield i          # hand back i, then pause here

for x in count_up(3):
    print(x)             # 1, then 2, then 3
```
