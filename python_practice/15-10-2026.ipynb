```python
# 1. Count Consonants  [Service / string]
# Return how many letters are consonants (letters that are not vowels).
# Example: "hello" -> 3
def count_consonants(s):
    return sum([1 for i in s if i.lower() not in 'aeiou'])

print(count_consonants("hello"))# -> 3 ; 
print(count_consonants("aeiou"))# -> 0

```

```python
# 2. Rotate List Right by K  [Service / logic]
# Rotate a list to the right by k positions.
# Example: ([1,2,3,4,5], 2) -> [4,5,1,2,3]
def rotate_right(nums, k):
    return nums[-k:]+nums[:-k]

print(rotate_right([1,2,3,4,5], 2))# -> [4,5,1,2,3]
```

```python
# 3. Is Palindrome  [Service / string]
# Return True if the string reads the same forwards and backwards,
# ignoring case. Example: "Madam" -> True
def is_palindrome(s):
    return s.lower() == s.lower()[::-1]

print(is_palindrome("Madam"))# -> True ; 
print(is_palindrome("hello"))# -> False
```

```python
# 4. Count Occurrences  [Service / logic]
# Return a dict of how many times each value appears.
# Example: [1,2,2,3,3,3] -> {1:1, 2:2, 3:3}
from collections import Counter
def count_occurrences(nums):
    counts = Counter(nums)
    return dict(counts)
print(count_occurrences([1,2,2,3,3,3]))# -> {1:1, 2:2, 3:3}
```

```python
# 5. Longest Word  [Service / string]
# Return the longest word in a sentence. On a tie, the first one.
# Example: "I love machine learning" -> "learning"... "machine" ties, first wins
def longest_word(sentence):
    lst = sorted(sentence.split(),key=len,reverse=True)
    return lst[0]
    
print(longest_word("I love data science and Learning Machine"))# -> "science"
```

```python
# 6. Normalize to 0-1  [AI screen / numpy]
# Min-max normalize a 1D array to the range [0, 1].
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
# write it on arr
arr = (arr-arr.min())/(arr.max()-arr.min())
arr
```

```python
# 7. Count Even Numbers  [AI screen / numpy]
# Return HOW MANY elements are even. (a count)
import numpy as np
data = np.array([1, 2, 3, 4, 5, 6, 7, 8])
# write it on data
count = np.sum(data % 2 == 0)
print(count)  # Output: 4
```

```python
# 13. Max Salary Per Department  [AI screen / groupby]
# Return each department and its highest salary.
import pandas as pd
staff = pd.DataFrame({
    "dept": ["IT","HR","IT","HR","Fin"],
    "salary": [60000, 45000, 70000, 48000, 90000],
})

staff.groupby('dept').max()
```

```python
# 14. Fill Missing With Mean  [AI screen / pandas]
# Fill missing scores with the MEAN of the score column.
import pandas as pd
import numpy as np
scores = pd.DataFrame({"name": ["A","B","C","D"], "score": [80, np.nan, 60, np.nan]})
scores['score'] = scores['score'].fillna(scores['score'].mean())
scores
```
