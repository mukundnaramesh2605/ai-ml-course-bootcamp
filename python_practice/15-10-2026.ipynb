```python
# 1. Count Consonants  [Service / string]
# Return how many letters are consonants (letters that are not vowels).
# Example: "hello" -> 3
def count_consonants(s):
    return sum([1 for i in s if i.lower() not in 'aeiou'])

print(count_consonants("hello"))# -> 3 ; 
print(count_consonants("aeiou"))# -> 0

```

```python
# 2. Rotate List Right by K  [Service / logic]
# Rotate a list to the right by k positions.
# Example: ([1,2,3,4,5], 2) -> [4,5,1,2,3]
def rotate_right(nums, k):
    return nums[-k:]+nums[:-k]

print(rotate_right([1,2,3,4,5], 2))# -> [4,5,1,2,3]
```

```python
# 3. Is Palindrome  [Service / string]
# Return True if the string reads the same forwards and backwards,
# ignoring case. Example: "Madam" -> True
def is_palindrome(s):
    return s.lower() == s.lower()[::-1]

print(is_palindrome("Madam"))# -> True ; 
print(is_palindrome("hello"))# -> False
```

```python
# 4. Count Occurrences  [Service / logic]
# Return a dict of how many times each value appears.
# Example: [1,2,2,3,3,3] -> {1:1, 2:2, 3:3}
from collections import Counter
def count_occurrences(nums):
    counts = Counter(nums)
    return dict(counts)
print(count_occurrences([1,2,2,3,3,3]))# -> {1:1, 2:2, 3:3}
```

```python
# 5. Longest Word  [Service / string]
# Return the longest word in a sentence. On a tie, the first one.
# Example: "I love machine learning" -> "learning"... "machine" ties, first wins
def longest_word(sentence):
    lst = sorted(sentence.split(),key=len,reverse=True)
    return lst[0]
    
print(longest_word("I love data science and Learning Machine"))# -> "science"
```

```python
# 1. Uppercase Result  [concept]
# Write a decorator that uppercases whatever string the function returns.
# @upper
# def greet(name): return f"hi {name}"
# greet("sam") -> "HI SAM"
def upper(func):
    def wrapper(*args):
        result = func(*args)
        return result.upper()
    return wrapper

@upper
def greet(name):
    return f"hi {name}"
greet("Mukundan")
```

```python
# 2. Repeat N Times  [intermediate]
# Write a decorator repeat(n) that runs the function n times and returns a
# list of the results. (a decorator that TAKES an argument)
# @repeat(3)
# def roll(): return 7
# roll() -> [7, 7, 7]
import functools

def repeat(n):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            return [func(*args, **kwargs) for _ in range(n)]
        return wrapper
    return decorator
@repeat(4)
def roll():
    return 7
roll()
```

```python
# 3. Cache Results  [intermediate]
# Write a decorator that caches results so the same argument isn't recomputed.
# Track a dict of arg -> result. On a repeat call, return the cached value.
# Add a print("computing") inside the function so you can SEE it only runs once.
from functools import wraps

def memoize(func):
    cache = {}

    @wraps(func)
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]

    return wrapper


@memoize
def square(n):
    print("computing")
    return n * n
square(4)
```

```python
# 4. Suppress Errors  [concept]
# Write a decorator that catches ANY exception the function raises and returns
# None instead of crashing.
# @safe
# def bad(): return 1/0
# bad() -> None (no crash)
from functools import wraps

def safe(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except Exception:
            return None

    return wrapper

@safe
def bad():
    return
```

```python
# 5. Countdown Generator  [concept]
# Write a generator that yields n, n-1, ..., 1.
# list(countdown(3)) -> [3, 2, 1]
def countdown(n):
    for i in range(n,0,-1):
        yield i
m = countdown(3)
print(next(m))
print(next(m))
print(next(m))
print(next(m))
```

```python
# 6. Take First K  [concept]
# Given any generator and k, return the first k values as a list, using next().
# first_k(countdown(10), 3) -> [10, 9, 8]
def first_k(gen, k):
    result = []
    for _ in range(k):
        try:
            result.append(next(gen))
        except StopIteration:
            break
    return result
first_k(countdown(10),3)
```

```python
# 8. Retry Once  [intermediate]
# Call func(). If it raises, catch it and try ONE more time. If it fails again,
# return "failed". If either call succeeds, return its result.
from functools import wraps

def retry_once(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except Exception:
            try:
                return func(*args, **kwargs)
            except Exception:
                return "failed"

    return wrapper
```

```python
# 9. Validate Age  [Service / exceptions]
# Convert input to int. If it's not a number, raise ValueError("not a number").
# If it's negative, raise ValueError("negative age"). Else return the int.
def validate_age(x):
    try:
        val = int(x)
    except (ValueError, TypeError):
        raise ValueError("not a number")

    if val < 0:
        raise ValueError("negative age")

    return val
# validate_age("25") -> 25 ; validate_age("ab") -> raises "not a number"
# validate_age("-3") -> raises "negative age"
```

```python
# Q1. Group by First Letter  [Service / data handling]
# Return a dict mapping each first letter to the list of words starting with it,
# preserving input order. Assume all words are non-empty, lowercase.
# Example: ["apple","avocado","banana"] -> {"a":["apple","avocado"], "b":["banana"]}
from collections import defaultdict
def group_by_first(words):
    grouped = defaultdict(list)
    for i in words:
        grouped[i.lower()[0]].append(i)
    return dict(grouped)
print(group_by_first(["apple","avocado","banana"]))# -> {"a":["apple","avocado"],"b":["banana"]}
print(group_by_first([]))# -> {}
```

```python
# Q2. Most Common Word  [AI screen / text]
# Return the most frequent word in a sentence, case-insensitive. On a tie,
# the one that appears first.
# Example: "The cat the DOG the" -> "the"
from collections import Counter
def most_common_word(sentence):
    counts = Counter(sentence.split())
    return [x for x,v in counts.items() if v == max(counts.values())][0]
print(most_common_word("The cat the DOG the"))# -> "the"
print(most_common_word("a b a b"))# -> "a"
```

```python
# Q3. Decorator With Argument: limit  [AI screen / decorators]
# Write a decorator limit(max_len) that truncates the function's returned
# STRING to at most max_len characters.
# @limit(3)
# def greet(): return "hello"
# greet() -> "hel"
def limit(max_len):
    def decorator(func):
        def wrapper(*args,**kwargs):
            result = func(*args,**kwargs)
            result = result[:max_len]
            return result
        return wrapper
    return decorator

@limit(3)
def greet(s):
    return s
# test: greet() with limit(3) on "hello" -> "hel"
# test: limit(10) on "hi" -> "hi"  (shorter than limit, unchanged)

print(greet("Hello"))
@limit(10)
def greet(s):
    return s
print(greet('hi'))
```
