```python
# 1. Count Consonants  [Service / string]
# Return how many letters are consonants (letters that are not vowels).
# Example: "hello" -> 3
def count_consonants(s):
    return sum([1 for i in s if i.lower() not in 'aeiou'])

print(count_consonants("hello"))# -> 3 ; 
print(count_consonants("aeiou"))# -> 0

```

```python
# 2. Rotate List Right by K  [Service / logic]
# Rotate a list to the right by k positions.
# Example: ([1,2,3,4,5], 2) -> [4,5,1,2,3]
def rotate_right(nums, k):
    return nums[-k:]+nums[:-k]

print(rotate_right([1,2,3,4,5], 2))# -> [4,5,1,2,3]
```

```python
# 3. Is Palindrome  [Service / string]
# Return True if the string reads the same forwards and backwards,
# ignoring case. Example: "Madam" -> True
def is_palindrome(s):
    return s.lower() == s.lower()[::-1]

print(is_palindrome("Madam"))# -> True ; 
print(is_palindrome("hello"))# -> False
```

```python
# 4. Count Occurrences  [Service / logic]
# Return a dict of how many times each value appears.
# Example: [1,2,2,3,3,3] -> {1:1, 2:2, 3:3}
from collections import Counter
def count_occurrences(nums):
    counts = Counter(nums)
    return dict(counts)
print(count_occurrences([1,2,2,3,3,3]))# -> {1:1, 2:2, 3:3}
```
