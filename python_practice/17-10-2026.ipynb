```python
# 1. Second Largest  [Service / logic]
# Return the second largest DISTINCT number in a list.
# [4, 1, 7, 7, 3] -> 4   (7 is largest, 4 is second)
def second_largest(nums):
    unique = list(set(nums))
    first=second=float('-inf')
    for i in unique:
        if i > first:
            second=first
            first=i
        elif i > second and i != first:
            second = i
    return second if second != float('-inf') else None
print(second_largest([4,1,7,7,3]))# -> 4 ; s
print(second_largest([5,5]))# -> None
```

```python
# 2. Title Case Except Small Words  [Service / string]
# Capitalize each word EXCEPT "a", "an", "the", "of", "in"
# (those stay lowercase), but ALWAYS capitalize the FIRST word.
# "the lord of the rings" -> "The Lord of the Rings"
def smart_title(s):
    stop_words = ["a", "an", "the", "of", "in"]
    return str(s.split()[0].title())+" "+" ".join([i.title() if i not in stop_words else i for i in s.split()[1:]])
print(smart_title("the lord of the rings"))# -> "The Lord of the Rings"
```

```python
# 3. Count Chars, Ignore Spaces  [Service / dict]
# Return a dict of character -> count, case-insensitive, SKIPPING spaces.
# "A a b" -> {"a": 2, "b": 1}
from collections import Counter
def char_freq(s):
    counts = Counter(s.lower().split())
    return dict(counts)
print(char_freq("A a b"))# -> {"a": 2, "b": 1}
```

```python
# 4. Longest Run  [Service / walk-and-track]
# Return the length of the longest run of the SAME value in a row.
# [1,1,2,2,2,3] -> 3   (three 2s in a row)
def longest_run(nums):
    if len(nums)==1:
        return 1
    count = 0
    best = 0 
    for i in range(1,len(nums)):
        if nums[i]==nums[i-1]:
            count+=1
            best=max(count,best)
        else:
            count=1
    return best
print(longest_run([1,1,2,2,2,3]))# -> 3 ;
print(longest_run([]))# -> 0 ; 
print(longest_run([5]))# -> 1

```

```python
# 5. Merge Two Dicts, Sum Shared Keys  [AI screen / dict]
# Merge two dicts. If a key is in both, SUM the values.
# {"a":1,"b":2}, {"b":3,"c":4} -> {"a":1,"b":5,"c":4}
def merge_sum(d1, d2):
    result = d1.copy()
    for i in d2:
        result[i] = result.get(i,0)+d2[i]
    return result
print(merge_sum({"a":1,"b":2}, {"b":3,"c":4}))# -> {"a":1,"b":5,"c":4}
```
