```python
# 1. Sum of Even-Indexed Elements  [logic]
# Return the sum of elements at even indices (0, 2, 4, ...).
def sum_even_index(nums):
    return sum([nums[i] for i in range(len(nums)) if i%2 == 0])
print(sum_even_index([10, 1, 20, 2, 30]))   # -> 60   (10 + 20 + 30)
print(sum_even_index([]))      
```

```python
# 2. Count Consonants  [string]
# Return how many consonants are in the string. Letters only, case-insensitive.
def count_consonants(s):
    return sum([1 for i in s.lower() if i not in 'aeiou' and i.isalpha()])
print(count_consonants("Hello World"))   # -> 7
print(count_consonants("AEIOU"))         # -> 0
print(count_consonants("123"))           # -> 0
```

```python
# 3. Merge Two Dicts Summing  [dict]
# Merge two dicts, summing values on shared keys. Handle negatives and zero.
def merge_sum(d1, d2):
    result = d1.copy()
    for k,v in d2.items():
        result[k] = result.get(k,0)+v
    return result
print(merge_sum({"a": 2, "b": 3}, {"b": 4, "c": 1}))   # -> {'a': 2, 'b': 7, 'c': 1}
print(merge_sum({"a": 5}, {"a": -5}))                  # -> {'a': 0}
```

```python
# 4. Is Subsequence  [two-pointer]
# Return True if a is a subsequence of b (same order, not necessarily contiguous).
def is_subsequence(a, b):
    it = iter(b)
    return all(char in it for char in a)
print(is_subsequence("ace", "abcde"))   # -> True
print(is_subsequence("aec", "abcde"))   # -> False
print(is_subsequence("", "abc"))        # -> True
```

```python
# 5. First Repeated Element  [hashing]
# Return the first value that appears a second time, scanning left to right.
# None if all unique.
def first_repeated(nums):
    for i in nums:
        if nums.count(i) > 1:
            return i
    return None
print(first_repeated([3, 1, 3, 2, 1]))   # -> 3
print(first_repeated([1, 2, 3]))         # -> None
```

```python
# 6. Run-Length Decode  [string]
# "a3b2c1" -> "aaabbc". Each letter is followed by its single-digit count.
import re
def rle_decode(s):
    return "".join([k*int(v) for k,v in list(zip(re.findall(r'[a-zA-Z]',s),re.findall(r'[0-9]',s)))])
print(rle_decode("a3b2c1"))   # -> "aaabbc"
print(rle_decode("x1"))       # -> "x"
print(rle_decode(""))         # -> ""
```

```python
# 7. Rotate List Right  [logic]
# Rotate the list right by k steps. k can exceed the length.
def rotate(nums, k):
    return nums[-k:]+nums[:-k] 
print(rotate([1, 2, 3, 4, 5], 2))   # -> [4, 5, 1, 2, 3]
print(rotate([1, 2, 3], 4))         # -> [3, 1, 2]
print(rotate([], 3))                # -> []
```

```python
# 8. Count Words of Each Length  [dict]
# Return a dict: word length -> how many words have that length.
def length_counts(words):
    result = {}
    for i in words:
        length = len(i)
        result[length] = result.get(length,0)+1
    return result
print(length_counts(["hi", "to", "cat", "a"]))   # -> {2: 2, 3: 1, 1: 1}
print(length_counts([]))   
```

```python
# 9. Longest Common Prefix  [string]
# Return the longest string that is a prefix of ALL words. "" if none or empty.
def longest_common_prefix(words):
    result=[]
    for chars in zip(*words):
        if len(set(chars)) == 1:
            result.append(chars[0])
        else:
            break
    return "".join(result)
print(longest_common_prefix(["flower", "flow", "flight"]))   # -> "fl"
print(longest_common_prefix(["dog", "cat"]))                 # -> ""
print(longest_common_prefix([]))                             # -> ""
```

```python
# 10. Majority Element  [hashing]
# Return the value appearing MORE than half the time, else None.
from collections import Counter
def majority(nums):
    threshold = len(nums) // 2
    counts = Counter(nums)
    for k,v in counts.items():
        if v > threshold : return k
    return None
print(majority([2, 2, 1, 2, 3]))   # -> 2
print(majority([1, 2, 3]))         # -> None
print(majority([7]))               # -> 7
```

```python
import numpy as np

# 11. Column Means  [axis]
# Return a 1-D array of the mean of each COLUMN.
def col_means(mat):
    return np.mean(mat,axis=0)
print(col_means(np.array([[1., 2.], [3., 4.], [5., 6.]])))   # -> [3. 4.]
```

```python
# 12. Count Nonzero  [masking]
# Return HOW MANY values are not equal to 0. Negatives count.
def count_nonzero(a):
    return np.sum(a!=0)
print(count_nonzero(np.array([0, -2, 0, 3, 0])))   # -> 2
```

```python
# 13. Clip to Both Bounds  [clip]
# Return the array clipped to [low, high] inclusive.
def clip_both(a, low, high):
    return np.clip(a, low, high)
print(clip_both(np.array([-5, 2, 8, 15]), 0, 10))   # -> [ 0  2  8 10]
```

```python

# 14. Index of Max Per Row  [argmax + axis]
# Return a 1-D array with the COLUMN index of the max value in each row.
def argmax_per_row(mat):
    return np.argmax(mat,axis=1)
print(argmax_per_row(np.array([[3, 9, 1], [8, 2, 8]])))   # -> [1 0]   (ties take first)
```

```python
# 15. Normalize Rows to Sum 1  [axis + broadcast]
# Divide each ROW by its own sum so each row sums to 1. Float.
def normalize_rows(mat):
    row_sums = np.sum(mat,axis=1,keepdims=True)
    return mat/row_sums
print(normalize_rows(np.array([[1., 1.], [1., 3.]])))   # -> [[0.5  0.5 ] [0.25 0.75]]
```

```python
# 16. Count Even Values  [masking]
# Return HOW MANY values are even. Zero and negatives count when divisible.
def count_even(a):
    return np.sum(a%2==0)
print(count_even(np.array([1, 2, 3, 4, -6, 0])))   # -> 4
```

```python
# 17. Replace Values Above Threshold  [where]
# Return the array with any value greater than thresh replaced by thresh.
# Do not mutate the input.
def cap_values(a, thresh):
    return np.where(a<thresh,a,thresh)
orig = np.array([5, 20, 8, 30])
print(cap_values(orig, 10))   # -> [ 5 10  8 10]
print(orig)                   # -> [ 5 20  8 30]   <- unchanged
```

```python
# 18. Row Range  [axis]
# Return a 1-D array: for each ROW, max minus min.
def row_range(mat):
    return np.max(mat,axis=1) - np.min(mat,axis=1)
print(row_range(np.array([[1, 5, 3], [10, 2, 8]])))   # -> [4 8]
```

```python
# 19. Count Rows All Positive  [axis + all]
# Return HOW MANY rows have every value strictly greater than 0.
def count_positive_rows(mat):
    return np.sum(np.all(mat > 0, axis=1))
print(count_positive_rows(np.array([[1, 2], [-1, 3], [4, 5]])))   # -> 2
```

```python
# 20. Running Max  [cumulative]
# Return a 1-D array where each element is the max seen so far, left to right.
def running_max(a):
    return np.maximum.accumulate(a)
print(running_max(np.array([1, 3, 2, 5, 4])))   # -> [1 3 3 5 5]
```
