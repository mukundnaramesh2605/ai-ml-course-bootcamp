```python
# ============ PYTHON (1-10) ============

# 1. Count Positive, Negative, Zero  [logic]
# Return a tuple (positives, negatives, zeros).
def count_signs(nums):
    return (sum([1 for i in nums if i>0]),sum([1 for i in nums if i<0]),sum([1 for i in nums if i==0]))
print(count_signs([1, -2, 0, 3, -4, 0]))   # -> (2, 2, 2)
print(count_signs([]))                     # -> (0, 0, 0)
```

```python
# 2. Remove Duplicates Keep Order  [logic]
# Remove duplicates, keep first-occurrence order. Do not use set for the result order.
def dedupe(nums):
    seen = []
    for i in nums:
        if i not in seen:
            seen.append(i)
    return seen
print(dedupe([3, 1, 3, 2, 1]))   # -> [3, 1, 2]
print(dedupe([]))                # -> []
```

```python
# 3. Capitalize Sentences  [string]
# Capitalize the first letter of each word, lowercase the rest. Collapse spaces.
def title_case(s):
    return " ".join([i[0].upper()+i[1:] for i in s.lower().split()])
print(title_case("hELLO   wORLD"))   # -> "Hello World"
print(title_case(""))                # -> ""
```

```python
# 4. Frequency Dictionary  [dict]
# Return a dict value -> count for the list.
def freq(nums):
    return {i:nums.count(i) for i in nums}
print(freq([1, 2, 2, 3, 3, 3]))   # -> {1: 1, 2: 2, 3: 3}
print(freq([]))                   # -> {}
```

```python
# 5. Pairs Summing to Target  [hashing]
# Return the COUNT of index pairs (i < j) where nums[i] + nums[j] == target.
def count_pairs(nums, target):
    seen = {}
    total = 0
    for i in nums:
        comp = target - i
        
        if comp in seen:
            total+=seen[comp]
        
        seen[i]=seen.get(i,0)+1

    return total
        
    
print(count_pairs([1, 2, 3, 2, 1], 3))   # -> 4
print(count_pairs([5, 5, 5], 10))        # -> 3
```
