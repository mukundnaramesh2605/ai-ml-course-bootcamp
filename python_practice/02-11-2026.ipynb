```python
# ============ PYTHON (1-10) ============

# 1. Count Positive, Negative, Zero  [logic]
# Return a tuple (positives, negatives, zeros).
def count_signs(nums):
    return (sum([1 for i in nums if i>0]),sum([1 for i in nums if i<0]),sum([1 for i in nums if i==0]))
print(count_signs([1, -2, 0, 3, -4, 0]))   # -> (2, 2, 2)
print(count_signs([]))                     # -> (0, 0, 0)
```
