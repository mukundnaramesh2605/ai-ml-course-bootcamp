```python
# 1. Clean and Count  [Service / data handling]
# Given a list of strings with stray whitespace and mixed case,
# return how many DISTINCT values there are after stripping and lowercasing.
# ["  Apple", "apple ", "BANANA", "banana"] -> 2
import re
def distinct_clean(items):
    words = re.findall(r"[a-zA-Z]+"," ".join(items))
    return len(set(word.lower() for word in words))
print(distinct_clean(["  Apple","apple ","BANANA","banana"]))# -> 2
```

```python
# 2. Parse Key-Value  [AI screen / data handling]
# Turn a list of "key=value" strings into a dict. Later keys overwrite earlier.
# ["a=1","b=2","a=3"] -> {"a":"3","b":"2"}
def parse_kv(pairs):
    result = {}
    for pair in pairs:
        key,value = pair.split("=",1)
        result[key] = value
    return result
print(parse_kv(["a=1","b=2","a=3"]))# -> {"a":"3","b":"2"}
```

```python
# 3. Fill Forward  [AI screen / data handling]
# Replace each None with the most recent non-None value before it.
# The first element is guaranteed not None.
# [1, None, None, 2, None, 3] -> [1,1,1,2,2,3]
def ffill(vals):
    non_none_values = [x for x in vals if x]
    fill = non_none_values[0]
    for i in range(len(vals)):
        if vals[i] is None:
            vals[i] = fill
        else:
            fill = vals[i]
    return vals
print(ffill([1,None,None,2,None,3]))# -> [1,1,1,2,2,3]
```
