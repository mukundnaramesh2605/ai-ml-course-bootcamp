```python
# 1. Clean and Count  [Service / data handling]
# Given a list of strings with stray whitespace and mixed case,
# return how many DISTINCT values there are after stripping and lowercasing.
# ["  Apple", "apple ", "BANANA", "banana"] -> 2
import re
def distinct_clean(items):
    words = re.findall(r"[a-zA-Z]+"," ".join(items))
    return len(set(word.lower() for word in words))
print(distinct_clean(["  Apple","apple ","BANANA","banana"]))# -> 2
```

```python
# 2. Parse Key-Value  [AI screen / data handling]
# Turn a list of "key=value" strings into a dict. Later keys overwrite earlier.
# ["a=1","b=2","a=3"] -> {"a":"3","b":"2"}
def parse_kv(pairs):
    result = {}
    for pair in pairs:
        key,value = pair.split("=",1)
        result[key] = value
    return result
print(parse_kv(["a=1","b=2","a=3"]))# -> {"a":"3","b":"2"}
```

```python
# 3. Fill Forward  [AI screen / data handling]
# Replace each None with the most recent non-None value before it.
# The first element is guaranteed not None.
# [1, None, None, 2, None, 3] -> [1,1,1,2,2,3]
def ffill(vals):
    non_none_values = [x for x in vals if x]
    fill = non_none_values[0]
    for i in range(len(vals)):
        if vals[i] is None:
            vals[i] = fill
        else:
            fill = vals[i]
    return vals
print(ffill([1,None,None,2,None,3]))# -> [1,1,1,2,2,3]
```

```python
# 4. Column from Records  [AI screen / data handling]
# Given a list of dicts, return a list of the "age" values,
# using None where a record has no "age" key.
# [{"age":30},{"name":"x"},{"age":25}] -> [30, None, 25]
def pluck_age(records):
    return [dcits.get('age',None) for dcits in records]
print(pluck_age([{"age":30},{"name":"x"},{"age":25}]))# -> [30, None, 25]
```

```python
# 5. Top K Frequent  [AI screen / data handling]
# Return the k most frequent values, most frequent first.
# On a tie, the value that appears first in the list wins.
# ["a","b","a","c","b","a"], 2 -> ["a","b"]
from collections import Counter
def top_k(items, k):
    counts = Counter(items)
    counts = dict(sorted(counts.items(),key=lambda x: x[1],reverse=True))
    return [i for i in counts][:k]
print(top_k(["a","b","a","c","b","a"], 2))# -> ["a","b"]
```

```python
# 1. Is Palindrome  [Service / string]
# Return True if the string reads the same forwards and backwards,
# ignoring case. "RaceCar" -> True
def is_palindrome(s):
    s=s.lower()
    return s == s[::-1]
print(is_palindrome("RaceCar"))# -> True ; 
print(is_palindrome("hello"))# -> False ; 
print(is_palindrome(""))# -> True
```
