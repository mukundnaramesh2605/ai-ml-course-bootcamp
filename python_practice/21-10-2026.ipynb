```python
# 1. Clean and Count  [Service / data handling]
# Given a list of strings with stray whitespace and mixed case,
# return how many DISTINCT values there are after stripping and lowercasing.
# ["  Apple", "apple ", "BANANA", "banana"] -> 2
import re
def distinct_clean(items):
    words = re.findall(r"[a-zA-Z]+"," ".join(items))
    return len(set(word.lower() for word in words))
print(distinct_clean(["  Apple","apple ","BANANA","banana"]))# -> 2
```
