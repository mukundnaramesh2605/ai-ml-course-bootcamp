```python
def sum_of_digits(n):
    sign = -1 if n <0 else 1
    return sum([int(i) for i in str(abs(n))])*sign

print(sum_of_digits(1234))      # normal
print(sum_of_digits(0))         # zero
print(sum_of_digits(5))         # single digit
print(sum_of_digits(-456))      # negative
print(sum_of_digits(1000000))   # trailing zeros
```

```python
# 2. Check if a string has all unique characters
def all_unique(s):
    return len(set(s)) == len(s)

print(all_unique("python"))     # all unique
print(all_unique("hello"))      # has repeats
print(all_unique(""))           # empty
print(all_unique("a"))          # single char
print(all_unique("aA"))         # case differs
```

```python
# 3. Return the intersection of two lists without duplicates
def list_intersection(a, b):
    return list(set(a).intersection(set(b)))

print(list_intersection([1, 2, 3], [2, 3, 4]))      # normal
print(list_intersection([1, 1, 2], [2, 2, 3]))      # duplicates in input
print(list_intersection([1, 2], [3, 4]))            # no overlap
print(list_intersection([], [1, 2]))                # one empty
print(list_intersection([5], [5]))                  # single match
```

```python
# 4. Count how many words are in a sentence
import re
def word_count(sentence):
    sentence = re.findall(r'[a-zA-Z0-9]+',sentence) 
    return len(sentence)

print(word_count("the quick brown fox"))    # normal
print(word_count(""))                       # empty
print(word_count("hello"))                  # single word
print(word_count("  extra   spaces  "))     # messy spacing
print(word_count("a b c d e"))              # many short words
```

```python
# 5. Return a dictionary of character frequencies in a string
def char_frequency(s):
    return {char:s.count(char) for char in s if char!= " "}

print(char_frequency("banana"))     # normal
print(char_frequency(""))           # empty
print(char_frequency("aaa"))        # all same
print(char_frequency("abc"))        # all unique
print(char_frequency("a b"))        # with space
```

```python
import numpy as np

# 1. Return the diagonal elements of a square matrix
def get_diagonal(mat):
    return np.diag(mat)

print(get_diagonal(np.array([[1, 2], [3, 4]])))             # normal
print(get_diagonal(np.array([[5]])))                        # single element
print(get_diagonal(np.array([[1, 0, 0], [0, 2, 0], [0, 0, 3]])))  # 3x3
print(get_diagonal(np.array([[-1, 2], [3, -4]])))           # negatives
print(get_diagonal(np.array([[0, 1], [1, 0]])))             # zeros on diagonal
```

```python
# 2. Count the number of elements greater than the mean
def count_above_mean(arr):
    return np.sum(arr>arr.mean())

print(count_above_mean(np.array([1, 2, 3, 4, 5])))      # normal
print(count_above_mean(np.array([5, 5, 5])))            # all equal
print(count_above_mean(np.array([10])))                 # single element
print(count_above_mean(np.array([-5, 0, 5])))           # negatives
print(count_above_mean(np.array([1, 100])))             # one large outlier
```
