```python
def sum_of_digits(n):
    sign = -1 if n <0 else 1
    return sum([int(i) for i in str(abs(n))])*sign

print(sum_of_digits(1234))      # normal
print(sum_of_digits(0))         # zero
print(sum_of_digits(5))         # single digit
print(sum_of_digits(-456))      # negative
print(sum_of_digits(1000000))   # trailing zeros
```
