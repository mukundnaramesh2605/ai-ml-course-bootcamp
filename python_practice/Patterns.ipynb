```python
# 1. Compress a String  [Service / string]   (consecutive + tail)
# Collapse runs of the same char into char+count. Keep order.
# Example: "aaabbc" -> "a3b2c1"
def compress2(s):
    if s == "":
        return ""
        
    result = []
    count = 1
    current = s[0]
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current = s[i]
    result.append(current+str(count))       
    return "".join(result)
print(compress2("aaabbc"))# -> "a3b2c1" ; 
print(compress2("aabbaa"))# -> "a2b2a2" ; 
print(compress2(""))# -> ""
```

```python
# 3. Count Streaks  [Service / logic]   (consecutive)
# Return how many groups of consecutive equal values there are.
# Example: [1,1,2,3,3,3,1] -> 4
def count_streaks(nums):
    if not nums:
        return 0
    count=1
    for i in range(1,len(nums)):
        if nums[i] != nums[i-1]:
            count+=1
    return count
        
print(count_streaks([1,1,2,3,3,3,1]))# -> 4 ; 
print(count_streaks([5,5,5])) #-> 1 ; 
count_streaks([])# -> 0

```

```python
# Q3. Valid Parentheses  [AI screen / logic]
# Return True if every bracket is closed correctly, in order.
# Brackets: ( ) [ ] { }
# Example: "(a[b]{c})" -> True ;  "(a[b)c]" -> False
def is_valid(s):
    brackets = {"}":"{","]":"[",")":"("}
    stack=[]
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0
        
print(is_valid("(a[b]{c})")) # -> True
# test: is_valid("(]") -> False
# test: is_valid("(") -> False
# test: is_valid("") -> True
```

```python
# 4. Longest Common Prefix  [AI screen / string]
# Return the longest prefix shared by ALL strings, or "" if none.
# ["flower","flow","flight"] -> "fl"
def common_prefix(strs):
    if not strs:
        return ""
    prefix = []
    for chars in zip(*strs):
        if len(set(chars)) == 1:
            prefix.append(chars[0])
        else:
            break
    return "".join(prefix)
print(common_prefix(["flower","flow","flight"]))# -> "fl" ; 
print(common_prefix(["a","b"]))# -> ""
print(common_prefix(["flower","flow","flowing"]))
```

```python
# 3. Missing Number  [logic]
# nums holds distinct numbers 0..n with one missing. Return it.
def missing_number(nums):
    n = len(nums)
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)
print(missing_number([3,0,1]))#-> 2 ; 
print(missing_number([0,1]))# -> 2 ; missing_number([0]) -> 1
```

```python
# 1. Factorial (iterative)  [screening]
# Return n! using a loop. 0! is 1.
def factorial(n):
    if n == 0:
        return 1
    fact_n = 1
    for i in range(1,n+1):
        fact_n*=i
    return fact_n
print(factorial(5))# -> 120 ; 
print(factorial(0))# -> 1
```

```python
# 2. Title Case  [string]
# Capitalize the first letter of each word, lowercase the rest.
# "hELLO wORLD" -> "Hello World"
def title_case(s):
    return " ".join([i[0].upper()+i[1:] for i in s.lower().split()])
print(title_case("hELLO wORLD"))# -> "Hello World" ; 
print(title_case(""))# -> ""
```

```python
# 3. Dedupe Keep Order  [logic]
# Remove duplicates, keep first-occurrence order.
# [3,1,3,2,1] -> [3,1,2]
def dedupe(nums):
    seen = []
    for i in nums:
        if i not in seen:
            seen.append(i)
    return seen
print(dedupe([3,1,3,2,1]))# -> [3,1,2] ; 
print(dedupe([]))# -> []
```

```python
# 4. First Unique Character Index  [string]
# Return the index of the first non-repeating char, or -1 if none.
# "leetcode" -> 0 ; "aabb" -> -1
def first_uniq_index(s):
    for idx,i in enumerate(s):
        if s.count(i) == 1:
            return idx
    return -1
print(first_uniq_index("leetcode"))# -> 0 ; 
print(first_uniq_index("loveleetcode"))# -> 2 ; 
print(first_uniq_index("aabb"))# -> -1
```
