```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
# 1. Count Uppercase Letters  [easy / string]
# Return HOW MANY characters are uppercase letters.
# "Hello World" -> 2
def count_upper(s):
    return sum([1 for i in s if i.isupper()])

print(count_upper("Hello World"))   # -> 2
print(count_upper(""))              # -> 0
print(count_upper("ABC"))           # -> 3
print(count_upper("123!"))          # -> 0   <- digits and symbols are not uppercase
print(count_upper("aBc D"))         # -> 2
```

```python
# 2. Repeated Characters  [easy / Counter]
# Return a dict char -> count for chars that appear AT LEAST twice. CASE-SENSITIVE ("A" and "a" differ).
# Keys in the order each char first appears.
# "aAbbA" -> {'A': 2, 'b': 2}
def repeated_chars(s):
    counts = Counter([i for i in s if s.count(i) > 1 and i!=' '])
    return dict(counts)

print(repeated_chars("aAbbA"))   # -> {'A': 2, 'b': 2}   <- don't lowercase: 'a' appears only once
print(repeated_chars(""))        # -> {}
print(repeated_chars("abc"))     # -> {}
print(repeated_chars("zzz"))     # -> {'z': 3}
print(repeated_chars("a a"))     # -> {'a': 2}
```

```python
# 3. Evens Minus Odds  [easy / logic]
# Return (sum of even numbers) - (sum of odd numbers).
# [1, 2, 3, 4] -> 2
def evens_minus_odds(nums):
    return sum([i for i in nums if i%2 == 0]) - sum([i for i in nums if i%2 != 0])

print(evens_minus_odds([1, 2, 3, 4]))   # -> 2
print(evens_minus_odds([]))             # -> 0
print(evens_minus_odds([5]))            # -> -5
print(evens_minus_odds([-2, -3]))       # -> 1   <- -2 - (-3) = 1; -3 is odd
print(evens_minus_odds([0]))            # -> 0
```

```python
# 4. Sort by Frequency  [medium / Counter + sort]
# Return the DISTINCT values sorted by count DESC; ties broken by value ASC.
# [4, 1, 1, 4, 2] -> [1, 4, 2]
def sort_by_freq(nums):
    counts = Counter(nums)
    return list(dict(sorted(counts.items(),key=lambda x:[-x[1],x[0]])).keys())

print(sort_by_freq([4, 1, 1, 4, 2]))        # -> [1, 4, 2]   <- 1 and 4 both appear twice, 1 is smaller
print(sort_by_freq([]))                     # -> []
print(sort_by_freq([7]))                    # -> [7]
print(sort_by_freq([3, 3, -1, -1, -1]))     # -> [-1, 3]
print(sort_by_freq([5, 4, 3]))              # -> [3, 4, 5]
```

```python
# 5. Group Names by Grade  [medium / dict]
# pairs is a list of (name, score). Grade: "A" if score >= 80, "B" if >= 60, else "C".
# Return a dict grade -> list of names in input order. Only grades that have names.
# Keys in the order each grade first appears.
def names_by_grade(pairs):
    result = {}
    for name, score in pairs:
        if score >= 80:
          grade = "A"
        elif score >= 60:
          grade = "B"
        else:
          grade = "C"
    
        if grade not in result:
          result[grade] = []
        
        result[grade].append(name)
    
    return result

print(names_by_grade([("Ravi", 85), ("Asha", 60), ("Kumar", 59)]))   # -> {'A': ['Ravi'], 'B': ['Asha'], 'C': ['Kumar']}
print(names_by_grade([]))                                          # -> {}
print(names_by_grade([("A", 80), ("B", 95)]))                      # -> {'A': ['A', 'B']}   <- 80 is an A
print(names_by_grade([("X", 0)]))                                  # -> {'C': ['X']}
print(names_by_grade([("P", 61), ("Q", 79), ("R", 81)]))           # -> {'B': ['P', 'Q'], 'A': ['R']}
```

```python
# 6. Sum of Main Diagonal  [easy / indexing]
# Return the sum of the main diagonal (top-left to bottom-right) of a square matrix.
def diag_sum(mat):
    return np.diag(mat).sum()

m6 = np.array([[4, 1, 2],
               [0, -5, 3],
               [7, 8, 9]])
print(diag_sum(m6))   # -> 8   <- 4 + (-5) + 9
```

```python
# 7. Index of First Negative  [easy / masking]
# Return the index of the FIRST negative value, or -1 if there is none.
def first_negative(a):
    mask = a < 0
    return int(np.argmax(mask)) if np.any(mask) else -1
a7 = np.array([3, 0, 5, 2])
print(first_negative(a7))   # -> -1   <- np.argmax on an all-False mask returns 0, not -1
```

```python
# 8. Range Per Row  [easy / axis]
# Return a 1-D array with (max - min) of each ROW.
def row_range(mat):
    row_max = np.max(mat,axis=1)
    row_min = np.min(mat,axis=1)
    return row_max-row_min

m8 = np.array([[3, 9, 1],
               [-2, -2, -2],
               [10, 0, 5]])
print(row_range(m8))   # -> [ 8  0 10]
```

```python
# 9. Count Above Column Mean  [medium / axis + mask]
# For each COLUMN, return HOW MANY values are STRICTLY above that column's mean.
def count_above_col_mean(mat):
    col_mean = np.mean(mat,axis=0,keepdims=True)
    return sum(mat > col_mean)
m9 = np.array([[1, 10, 5],
               [2, 20, 5],
               [9, 30, 5]])
print(count_above_col_mean(m9))   # -> [1 1 0]   <- column 2 is all 5s, nothing is strictly above
```

```python
# 10. Keep Only Allowed Values  [medium / np.isin]
# Values in allowed stay as they are; every other value becomes 0. Return a NEW array.
def keep_allowed(a, allowed):
    return np.where(np.isin(a,allowed),a,0)

a10 = np.array([1, 5, 2, 3, 0, -1, 3])
print(keep_allowed(a10, [1, 2, 3]))   # -> [1 0 2 3 0 0 3]
```

```python
# 11. Filter by City List  [easy / isin]
# Keep rows where city is EXACTLY "Chennai" or "Pune". Do NOT clean or lowercase the data.
# Columns EXACTLY: name, city. Reset index.
def in_cities(df):
    return df[df['city'].isin(['Chennai','Pune'])].reset_index(drop=True)

df11 = pd.DataFrame({
    "name": ["A", "B", "C", "D", "E"],
    "city": ["Chennai", "Delhi", "chennai", "Pune", "Chennai "],
})
print(in_cities(df11))
# -> A Chennai | D Pune   <- "chennai" and "Chennai " (trailing space) are not exact matches
```

```python
# 12. Total Quantity Per Product  [easy / agg]
# ONE row per product with total qty. Columns EXACTLY: product, total_qty.
# Sort by total_qty DESC, ties by product ASC. Reset index.
def product_totals(df):
    df2 = df.copy()
    return df2.groupby('product',as_index=False).agg(total_qty=('qty','sum')).sort_values('total_qty',ascending=False).reset_index(drop=True)

df12 = pd.DataFrame({
    "product": ["pen", "book", "pen", "bag", "book", "bag"],
    "qty":     [5, 3, 2, 4, 4, 1],
})
print(product_totals(df12))
# -> book 7 | pen 7 | bag 5   <- book and pen tie at 7, book first alphabetically
```

```python
# 13. Dept Max on Every Row  [medium / transform]
# Add "dept_max" = the highest salary in that row's dept. Keep ALL rows in original order.
# Columns: name, dept, salary, dept_max. Return a NEW df.
def add_dept_max(df):
    df2 = df.copy()
    df2['dept_max'] = df.groupby('dept')['salary'].transform('max')
    return df2

df13 = pd.DataFrame({
    "name":   ["A", "B", "C", "D", "E"],
    "dept":   ["IT", "HR", "IT", "HR", "FIN"],
    "salary": [70, 40, 90, 55, 60],
})
print(add_dept_max(df13))
# -> dept_max: 90, 55, 90, 55, 60   <- 5 rows out, not 3 (this is transform, not agg)
```
