# ===== CORE PYTHON (1-7) =====

```python
# 1. Count Even and Odd
# Given a list, return how many even and how many odd numbers it has.
# Example: [1, 2, 3, 4, 5, 6] -> even: 3, odd: 3
nums = [1, 2, 3, 4, 5, 6]
from collections import Counter
counts = Counter("Even" if x%2 == 0 else "Odd" for x in nums)
print(counts)
```

```python
# 2. Title Case a Sentence
# Capitalize the first letter of each word without using .title().
# Example: "hello world from python" -> "Hello World From Python"
sentence = "hello world from python"
print(" ".join([x.capitalize() for x in sentence.split()]))
```

```python
# 3. Find Common Elements
# Return elements present in both lists.
# Example: [1, 2, 3, 4], [3, 4, 5, 6] -> [3, 4]
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]
print(f"Common Elements -> {list(set(a)&set(b))}")
```

```python
# 4. Count Vowels and Consonants
# Given a string, count vowels and consonants separately (letters only).
# Example: "hello" -> vowels: 2, consonants: 3
text = "hello"
from collections import Counter
counts = Counter("vowels" if x.lower() in 'aeiou' else 'consonants' for x in text if x.isalpha())
print(counts)
```

```python
# 5. Sum of Even Numbers
# Return the sum of only the even numbers in a list.
# Example: [1, 2, 3, 4, 5, 6] -> 12
values = [1, 2, 3, 4, 5, 6]
print(f"Sum of even Nos -> :{sum([x for x in values if x%2 == 0])}")
```

```python
# 6. Dictionary From Two Lists
# Build a dictionary pairing keys with values.
# Example: ["a", "b", "c"], [1, 2, 3] -> {"a": 1, "b": 2, "c": 3}
keys = ["a", "b", "c"]
vals = [1, 2, 3]
{k:v for k,v in zip(keys,values)}
```

```python
# 7. Remove Punctuation
# Return the string with all punctuation removed.
# Example: "Hello, World! How are you?" -> "Hello World How are you"
line = "Hello, World! How are you?"
import re
new_line = re.sub(r'[^a-zA-Z0-9 ]','',line)
new_line
```

```python
# 8. Count Values Above Mean
# Return how many elements in the array are greater than the mean.
# Example: [10, 20, 30, 40, 50] -> 2
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
arr[arr>arr.mean()]
```

```python
# 9. Element-wise Operations
# Given two arrays, return their element-wise sum, difference, and product.
import numpy as np
x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7, 8])
print(f"Sum:{x+y}\nDifference:{abs(x-y)}\nProduct:{x*y}")
```
