```python
# 1. Count Even and Odd
# Given a list, return how many even and how many odd numbers it has.
# Example: [1, 2, 3, 4, 5, 6] -> even: 3, odd: 3
nums = [1, 2, 3, 4, 5, 6]
from collections import Counter
counts = Counter("Even" if x%2 == 0 else "Odd" for x in nums)
print(counts)
```

```python
# 2. Title Case a Sentence
# Capitalize the first letter of each word without using .title().
# Example: "hello world from python" -> "Hello World From Python"
sentence = "hello world from python"
print(" ".join([x.capitalize() for x in sentence.split()]))
```

```python
# 3. Find Common Elements
# Return elements present in both lists.
# Example: [1, 2, 3, 4], [3, 4, 5, 6] -> [3, 4]
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]
print(f"Common Elements -> {list(set(a)&set(b))}")
```

```python
# 4. Count Vowels and Consonants
# Given a string, count vowels and consonants separately (letters only).
# Example: "hello" -> vowels: 2, consonants: 3
text = "hello"
from collections import Counter
counts = Counter("vowels" if x.lower() in 'aeiou' else 'consonants' for x in text if x.isalpha())
print(counts)
```

```python
# 5. Sum of Even Numbers
# Return the sum of only the even numbers in a list.
# Example: [1, 2, 3, 4, 5, 6] -> 12
values = [1, 2, 3, 4, 5, 6]
print(f"Sum of even Nos -> :{sum([x for x in values if x%2 == 0])}")
```

```python
# 6. Dictionary From Two Lists
# Build a dictionary pairing keys with values.
# Example: ["a", "b", "c"], [1, 2, 3] -> {"a": 1, "b": 2, "c": 3}
keys = ["a", "b", "c"]
vals = [1, 2, 3]
{k:v for k,v in zip(keys,vals)}
```

```python
# 7. Remove Punctuation
# Return the string with all punctuation removed.
# Example: "Hello, World! How are you?" -> "Hello World How are you"
line = "Hello, World! How are you?"
import re
new_line = re.sub(r'[^a-zA-Z0-9 ]','',line)
new_line
```

```python
# 20. Apply a Custom Function
# Add a "grade" column: A if score >= 90, B if >= 75, else C.
import pandas as pd
results = pd.DataFrame({
    "student": ["P", "Q", "R", "S"],
    "score": [92, 78, 65, 88],
})
def grader(mark):
    if mark >= 90:
        return "A"
    elif mark <90 and mark >= 75:
        return "B"
    else:
        return "C"

results['grade'] = results['score'].apply(grader)
results
```

```python
# 1. First Non-Repeating Character
# Return the first character that appears exactly once in the string.
# Example: "swiss" -> "w"
from collections import defaultdict
s = "swiss"
char_count = defaultdict(list)
for char in s:
    char_count[char] = char_count.get(char,0)+1
for i,j in char_count.items():
    if j==1:
        print(f'First Non repeating character is "{i}"')
        break
```

```python
# 2. Merge Overlapping Counts
# Given a list of (name, score) tuples, sum scores per name.
# Example: [("a", 10), ("b", 5), ("a", 7)] -> {"a": 17, "b": 5}
from collections import Counter
pairs = [("a", 10), ("b", 5), ("a", 7), ("c", 3), ("b", 2)]
overlapping = {}
for i,j in pairs:
    overlapping[i] = overlapping.get(i,0)+j
print(overlapping)
```

```python
# 3. Running Maximum
# Return a list where each element is the max seen so far.
# Example: [3, 1, 4, 1, 5, 9, 2] -> [3, 3, 4, 4, 5, 9, 9]
nums = [3, 1, 4, 1, 5, 9, 2]
current_max = nums[0]
lst = []
for i in nums:
    current_max = max(current_max,i)
    lst.append(current_max)
print(lst)
```

```python
# ===== NUMPY (4-6) =====

# 4. Row-wise Normalization
# Normalize each row of the matrix so its values sum to 1.
import numpy as np
matrix = np.array([[1, 1, 2], [2, 3, 5], [4, 4, 2]])
row_sums = matrix.sum(axis=1,keepdims=True)
normalized = np.divide(matrix,row_sums,np.zeros_like(matrix,dtype=np.float64),where=row_sums!=0)
normalized
```

```python
# 6. Pairwise Euclidean Distance
# Given points, return the distance between the first point and every other point.
# Example: [[0,0],[3,4],[6,8]] -> [0.0, 5.0, 10.0]
import numpy as np
points = np.array([[0, 0], [3, 4], [6, 8]])
distances = np.linalg.norm(points - points[0], axis=1)
distances
```

```python
# 7. Merge Two DataFrames
# Join orders with customers on customer_id. Keep all orders.
import pandas as pd
customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Arjun", "Priya", "Karan"],
})
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [1, 2, 1, 3],
    "amount": [500, 300, 700, 200],
})
pd.merge(left=customers,right=orders,how='right')
```

```python
# 8. Group, Aggregate, Multiple Metrics
# Per region, return total and average amount in one result.
import pandas as pd
sales = pd.DataFrame({
    "region": ["S", "N", "S", "W", "N", "S"],
    "amount": [250, 400, 300, 150, 500, 200],
})
sales.groupby('region').agg(total_sales=('amount','sum'),average_amount=('amount','mean')).reset_index()
```

```python
# 9. Pivot Table
# Reshape so rows are "product", columns are "month", values are "units" sold.
import pandas as pd
data = pd.DataFrame({
    "product": ["A", "A", "B", "B", "A", "B"],
    "month": ["Jan", "Feb", "Jan", "Feb", "Jan", "Feb"],
    "units": [10, 15, 5, 8, 12, 6],
})
pd.pivot_table(
    data = data,
    values = 'units',
    index= 'product',
    columns='month'
)
```

```python
# 10. Rank Within Group
# Add a column ranking employees by salary within each department (1 = highest).
import pandas as pd
emp = pd.DataFrame({
    "name": ["A", "B", "C", "D", "E"],
    "dept": ["IT", "IT", "HR", "HR", "IT"],
    "salary": [50000, 70000, 45000, 48000, 60000],
})
emp['rank'] = emp.groupby('dept')['salary'].rank(method='dense',ascending=False)
emp
```
