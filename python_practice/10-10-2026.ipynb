```python
# 1. Count Even and Odd
# Given a list, return how many even and how many odd numbers it has.
# Example: [1, 2, 3, 4, 5, 6] -> even: 3, odd: 3
nums = [1, 2, 3, 4, 5, 6]
from collections import Counter
counts = Counter("Even" if x%2 == 0 else "Odd" for x in nums)
print(counts)
```

```python
# 2. Title Case a Sentence
# Capitalize the first letter of each word without using .title().
# Example: "hello world from python" -> "Hello World From Python"
sentence = "hello world from python"
print(" ".join([x.capitalize() for x in sentence.split()]))
```
