```python
# 1. Sum of Even Numbers  [Service / logic]
# Return the sum of only the even numbers in a list.
# [1,2,3,4,5,6] -> 12
def sum_evens(nums):
    return sum([x for x in nums if x%2==0])
print(sum_evens([1,2,3,4,5,6]))# -> 12 ; 
print(sum_evens([1,3,5]))# -> 0 ; sum_evens([]) -> 0

```

```python
# 2. First Non-Repeating Char  [AI screen / string]
# Return the first character that appears exactly once, case-sensitive.
# If none, return None.
# "aabbc" -> "c" ; "aabb" -> None
def first_unique(s):
    for i in s:
        if s.count(i)==1:
            return i
    return None
print(first_unique("aabbc"))# -> "c" ; 
print(first_unique("aabb"))# -> None ; 
print(first_unique(""))# -> None
```

```python
# 3. Dedupe Keep Order  [Service / logic]
# Return the list with duplicates removed, keeping FIRST occurrence order.
# [3,1,3,2,1] -> [3,1,2]
def dedupe(nums):
    return (list(dict.fromkeys(nums)))
print(dedupe([3,1,3,2,1]))# -> [3,1,2] ; 
print(dedupe([]))# -> []
```

```python
# 5. Merge Sorted Lists  [AI screen / logic]
# Given two already-sorted lists, return one merged sorted list.
# [1,3,5], [2,4,6] -> [1,2,3,4,5,6]
import heapq
def merge_sorted(a, b):
    return list(heapq.merge(a,b))
print(merge_sorted([1,3,5],[2,4,6]))# -> [1,2,3,4,5,6] ; 
print(merge_sorted([],[1]))# -> [1]
```

```python
# 6. Percentage of Total  [AI screen / numpy]
# Return an array where each value is its percentage of the total sum.
# [10, 30, 60] -> [10.0, 30.0, 60.0]
import numpy as np
arr = np.array([10, 30, 60])
(arr/arr.sum())*100
```

```python
# 7. Replace Above Mean with Mean  [AI screen / numpy]
# Return the array where every value ABOVE the mean is replaced by the mean.
# Values at or below the mean stay.
import numpy as np
a = np.array([1, 2, 3, 10])
a[a>a.mean()] = a.mean()
a
```

```python
# 8. Cumulative Sum  [AI screen / numpy]
# Return the running cumulative sum of a 1D array.
# [1,2,3,4] -> [1,3,6,10]
import numpy as np
x = np.array([1, 2, 3, 4])
np.cumsum(x)
```

```python
# 9. Count Values in Range  [AI screen / data handling]
# Return HOW MANY values are between 20 and 40 INCLUSIVE. (a count)
import numpy as np
data = np.array([10, 20, 30, 40, 50, 25])
data[(data>=20) & (data <=40)].shape[0]
```

```python
# 10. Column with Largest Mean  [AI screen / numpy]
# Return the INDEX of the column whose mean is the largest.
import numpy as np
grid = np.array([[1, 8], [2, 9], [3, 7]])
np.mean(grid,axis=0).argmax()
```

```python
# 11. Filter, Select, Sort  [AI screen / pandas]
# Return name and age for rows where age >= 30,
# columns name and age ONLY, sorted by age ASCENDING.
import pandas as pd
df = pd.DataFrame({
    "name": ["A","B","C","D"],
    "age": [25, 45, 30, 60],
    "city": ["P","Q","R","S"],
})

df[df['age']>=30][['name','age']].sort_values(by='age',ascending=True)
```
