```python
# 1. Sum of Even Numbers  [Service / logic]
# Return the sum of only the even numbers in a list.
# [1,2,3,4,5,6] -> 12
def sum_evens(nums):
    return sum([x for x in nums if x%2==0])
print(sum_evens([1,2,3,4,5,6]))# -> 12 ; 
print(sum_evens([1,3,5]))# -> 0 ; sum_evens([]) -> 0

```

```python
# 2. First Non-Repeating Char  [AI screen / string]
# Return the first character that appears exactly once, case-sensitive.
# If none, return None.
# "aabbc" -> "c" ; "aabb" -> None
def first_unique(s):
    for i in s:
        if s.count(i)==1:
            return i
    return None
print(first_unique("aabbc"))# -> "c" ; 
print(first_unique("aabb"))# -> None ; 
print(first_unique(""))# -> None
```
