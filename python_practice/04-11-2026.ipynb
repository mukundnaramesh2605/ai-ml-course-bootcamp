```python
# ============ PYTHON (1-10) ============

# 1. Count Even and Odd  [logic]
# Return a tuple (evens, odds). Negatives count by parity.
def count_even_odd(nums):
    return (len(list(filter(lambda x: x%2==0,nums))),len(list(filter(lambda x: x%2!=0,nums))))
print(count_even_odd([1, 2, 3, 4, -6]))   # -> (3, 2)
print(count_even_odd([]))                 # -> (0, 0)
```

```python
# 2. First Non-Repeating Char  [string]
# Return the first character that appears exactly once. None if none.
def first_unique(s):
    for i in s:
        if s.count(i) == 1:
            return i
    return None
print(first_unique("swiss"))   # -> "w"
print(first_unique("aabb"))    # -> None
```

```python
# 3. Sum of Unique Values  [set]
# Return the sum of values that appear EXACTLY once.
def sum_unique(nums):
    return sum(list(filter(lambda x: nums.count(x)==1,nums)))
print(sum_unique([1, 2, 2, 3, 4, 4]))   # -> 4   (1 + 3)
print(sum_unique([5, 5]))               # -> 0
```

```python
# 4. Title Case  [string]
# Capitalize the first letter of each word, lowercase the rest. Collapse spaces.
def title_case(s):
    return " ".join([x[0].upper()+x[1:].lower() for x in s.split()])
print(title_case("hELLO   wORLD"))   # -> "Hello World"
print(title_case(""))                # -> ""
```

```python
# 5. Pairs Summing to Target  [hashing]
# Return the COUNT of index pairs (i < j) where nums[i] + nums[j] == target.
def count_pairs(nums, target):
    seen = {}
    count = 0
    for num in nums:
        comp = target - num
        if comp in seen:
            count +=seen[comp]
        seen[num] = seen.get(num, 0) + 1
    return count
print(count_pairs([1, 2, 3, 2, 1], 3))   # -> 4
print(count_pairs([5, 5, 5], 10))        # -> 3
```

```python
# 6. Move Zeros to End  [two-pointer]
# Move all 0s to the end, keep order of the rest. Negatives stay in order.
def move_zeros(nums):
    write = 0
    for i in range(len(nums)):
        if nums[i]!=0:
            nums[write],nums[i] = nums[i],nums[write]
            write+=1
    return nums
print(move_zeros([0, -1, 0, 3, 12]))   # -> [-1, 3, 12, 0, 0]
print(move_zeros([0, 0, 1]))           # -> [1, 0, 0]
```

```python
# 7. Valid Parentheses  [stack]
# Return True if (), [], {} are correctly matched and nested.
def is_valid(s):
    bracket = {"]":"[","}":"{",")":"("}
    stack = []
    for i in s:
        if i in bracket.values():
            stack.append(i)
        elif i in bracket:
            if not stack or stack[-1] != bracket[i]:
                return False
            stack.pop()
    return len(stack) == 0
print(is_valid("([{}])"))   # -> True
print(is_valid("([)]"))     # -> False
print(is_valid("("))        # -> False
```

```python
# 8. Compress String  [string]
# Run-length encode: "aaabbc" -> "a3b2c1". Empty -> "".
def compress(s):
    if s == "":
        return ""
    count=1
    current=s[0]
    result = []
    for i in range(1,len(s)):
        if s[i-1] == s[i]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current=s[i]
    result.append(current+str(count))
    return "".join(result)        
print(compress("aaabbc"))   # -> "a3b2c1"
print(compress("abc"))      # -> "a1b1c1"
print(compress(""))         # -> ""
```

```python
# 9. Max Subarray Sum  [logic]
# Return the largest sum of any contiguous non-empty subarray (Kadane).
def max_subarray(nums):
    current = best = nums[0]
    for num in nums[1:]:
        current = max(num,current+num)
        best = max(current,best)
    return best
print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))   # -> 6
print(max_subarray([-1, -2, -3]))                       # -> -1
```

```python
# 10. Group Anagrams  [dict]
# Group words that are anagrams. Return a list of groups, first-appearance order.
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for i in words:
        grouped["".join(sorted(i))].append(i)
    return list(grouped.values())
print(group_anagrams(["eat", "tea", "tan", "ate", "nat"]))
# -> [['eat', 'tea', 'ate'], ['tan', 'nat']]
print(group_anagrams([]))   # -> []
```

```python
# ============ NUMPY (11-20) ============
import numpy as np

# 11. Row Sums  [axis]
# Return a 1-D array of the sum of each ROW.
def row_sums(mat):
    return np.sum(mat,axis=1)
print(row_sums(np.array([[1, 2, 3], [4, 5, 6]])))   # -> [ 6 15]
```

```python
# 12. Count Above Mean  [masking]
# Return HOW MANY values are strictly greater than the array's mean.
def above_mean(a):
    return np.sum(a>a.mean())
print(above_mean(np.array([1, 2, 3, 10])))   # -> 1
```

```python
# 13. Replace Negatives With Zero  [where]
# Return the array with every negative set to 0. Do not mutate the input.
def clip_negatives(a):
    return np.where(a<0,0,a)
orig = np.array([-1, 5, -3, 0, 2])
print(clip_negatives(orig))   # -> [0 5 0 0 2]
print(orig)                   # -> [-1  5 -3  0  2]   <- unchanged
```

```python
# 14. Min-Max Normalize  [normalize]
# Scale a 1-D array to 0..1 using (x - min) / (max - min). Float.
def min_max(a):
    return (a - a.min())/(a.max() - a.min())
print(min_max(np.array([10., 20., 30.])))   # -> [0.  0.5 1. ]
```

```python
# 15. Index of Min Per Row  [argmin + axis]
# Return a 1-D array with the COLUMN index of the min value in each row.
def argmin_per_row(mat):
    return np.argmin(mat,axis=1)
print(argmin_per_row(np.array([[3, 9, 1], [8, 2, 8]])))   # -> [2 1]
```

```python
# 16. Count Even Values  [masking]
# Return HOW MANY values are even. Zero and negatives count when divisible.
def count_even(a):
    return np.sum(a%2==0)
print(count_even(np.array([1, 2, 3, 4, -6, 0])))   # -> 4
```

```python
# 17. Standardize 1D  [normalize]
# Return (x - mean) / std for a 1-D array. Population std (ddof=0).
def zscore(a):
    return ((a - a.mean())/np.std(a,ddof=0)).round(4)
print(zscore(np.array([2., 4., 6., 8.])))   # -> [-1.3416 -0.4472 0.4472 1.3416]
```

```python
# 18. Column Range  [axis]
# Return a 1-D array: for each COLUMN, its max minus its min.
def col_range(mat):
    return np.max(mat,axis=0) - np.min(mat,axis=0)
print(col_range(np.array([[1, 5], [3, 2], [7, 9]])))   # -> [6 7]
```

```python
# 19. Count Rows With Any Negative  [axis + any]
# Return HOW MANY rows contain at least one negative value.
def rows_with_negative(mat):
    return np.sum(np.any(mat<0 ,axis=1))
print(rows_with_negative(np.array([[1, 2], [-1, 3], [4, -5]])))   # -> 2
```

```python
# 20. Elementwise Max of Two Arrays  [vectorize]
# Return the elementwise maximum of two equal-length arrays.
def elem_max(x, y):
    return np.maximum(x,y)
print(elem_max(np.array([1, 8, 3]), np.array([4, 2, 3])))   # -> [4 8 3]
```

```python
# ============ PANDAS (21-30) ============
import pandas as pd

# 21. Filter by Condition  [masking]
# Return rows where "sales" > 100. Keep all columns, original index.
def big_sales(df):
    return df[df['sales']>100]
df21 = pd.DataFrame({"item":["A","B","C","D"], "sales":[50, 150, 200, 80]})
print(big_sales(df21))   # B and C
```
