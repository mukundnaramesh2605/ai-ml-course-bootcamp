```python
# ============ PYTHON (1-10) ============

# 1. Count Even and Odd  [logic]
# Return a tuple (evens, odds). Negatives count by parity.
def count_even_odd(nums):
    return (len(list(filter(lambda x: x%2==0,nums))),len(list(filter(lambda x: x%2!=0,nums))))
print(count_even_odd([1, 2, 3, 4, -6]))   # -> (3, 2)
print(count_even_odd([]))                 # -> (0, 0)
```

```python
# 2. First Non-Repeating Char  [string]
# Return the first character that appears exactly once. None if none.
def first_unique(s):
    for i in s:
        if s.count(i) == 1:
            return i
    return None
print(first_unique("swiss"))   # -> "w"
print(first_unique("aabb"))    # -> None
```

```python
# 3. Sum of Unique Values  [set]
# Return the sum of values that appear EXACTLY once.
def sum_unique(nums):
    return sum(list(filter(lambda x: nums.count(x)==1,nums)))
print(sum_unique([1, 2, 2, 3, 4, 4]))   # -> 4   (1 + 3)
print(sum_unique([5, 5]))               # -> 0
```

```python
# 4. Title Case  [string]
# Capitalize the first letter of each word, lowercase the rest. Collapse spaces.
def title_case(s):
    return " ".join([x[0].upper()+x[1:].lower() for x in s.split()])
print(title_case("hELLO   wORLD"))   # -> "Hello World"
print(title_case(""))                # -> ""
```

```python
# 5. Pairs Summing to Target  [hashing]
# Return the COUNT of index pairs (i < j) where nums[i] + nums[j] == target.
def count_pairs(nums, target):
    seen = {}
    count = 0
    for num in nums:
        comp = target - num
        if comp in seen:
            count +=seen[comp]
        seen[num] = seen.get(num, 0) + 1
    return count
print(count_pairs([1, 2, 3, 2, 1], 3))   # -> 4
print(count_pairs([5, 5, 5], 10))        # -> 3
```

```python
# 6. Move Zeros to End  [two-pointer]
# Move all 0s to the end, keep order of the rest. Negatives stay in order.
def move_zeros(nums):
    write = 0
    for i in range(len(nums)):
        if nums[i]!=0:
            nums[write],nums[i] = nums[i],nums[write]
            write+=1
    return nums
print(move_zeros([0, -1, 0, 3, 12]))   # -> [-1, 3, 12, 0, 0]
print(move_zeros([0, 0, 1]))           # -> [1, 0, 0]
```

```python
# 7. Valid Parentheses  [stack]
# Return True if (), [], {} are correctly matched and nested.
def is_valid(s):
    bracket = {"]":"[","}":"{",")":"("}
    stack = []
    for i in s:
        if i in bracket.values():
            stack.append(i)
        elif i in bracket:
            if not stack or stack[-1] != bracket[i]:
                return False
            stack.pop()
    return len(stack) == 0
print(is_valid("([{}])"))   # -> True
print(is_valid("([)]"))     # -> False
print(is_valid("("))        # -> False
```

```python
# 8. Compress String  [string]
# Run-length encode: "aaabbc" -> "a3b2c1". Empty -> "".
def compress(s):
    if s == "":
        return ""
    count=1
    current=s[0]
    result = []
    for i in range(1,len(s)):
        if s[i-1] == s[i]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current=s[i]
    result.append(current+str(count))
    return "".join(result)        
print(compress("aaabbc"))   # -> "a3b2c1"
print(compress("abc"))      # -> "a1b1c1"
print(compress(""))         # -> ""
```

```python
# 9. Max Subarray Sum  [logic]
# Return the largest sum of any contiguous non-empty subarray (Kadane).
def max_subarray(nums):
    current = best = nums[0]
    for num in nums[1:]:
        current = max(num,current+num)
        best = max(current,best)
    return best
print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))   # -> 6
print(max_subarray([-1, -2, -3]))                       # -> -1
```

```python
# 10. Group Anagrams  [dict]
# Group words that are anagrams. Return a list of groups, first-appearance order.
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for i in words:
        grouped["".join(sorted(i))].append(i)
    return list(grouped.values())
print(group_anagrams(["eat", "tea", "tan", "ate", "nat"]))
# -> [['eat', 'tea', 'ate'], ['tan', 'nat']]
print(group_anagrams([]))   # -> []
```

```python
# ============ NUMPY (11-20) ============
import numpy as np

# 11. Row Sums  [axis]
# Return a 1-D array of the sum of each ROW.
def row_sums(mat):
    return np.sum(mat,axis=1)
print(row_sums(np.array([[1, 2, 3], [4, 5, 6]])))   # -> [ 6 15]
```

```python
# 12. Count Above Mean  [masking]
# Return HOW MANY values are strictly greater than the array's mean.
def above_mean(a):
    return np.sum(a>a.mean())
print(above_mean(np.array([1, 2, 3, 10])))   # -> 1
```

```python
# 13. Replace Negatives With Zero  [where]
# Return the array with every negative set to 0. Do not mutate the input.
def clip_negatives(a):
    return np.where(a<0,0,a)
orig = np.array([-1, 5, -3, 0, 2])
print(clip_negatives(orig))   # -> [0 5 0 0 2]
print(orig)                   # -> [-1  5 -3  0  2]   <- unchanged
```

```python
# 14. Min-Max Normalize  [normalize]
# Scale a 1-D array to 0..1 using (x - min) / (max - min). Float.
def min_max(a):
    return (a - a.min())/(a.max() - a.min())
print(min_max(np.array([10., 20., 30.])))   # -> [0.  0.5 1. ]
```

```python
# 15. Index of Min Per Row  [argmin + axis]
# Return a 1-D array with the COLUMN index of the min value in each row.
def argmin_per_row(mat):
    return np.argmin(mat,axis=1)
print(argmin_per_row(np.array([[3, 9, 1], [8, 2, 8]])))   # -> [2 1]
```

```python
# 16. Count Even Values  [masking]
# Return HOW MANY values are even. Zero and negatives count when divisible.
def count_even(a):
    return np.sum(a%2==0)
print(count_even(np.array([1, 2, 3, 4, -6, 0])))   # -> 4
```

```python
# 17. Standardize 1D  [normalize]
# Return (x - mean) / std for a 1-D array. Population std (ddof=0).
def zscore(a):
    return ((a - a.mean())/np.std(a,ddof=0)).round(4)
print(zscore(np.array([2., 4., 6., 8.])))   # -> [-1.3416 -0.4472 0.4472 1.3416]
```

```python
# 18. Column Range  [axis]
# Return a 1-D array: for each COLUMN, its max minus its min.
def col_range(mat):
    return np.max(mat,axis=0) - np.min(mat,axis=0)
print(col_range(np.array([[1, 5], [3, 2], [7, 9]])))   # -> [6 7]
```

```python
# 19. Count Rows With Any Negative  [axis + any]
# Return HOW MANY rows contain at least one negative value.
def rows_with_negative(mat):
    return np.sum(np.any(mat<0 ,axis=1))
print(rows_with_negative(np.array([[1, 2], [-1, 3], [4, -5]])))   # -> 2
```

```python
# 20. Elementwise Max of Two Arrays  [vectorize]
# Return the elementwise maximum of two equal-length arrays.
def elem_max(x, y):
    return np.maximum(x,y)
print(elem_max(np.array([1, 8, 3]), np.array([4, 2, 3])))   # -> [4 8 3]
```

```python
# ============ PANDAS (21-30) ============
import pandas as pd

# 21. Filter by Condition  [masking]
# Return rows where "sales" > 100. Keep all columns, original index.
def big_sales(df):
    return df[df['sales']>100]
df21 = pd.DataFrame({"item":["A","B","C","D"], "sales":[50, 150, 200, 80]})
print(big_sales(df21))   # B and C
```

```python
# 22. Count Per Group as DataFrame  [groupby]
# Return a DataFrame "city" and "count": rows per city, highest first. Reset index.
def city_counts(df):
    return df.groupby('city',as_index=False).agg(count = ('city','count')).sort_values('count',ascending=False).reset_index(drop=True)
df22 = pd.DataFrame({"name":["A","B","C","D","E"], "city":["CHN","MUM","CHN","CHN","MUM"]})
print(city_counts(df22))   # CHN 3, MUM 2
```

```python
# 23. Fill Missing With Mean  [missing]
# Fill NaN in "age" with the column mean. Return the df.
def fill_mean(df):
    df['age'] = df['age'].fillna(df['age'].mean())
    return df
df23 = pd.DataFrame({"name":["A","B","C"], "age":[20.0, np.nan, 40.0]})
print(fill_mean(df23))   # B -> 30.0
```

```python
# 24. Highest Paid Per Dept  [groupby]
# Return the highest-paid row per dept. Columns: name, dept, salary. Reset index.
def top_paid(df):
    return df.sort_values('salary',ascending=False).drop_duplicates(subset=['dept']).reset_index(drop=True)
df24 = pd.DataFrame({"name":["A","B","C","D"], "dept":["IT","IT","HR","HR"], "salary":[50, 90, 60, 55]})
print(top_paid(df24))   # IT -> B(90), HR -> C(60)
```

```python
# 25. Left Merge and Fill  [merge]
# Left-merge orders with prices on "item". Missing prices become 0 (numeric).
# Return the df.
def merge_fill(orders, prices):
    return pd.merge(orders,prices,on='item',how='left').fillna(0)
orders25 = pd.DataFrame({"item":["pen","book","lamp"], "qty":[2, 1, 3]})
prices25 = pd.DataFrame({"item":["pen","book"], "price":[10, 50]})
print(merge_fill(orders25, prices25))   # lamp price -> 0
```

```python
# 26. Distinct Count Per Group  [groupby + nunique]
# Return a Series: for each "team", how many DISTINCT "project" values it has.
def distinct_projects(df):
    return df.groupby('team')['project'].nunique()
df26 = pd.DataFrame({"team":["A","A","A","B"], "project":["x","x","y","z"]})
print(distinct_projects(df26))   # A -> 2, B -> 1
```

```python
# 27. Add Running Total Per Group  [groupby transform]
# Add "cum" = cumulative SUM of "amount" within each "cust", row order. Return the df.
def running_total(df):
    df['cum'] = df.groupby('cust')['amount'].transform('cumsum')
    return df
df27 = pd.DataFrame({"cust":["X","X","Y","X"], "amount":[10, 20, 5, 30]})
print(running_total(df27))   # X: 10, 30, 60 ; Y: 5
```

```python
# 28. Crosstab Counts  [crosstab]
# Return a crosstab of "region" (rows) by "status" (cols) with counts.
def region_status(df):
    return pd.crosstab(df['region'],df['status'])
df28 = pd.DataFrame({"region":["N","N","S","S","S"], "status":["win","loss","win","win","loss"]})
print(region_status(df28))   # N: loss 1 win 1 ; S: loss 1 win 2
```

```python
# 29. Bin Into Grades  [cleaning]
# Add "grade": score >= 80 -> "A", 50..79 -> "B", below 50 -> "C". Return the df.
def grade(df):
    bins = [-1, 49, 79, float('inf')]
    labels = ["C", "B", "A"]
    df["grade"] = pd.cut(df["score"], bins=bins, labels=labels)
    return df
df29 = pd.DataFrame({"name":["A","B","C"], "score":[85, 60, 40]})
print(grade(df29))   # A, B, C
```

```python
# 30. Rows Above Group Mean  [transform + filter]
# Return rows where "score" is strictly above the row's own "team" mean.
# Keep all columns, reset index.
def above_team_mean(df):
    avg = df.groupby('team')['score'].transform('mean')
    return df[df['score'] > avg].reset_index(drop=True)
df30 = pd.DataFrame({"team":["A","A","B","B"], "score":[10, 30, 40, 50]})
print(above_team_mean(df30))   # A mean 20 -> score 30 ; B mean 45 -> score 50
```

```python
# 1. Parse Log Levels  [dict]
# Count how many log lines are at each level. Level is the first word of each line.
def count_levels(lines):
    grouped = {}
    for i in lines:
        grouped[i.split()[0]] = grouped.get(i.split()[0],0)+1
    return grouped
logs = ["INFO server started", "WARN low memory", "INFO request ok", "ERROR timeout"]
print(count_levels(logs))   # -> {'INFO': 2, 'WARN': 1, 'ERROR': 1}
print(count_levels([]))     # -> {}
```

```python
# 2. Validate IPv4 Octets  [string/logic]
# True if the string is 4 dot-separated integers, each 0..255.
def valid_ip(s):
    s= [int(i) for i in s.split('.')]
    if len(s) !=4:
        return False
    for i in s:
        if i <0 or i>255:
            return False
    return True
print(valid_ip("192.168.1.1"))   # -> True
print(valid_ip("256.1.1.1"))     # -> False  (256 out of range)
print(valid_ip("1.2.3"))         # -> False  (only 3 parts)
```

```python
# 3. Temperature Swing  [logic]
# Given daily temps, return the largest single-day increase (next - current).
# 0 if it never rises.
def max_rise(temps):
    current = best = 0
    for i in range(1,len(temps)):
        current = temps[i] - temps[i-1]
        best = max(best,current)
    return best
print(max_rise([30, 32, 28, 35, 33]))   # -> 7   (28 -> 35)
print(max_rise([40, 38, 35]))           # -> 0   (only falls)
print(max_rise([20]))                   # -> 0
```

```python
# 4. Shopping Cart Total  [dict]
# Given a cart {item: qty} and prices {item: price}, return the total cost.
# Items missing from prices cost 0.
def cart_total(cart, prices):
    return sum([prices.get(k,0)*v for k,v in cart.items()])
cart = {"milk": 2, "bread": 1, "gum": 5}
prices = {"milk": 3.0, "bread": 2.5}
print(cart_total(cart, prices))   # -> 8.5   (6 + 2.5 + 0)
print(cart_total({}, prices))     # -> 0
```

```python
# 5. Deduplicate Emails  [set/string]
# Normalize emails (lowercase, strip surrounding spaces) and return the count of
# distinct addresses.
def unique_emails(emails):
    return len(set([x.lower().strip() for x in emails]))
print(unique_emails([" Alice@X.com", "alice@x.com ", "BOB@x.com"]))   # -> 2
print(unique_emails([]))                                              # -> 0
```

```python
# 6. Daily Returns  [vectorize]
# Given closing prices, return the daily percent change: (today - yesterday)/yesterday.
# Output length is len(prices) - 1. Round to 4 decimals.
def daily_returns(prices):
    returns = (prices[1:] - prices[:-1]) / prices[:-1]
    return np.round(returns, 4)
print(daily_returns(np.array([100., 110., 99.])))   # -> [ 0.1   -0.1 ]
```

```python
# 7. Clip Sensor Noise  [clip + count]
# Clip readings to [0, 100], then return how many were out of range (changed).
def clip_and_count(readings):
    return np.sum(readings != np.clip(readings,0,100))
print(clip_and_count(np.array([-3, 50, 120, 0, 100])))   # -> 2
```

```python
# 8. Pixel Brightness Mask  [masking]
# Given a 2-D grayscale image, return HOW MANY pixels are brighter than threshold.
def bright_pixels(img, thresh):
    return np.sum(img>thresh)
img = np.array([[10, 200], [150, 30]])
print(bright_pixels(img, 100))   # -> 2   (200, 150)
```

```python
# 9. Normalize Feature Columns  [axis + broadcast]
# Min-max scale each COLUMN of a 2-D feature matrix to 0..1.
def scale_features(mat):
    col_min = mat.min(axis=0)
    col_max = mat.max(axis=0)
    col_range = col_max - col_min
    col_range[col_range == 0] = 1.0
    return (mat - col_min) / col_range
print(scale_features(np.array([[1., 10.], [2., 20.], [3., 30.]])))
# col0 -> [0, 0.5, 1] ; col1 -> [0, 0.5, 1]
```

```python
# 10. Rolling 3-Day Average  [window]
# Given a 1-D array, return the 3-element moving average. Output length len - 2.
def rolling3(a):
    return np.convolve(a, np.ones(3)/3, mode='valid')
print(rolling3(np.array([2., 4., 6., 8.])))   # -> [4. 6.]
```

```python
# 11. Active Users Last 7 Days  [datetime + filter]
# Return the count of DISTINCT user_ids whose "last_seen" is on or after the cutoff.
def active_count(df, cutoff):
    last_seen = pd.to_datetime(df["last_seen"])
    cutoff_dt = pd.to_datetime(cutoff)
    return df.loc[last_seen >= cutoff_dt, "user_id"].nunique()
df11 = pd.DataFrame({
    "user_id": [1, 2, 2, 3],
    "last_seen": ["2026-11-01", "2026-11-03", "2026-10-20", "2026-11-04"],
})
print(active_count(df11, "2026-11-01"))   # -> 3   (users 1, 2, 3 each have a qualifying row)
```

```python
# 12. Revenue by Product Category  [groupby]
# Return a DataFrame "category" and "revenue" (sum of price*qty), sorted desc. Reset index.
def category_revenue(df):
    df['revenue'] = df['price']*df['qty']
    return df[['category','revenue']].sort_values('revenue',ascending=False).reset_index(drop=True)
df12 = pd.DataFrame({
    "category": ["books", "toys", "books", "toys"],
    "price": [10, 5, 20, 8],
    "qty": [2, 3, 1, 2],
})
print(category_revenue(df12))   # books 40, toys 31 -> books 40, toys 31
```

```python
# 13. Flag Churn Risk  [transform + feature eng]
# Add "at_risk" = True where a customer's "logins" is below their own "plan" average.
# Return the df.
def flag_risk(df):
    avg = df.groupby('plan')['logins'].transform('mean').astype(int)
    df['at_risk'] =df['logins'] < avg
    return df
df13 = pd.DataFrame({
    "plan": ["free", "free", "pro", "pro"],
    "logins": [2, 10, 5, 15],
})
print(flag_risk(df13))   # free avg 6: 2 True, 10 False ; pro avg 10: 5 True, 15 False

```

```python
# 14. Merge Orders With Customers  [merge]
# Left-merge orders with customers on "cust_id". Orders with no matching customer
# get name "Guest". Return the df.
def label_orders(orders, customers):
    merged = pd.merge(orders,customers,on='cust_id',how='left').fillna('Guest')
    return merged[merged['name']=='Guest']
orders14 = pd.DataFrame({"order_id": [1, 2, 3], "cust_id": [10, 20, 99]})
customers14 = pd.DataFrame({"cust_id": [10, 20], "name": ["Ravi", "Priya"]})
print(label_orders(orders14, customers14))   # cust 99 -> "Guest"
```

```python
# 15. Top Spender Per City  [groupby]
# Return the highest-spending customer row per city. Columns: city, name, spend.
# Reset index.
def top_spender(df):
    return df.sort_values('spend',ascending=False).drop_duplicates(subset=['city']).reset_index(drop=True)
df15 = pd.DataFrame({
    "city": ["CHN", "CHN", "BLR", "BLR"],
    "name": ["A", "B", "C", "D"],
    "spend": [500, 900, 700, 650],
})
print(top_spender(df15))   # CHN -> B(900), BLR -> C(700)
```
