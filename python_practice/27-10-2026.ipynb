```python
# 1. Sum of Digits  [Service / logic]
# Return the sum of the digits of a non-negative integer.
# 1234 -> 10
def digit_sum(n):
    return sum([int(x) for x in str(n)])
print(digit_sum(1234))# -> 10 ; 
print(digit_sum(0))# -> 0
```

```python
# 2. Count Words  [Service / string]
# Return how many words are in a sentence.
# "I am learning" -> 3
def word_count(s):
    return len(s.split())
print(word_count("I am learning"))# -> 3 ; 
print(word_count(""))# -> 0
```

```python
# 3. Remove Vowels  [Service / string]
# Return the string with all vowels removed (case-insensitive).
# "Hello" -> "Hll"
import re
def remove_vowels(s):
    without_vowels = re.findall(r'[^aeiouAEIOU]',s)
    return "".join(without_vowels)
print(remove_vowels("Hello"))# -> "Hll" ; remove_vowels("xyz") -> "xyz"
```
