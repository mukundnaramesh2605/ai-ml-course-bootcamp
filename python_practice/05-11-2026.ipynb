```python
# 1. Type: hashing
# Return the indices [i, j] (i < j) of the two numbers that add up to target.
# Exactly one pair is guaranteed when a solution exists; if none, return [].
def two_sum(nums, target):
    seen = {}
    for idx,num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return [seen[comp],idx]
        seen[num] = idx
    return []

print(two_sum([2, 7, 11, 15], 9))     # [0, 1]
print(two_sum([3, 3], 6))             # [0, 1]
print(two_sum([-3, 4, 1, -1], 0))     # [2, 3]
print(two_sum([1, 2, 3], 100))        # []
```
