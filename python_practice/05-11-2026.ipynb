```python
# 1. Type: hashing
# Return the indices [i, j] (i < j) of the two numbers that add up to target.
# Exactly one pair is guaranteed when a solution exists; if none, return [].
def two_sum(nums, target):
    seen = {}
    for idx,num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return [seen[comp],idx]
        seen[num] = idx
    return []

print(two_sum([2, 7, 11, 15], 9))     # [0, 1]
print(two_sum([3, 3], 6))             # [0, 1]
print(two_sum([-3, 4, 1, -1], 0))     # [2, 3]
print(two_sum([1, 2, 3], 100))        # []
```

```python
# 2. Type: hashing
# Return True if s and t are anagrams. Case-insensitive, ignore spaces.
# Different letter counts -> False.
def is_anagram(s, t):
    s = [i for i in s.lower() if i.isalnum()]
    t = [i for i in t.lower() if i.isalnum()]
    return sorted(s) == sorted(t)
print(is_anagram("Listen", "Silent"))         # True
print(is_anagram("Dormitory", "Dirty Room"))  # True
print(is_anagram("aab", "abb"))               # False
print(is_anagram("abc", "abcd"))              # False
```

```python
# 3. Type: stack
# Return True if the brackets ()[]{} are validly matched and nested.
def is_valid(s):
    brackets = {"}":"{","]":"[",")":"("}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0

print(is_valid("()[]{}"))   # True
print(is_valid("([{}])"))   # True
print(is_valid("(]"))       # False
print(is_valid("("))        # False
print(is_valid(""))         # True
print(is_valid("]"))        # False
```

```python
# 4. Type: two-pointer
# Move all zeros to the end in place, keeping the order of non-zero values.
# Return the modified list.
def move_zeros(nums):
    non_zero_ptr = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[non_zero_ptr],nums[i] = nums[i],nums[non_zero_ptr]
            non_zero_ptr+=1
    return nums

print(move_zeros([0, 1, 0, 3, 12]))    # [1, 3, 12, 0, 0]
print(move_zeros([-5, 0, -2, 0, 4]))   # [-5, -2, 4, 0, 0]
print(move_zeros([1, 2, 3]))           # [1, 2, 3]
print(move_zeros([0, 0, 0]))           # [0, 0, 0]
```

```python
# 5. Type: single-pass / math
# nums holds n distinct numbers taken from 0..n (one is missing). Return the missing one.
def missing_number(nums):
    n = len(nums)
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)
print(missing_number([3, 0, 1]))                    # 2
print(missing_number([0, 1, 2]))                    # 3
print(missing_number([1, 2, 3]))                    # 0
print(missing_number([9, 6, 4, 2, 3, 5, 7, 0, 1]))  # 8
```

```python
# 6. Type: single-pass
# Compress each run of a repeated character into char+count.
# Every character gets a count, including 1. Case-sensitive. "aaabbc" -> "a3b2c1".
def compress(s):
    if s == "": return ""
    count = 1
    current = s[0]
    result = []
    for i in range(1,len(s)):
        if s[i-1] == s[i]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current=s[i]
    
    result.append(current+str(count))
    return "".join(result)
print(compress("aaabbc"))   # "a3b2c1"
print(compress("abcd"))     # "a1b1c1d1"
print(compress("aaAA"))     # "a2A2"
print(compress("x"))        # "x1"
print(compress(""))         # ""
```

```python
# 7. Type: string scan
# Return the longest common prefix shared by all strings. "" if there is none.
def longest_common_prefix(words):
    result = []
    for char in zip(*words):
        if len(set(char)) == 1:
            result.append(char[0])
        else:
            break
    return "".join(result)

print(longest_common_prefix(["flower", "flow", "flight"]))      # "fl"
print(longest_common_prefix(["dog", "cat", "car"]))            # ""
print(longest_common_prefix(["same", "same"]))                 # "same"
print(longest_common_prefix(["interview", "intern", "inter"])) # "inter"
print(longest_common_prefix(["abc", "", "ab"]))                # ""
```

```python
# 8. Type: hashing
# Return the index of the first character that appears exactly once. -1 if none.
# Case-sensitive.
def first_unique(s):
    for i in range(len(s)):
        if s.count(s[i]) == 1:
            return i
    return -1

print(first_unique("leetcode"))      # 0
print(first_unique("loveleetcode"))  # 2
print(first_unique("aabb"))          # -1
print(first_unique("aabbc"))         # 4
```

```python
# 9. Type: single-pass (Kadane)
# Return the largest sum of any contiguous, non-empty subarray.
def max_subarray(nums):
    current = best = nums[0]
    for i in (nums[1:]):
        current = max(i, current+i)
        best = max(current,best)
    return best

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # 6
print(max_subarray([-5, -2, -8, -1]))                 # -1
print(max_subarray([5]))                              # 5
print(max_subarray([3, -1, 4]))                       # 6
```

```python
# 10. Type: hashing
# Group words that are anagrams of each other.
# Return a list of groups; sort each group alphabetically, and sort the
# outer list by each group's first word.
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for word in words:
        grouped["".join(sorted(word))].append(word)

    for k,v in grouped.items():
        grouped[k] = sorted(v)
    return list(sorted(grouped.values()))

print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
# [['ate', 'eat', 'tea'], ['bat'], ['nat', 'tan']]
print(group_anagrams(["abc", "cba", "xyz"]))
# [['abc', 'cba'], ['xyz']]
print(group_anagrams(["a"]))
# [['a']]
```

```python
import numpy as np

# 1. Type: axis reduction
# Return a 1D array holding the mean of each COLUMN.
def col_means(arr):
    return np.mean(arr,axis=0)

print(col_means(np.array([[1, 2, 3], [4, 5, 6]])))      # [2.5 3.5 4.5]
print(col_means(np.array([[2, 4], [6, 8], [10, 12]])))  # [6. 8.]
```

```python
# 2. Type: axis reduction
# Return a 1D array holding the maximum of each ROW.
def row_max(arr):
    return np.max(arr,axis=1)

print(row_max(np.array([[1, 9, 3], [7, 2, 5]])))   # [9 7]
print(row_max(np.array([[-1, -5], [-3, -2]])))     # [-1 -2]
```

```python
# 3. Type: boolean masking
# Return how many elements are STRICTLY greater than threshold.
def count_above(arr, threshold):
    return np.sum(arr>threshold)

print(count_above(np.array([1, 5, 3, 8, 5]), 5))   # 2
print(count_above(np.array([-2, -1, 0, 3]), -1))   # 2
```

```python
# 4. Type: normalization
# Scale the 1D array to the range [0, 1] using (x - min) / (max - min).
# If all values are equal, return an array of zeros of the same shape.
def min_max_normalize(arr):
    data_range = arr.max() - arr.min()
    if data_range == 0:
        return np.zeros_like(arr, dtype=float)
    normalised = ((arr - arr.min()) /( arr.max() - arr.min())).round(3)
    return normalised

print(min_max_normalize(np.array([10, 20, 30, 40])))  # [0.    0.333 0.667 1.   ]
print(min_max_normalize(np.array([5, 5, 5])))         # [0. 0. 0.]
```

```python
# 5. Type: normalization
# Return the z-score of each element: (x - mean) / std.
# Use population std (np.std default, ddof=0).
def zscore(arr):
    return ((arr - arr.mean())/np.std(arr,ddof=0)).round(3)

print(zscore(np.array([2, 4, 6, 8])))   # [-1.342 -0.447  0.447  1.342]
print(zscore(np.array([1, 1, 2, 2])))   # [-1. -1.  1.  1.]
```

```python
# 6. Type: where / clip
# Return a new array with every value limited to [low, high].
# Values below low become low; values above high become high.
def clip_range(arr, low, high):
    return np.clip(arr,low,high)

print(clip_range(np.array([-3, 0, 5, 12, 7]), 0, 10))  # [ 0  0  5 10  7]
print(clip_range(np.array([1, 2, 3]), 0, 10))          # [1 2 3]
```

```python
# 7. Type: where
# Return a new array where every negative value is replaced with 0,
# and all other values are left unchanged.
def zero_negatives(arr):
    return np.where(arr<0,0,arr)

print(zero_negatives(np.array([-1, 4, -7, 0, 9])))   # [0 4 0 0 9]
print(zero_negatives(np.array([2, 3, 5])))           # [2 3 5]
```

```python
# 8. Type: argmax / axis reduction
# Return the index of the ROW with the largest sum.
# On a tie, return the first such row.
def max_sum_row(arr):
    return np.argmax(np.sum(arr,axis=1))

print(max_sum_row(np.array([[1, 2], [5, 5], [3, 3]])))    # 1
print(max_sum_row(np.array([[-1, -1], [2, 2], [4, 0]])))  # 1
```

```python
# 9. Type: reshape
# Reshape the 1D array into a 2D array with exactly 2 columns
# (let numpy infer the number of rows). Return it.
def to_two_cols(arr):
    return np.reshape(arr,(-1,2))

print(to_two_cols(np.array([1, 2, 3, 4, 5, 6])))
# [[1 2]
#  [3 4]
#  [5 6]]
```

```python
# 10. Type: axis ops / broadcasting
# Return a float array where each ROW is divided by its own sum,
# so every row sums to 1.
def row_proportions(arr):
    row_sum = np.sum(arr,axis=1,keepdims=True)
    return np.divide(arr,row_sum)

print(row_proportions(np.array([[1, 1, 2], [2, 3, 5]])))
# [[0.25 0.25 0.5 ]
#  [0.2  0.3  0.5 ]]
```

```python
import pandas as pd
import numpy as np

# 1. Type: missing-value handling / groupby transform
# Fill missing values in 'score' with the MEDIAN score of that row's 'team'.
# Return the full DataFrame.
def fill_group_median(df):
    df['score'] = df['score'].fillna(df.groupby('team')['score'].transform('median'))
    return df

df1 = pd.DataFrame({
    "team": ["A", "A", "A", "B", "B"],
    "score": [10, np.nan, 20, 7, np.nan]
})
print(fill_group_median(df1))
# A median(10,20)=15 -> row1 = 15 ; B median = 7 -> row4 = 7
```
