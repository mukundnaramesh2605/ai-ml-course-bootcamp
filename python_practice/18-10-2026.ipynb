```python
# 1. Most Frequent Word  [Service / string]
# Return the word that appears most often, case-insensitive.
# If there's a tie, return the one that appears FIRST in the text.
# "the cat the dog" -> "the"
from collections import Counter
def most_frequent(s):
    counts = Counter(s.split())
    return [k for k,v in counts.items() if v==max(counts.values())][0]
print(most_frequent("the cat the dog"))# -> "the" ; 
print(most_frequent("a b a b"))# -> "a"
```

```python
# 2. Group by First Letter  [Service / dict]
# Return a dict: first letter -> list of words starting with it,
# words kept in their original order.
# ["apple","banana","avocado"] -> {"a":["apple","avocado"], "b":["banana"]}
from collections import defaultdict
def group_by_first(words):
    grouped = defaultdict(list)
    for i in words:
        grouped[i.lower()[0]].append(i)
    return dict(grouped)
print(group_by_first(["apple","banana","avocado"]))# -> {"a":["apple","avocado"],"b":["banana"]}

```

```python
# 3. Running Difference  [Service / logic]
# Return a list where each element is the difference from the PREVIOUS
# element. The first element has no previous, so start the output at index 1.
# [10, 7, 7, 2] -> [-3, 0, -5]
def diffs(nums):
    return [nums[i]-nums[i-1] for i in range(1,len(nums))]
print(diffs([10,7,7,2]))# -> [-3,0,-5] ; 
print(diffs([5]))# -> [] ; diffs([]) -> []
```
