```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
# 1. Censor Banned Words  [easy / string]
# Replace every word that is in banned (CASE-INSENSITIVE exact match) with "*" of the SAME LENGTH.
# KEEP ALL SPACES EXACTLY as in the input. Punctuation stays part of the word.
# "Bad  day, bad dog", ["bad"] -> "***  day, *** dog"
def censor(s, banned):
    banned = [i.lower() for i in banned]
    return " ".join([word if word.lower() not in banned else '***' for word in s.split()])

print(repr(censor("Bad  day, bad dog", ["bad"])))   # -> '***  day, *** dog'   <- double space must survive
print(repr(censor("", ["x"])))                      # -> ''
print(repr(censor("hello world", [])))              # -> 'hello world'
print(repr(censor("BAD bad Bad", ["bad"])))         # -> '*** *** ***'
print(repr(censor(" cat ", ["cat"])))               # -> ' *** '   <- leading/trailing spaces kept
```

```python
# 2. Word Length Buckets  [easy / dict]
# Return a dict with keys EXACTLY "short", "medium", "long" (in that order), ALWAYS all three.
# short: len <= 3, medium: 4 to 6 INCLUSIVE, long: len >= 7. Values are COUNTS.
# ["a","tree","giraffe","cat"] -> {'short': 2, 'medium': 1, 'long': 1}
def length_buckets(words):
    bucket = {"short":0,"medium":0,"long":0}
    for word in words:
        if len(word) <= 3:
            bucket['short']+=1
            
        elif len(word) >= 4 and len(word) <=6 :
            bucket['medium']+=1
            
        else:
            bucket['long']+=1
        
    return bucket
print(length_buckets(["a", "tree", "giraffe", "cat"]))           # -> {'short': 2, 'medium': 1, 'long': 1}
print(length_buckets([]))                                        # -> {'short': 0, 'medium': 0, 'long': 0}   <- keys still present
print(length_buckets(["elephant"]))                              # -> {'short': 0, 'medium': 0, 'long': 1}
print(length_buckets(["abc", "abcd", "abcdef", "abcdefg"]))      # -> {'short': 1, 'medium': 2, 'long': 1}   <- boundaries 3/4, 6/7
print(length_buckets([""]))                                      # -> {'short': 1, 'medium': 0, 'long': 0}
```

```python
# 3. Top N Longest Distinct Words  [medium / sort + slice]
# Return the n longest DISTINCT words. Sort by length DESC, ties ALPHABETICAL ASC. Then take the FIRST n.
# ["pear","grape","fig","apple","kiwi"], 2 -> ['apple', 'grape']
def top_longest(words, n):
    return sorted(words,key=lambda x: [-len(x),x])[:2]

print(top_longest(["pear", "grape", "fig", "apple", "kiwi"], 2))  # -> ['apple', 'grape']   <- tie on length 5, grape comes first in input
print(top_longest([], 3))                                         # -> []
print(top_longest(["sun"], 5))                                    # -> ['sun']   <- n bigger than list
print(top_longest(["bb", "aa", "cc", "aa"], 2))                   # -> ['aa', 'bb']   <- duplicate "aa" counted once
print(top_longest(["x", "yy"], 0))                                # -> []
```

```python
# 4. Prefix Sums  [easy / list]
# Return a NEW list where element i is the sum of nums[0..i] INCLUSIVE. Do not use itertools.
# [1, 2, 3] -> [1, 3, 6]
def prefix_sums(nums):
    if not nums:
        return []
    result = [nums[0]]
    for i in nums[1:]:
        result.append(result[-1]+i)
    return result

print(prefix_sums([1, 2, 3]))         # -> [1, 3, 6]
print(prefix_sums([]))                # -> []
print(prefix_sums([5]))               # -> [5]
print(prefix_sums([3, -3, 4, -4]))    # -> [3, 0, 4, 0]   <- negatives bring the total back to 0
print(prefix_sums([0, 0, 0]))         # -> [0, 0, 0]
```

```python
# 5. Words Seen Exactly K Times  [medium / Counter + order]
# Return DISTINCT words whose COUNT is EXACTLY k, in FIRST-APPEARANCE order (NOT sorted). Case-SENSITIVE.
# ["b","a","b","c","a","d"], 2 -> ['b', 'a']
def exactly_k(words, k):
    counts = Counter(words)
    return [ke for ke,v in counts.items() if v == k]

print(exactly_k(["b", "a", "b", "c", "a", "d"], 2))   # -> ['b', 'a']   <- sorted() would give ['a', 'b'], wrong
print(exactly_k([], 1))                                # -> []
print(exactly_k(["x"], 1))                             # -> ['x']
print(exactly_k(["Go", "go", "go"], 1))                # -> ['Go']   <- "Go" and "go" are different
print(exactly_k(["a", "a"], 3))                        # -> []   <- no match
```

```python
# 6. Count In Range  [easy / mask]
# Return the COUNT of values with lo <= x <= hi (INCLUSIVE both ends) as an int.
def count_in_range(a, lo, hi):
    mask = (a>=lo) & (a<=hi)
    return sum(mask)

a26 = np.array([1, 5, 10, -2, 15, 10, 4])
print(count_in_range(a26, 5, 10))   # -> 3   <- 5, 10, 10 (both boundaries count, 10 twice)
```

```python
# 7. Running Total Per Row  [easy / cumsum axis]
# Return a 2D array of the same shape where each row is the running total ACROSS that row.
def row_running_total(a):
    return np.cumsum(a,axis=1)

a27 = np.array([[1, 2, 3],
                [4, -1, 0]])
print(row_running_total(a27))
# -> [[1 3 6]
#     [4 3 3]]   <- per ROW, not down the columns
```

```python
# 8. Row Index Of Column Minimum  [easy / argmin axis]
# Return a 1D array: for each COLUMN, the ROW index of its minimum. Ties -> FIRST row.
def col_min_rows(a):
    return np.argmin(a,axis=0)

a28 = np.array([[3, 7, 2],
                [1, 9, 2],
                [1, 4, 5]])
print(col_min_rows(a28))   # -> [1 2 0]   <- col 0 has 1 twice (rows 1, 2), col 2 has 2 twice (rows 0, 1)
```

```python
# 9. Row Percent Share  [medium / keepdims + zero guard]
# Return each value as PERCENT of its row total, ROUNDED to 1 decimal, float array same shape.
# If a row total is 0, that row is ALL 0.0 (no nan, no warning).
def row_percent(a):
    row_totals = np.sum(a, axis=1, keepdims=True)
    pct = np.divide(
        a * 100.0,
        row_totals,
        out=np.zeros_like(a, dtype=float),
        where=row_totals != 0
    )
    return np.round(pct, 1)

a29 = np.array([[1, 1, 2],
                [0, 0, 0],
                [5, 0, 15]])
print(row_percent(a29))
# -> [[25. 25. 50.]
#     [ 0.  0.  0.]   <- zero-total row
#     [25.  0. 75.]]
```

```python
# 10. Sum Of Top 2 Per Row  [medium / sort + slice]
# Return a 1D array: for each row, the sum of its 2 LARGEST values (duplicates count separately).
def top2_row_sum(a):
    sorted_rows = np.sort(a, axis=1)
    return np.sum(sorted_rows[:, -2:], axis=1)

a30 = np.array([[4, 9, 1],
                [-5, -1, -3],
                [7, 7, 2]])
print(top2_row_sum(a30))   # -> [13 -4 14]   <- row 1 all negative (-1 + -3), row 2 has 7 twice
```

```python
# 11. Names Starting With A  [easy / str filter]
# Return rows where name starts with UPPERCASE "A" (CASE-SENSITIVE).
# Columns EXACTLY ["name", "city"], keep original row order, RESET index.
def names_with_a(df):
    return df[df['name'].str.startswith('A')][['name','city']].reset_index(drop=True)

df31 = pd.DataFrame({
    "name": ["Asha", "arun", "Bala", "Anil", "Ravi"],
    "city": ["Chennai", "Pune", "Delhi", "Mumbai", "Chennai"],
    "age":  [25, 30, 28, 35, 22],
})
print(names_with_a(df31))
# ->    name     city
#    0  Asha  Chennai
#    1  Anil   Mumbai      <- "arun" is lowercase, NOT included; no "age" column
```

```python
# 12. Salary Band  [easy / between + sort]
# Keep rows with 30000 <= salary <= 70000 (INCLUSIVE). Sort salary DESC, ties name ASC.
# Columns ["name", "salary"], RESET index.
def salary_band(df):
    return df[(df['salary']>=30000)&(df['salary']<=70000)].sort_values(['salary','name'],ascending=[False,True]).reset_index(drop=True)

df32 = pd.DataFrame({
    "name":   ["Zara", "Mani", "Kavi", "Arjun", "Devi"],
    "salary": [50000, 70000, 30000, 50000, 70001],
})
print(salary_band(df32))
# ->     name  salary
#    0   Mani   70000
#    1  Arjun   50000   <- tie: Arjun before Zara even though Zara is first in input
#    2   Zara   50000
#    3   Kavi   30000      <- Devi 70001 is out
```

```python
# 13. Decode Status Codes  [easy / replace vs map]
# Return a NEW DataFrame with status codes mapped: "P"->"Pass", "F"->"Fail", "A"->"Absent".
# Codes NOT in the mapping stay UNCHANGED. Columns ["id", "status"]. Do NOT modify df.
def decode_status(df):
    df2 = df.copy()
    mapping = {"P":'Pass','F':'Fail','A':'Absent'}
    df2['stats']=df['status'].map(lambda x:mapping.get(x,x))
    return df2

df33 = pd.DataFrame({"id": [1, 2, 3, 4, 5], "status": ["P", "F", "P", "A", "X"]})
print(decode_status(df33))
# ->    id  status
#    0   1    Pass
#    1   2    Fail
#    2   3    Pass
#    3   4  Absent
#    4   5       X     <- unmapped "X" stays "X", not NaN
```
