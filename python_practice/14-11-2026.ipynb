```python
# 1. Longest Strictly Increasing Run  [walk-and-track]
# Return the LENGTH of the longest contiguous run where each value is STRICTLY greater than the one before.
# [1,2,3,1,2] -> 3
def longest_increasing_run(nums):
    if not nums:
        return 0
    count = best = 1
    for i in range(1,len(nums)):
        if nums[i] > nums[i-1]:
            count+=1
        else:
            best = max(best,count)
            count=1
    return max(best,count)

print(longest_increasing_run([1, 2, 3, 1, 2]))      # -> 3
print(longest_increasing_run([5, 4, 3]))            # -> 1
print(longest_increasing_run([]))                   # -> 0
print(longest_increasing_run([1, 2, 2, 3, 4, 5]))   # -> 4   <- equal values break the run, and the best run is the LAST one
print(longest_increasing_run([7]))                  # -> 1
```

```python
# 2. Summary Ranges  [walk-and-track]
# nums is sorted and DISTINCT. Return a list of strings: "a->b" for a consecutive run, "a" for a single value.
# [0,1,2,4,5,7] -> ["0->2", "4->5", "7"]
def summary_ranges(nums):
    if not nums:
        return []
    ranges = []
    start = nums[0]
    for i in range(1, len(nums)):
        if nums[i] != nums[i - 1] + 1:
            end = nums[i - 1]
            ranges.append(f"{start}->{end}" if start != end else str(start))
            start = nums[i]
    end = nums[-1]
    ranges.append(f"{start}->{end}" if start != end else str(start))
    return ranges
    

print(summary_ranges([0, 1, 2, 4, 5, 7]))   # -> ['0->2', '4->5', '7']   <- don't drop the last group
print(summary_ranges([]))                   # -> []
print(summary_ranges([3]))                  # -> ['3']
print(summary_ranges([-3, -2, 0]))          # -> ['-3->-2', '0']
print(summary_ranges([1, 3, 5]))            # -> ['1', '3', '5']
```

```python
# 3. Kth Distinct String  [hashing]
# Return the k-th (1-based) string that appears EXACTLY once, in original order. "" if there are fewer than k.
# ["d","b","c","b","c","a"], 2 -> "a"
def kth_distinct(arr, k):
    seen = [i for i in arr if arr.count(i)==1]
    return seen[k-1] if seen else ''

print(kth_distinct(["d", "b", "c", "b", "c", "a"], 2))   # -> 'a'
print(kth_distinct(["a", "b", "a"], 1))                  # -> 'b'   <- "a" repeats, so it doesn't count
print(kth_distinct([], 1))                               # -> ''
print(kth_distinct(["x", "y", "z"], 3))                  # -> 'z'
print(kth_distinct(["p", "p"], 1))                       # -> ''
```

```python
# 4. Isomorphic Strings  [hashing]
# True if each char in s can be replaced to get t, with a ONE-TO-ONE mapping (no two chars map to the same char).
# "egg", "add" -> True
from collections import defaultdict
def is_isomorphic(s, t):
    return len(s) == len(t) and len(set(zip(s, t))) == len(set(s)) == len(set(t))
    

print(is_isomorphic("egg", "add"))     # -> True
print(is_isomorphic("foo", "bar"))     # -> False
print(is_isomorphic("badc", "baba"))   # -> False   <- b->b and d->b: two chars map to one
print(is_isomorphic("", ""))           # -> True
print(is_isomorphic("ab", "a"))        # -> False
```

```python
# 5. Decode Run-Length  [string]
# Reverse of RLE: each lowercase letter is followed by its count (count can be MULTI-DIGIT or 0).
# "a3b2c1" -> "aaabbc"
import re
def decode(s):
    nums = re.findall(r'[0-9]+',s)
    alpha = re.findall(r'[a-zA-Z]+',s)
    return "".join(i*int(j) for i,j in zip(alpha,nums))
print(decode("a3b2c1"))   # -> 'aaabbc'
print(decode("a10"))      # -> 'aaaaaaaaaa'   <- 10, not 1 then 0
print(decode(""))         # -> ''
print(decode("x1y1"))     # -> 'xy'
print(decode("z0a2"))     # -> 'aa'
```

```python
# 6. Running Max  [cumulative]
# Return an array where position i is the max of a[0..i]. No Python loops.
def running_max(a):
    return np.maximum.accumulate(a)

a6 = np.array([3, 1, 4, 1, 5, -9, 2])
print(running_max(a6))   # -> [3 3 4 4 5 5 5]
```

```python
# 7. Replace Outliers With Median  [masking]
# Values whose distance from the mean is STRICTLY more than 1.5 * std (ddof=0) become the median of the ORIGINAL array.
# Return a NEW float array. Do NOT modify the input.
def replace_outliers(a):
    arr = np.asarray(a, dtype=float)
    mean = np.mean(arr)
    std = np.std(arr, ddof=0)
    median = np.median(arr)
    
    outlier_mask = np.abs(arr - mean) > 1.5 * std
    return np.where(outlier_mask, median, arr)

a7 = np.array([10., 12., 11., 13., 100., 12., -50.])
print(replace_outliers(a7))   # -> [10. 12. 11. 13. 12. 12. 12.]   <- both 100 and -50 are outliers; median is 12
```
