```python
# 1. Longest Strictly Increasing Run  [walk-and-track]
# Return the LENGTH of the longest contiguous run where each value is STRICTLY greater than the one before.
# [1,2,3,1,2] -> 3
def longest_increasing_run(nums):
    if not nums:
        return 0
    count = best = 1
    for i in range(1,len(nums)):
        if nums[i] > nums[i-1]:
            count+=1
        else:
            best = max(best,count)
            count=1
    return max(best,count)

print(longest_increasing_run([1, 2, 3, 1, 2]))      # -> 3
print(longest_increasing_run([5, 4, 3]))            # -> 1
print(longest_increasing_run([]))                   # -> 0
print(longest_increasing_run([1, 2, 2, 3, 4, 5]))   # -> 4   <- equal values break the run, and the best run is the LAST one
print(longest_increasing_run([7]))                  # -> 1
```

```python
# 2. Summary Ranges  [walk-and-track]
# nums is sorted and DISTINCT. Return a list of strings: "a->b" for a consecutive run, "a" for a single value.
# [0,1,2,4,5,7] -> ["0->2", "4->5", "7"]
def summary_ranges(nums):
    if not nums:
        return []
    ranges = []
    start = nums[0]
    for i in range(1, len(nums)):
        if nums[i] != nums[i - 1] + 1:
            end = nums[i - 1]
            ranges.append(f"{start}->{end}" if start != end else str(start))
            start = nums[i]
    end = nums[-1]
    ranges.append(f"{start}->{end}" if start != end else str(start))
    return ranges
    

print(summary_ranges([0, 1, 2, 4, 5, 7]))   # -> ['0->2', '4->5', '7']   <- don't drop the last group
print(summary_ranges([]))                   # -> []
print(summary_ranges([3]))                  # -> ['3']
print(summary_ranges([-3, -2, 0]))          # -> ['-3->-2', '0']
print(summary_ranges([1, 3, 5]))            # -> ['1', '3', '5']
```
