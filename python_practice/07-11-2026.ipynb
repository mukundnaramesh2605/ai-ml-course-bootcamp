```python
# 1. Reverse a string without slicing or reverse()
def reverse_string(s):
    # your code here
    result = ""
    for char in s:
        result = char+result
    return result

print(reverse_string("hello"))
print(reverse_string(""))           # empty
print(reverse_string("a"))          # single char
print(reverse_string("racecar"))    # palindrome
print(reverse_string("12345"))      # digits
```

```python
# 2. Check if a string is a palindrome (ignore spaces and case)
def is_palindrome(s):
    s = "".join([i for i in s.lower() if i.isalnum()])
    return s == s[::-1]

print(is_palindrome("Race car"))                        # spaces + case
print(is_palindrome(""))                                # empty
print(is_palindrome("a"))                               # single char
print(is_palindrome("Was it a car or a cat I saw"))     # full sentence
print(is_palindrome("hello"))                           # not a palindrome
```

```python
# 3. First non-repeating character in a string
def first_unique_char(s):
    if s == "":
        return s
    for i in s:
        if s.count(i) == 1:
            return i
    return None

print(first_unique_char("swiss"))     # normal
print(first_unique_char(""))          # empty
print(first_unique_char("aabbcc"))    # none unique
print(first_unique_char("x"))         # single char
print(first_unique_char("aabbc"))     # unique at end
```

```python
# 4. Word frequency, return top 2
from collections import Counter
def top_two_words(sentence):
    counts = Counter(sentence.lower().split())
    counts = dict(sorted(counts.items(),key=lambda x: x[1],reverse=True))
    return list(counts.keys())[:2]
print(top_two_words("the cat sat on the mat the cat ran"))  # normal
print(top_two_words("hello"))                               # single word
print(top_two_words(""))                                    # empty
print(top_two_words("a b a b c"))                           # tie
print(top_two_words("one two"))                             # exactly two
```

```python
# 6. Check if two strings are anagrams
def are_anagrams(a, b):
    a = sorted([i for i in a if i.isalnum()])
    b = sorted([i for i in b if i.isalnum()])
    return a == b

print(are_anagrams("listen", "silent"))  # normal
print(are_anagrams("", ""))              # both empty
print(are_anagrams("a", "b"))            # single char, not anagram
print(are_anagrams("rat", "car"))        # same length, not anagram
print(are_anagrams("aabb", "bbaa"))      # with repeats
```

```python
# 7. Remove duplicates, preserve order
def dedupe(lst):
    result = []
    for i in lst:
        if i not in result:
            result.append(i)
    return result

print(dedupe([1, 3, 2, 3, 1, 4]))  # normal
print(dedupe([]))                  # empty
print(dedupe([1, 1, 1]))           # all same
print(dedupe([1, 2, 3]))           # no duplicates
print(dedupe([7]))                 # single element
```

```python
# 8. All pairs that sum to target
def pair_sum(nums, target):
    seen = set()
    for num in nums:
        comp = target - num
        if comp in nums:
            seen.add(tuple(sorted((comp,num))))
    return seen

print(pair_sum([2, 4, 3, 5, 7, 8, 1], 9))  # normal
print(pair_sum([], 5))                      # empty
print(pair_sum([1, 2, 3], 100))             # no pair
print(pair_sum([3, 3, 3], 6))               # duplicates
print(pair_sum([-1, 5, -3, 4], 1))          # negatives
```

```python
# 9. Flatten a nested list of arbitrary depth
def flatten(nested):
    result = []
    for i in nested:
        if isinstance(i,list):
            result.extend(flatten(i))
        else:
            result.append(i)
    return result

print(flatten([1, [2, [3, 4], 5], [6, [7, [8]]]]))  # deep
print(flatten([]))                                  # empty
print(flatten([1, 2, 3]))                           # already flat
print(flatten([[], [1], []]))                       # empty sublists
print(flatten([[[[9]]]]))                           # single deep value
```

```python
# 10. Find the missing number in 1..n
def find_missing(nums, n):
    expected_sum = n * (n+1)//2
    return expected_sum - sum(nums)

print(find_missing([1, 2, 4, 5, 6], 6))  # missing middle
print(find_missing([2, 3, 4, 5], 5))     # missing first
print(find_missing([1, 2, 3, 4], 5))     # missing last
print(find_missing([], 1))               # n=1, missing only value
print(find_missing([1, 3], 3))           # small case
```

```python
import numpy as np

# 1. 3x3 random ints, replace values > mean with 0
def replace_above_mean(seed):
    np.random.seed(seed)
    arr = np.random.randint(0, 10, size=(3, 3))
    arr[arr > arr.mean()] = 0
    return arr

print(replace_above_mean(0))   # seed 0
print(replace_above_mean(1))   # seed 1
print(replace_above_mean(42))  # seed 42
```

```python
# 2. Min-max normalize a 1D array to [0, 1]
def min_max_normalize(arr):
    arr = np.asarray(arr, dtype=float)
    arr_min = arr.min()
    arr_max = arr.max()
    denom = arr_max - arr_min
    
    # Prevent division by zero if all elements are identical or there is only a single element
    if denom == 0:
        return np.zeros_like(arr)
    
    return (arr - arr_min) / denom

print(min_max_normalize(np.array([10, 20, 30, 40, 50])))  # normal
print(min_max_normalize(np.array([5, 5, 5])))             # all same (watch div by zero)
print(min_max_normalize(np.array([7])))                   # single element
print(min_max_normalize(np.array([-10, 0, 10])))          # negatives
print(min_max_normalize(np.array([0.5, 0.25, 1.0])))      # floats
```

```python
# 3. Row with the maximum sum in a 2D array
def max_sum_row(mat):
    return np.argmax(np.sum(mat, axis=1))
print(max_sum_row(np.array([[1, 2, 3], [9, 8, 7], [4, 5, 6]])))  # normal
print(max_sum_row(np.array([[5]])))                              # single element
print(max_sum_row(np.array([[1, 1], [1, 1]])))                   # tie
print(max_sum_row(np.array([[-1, -2], [-3, -4]])))               # all negative
print(max_sum_row(np.array([[0, 0, 0], [0, 0, 1]])))             # mostly zeros
```

```python
# 4. Replace NaN values with the column mean
def fill_nan_with_col_mean(mat):
    out = mat.copy()
    
    # 2. Compute the mean of each column ignoring NaN values
    col_means = np.nanmean(out, axis=0)
    
    # 3. Handle columns that are completely NaN (their nanmean will be NaN)
    # Replace those with 0.0 or a default fallback value
    col_means = np.nan_to_num(col_means, nan=0.0)
    # 4. Find where the NaNs are located
    inds = np.where(np.isnan(out))
    
    # 5. Replace each NaN with the mean corresponding to its column index
    out[inds] = col_means[inds[1]]
    
    return out

print(fill_nan_with_col_mean(np.array([[1.0, 2.0], [np.nan, 4.0], [5.0, np.nan]])))  # normal
print(fill_nan_with_col_mean(np.array([[1.0, 2.0], [3.0, 4.0]])))                    # no NaN
print(fill_nan_with_col_mean(np.array([[np.nan, 1.0], [np.nan, 3.0]])))              # whole column NaN
print(fill_nan_with_col_mean(np.array([[np.nan]])))                                  # single NaN
print(fill_nan_with_col_mean(np.array([[10.0, np.nan, 30.0]])))    
```

```python
# 5. Element-wise product via broadcasting
def broadcast_product(a, b):
    return a*b

print(broadcast_product(np.array([[1], [2], [3]]), np.array([10, 20, 30])))  # column x row
print(broadcast_product(np.array([1, 2, 3]), np.array([2])))                 # scalar-like
print(broadcast_product(np.array([[1, 2]]), np.array([[3], [4]])))           # 1x2 and 2x1
print(broadcast_product(np.array([0, 0]), np.array([5, 9])))                 # zeros
print(broadcast_product(np.array([-1, 2]), np.array([3, -4])))               # negatives
```

```python
# 6. Indices of top 3 largest values
def top_three_indices(arr):
    return np.argsort(arr)[::-1][:3]

print(top_three_indices(np.array([5, 1, 9, 3, 7, 2, 8])))  # normal
print(top_three_indices(np.array([1, 2, 3])))              # exactly 3
print(top_three_indices(np.array([4, 4, 4, 4])))           # all equal
print(top_three_indices(np.array([10, 20])))               # fewer than 3
print(top_three_indices(np.array([-1, -5, -2, -9])))       # negatives
```

```python
# 7. Count values between 10 and 50
def count_in_range(arr):
    return np.sum((arr>10)&(arr<50))

print(count_in_range(np.array([5, 15, 25, 55, 40, 8, 50])))  # normal
print(count_in_range(np.array([10, 50])))                    # boundary values
print(count_in_range(np.array([1, 2, 3])))                   # none in range
print(count_in_range(np.array([20, 30, 40])))                # all in range
print(count_in_range(np.array([]).astype(int)))              # empty
```

```python
# 8. Reshape 12 elements to 3x4 and transpose
def reshape_transpose(arr):
    return np.transpose(np.reshape(arr,(3,4)))

print(reshape_transpose(np.arange(12)))        # 0..11
print(reshape_transpose(np.arange(1, 13)))     # 1..12
print(reshape_transpose(np.ones(12)))          # all ones
print(reshape_transpose(np.arange(12) * 10))   # multiples of 10
```

```python
# 9. Z-score standardize (zero mean, unit variance)
def z_score(arr):
    std = np.std(arr, ddof=0)

  # Check if variance/standard deviation is zero to prevent division by zero
    if std == 0:
        return np.zeros_like(arr, dtype=float)
    return (arr - arr.mean()) / np.std(arr,ddof=0)

print(z_score(np.array([2, 4, 6, 8, 10])))  # normal
print(z_score(np.array([5, 5, 5])))         # zero variance (watch div by zero)
print(z_score(np.array([1, 2])))            # two elements
print(z_score(np.array([-3, 0, 3])))        # negatives
print(z_score(np.array([100])))             # single element
```

```python
# 10. Merge two arrays, return unique sorted values
def merge_unique(a, b):
    return np.union1d(a, b)

print(merge_unique(np.array([3, 1, 2, 5]), np.array([5, 4, 2, 6])))  # overlap
print(merge_unique(np.array([1, 2]), np.array([3, 4])))              # no overlap
print(merge_unique(np.array([1, 1, 1]), np.array([1, 1])))           # all same
print(merge_unique(np.array([]), np.array([2, 1])))                  # one empty
print(merge_unique(np.array([-1, 0]), np.array([0, -1])))            # negatives + dup
```

```python
import pandas as pd
import numpy as np

# 1. Rows where a column is above its own average
def above_average(df, col):
    return df[df[col] > df[col].mean()]

print(above_average(pd.DataFrame({"value": [10, 20, 30, 40, 50]}), "value"))  # normal
print(above_average(pd.DataFrame({"value": [5, 5, 5]}), "value"))             # all equal
print(above_average(pd.DataFrame({"value": [100]}), "value"))                 # single row
print(above_average(pd.DataFrame({"value": [-10, 0, 10]}), "value")) 
```

```python
# 2. Fill missing numeric values with median, grouped by category
def fill_median_by_group(df, group_col, num_col):
    df[num_col] = df[num_col].fillna(df.groupby(group_col)[num_col].transform('median'))
    return df

print(fill_median_by_group(pd.DataFrame({"cat": ["X", "X", "Y", "Y"], "num": [1, np.nan, 10, np.nan]}), "cat", "num"))  # normal
print(fill_median_by_group(pd.DataFrame({"cat": ["X", "X"], "num": [2, 4]}), "cat", "num"))                            # no NaN
print(fill_median_by_group(pd.DataFrame({"cat": ["X", "Y"], "num": [np.nan, 5]}), "cat", "num"))                       # group all NaN
print(fill_median_by_group(pd.DataFrame({"cat": ["A", "A", "A"], "num": [1, np.nan, 3]}), "cat", "num"))               # single group
```

```python
# 3. Top 5 customers by total purchase amount
def top_customers(df):
    grouped = df.groupby('customer',as_index=False).agg(total =('amount','sum')).sort_values('total',ascending=False).head(5)
    return grouped

print(top_customers(pd.DataFrame({"customer": ["c1","c2","c1","c3","c2","c4","c5","c6"], "amount": [100,200,150,300,50,400,250,120]})))  # 6 customers
print(top_customers(pd.DataFrame({"customer": ["c1","c2"], "amount": [10, 20]})))                                                        # fewer than 5
print(top_customers(pd.DataFrame({"customer": ["c1","c1","c1"], "amount": [5, 5, 5]})))                                                  # single customer
print(top_customers(pd.DataFrame({"customer": ["a","b","c"], "amount": [50, 50, 50]})))                                                  # all tied
```

```python
# 4. Count nulls per column, drop columns with >50% nulls
def drop_sparse_columns(df):
    # your code here
    cols_to_drop = df.columns[df.isnull().mean() > 0.5]
    return df.drop(columns=cols_to_drop)

print(drop_sparse_columns(pd.DataFrame({"a": [1,2,3,4], "b": [1,np.nan,np.nan,np.nan], "c": [np.nan,2,3,4]})))  # one sparse col
print(drop_sparse_columns(pd.DataFrame({"a": [1,2], "b": [3,4]})))                                              # no nulls
print(drop_sparse_columns(pd.DataFrame({"a": [np.nan,np.nan], "b": [np.nan,np.nan]})))                          # all sparse
print(drop_sparse_columns(pd.DataFrame({"a": [1,np.nan], "b": [1,2]})))    
```
