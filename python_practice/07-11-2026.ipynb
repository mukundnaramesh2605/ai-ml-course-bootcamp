```python
# 1. Reverse a string without slicing or reverse()
def reverse_string(s):
    # your code here
    result = ""
    for char in s:
        result = char+result
    return result

print(reverse_string("hello"))
print(reverse_string(""))           # empty
print(reverse_string("a"))          # single char
print(reverse_string("racecar"))    # palindrome
print(reverse_string("12345"))      # digits
```
