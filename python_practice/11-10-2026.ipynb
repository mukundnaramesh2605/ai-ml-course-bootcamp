```python
# 1. Count Positive, Negative, Zero  [Service / screening]
# Given a list, return counts of positive, negative, and zero values.
# Example: [1, -2, 0, 5, -3, 0] -> positive: 2, negative: 2, zero: 2
from collections import Counter
nums = [1, -2, 0, 5, -3, 0]
Counter(["Positive" if i >0 else 'Ngeative' if i!=0 else "Zero" for i in nums])
```

```python
# 2. Reverse Words in a Sentence  [Service / screening]
# Reverse the order of words, not the letters.
# Example: "I love python" -> "python love I"
sentence = "I love python"
print(" ".join(sentence.split()[::-1]))
```
