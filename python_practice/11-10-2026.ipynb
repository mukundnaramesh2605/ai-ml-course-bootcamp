```python
# 1. Count Positive, Negative, Zero  [Service / screening]
# Given a list, return counts of positive, negative, and zero values.
# Example: [1, -2, 0, 5, -3, 0] -> positive: 2, negative: 2, zero: 2
from collections import Counter
nums = [1, -2, 0, 5, -3, 0]
Counter(["Positive" if i >0 else 'Ngeative' if i!=0 else "Zero" for i in nums])
```

```python
# 2. Reverse Words in a Sentence  [Service / screening]
# Reverse the order of words, not the letters.
# Example: "I love python" -> "python love I"
sentence = "I love python"
print(" ".join(sentence.split()[::-1]))
```

```python
# 3. Check if List is Sorted  [Service / logic round]
# Return True if the list is sorted in ascending order.
# Example: [1, 2, 3, 4] -> True, [1, 3, 2] -> False
data = [1, 2, 3, 4]
def chech_list_sorted(data):
    length = len(data)
    if length<= 1:
        return True
    for i in range(length -1):
        if data[i] > data[i+1]:
            return False
    return True
print(chech_list_sorted(data))
print(chech_list_sorted([1,3,2]))

# or #
print(all(data[i]<=data[i+1] for i in range(len(data)-1)))
```

```python
# 4. Count Words by Length
# Return a dict mapping word length to how many words have that length.
# Example: "the cat sat on a mat" -> {3: 4, 2: 1, 1: 1}
from collections import Counter
text = "the cat sat on a mat"
Counter([len(word) for word in text.split()])
```

```python
# 5. Find Pairs That Sum to Target  [AI screen / logic]
# Return all index pairs where values add up to the target.
# Example: [2, 4, 3, 7], target 6 -> [(0, 1)]  (2+4)
arr = [2, 4, 3, 7]
target = 6
lst = []
for i in range(len(arr)):
    for j in range(i,len(arr)):
        if (arr[i]+arr[j]) == target and i!=j:
            lst.append((i,j))
print(lst)
```

```python
# 6. Flatten a Dictionary of Lists  [Service / data handling]
# Combine all list values into one flat list.
# Example: {"a": [1, 2], "b": [3, 4]} -> [1, 2, 3, 4]
d = {"a": [1, 2], "b": [3, 4], "c": [5]}
print([x for i in d.values() for x in i])
```

```python
# 7. Most Common Word  [AI screen / text]
# Return the word that appears most often, case-insensitive.
# Example: "the dog the cat the bird" -> "the"
from collections import Counter
line = "the dog the cat the bird"
counts = Counter(line.lower().split())
print(*[k for k,v in counts.items() if v == max(counts.values())])
```

```python
# 8. Remove Keys With None Values  [Service / data cleaning]
# Return a new dict dropping any key whose value is None.
# Example: {"a": 1, "b": None, "c": 3} -> {"a": 1, "c": 3}
record = {"a": 1, "b": None, "c": 3, "d": None}
print({k:v for k,v in record.items() if v})
```

```python
# 9. Count Common Characters  [Service / logic round]
# Return how many characters appear in both strings (unique count).
# Example: "hello", "world" -> 2  (l and o)
s1 = "hello"
s2 = "world"
print(len(set(s1) & set(s2)))
```

```python
# 10. Group Numbers by Even/Odd  [AI screen / logic]
# Return a dict with evens and odds separated, order preserved.
# Example: [1, 2, 3, 4, 5] -> {"even": [2, 4], "odd": [1, 3, 5]}
from collections import defaultdict
values = [1, 2, 3, 4, 5]
groups = defaultdict(list)
for i in values:
    if i % 2 ==0:
        groups['even'].append(i)
    else:
        groups['odd'].append(i)
print(groups)
```

```python
# 3. Character Frequency Sort  [Service / text]
# Return characters sorted by frequency, highest first. Ties broken alphabetically.
# Example: "tree" -> ['e', 'r', 't']  (e:2, then r,t alphabetical)
from collections import Counter
s = "aabbbcc"
counts = Counter(s)
print([k for k,v in sorted(counts.items(),key=lambda x : x[1],reverse=True)])
```
