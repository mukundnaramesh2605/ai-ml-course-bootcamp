```python
import numpy as np
import pandas as pd
from collections import Counter, defaultdict
```

```python
# 1. Mask Middle Letters  [easy / string]
# Keep the first and last letter of each word, replace every letter in between with "*".
# Words of length 2 or less stay as they are. KEEP SPACES EXACTLY as given.
# "hello world" -> "h***o w***d"
def mask_middle(s):
    return " ".join([word[0]+len(word[1:-1])*'*'+word[-1] if len(word) > 2 else word for word in s.split(" ")])

print(repr(mask_middle("hello world")))  # -> 'h***o w***d'
print(repr(mask_middle("I am  here")))   # -> 'I am  h**e'   <- double space must survive
print(repr(mask_middle("")))             # -> ''
print(repr(mask_middle("ok")))           # -> 'ok'           <- length 2, unchanged
print(repr(mask_middle("a   b")))        # -> 'a   b'
```

```python
# 2. Word Stats  [easy / dict]
# Words are separated by ANY whitespace. Return a dict with EXACT keys "words", "longest", "shortest".
# words = COUNT of words. Ties for longest/shortest: FIRST word that appears.
# Empty or whitespace-only -> {'words': 0, 'longest': '', 'shortest': ''}
def word_stats(s):
    grouped = {'words': 0, 'longest': '', 'shortest': ''}
    if not s:
        return grouped
    s= s.split()
    counts = {k:s.count(k) for k in s}
    max_val = max(counts.values())
    min_val = min(counts.values())
    grouped['words']=len(s)
    for k,v in counts.items():
        if v == max_val:   
            grouped['longest']=k
            break
    for k,v in counts.items(): 
        if v == min_val:
            grouped['shortest']=k
            break
    return grouped
print(word_stats("big apple pie"))  # -> {'words': 3, 'longest': 'apple', 'shortest': 'big'}
print(word_stats(""))               # -> {'words': 0, 'longest': '', 'shortest': ''}
print(word_stats("solo"))           # -> {'words': 1, 'longest': 'solo', 'shortest': 'solo'}
print(word_stats("cat dog ox"))     # -> {'words': 3, 'longest': 'cat', 'shortest': 'ox'}   <- cat/dog tie, first wins
print(word_stats("  hi  "))         # -> {'words': 1, 'longest': 'hi', 'shortest': 'hi'}
```

```python
# 3. Letters In Every Word  [medium / set]
# Return the DISTINCT characters that appear in EVERY word of the list, sorted ASC.
# ["apple", "ample", "maple"] -> ['a', 'e', 'l', 'p']
def common_chars(words):
    if not words:
        return []
    common = set(words[0])
    for word in words[1:]:
        common &= set(word)
        
    return sorted(common)
        
print(common_chars(["apple", "ample", "maple"]))  # -> ['a', 'e', 'l', 'p']
print(common_chars([]))                           # -> []
print(common_chars(["zebra"]))                    # -> ['a', 'b', 'e', 'r', 'z']   <- single word
print(common_chars(["abc", "def"]))               # -> []                          <- no match
print(common_chars(["aab", "aba", "baa"]))        # -> ['a', 'b']                  <- DISTINCT
```
