```python
# 1. Two Sum (Easy)
# Return indices of two numbers that add up to target.
def two_sum(nums, target):
    seen = {}
    for i, n in enumerate(nums):
        if target - n in seen:
            return [seen[target - n], i]
        seen[n] = i

print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
```

```python
# 2. Reverse a String (Easy)
# Reverse without using built-in reverse.
def reverse_string(s):
    chars = list(s)
    left, right = 0, len(chars) - 1
    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1
    return "".join(chars)

print(reverse_string("hello"))  # olleh
```

```python
# 3. FizzBuzz (Easy)
def fizzbuzz(n):
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            out.append("FizzBuzz")
        elif i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(str(i))
    return out

print(fizzbuzz(15))
```

```python
# 4. Valid Palindrome (Easy-Medium)
# Ignore case and non-alphanumeric characters.
def is_palindrome(s):
    cleaned = [c.lower() for c in s if c.isalnum()]
    return cleaned == cleaned[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))  # True
```

```python
# 5. Group Anagrams (Medium)
from collections import defaultdict

def group_anagrams(words):
    groups = defaultdict(list)
    for w in words:
        key = "".join(sorted(w))
        groups[key].append(w)
    return list(groups.values())

print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

```python
def group_anagrams2(words):
    group = defaultdict(list)

    for w in words:
        key = ''.join(sorted(w))
        group[key].append(w)
    return list(group.values())

print(group_anagrams2(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

```python
# 6. Longest Substring Without Repeating Characters (Medium)
def length_of_longest_substring(s):
    seen = {}
    start = 0
    best = 0
    for i, c in enumerate(s):
        if c in seen and seen[c] >= start:
            start = seen[c] + 1
        seen[c] = i
        best = max(best, i - start + 1)
    return best

print(length_of_longest_substring("abcabcbb"))  # 3
```

```python
# 1. Softmax (Easy-Medium)
# Convert a vector of scores into probabilities. Subtract max for numerical stability.
import numpy as np

def softmax(x):
    x = np.array(x, dtype=float)
    shifted = x - np.max(x)
    exp = np.exp(shifted)
    return exp / exp.sum()

print(softmax([2.0, 1.0, 0.1]))  # ~[0.659, 0.242, 0.099]
```

```python
# 2. Remove Duplicates While Preserving Order — Easy
# Given a list, remove duplicate values while keeping the first occurrence order.
nums = [4, 2, 4, 3, 2, 1, 3]

def remove_duplicates(nums):
    seen = []
    for i in nums:
        if i not in seen:
            seen.append(i)
    return seen

remove_duplicates(nums)
# Output: [4, 2, 3, 1]
```

```python
"""3. Count Word Frequency — Easy
Given a sentence, return a dictionary containing the frequency of each word."""

text = "machine learning is fun and machine learning is powerful"
from collections import Counter
def word_counter(text):
    count = Counter(text.split())
    return count

word_counter(text)

# Output:
# {
#   "machine": 2,
#   "learning": 2,
#   "is": 2,
#   "fun": 1,
#   "and": 1,
#   "powerful": 1
# }
```

```python
"""4. Find the First Non-Repeating Character — Easy
Given a string, return the first character that appears only once."""
s = "aabbccddee"

def first_non_repeating_char(s):
    first_non_repeated_char = None
    for i in s:
        if s.count(i) == 1:
            first_non_repeated_char = i
            break
    return first_non_repeated_char

first_non_repeating_char(s)
# Output: "c"
```

```python
"""5. Find Missing Number — Easy
You are given n distinct numbers from 0 to n. Find the missing number."""
nums = [3, 0, 1]

def find_missing_number(nums):
    min_n,max_n = min(nums),max(nums)
    full_list = set(range(min_n,max_n+1))
    return list(full_list ^ set(nums))
# Output: 2
find_missing_number(nums)
```

```python
"""6. Move All Zeros to the End — Easy/Medium
Move all zeros to the end of an array without changing the relative order of non-zero elements."""
nums = [2, 3, 12]

def move_zeroes_to_end(nums):
    return [x for x in nums if x!=0] + [0] * nums.count(0)

    
move_zeroes_to_end(nums)    
# Output:[1, 3, 12, 0, 0]
```
