```python
# 1. Count Even and Odd  [screening]
# Return a tuple (evens, odds) counting even and odd numbers. Handle negatives.
def count_even_odd(nums):
    return (sum([1 for i in nums if i%2==0]), sum([1 for i in nums if i%2!=0]))
print(count_even_odd([1,2,3,4,-6]))   # -> (3, 2)
print(count_even_odd([]))
```

```python
# 2. Second Largest Distinct  [logic]
# Return the second largest DISTINCT value, or None if there isn't one.
def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in set(nums):
        if num > first:
            second=first
            first = num
        elif num > second and  num != first:
            second = num
    return second if second != neg_inf else None
print(second_largest([4,1,4,3]))      # -> 3
print(second_largest([7,7]))          # -> None
print(second_largest([9,2,5,5,9]))    # -> 5
```
