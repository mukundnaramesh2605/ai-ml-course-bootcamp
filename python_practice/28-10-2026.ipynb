```python
# 1. Count Even and Odd  [screening]
# Return a tuple (evens, odds) counting even and odd numbers. Handle negatives.
def count_even_odd(nums):
    return (sum([1 for i in nums if i%2==0]), sum([1 for i in nums if i%2!=0]))
print(count_even_odd([1,2,3,4,-6]))   # -> (3, 2)
print(count_even_odd([]))
```

```python
# 2. Second Largest Distinct  [logic]
# Return the second largest DISTINCT value, or None if there isn't one.
def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in set(nums):
        if num > first:
            second=first
            first = num
        elif num > second and  num != first:
            second = num
    return second if second != neg_inf else None
print(second_largest([4,1,4,3]))      # -> 3
print(second_largest([7,7]))          # -> None
print(second_largest([9,2,5,5,9]))    # -> 5
```

```python
# 3. Character Frequency Sorted  [dict]
# Return a list of (char, count) sorted by count descending, then char ascending.
# Ignore spaces. Case-insensitive.
def char_freq(s):
    if s =="":
        return ""
    s = s.lower()
    return list(sorted(set((char,s.count(char)) for char in s if char.isalnum()),key=lambda x: x[1],reverse=True))
print(char_freq("aA bb c"))           # -> [('b', 2), ('a', 2), ('c', 1)]
print(char_freq(""))    
```

```python
# 4. Sum of Unique  [set]
# Return the sum of values that appear EXACTLY once.
def sum_unique(nums):
    return sum([num for num in nums if nums.count(num) == 1])
print(sum_unique([1,2,2,3,4,4]))      # -> 4   (1 + 3)
print(sum_unique([5,5]))  
```
