```python
def compress(s):
    count=1
    cuurent = s[0]
    result = []
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            result.append(cuurent+str(count))
            count=1
            cuurent=s[i]
    result.append(cuurent+str(count))
    return "".join(result)
compress("aaabbaaccbb")
```

```python
def valid_paranthesis(s):
    brackets = {"}":"{",")":"(","]":"["}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets: 
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0

valid_paranthesis("(a[b]{c})")
             
```

```python
# 1. Flatten Any Depth  [AI screen / recursion]
# Flatten a list nested to any depth into a single flat list.
# Example: [1, [2, [3, [4, 5]]], 6] -> [1, 2, 3, 4, 5, 6]
def flatten(lst):
    result = []
    for i in lst:
        if isinstance(i,list):
            result.extend(flatten(i))
        else:
            result.append(i)
    return result
print(flatten([1, [2, [3, [4, 5]]], 6]))# -> [1,2,3,4,5,6]
# test: flatten([]) -> []
# test: flatten([[[[1]]]]) -> [1]
```

```python
# 2. Sum All Numbers  [AI screen / recursion]
# Sum every number in a deeply nested list.
# Example: [1, [2, [3, 4]], [[5]]] -> 15
def deep_sum(lst):
    return sum(flatten(lst))
print(deep_sum([1, [2, [3, 4]], [[5]]]))# -> 15
print(deep_sum([]))# -> 0
```

```python
# 3. Max in Nested List  [AI screen / recursion]
# Return the largest number anywhere in a deeply nested list. Assume at least one number.
# Example: [1, [9, [3]], [[7], 2]] -> 9
def deep_max(lst):
    return max(flatten(lst))
print(deep_max([1, [9, [3]], [[7], 2]]))# -> 9
print(deep_max([[[42]]]))# -> 42
```
