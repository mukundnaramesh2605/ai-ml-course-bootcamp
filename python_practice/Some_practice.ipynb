```python
def compress(s):
    count=1
    cuurent = s[0]
    result = []
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            result.append(cuurent+str(count))
            count=1
            cuurent=s[i]
    result.append(cuurent+str(count))
    return "".join(result)
compress("aaabbaaccbb")
```

```python
def valid_paranthesis(s):
    brackets = {"}":"{",")":"(","]":"["}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets: 
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0

valid_paranthesis("(a[b]{c})")
             
```

```python
arr = [1,2,3,4,5]
k=7
left= 0
right = len(arr)-1

while left<right:
    mid = left + (right - left) // 2
    if arr[mid] == k:
        print(mid)
        break
    elif arr[mid] < k:
        left = mid +1
    else:
        right = mid-1
```
