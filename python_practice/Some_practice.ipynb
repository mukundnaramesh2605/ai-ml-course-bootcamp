```python
def compress(s):
    count=1
    cuurent = s[0]
    result = []
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            result.append(cuurent+str(count))
            count=1
            cuurent=s[i]
    result.append(cuurent+str(count))
    return "".join(result)
compress("aaabbaaccbb")
```
