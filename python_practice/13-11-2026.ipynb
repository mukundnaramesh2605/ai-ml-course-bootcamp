```python
# Q1. Second Largest Number
# Return the second largest DISTINCT number in the list.
# Return None if it doesn't exist.

def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in set(nums):
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    return second if second != neg_inf else None
print(second_largest([10, 5, 8, 10]))     # expected: 8
print(second_largest([3]))                # expected: None
print(second_largest([7, 7, 7]))          # expected: None
print(second_largest([-1, -5, -3]))       # expected: -3
print(second_largest([]))                 # expected: None
```
