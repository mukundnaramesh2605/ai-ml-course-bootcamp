```python
# Q1. Second Largest Number
# Return the second largest DISTINCT number in the list.
# Return None if it doesn't exist.

def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in set(nums):
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    return second if second != neg_inf else None
print(second_largest([10, 5, 8, 10]))     # expected: 8
print(second_largest([3]))                # expected: None
print(second_largest([7, 7, 7]))          # expected: None
print(second_largest([-1, -5, -3]))       # expected: -3
print(second_largest([]))                 # expected: None
```

```python
# Q2. Balanced Brackets
# Return True if every (), [], {} is properly opened and closed in order.
# Ignore non-bracket characters.

def is_balanced(s):
    brackets = {"]":"[","}":"{",")":"("}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0

print(is_balanced("{[()]}"))     # expected: True
print(is_balanced("([)]"))       # expected: False
print(is_balanced(""))           # expected: True
print(is_balanced("((("))        # expected: False
print(is_balanced("a(b)c]"))     # expected: False
```

```python
# Q3. First Non-Repeating Character
# Return the first character that appears exactly once (case-sensitive).
# Return None if there is none.

def first_non_repeating(s):
    for i in s:
        if s.count(i) == 1: return i
    return None

print(first_non_repeating("swiss"))   # expected: w
print(first_non_repeating("aabb"))    # expected: None
print(first_non_repeating(""))        # expected: None
print(first_non_repeating("a"))       # expected: a
print(first_non_repeating("Aa"))      # expected: A
```
