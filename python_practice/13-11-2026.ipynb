```python
# Q1. Second Largest Number
# Return the second largest DISTINCT number in the list.
# Return None if it doesn't exist.

def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in set(nums):
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    return second if second != neg_inf else None
print(second_largest([10, 5, 8, 10]))     # expected: 8
print(second_largest([3]))                # expected: None
print(second_largest([7, 7, 7]))          # expected: None
print(second_largest([-1, -5, -3]))       # expected: -3
print(second_largest([]))                 # expected: None
```

```python
# Q2. Balanced Brackets
# Return True if every (), [], {} is properly opened and closed in order.
# Ignore non-bracket characters.

def is_balanced(s):
    brackets = {"]":"[","}":"{",")":"("}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0

print(is_balanced("{[()]}"))     # expected: True
print(is_balanced("([)]"))       # expected: False
print(is_balanced(""))           # expected: True
print(is_balanced("((("))        # expected: False
print(is_balanced("a(b)c]"))     # expected: False
```

```python
# Q3. First Non-Repeating Character
# Return the first character that appears exactly once (case-sensitive).
# Return None if there is none.

def first_non_repeating(s):
    for i in s:
        if s.count(i) == 1: return i
    return None

print(first_non_repeating("swiss"))   # expected: w
print(first_non_repeating("aabb"))    # expected: None
print(first_non_repeating(""))        # expected: None
print(first_non_repeating("a"))       # expected: a
print(first_non_repeating("Aa"))      # expected: A
```

```python
# Q4. Group Anagrams
# Group words that are anagrams of each other.
# Keep groups in order of first appearance, and words in original order inside each group.

def group_anagrams(words):
    grouped = {}
    for i in words:
        key = "".join(sorted(i))
        if key not in grouped:
            grouped[key] = []
        grouped[key].append(i)
    return list(grouped.values())

print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))  # expected: [['eat','tea','ate'], ['tan','nat'], ['bat']]
print(group_anagrams([]))                                          # expected: []
print(group_anagrams([""]))                                        # expected: [['']]
print(group_anagrams(["a"]))                                       # expected: [['a']]
print(group_anagrams(["abc", "bca", "xyz"]))                       # expected: [['abc','bca'], ['xyz']]
```

```python
# Q5. Longest Substring Without Repeating Characters
# Return the LENGTH of the longest substring with all unique characters.

def longest_unique_substring(s):
    seen = set()
    left = 0
    max_length = 0

    for right in range(len(s)):

        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])

        max_length = max(max_length, right - left + 1)

    return max_length

print(longest_unique_substring("abcabcbb"))  # expected: 3
print(longest_unique_substring("bbbb"))      # expected: 1
print(longest_unique_substring(""))          # expected: 0
print(longest_unique_substring("pwwkew"))    # expected: 3
print(longest_unique_substring(" "))         # expected: 1
```

```python
# Q6. Min-Max Scaling (column-wise)
# Scale each column to the range [0, 1].
# If a column has the same value everywhere, set that column to 0.
# No loops. Use vectorized NumPy.

def min_max_scale(X):
    X_min = np.min(X, axis=0)
    X_max = np.max(X, axis=0)
    denom = X_max - X_min
    return np.divide(
      X - X_min, denom, out=np.zeros_like(X, dtype=float), where=denom != 0)

print(min_max_scale(np.array([[1, 10], [2, 20], [3, 30]])))   # expected: [[0,0],[0.5,0.5],[1,1]]
print(min_max_scale(np.array([[5, 1], [5, 2], [5, 3]])))      # expected: [[0,0],[0,0.5],[0,1]]  (constant column)
print(min_max_scale(np.array([[4, 8]])))                      # expected: [[0,0]]  (single row)
print(min_max_scale(np.array([[-2], [0], [2]])))              # expected: [[0],[0.5],[1]]
```
