```python
# Q1. Second Largest Number
# Return the second largest DISTINCT number in the list.
# Return None if it doesn't exist.

def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in set(nums):
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    return second if second != neg_inf else None
print(second_largest([10, 5, 8, 10]))     # expected: 8
print(second_largest([3]))                # expected: None
print(second_largest([7, 7, 7]))          # expected: None
print(second_largest([-1, -5, -3]))       # expected: -3
print(second_largest([]))                 # expected: None
```

```python
# Q2. Balanced Brackets
# Return True if every (), [], {} is properly opened and closed in order.
# Ignore non-bracket characters.

def is_balanced(s):
    brackets = {"]":"[","}":"{",")":"("}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0

print(is_balanced("{[()]}"))     # expected: True
print(is_balanced("([)]"))       # expected: False
print(is_balanced(""))           # expected: True
print(is_balanced("((("))        # expected: False
print(is_balanced("a(b)c]"))     # expected: False
```

```python
# Q3. First Non-Repeating Character
# Return the first character that appears exactly once (case-sensitive).
# Return None if there is none.

def first_non_repeating(s):
    for i in s:
        if s.count(i) == 1: return i
    return None

print(first_non_repeating("swiss"))   # expected: w
print(first_non_repeating("aabb"))    # expected: None
print(first_non_repeating(""))        # expected: None
print(first_non_repeating("a"))       # expected: a
print(first_non_repeating("Aa"))      # expected: A
```

```python
# Q4. Group Anagrams
# Group words that are anagrams of each other.
# Keep groups in order of first appearance, and words in original order inside each group.

def group_anagrams(words):
    grouped = {}
    for i in words:
        key = "".join(sorted(i))
        if key not in grouped:
            grouped[key] = []
        grouped[key].append(i)
    return list(grouped.values())

print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))  # expected: [['eat','tea','ate'], ['tan','nat'], ['bat']]
print(group_anagrams([]))                                          # expected: []
print(group_anagrams([""]))                                        # expected: [['']]
print(group_anagrams(["a"]))                                       # expected: [['a']]
print(group_anagrams(["abc", "bca", "xyz"]))                       # expected: [['abc','bca'], ['xyz']]
```

```python
# Q5. Longest Substring Without Repeating Characters
# Return the LENGTH of the longest substring with all unique characters.

def longest_unique_substring(s):
    seen = set()
    left = 0
    max_length = 0

    for right in range(len(s)):

        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])

        max_length = max(max_length, right - left + 1)

    return max_length

print(longest_unique_substring("abcabcbb"))  # expected: 3
print(longest_unique_substring("bbbb"))      # expected: 1
print(longest_unique_substring(""))          # expected: 0
print(longest_unique_substring("pwwkew"))    # expected: 3
print(longest_unique_substring(" "))         # expected: 1
```

```python
# Q6. Min-Max Scaling (column-wise)
# Scale each column to the range [0, 1].
# If a column has the same value everywhere, set that column to 0.
# No loops. Use vectorized NumPy.

def min_max_scale(X):
    X_min = np.min(X, axis=0)
    X_max = np.max(X, axis=0)
    denom = X_max - X_min
    return np.divide(
      X - X_min, denom, out=np.zeros_like(X, dtype=float), where=denom != 0)

print(min_max_scale(np.array([[1, 10], [2, 20], [3, 30]])))   # expected: [[0,0],[0.5,0.5],[1,1]]
print(min_max_scale(np.array([[5, 1], [5, 2], [5, 3]])))      # expected: [[0,0],[0,0.5],[0,1]]  (constant column)
print(min_max_scale(np.array([[4, 8]])))                      # expected: [[0,0]]  (single row)
print(min_max_scale(np.array([[-2], [0], [2]])))              # expected: [[0],[0.5],[1]]
```

```python
# Q7. Fill NaN with Column Mean
# Replace each NaN with its column's mean (ignoring NaNs).
# If a whole column is NaN, fill it with 0.
# Do NOT modify the original array.

def fill_nan_col_mean(X):
    X_out = X.astype(float, copy=True)
    col_means = np.nanmean(X_out, axis=0)
    col_means = np.nan_to_num(col_means, nan=0.0)
    nan_mask = np.isnan(X_out)
    X_out[nan_mask] = np.take(col_means, np.where(nan_mask)[1])
    return X_out
print(fill_nan_col_mean(np.array([[1, np.nan], [3, 4], [np.nan, 8]])))  # expected: [[1,6],[3,4],[2,8]]
print(fill_nan_col_mean(np.array([[1.0, 2.0], [3.0, 4.0]])))           # expected: unchanged
print(fill_nan_col_mean(np.array([[np.nan, 1], [np.nan, 3]])))          # expected: [[0,1],[0,3]]  (all-NaN column)
print(fill_nan_col_mean(np.array([[np.nan]])))                          # expected: [[0]]
```

```python
# Q8. One-Hot Encoding
# Convert integer labels into a one-hot matrix of shape (n, num_classes).
# If num_classes is None, infer it as max(label) + 1.
# Handle empty input.

def one_hot(labels, num_classes=None):
  labels = np.asarray(labels, dtype=int)
  if num_classes is None:
    num_classes = int(labels.max()) + 1 if labels.size > 0 else 0
  out = np.zeros((labels.size, num_classes), dtype=int)
  if labels.size > 0:
    out[np.arange(labels.size), labels] = 1
  return out

print(one_hot(np.array([0, 2, 1]), 3))        # expected: [[1,0,0],[0,0,1],[0,1,0]]
print(one_hot(np.array([1, 1]), 2))           # expected: [[0,1],[0,1]]
print(one_hot(np.array([0, 3])))              # expected: 4 columns, inferred
print(one_hot(np.array([], dtype=int), 3))    # expected: empty array with shape (0, 3)
```

```python
# Q9. Cosine Similarity (used for embeddings / RAG)
# Return the cosine similarity between two 1D vectors, rounded to 4 decimals.
# If either vector is all zeros, return 0.0.

def cosine_similarity(a, b):
  norm_a = np.linalg.norm(a)
  norm_b = np.linalg.norm(b)
  if norm_a == 0 or norm_b == 0:
    return 0.0
  return round(float(np.dot(a, b) / (norm_a * norm_b)), 4)

print(cosine_similarity(np.array([1, 0]), np.array([0, 1])))           # expected: 0.0
print(cosine_similarity(np.array([1, 2, 3]), np.array([2, 4, 6])))     # expected: 1.0
print(cosine_similarity(np.array([1, 0]), np.array([-1, 0])))          # expected: -1.0
print(cosine_similarity(np.array([0, 0]), np.array([1, 1])))           # expected: 0.0  (zero vector)
print(cosine_similarity(np.array([3, 4]), np.array([4, 3])))           # expected: 0.96
```

```python
# Q10. Top-K Indices (like retrieving top-k results)
# Return indices of the k highest scores, highest first, as a Python list.
# On ties, the smaller index comes first.
# If k > len(scores), return all. If k <= 0 or the array is empty, return [].

def top_k_indices(scores, k):
  scores = np.asarray(scores)
  if k <= 0 or scores.size == 0:
    return []
  return np.argsort(-scores, kind="stable")[:k].tolist()

print(top_k_indices(np.array([0.1, 0.9, 0.4, 0.7]), 2))   # expected: [1, 3]
print(top_k_indices(np.array([0.5, 0.2]), 5))             # expected: [0, 1]  (k > length)
print(top_k_indices(np.array([0.3, 0.3, 0.1]), 2))        # expected: [0, 1]  (tie)
print(top_k_indices(np.array([0.8, 0.6]), 0))             # expected: []
print(top_k_indices(np.array([]), 3))                     # expected: []
```
