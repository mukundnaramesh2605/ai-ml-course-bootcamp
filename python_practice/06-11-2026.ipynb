```python
# 1. two-sum  [tag: hashing]
# Return the indices (i, j), i < j, of the two numbers that add to target.
# Exactly one answer exists. Negatives and duplicates are in play.
def two_sum(nums, target):
    seen = {}
    for idx,num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return [seen[comp],idx]
        seen[num] = idx
    return None

print(two_sum([2, 7, 11, 15], 9))        # a pair early in the list
print(two_sum([-3, 4, 1, -1, 8], -4))    # needs negatives
print(two_sum([3, 3], 6))                # duplicate values, distinct indices
```

```python
# 2. valid parentheses  [tag: stack]
# Return True if every bracket in ()[]{} is closed by the right type in the right order.
def is_valid(s):
    brackets = {']':'[','}':'{',')':'('}
    stack = []
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0
print(is_valid("()[]{}"))   # all matched
print(is_valid("([)]"))     # interleaved, should fail
print(is_valid("("))        # unclosed
print(is_valid(""))         # empty is valid
```

```python
# 3. first unique character  [tag: hashing]
# Return the first character that appears exactly once. If none, return "".
def first_unique(s):
    counts = {}
    for char in s:
        counts[char]=counts.get(char,0)+1
    for k,v in counts.items():
        if v == 1:
            return k
    return ""
    

print(first_unique("swiss"))      # 'w'
print(first_unique("aabbcc"))     # "" (no unique)
print(first_unique("loveleetcode"))
print(first_unique("x"))          # single char
```

```python
# 4. move zeros  [tag: two-pointer]
# Return a new list with all 0s moved to the end, non-zero order preserved.
# Do not mutate the input.
def move_zeros(nums):
    zero_ptr = 0
    for i in range(len(nums)):
        if nums[i]!=0:
            nums[zero_ptr],nums[i] = nums[i],nums[zero_ptr]
            zero_ptr+=1
    return nums
    

print(move_zeros([0, 1, 0, 3, 12]))
print(move_zeros([0, 0, 1]))          # multiple leading zeros
print(move_zeros([4, 5, 6]))          # no zeros, unchanged
print(move_zeros([-2, 0, -5, 0]))     # negatives preserved
```

```python
# 5. run-length compress  [tag: single-pass]
# Return the run-length string: "aaabbc" -> "a3b2c1". Every char gets a count,
# including count 1. Case-sensitive.
def compress(s):
    if s == "":
        return ""
    count = 1 
    current = s[0]
    result = []
    for i in range(1,len(s)):
        if s[i-1] == s[i]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current=s[i]
    
    result.append(current+str(count))
    return "".join(result)
print(compress("aaabbc"))    # a3b2c1
print(compress("abc"))       # a1b1c1
print(compress("aaAA"))      # a2A2 (case matters)
print(compress(""))          # ""
```

```python
# 6. missing number  [tag: single-pass]
# nums holds n distinct numbers from 0..n with exactly one missing. Return it.
def missing_number(nums):
    n = len(nums)
    expected_sum = n * (n+1) //2 
    return expected_sum - sum(nums)
print(missing_number([3, 0, 1]))        # 2
print(missing_number([0, 1, 2, 3]))     # 4 (missing is the top)
print(missing_number([1, 2, 3]))        # 0 (missing is zero)
```

```python
# 7. max subarray sum  [tag: Kadane]
# Return the largest sum of any contiguous non-empty subarray.
def max_subarray(nums):
    current = best = nums[0]
    for i in nums[1:]:
        current = max(i,current+i)
        best = max(current,best)
    return best

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # 6
print(max_subarray([-5, -2, -8]))    # -2 (all negative, pick the least bad)
print(max_subarray([3, 4, 5]))       # 12 (all positive)
```
