```python
# 1. two-sum  [tag: hashing]
# Return the indices (i, j), i < j, of the two numbers that add to target.
# Exactly one answer exists. Negatives and duplicates are in play.
def two_sum(nums, target):
    seen = {}
    for idx,num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return [seen[comp],idx]
        seen[num] = idx
    return None

print(two_sum([2, 7, 11, 15], 9))        # a pair early in the list
print(two_sum([-3, 4, 1, -1, 8], -4))    # needs negatives
print(two_sum([3, 3], 6))                # duplicate values, distinct indices
```
