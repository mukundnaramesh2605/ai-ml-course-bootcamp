```python
# 1. Sum Positive Only  [logic]
# Return the sum of only the positive numbers (strictly > 0).
def sum_positive(nums):
    return sum([x for x in nums if x>0])
print(sum_positive([3, -1, 4, -2, 0]))   # -> 7
print(sum_positive([-1, -2]))            # -> 0
print(sum_positive([]))       
```

```python
# 2. Count Unique Words  [set]
# Return how many DISTINCT words are in a sentence, case-insensitive.
def unique_word_count(s):
    return len(set(s.lower().split()))
print(unique_word_count("the cat The CAT dog"))   # -> 3
print(unique_word_count("")) 
```

```python
# 3. Running Total  [logic]
# Return a list where each element is the cumulative sum up to that point.
def running_total(nums):
    if not nums:
        return []
    result = [nums[0]]
    for i in nums[1:]:
        result.append(i+result[-1])
    return result
print(running_total([1, 2, 3, 4]))   # -> [1, 3, 6, 10]
print(running_total([]))    
```

```python
# 4. Most Frequent Element  [dict]
# Return the most frequent value. On a tie, the one that appears first.
from collections import Counter
def most_frequent(nums):
    counts = Counter(nums)
    return sorted(nums,key=counts.get,reverse=True)[0]
print(most_frequent([1, 2, 2, 3, 3]))   # -> 2   (2 and 3 tie, 2 appears first)
print(most_frequent([5]))               # -> 5
```

```python
# 5. Is Subsequence  [two-pointer]
# Return True if a is a subsequence of b (same order, not necessarily contiguous).
def is_subsequence(a, b):
    it = iter(b)
    return all(char in it for char in a)
print(is_subsequence("ace", "abcde"))   # -> True
print(is_subsequence("aec", "abcde"))   # -> False   <- order matters
print(is_subsequence("", "abc"))      
```

```python
# 6. Group Numbers by Parity  [dict]
# Return a dict {"even": [...], "odd": [...]} keeping original order. Negatives too.
def group_parity(nums):
    if not nums:
        return []
    grouped = {'even':[],'odd':[]}
    for i in nums:
        grouped['even'].append(i) if i %2==0 else grouped['odd'].append(i)
    return grouped
print(group_parity([1, 2, 3, 4, -5]))   # -> {'even': [2, 4], 'odd': [1, 3, -5]}
print(group_parity([])) 
```

```python
# 7. Count Pairs With Difference K  [hashing]
# Return how many unordered pairs (i, j) have abs(nums[i] - nums[j]) == k.
# Count distinct index pairs.
def count_pairs_diff(nums, k):
    freq = {}
    count = 0
    for num in nums:
        if k==0:
            count += freq.get(num, 0)
        else:
            count += freq.get(num-k, 0)
            count += freq.get(num-k, 0)
        freq[num] = freq.get(num, 0) + 1
    return count
print(count_pairs_diff([1, 5, 3, 4, 2], 2))   # -> 3   (1,3),(5,3),(2,4)
print(count_pairs_diff([1, 1, 1], 0))         # -> 3   (all three pairs)
```

```python
# 8. Longest Word Length  [string]
# Return the length of the longest word. 0 for an empty string.
def longest_len(s):
    return len(max(s.split(),key=len)) if s.split() else ""
print(longest_len("the quick brown fox"))   # -> 5
print(longest_len(""))    
```

```python
# 9. Remove Every Nth  [logic]
# Return the list with every nth element removed (1-indexed). n >= 1.
def remove_every_nth(lst, n):
    return [lst[i] for i in range(0,len(lst),n)]
print(remove_every_nth([1, 2, 3, 4, 5, 6], 2))   # -> [1, 3, 5]   (drop 2nd, 4th, 6th)
print(remove_every_nth([1, 2, 3], 1))  
```

```python
# 10. Balanced Brackets  [stack]
# Return True if (), [], {} are correctly matched and nested.
def is_balanced(s):
    brackets = {"}":"{","]":"[",")":"("}
    stack=[]
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0
print(is_balanced("([{}])"))   # -> True
print(is_balanced("([)]"))     # -> False
print(is_balanced("("))        # -> False
```

```python
import numpy as np

# 11. Column Sums  [axis]
# Return a 1-D array of the sum of each COLUMN.
def col_sums(mat):
    return np.sum(mat,axis=0)
print(col_sums(np.array([[1, 2], [3, 4], [5, 6]])))   # -> [ 9 12]
```

```python
# 12. Clip to Lower Bound  [where]
# Return the array with any value BELOW floor set to floor. Values above stay.
def clip_floor(a, floor):
    return np.where(a > floor,a ,floor)
print(clip_floor(np.array([-5, 2, 8, -1]), 0))   # -> [0 2 8 0]
```

```python
# 13. Row Means  [axis]
# Return a 1-D array of the mean of each ROW.
def row_means(mat):
    return np.mean(mat,axis=1)
print(row_means(np.array([[2., 4.], [6., 8.]])))   # -> [3. 7.]
```

```python
# 14. Count Above Threshold  [masking]
# Return HOW MANY values are strictly greater than threshold. Negatives allowed.
def count_above(a, thresh):
    return ((a>thresh).sum())
print(count_above(np.array([-5, 10, 3, 20, -1]), 0))   # -> 3
```

```python
# 15. Index of Min Per Column  [argmin + axis]
# Return a 1-D array with the ROW index of the min value in each column.
def argmin_per_col(mat):
    return np.argmin(mat,axis=0)
print(argmin_per_col(np.array([[3, 9], [1, 5], [7, 2]])))   # -> [1 2]
```

```python
# 16. Reverse Each Row  [slicing]
# Return the array with each ROW reversed left-to-right.
def reverse_rows(mat):
    return np.fliplr(mat)
print(reverse_rows(np.array([[1, 2, 3], [4, 5, 6]])))   # -> [[3 2 1] [6 5 4]]
```

```python
# 17. Z-Score Standardize 1D  [normalize]
# Return (x - mean) / std for a 1-D array. Use population std (ddof=0).
def zscore(a):
    return (a - a.mean())/np.std(a,ddof=0)
print(zscore(np.array([2., 4., 6., 8.])))   # -> [-1.3416 -0.4472 0.4472 1.3416]

```

```python
# 18. Count Even Values  [masking]
# Return HOW MANY values are even. Negatives and zero count as even when divisible.
def count_even(a):
    even_mask = (a % 2 == 0)
    return np.sum(even_mask)
print(count_even(np.array([1, 2, 3, 4, -6, 0])))   # -> 4   (2, 4, -6, 0)
```

```python
# 19. Diagonal Sum  [diagonal]
# Return the sum of the main diagonal of a square matrix.
def diag_sum(mat):
    return np.sum(np.diag(mat))
print(diag_sum(np.array([[1, 2], [3, 4]])))   # -> 5   (1 + 4)
```

```python
# 20. Replace Above Mean With Zero  [where]
# Return the array with every value strictly above the mean set to 0.
def zero_above_mean(a):
    mean = np.mean(a)
    a[a>mean] = 0
    return a
print(zero_above_mean(np.array([1, 2, 3, 10])))   # -> [1 2 3 0]   (mean=4, only 10 above)
```

```python
import pandas as pd

# 21. Filter by Two Conditions  [masking]
# Return rows where age >= 25 AND dept == "IT". Keep all columns, original index.
def filter_it_adults(df21):
    return df21[(df21['dept']=='IT') & (df21['age']>=25)]
df21 = pd.DataFrame({"name":["A","B","C","D"], "age":[30, 22, 40, 28], "dept":["IT","IT","HR","IT"]})
print(filter_it_adults(df21))   # A and D (IT and >= 25); B is IT but 22; C is 40 but HR
```

```python
# 22. Add Rank Column  [rank]
# Add "rank" = dense rank of salary, highest salary is rank 1. Return the df.
def add_rank(df):
    df["rank"] = df["salary"].rank(method="dense", ascending=False).astype(int)
    return df
df22 = pd.DataFrame({"name":["A","B","C"], "salary":[50, 90, 70]})
print(add_rank(df22))   # B rank 1, C rank 2, A rank 3
```

```python
# 23. Group Sum  [groupby]
# Total "amount" per "cust" as a DataFrame, columns "cust" and "total", reset index.
def cust_totals(df):
    return df.groupby('cust',as_index=False).agg(total = ('amount','sum')).reset_index(drop=True)
    #return df
df23 = pd.DataFrame({"cust":["X","Y","X","Y"], "amount":[10, 20, 30, 5]})
print(cust_totals(df23))   # X 40, Y 25
```

```python
# 24. Fill Missing With Group Median  [transform]
# Fill NaN in "score" with the MEDIAN of that row's "team". Return the df.
def fill_group_median(df):
    df['score'] = df['score'].fillna(df.groupby('team')['score'].transform('median'))
    return df
df24 = pd.DataFrame({"team":["A","A","A","B"], "score":[10.0, 20.0, np.nan, 50.0]})
print(fill_group_median(df24))   # A median of [10,20] = 15.0 -> fills the NaN
```

```python
# 25. Merge Keep All Left  [merge]
# Left-merge students with grades on "id". Students with no grade get "N/A". Return the df.
def merge_grades(students, grades):
    return pd.merge(students,grades,on='id',how='left',).fillna('N/A')
students25 = pd.DataFrame({"id":[1, 2, 3], "name":["A","B","C"]})
grades25 = pd.DataFrame({"id":[1, 3], "grade":["A+", "B"]})
print(merge_grades(students25, grades25))   # id 2 -> grade "N/A"
```

```python
# 26. Top N Per Group  [groupby]
# Return the top 2 highest-scoring rows per "subject", score descending within group.
# Columns: subject, student, score. Reset index.
def top2_per_subject(df):
    return df.sort_values('score',ascending=False).groupby('subject').head(2).reset_index(drop=True)
df26 = pd.DataFrame({
    "subject":["Math","Math","Math","Sci","Sci"],
    "student":["A","B","C","D","E"],
    "score":[80, 95, 70, 60, 88],
})
print(top2_per_subject(df26))   # Math: B(95), A(80) ; Sci: E(88), D(60)
```

```python
# 27. Pivot Count  [pivot]
# Pivot region (rows) x status (cols) counting rows. Missing combos -> 0.
def status_pivot(df):
    return pd.pivot_table(data=df,index='region',columns='status',aggfunc='size',fill_value=0)
df27 = pd.DataFrame({
    "region":["N","N","S","S","S"],
    "status":["win","loss","win","win","loss"],
})
display(status_pivot(df27))   # N: win 1 loss 1 ; S: win 2 loss 1
```

```python
# 28. One-Hot Encode  [preprocessing]
# Add one-hot columns for "color" (values red, blue). Return the df with the
# original plus the new 0/1 columns. Use get_dummies with int output.
def one_hot(df):
    dummies = pd.get_dummies(df['color'], dtype=int)
    return pd.concat([df, dummies], axis=1)
df28 = pd.DataFrame({"id":[1, 2, 3], "color":["red", "blue", "red"]})
print(one_hot(df28))   # columns id, color, then blue/red 0-1 flags
```

```python
# 29. Drop Columns With Any Missing  [cleaning]
# Return the df with any COLUMN that contains a NaN dropped entirely.
def drop_bad_cols(df):
    return df.dropna(axis=1)
df29 = pd.DataFrame({"a":[1, 2, 3], "b":[4, np.nan, 6], "c":[7, 8, 9]})
print(drop_bad_cols(df29))   # column b dropped, a and c survive
```

```python
# 30. Rolling Mean  [window]
# Add "roll3" = the 3-period rolling mean of "value". First two rows are NaN. Return the df.
def rolling_mean(df):
    df['roll3'] = df['value'].rolling(3).mean()
    return df
df30 = pd.DataFrame({"day":[1,2,3,4,5], "value":[10, 20, 30, 40, 50]})
print(rolling_mean(df30))   # roll3 at day 3 = 20.0, day 4 = 30.0, day 5 = 40.0
```

```python
# 1. Count Divisible  [logic]
# Return how many numbers are divisible by d. Handle negatives and zero.
def count_divisible(nums, d):
    return sum([1 for i in nums if i%d==0])
print(count_divisible([10, 15, 20, 7, -30], 5))   # -> 4   (10, 15, 20, -30)
print(count_divisible([1, 2, 3], 10))             # -> 0
```

```python
# 2. Reverse Words Keep Order of Letters  [string]
# Reverse the ORDER of words, but keep each word's letters as-is. Collapse spaces.
def reverse_word_order(s):
    return " ".join(s.split()[::-1]) if s.split() else ""
print(reverse_word_order("the quick  brown"))   # -> "brown quick the"
print(reverse_word_order(""))                   # -> ""
```

```python
# 3. Dict From Two Lists  [dict]
# Build a dict pairing keys to values. If keys is longer, extra keys map to None.
def pair_up(keys, values):
    len_keys, len_values = len(keys) , len(values)
    if len_keys >len_values : values.extend(['None']*(len_keys - len_values))
    return dict(zip(keys,values))
print(pair_up(["a", "b", "c"], [1, 2]))   # -> {'a': 1, 'b': 2, 'c': None}
print(pair_up(["x"], [9, 8]))  
```

```python
# 4. Second Smallest Distinct  [logic]
# Return the second smallest DISTINCT value, or None if there isn't one.
def second_smallest(nums):
    first = second = float('inf')
    for i in set(nums):
        if i < first:
            second = first
            first = i
        elif i < second and i != first:
            second =i
    return second if second != float('inf') else None
print(second_smallest([4, 1, 1, 3]))   # -> 3
print(second_smallest([2, 2]))         # -> None
print(second_smallest([5, 1, 2, 3]))   # -> 2
```

```python
# 5. Flatten and Dedupe  [set + list]
# Flatten a list of lists one level, remove duplicates, keep first-occurrence order.
def flatten_unique(nested):
    return list(set(x for i in nested for x in i))
print(flatten_unique([[1, 2], [2, 3], [3, 4]]))   # -> [1, 2, 3, 4]
print(flatten_unique([]))                         # -> []
```

```python
# 6. Normalize Columns Min-Max  [numpy / axis]
# Scale each COLUMN to 0..1 using (x - col_min) / (col_max - col_min). Float array.
def normalize_cols(mat):
    col_min = np.min(mat,axis=0)
    col_max = np.max(mat,axis=0)
    denom = col_max - col_min
    denom[denom == 0] = 1.0
    return ((mat-col_min)/denom)
print(normalize_cols(np.array([[10., 1.], [20., 3.], [30., 5.]])))
# col 0 -> [0, 0.5, 1] ; col 1 -> [0, 0.5, 1]
```

```python
# 7. Mask Out of Range  [numpy / where]
# Return the array with any value outside [low, high] set to -1. Inclusive bounds.
def mask_range(a, low, high):
    return np.where((a>=low) & (a<=high),a,-1)
print(mask_range(np.array([-5, 0, 5, 10, 15]), 0, 10))   # -> [-1  0  5 10 -1]
```

```python
# 8. Cumulative Product Per Row  [numpy / axis]
# Return an array where each ROW is its running cumulative product.
def row_cumprod(mat):
    return np.cumulative_prod(mat,axis=1)
print(row_cumprod(np.array([[1, 2, 3], [2, 2, 2]])))   # -> [[1 2 6] [2 4 8]]
```

```python
# 9. Count Per Group, Keep Big Groups  [pandas / groupby + filter]
# Return a Series dept -> headcount, but ONLY for depts with 2 or more people.
# Depts with 1 person must not appear.
def big_dept_counts(df):
    filtered_df = df.groupby('dept').filter(lambda x: len(x) >= 2)
    return filtered_df['dept'].value_counts()
df9 = pd.DataFrame({"name":["A","B","C","D","E"], "dept":["IT","IT","HR","IT","FIN"]})
print(big_dept_counts(df9))   # IT -> 3 only ; HR and FIN dropped
```

```python
# 10. Pivot Mean With Zero Fill  [pandas / pivot]
# Pivot region (rows) x product (cols), MEAN of amount. Missing combos -> 0.
def mean_pivot(df):
    return pd.pivot_table(data=df,index='region',columns='product',values='amount',aggfunc='mean',fill_value=0)
df10 = pd.DataFrame({
    "region":["N","N","S","S"],
    "product":["X","X","Y","Y"],
    "amount":[10, 20, 30, 50],
})
print(mean_pivot(df10))   # N/X mean = 15, S/Y mean = 40, N/Y and S/X -> 0
```
