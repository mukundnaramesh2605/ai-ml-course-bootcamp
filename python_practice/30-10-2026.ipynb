```python
import pandas as pd
import numpy as np
```

```python
# 1. Count Greater Than Average  [logic]
# Return how many numbers are strictly greater than the list's average.
def count_above_avg(nums):
    return sum([1 for i in nums if i > (sum(nums)/len(nums))])
print(count_above_avg([1, 2, 3, 10]))   # -> 1   (avg 4, only 10)
print(count_above_avg([5, 5, 5]))
```

```python
# 2. First Non-Repeating Char  [string]
# Return the first character that appears exactly once. None if none.
def first_unique(s):
    for i in s:
        if s.count(i) == 1:
            return i
    return None
print(first_unique("swiss"))   # -> "w"
print(first_unique("aabb"))    # -> None
```

```python
# 3. Merge Dicts Keep Max  [dict]
# Merge two dicts. On a shared key, keep the LARGER value.
def merge_max(d1, d2):
    result = d1.copy()
    for k,v in d2.items():
        result[k] = max(result.get(k,0),v)
    return result
print(merge_max({"a": 3, "b": 1}, {"b": 5, "c": 2}))   # -> {'a': 3, 'b': 5, 'c': 2}
print(merge_max({}, {"x": 1})) 
```

```python
# 4. Count Vowel Groups  [string]
# Return how many maximal runs of consecutive vowels appear. Case-insensitive.
import re
def vowel_groups(s):
    return len(re.findall(r'[aeiou]+',s.lower()))
print(vowel_groups("beautiful"))   # -> 3   (eau, i, u)
print(vowel_groups("xyz"))   
```

```python
# 5. Partition by Pivot  [two-pointer]
# Return a list: all values < pivot first (original order), then all >= pivot.
def partition(nums, pivot):
    return [i for i in nums if i < pivot] + [i for i in nums if i >= pivot]
print(partition([3, 1, 4, 1, 5, 9, 2], 4))   # -> [3, 1, 1, 2, 4, 5, 9]
print(partition([], 0)) 
```

```python
# 6. Longest Run of Same Char  [walk-and-track]
# Return the length of the longest run of the same character in a row.
def longest_run(s):
    if s == "":
        return 0
    count = best = 1
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            best=max(count,best)
            count=1
    return best
print(longest_run("aabbbbc"))   # -> 4
print(longest_run("abcabc"))    # -> 1   <- each char appears twice but never in a row
print(longest_run(""))          # -> 0
```

```python
# 7. Pairs That Sum to Target  [hashing]
# Return the COUNT of unordered index pairs (i < j) where nums[i] + nums[j] == target.
def count_sum_pairs(nums, target):
    counts = {}
    pair_count = 0
    for i,num in enumerate(nums):
        complement  = target - num
        if complement in counts:
            pair_count += counts[complement]
        counts[num] = counts.get(num, 0) + 1
    return pair_count
print(count_sum_pairs([1, 2, 3, 2, 1], 3))   # -> 4   (1+2 appears 4 ways)
print(count_sum_pairs([5, 5, 5], 10))        # -> 3
```

```python
# 8. Transpose a Matrix  [list]
# Given a list of lists (rectangular), return its transpose. No numpy.
def transpose(mat):
    return [list(i) for i in zip(*mat)]
print(transpose([[1, 2, 3], [4, 5, 6]]))   # -> [[1, 4], [2, 5], [3, 6]]
print(transpose([[1], [2], [3]]))  
```

```python
# 9. Caesar Shift  [string]
# Shift each lowercase letter forward by k (wrapping z->a). Leave non-letters as-is.
def caesar(s, k):
    res = []
    for char in s:
        if 'a' <= char <= 'z':
            shifted_char = chr((ord(char) - ord('a') + k) % 26 + ord('a'))
            res.append(shifted_char)
        else:
            res.append(char)
            
    return "".join(res)

print(caesar("abc", 2))     # -> "cde"
print(caesar("xyz", 3))     # -> "abc"   <- wraps around
print(caesar("a b!", 1))    # -> "b c!"
```

```python
# 10. Valid Number String  [logic]
# Return True if the string is a valid non-negative integer (digits only, non-empty).
def is_valid_number(s):
    return s.isdigit() and int(s)>0
print(is_valid_number("123"))   # -> True
print(is_valid_number("12a"))   # -> False
print(is_valid_number(""))      # -> False
```

```python
# 11. Column Std  [axis]
# Return a 1-D array of the std (population, ddof=0) of each COLUMN.
def col_std(mat):
    return np.std(mat,axis=0,ddof=0)
print(col_std(np.array([[1., 3.], [3., 7.]])))   # -> [1. 2.]
```

```python
# 12. Clip to Both Bounds  [clip]
# Return the array clipped to [low, high] inclusive.
def clip_both(a, low, high):
    return np.clip(a,low,high)
print(clip_both(np.array([-5, 2, 8, 15]), 0, 10))   # -> [ 0  2  8 10]
```

```python
# 13. Count Rows With Any Negative  [axis + any]
# Return HOW MANY rows contain at least one negative value.
def rows_with_negative(mat):
     return (mat < 0).any(axis=1).sum()
print(rows_with_negative(np.array([[1, 2], [-1, 3], [4, -5]])))   # -> 2
```

```python
# 14. Normalize to Unit Sum  [normalize]
# Return the 1-D array divided by its total so it sums to 1. Float.
def to_unit_sum(a):
    return a / a.sum()
print(to_unit_sum(np.array([2., 2., 4.])))   # -> [0.25 0.25 0.5 ]
```

```python
# 15. Pairwise Row Means Into Column  [axis + reshape]
# Return the row means as a COLUMN vector (shape n x 1).
def row_means_col(mat):
    return mat.mean(axis=1).reshape(-1, 1)
print(row_means_col(np.array([[2., 4.], [6., 8.]])))   # -> [[3.] [7.]]
```

```python
# 16. Replace Max With Zero  [argmax]
# Return the array with the single maximum value set to 0. First max if tied.
def zero_out_max(a):
    a[np.argmax(a)] = 0
    return a
print(zero_out_max(np.array([3, 7, 2, 7])))   # -> [3 0 2 7]   (first 7 only)
```

```python
# 17. Boolean Mask Count in Range  [masking]
# Return HOW MANY values are in the half-open range [low, high): low <= x < high.
def count_half_open(a, low, high):
    return ((a>=low)&(a<high)).sum()
print(count_half_open(np.array([0, 5, 10, 15]), 0, 10))   # -> 2   (0 and 5; 10 excluded)
```

```python
# 18. Elementwise Max of Two Arrays  [vectorize]
# Return the elementwise maximum of two equal-length arrays.
def elem_max(x, y):
    return np.maximum(x,y)
print(elem_max(np.array([1, 8, 3]), np.array([4, 2, 3])))   # -> [4 8 3]
```
