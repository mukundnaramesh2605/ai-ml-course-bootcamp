```python
# 1. Count Greater Than Average  [logic]
# Return how many numbers are strictly greater than the list's average.
def count_above_avg(nums):
    return sum([1 for i in nums if i > (sum(nums)/len(nums))])
print(count_above_avg([1, 2, 3, 10]))   # -> 1   (avg 4, only 10)
print(count_above_avg([5, 5, 5]))
```

```python
# 2. First Non-Repeating Char  [string]
# Return the first character that appears exactly once. None if none.
def first_unique(s):
    for i in s:
        if s.count(i) == 1:
            return i
    return None
print(first_unique("swiss"))   # -> "w"
print(first_unique("aabb"))    # -> None
```

```python
# 3. Merge Dicts Keep Max  [dict]
# Merge two dicts. On a shared key, keep the LARGER value.
def merge_max(d1, d2):
    result = d1.copy()
    for k,v in d2.items():
        result[k] = max(result.get(k,0),v)
    return result
print(merge_max({"a": 3, "b": 1}, {"b": 5, "c": 2}))   # -> {'a': 3, 'b': 5, 'c': 2}
print(merge_max({}, {"x": 1})) 
```

```python
# 4. Count Vowel Groups  [string]
# Return how many maximal runs of consecutive vowels appear. Case-insensitive.
import re
def vowel_groups(s):
    return len(re.findall(r'[aeiou]+',s.lower()))
print(vowel_groups("beautiful"))   # -> 3   (eau, i, u)
print(vowel_groups("xyz"))   
```
