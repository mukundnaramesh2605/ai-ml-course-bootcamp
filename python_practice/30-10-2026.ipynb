```python
# 1. Count Greater Than Average  [logic]
# Return how many numbers are strictly greater than the list's average.
def count_above_avg(nums):
    return sum([1 for i in nums if i > (sum(nums)/len(nums))])
print(count_above_avg([1, 2, 3, 10]))   # -> 1   (avg 4, only 10)
print(count_above_avg([5, 5, 5]))
```

```python
# 2. First Non-Repeating Char  [string]
# Return the first character that appears exactly once. None if none.
def first_unique(s):
    for i in s:
        if s.count(i) == 1:
            return i
    return None
print(first_unique("swiss"))   # -> "w"
print(first_unique("aabb"))    # -> None
```
