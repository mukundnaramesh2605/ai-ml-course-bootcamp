```python
# 1. Return the number of times each element appears, only for duplicates
def duplicates_only(lst):
    return {i:lst.count(i) for i in lst if lst.count(i) > 1}

print(duplicates_only([1, 2, 2, 3, 3, 3]))      # normal
print(duplicates_only([1, 2, 3]))               # no duplicates
print(duplicates_only([]))                      # empty
print(duplicates_only([5, 5, 5, 5]))            # all same
print(duplicates_only(["a", "b", "a", "c"]))    # strings
```

```python
# 2. Check if a string is a rotation of another (e.g. "abc" -> "cab")
def is_rotation(a, b):
    return b in (a+a)

print(is_rotation("abcde", "cdeab"))    # valid rotation
print(is_rotation("abc", "acb"))        # not a rotation
print(is_rotation("", ""))              # both empty
print(is_rotation("a", "a"))            # single char
print(is_rotation("abc", "abcd"))       # different lengths
```

```python
# 3. Return the running maximum of a list
def running_max(lst):
    if not lst:
        return []
    result = [lst[0]]
    
    for i in lst[1:]:
        result.append(max(i,result[-1]))
    return result
print(running_max([1, 3, 2, 5, 4]))     # normal
print(running_max([5, 4, 3]))           # decreasing
print(running_max([]))                  # empty
print(running_max([7]))                 # single element
print(running_max([-1, -5, -2]))        # negatives
```

```python
# 4. Group a list of words by their first letter
from collections import defaultdict
def group_by_first(words):
    grouped = defaultdict(list)
    for word in words:
        grouped[word[0]].append(word)
    return list(grouped.values())
print(group_by_first(["apple", "ant", "bat", "cat"]))   # normal
print(group_by_first([]))                               # empty
print(group_by_first(["one"]))                          # single word
print(group_by_first(["dog", "deer", "duck"]))          # all same letter
print(group_by_first(["x", "y", "z"]))                  # single chars
```

```python
# 5. Find the longest run of consecutive equal values
def longest_run(lst):
    if not lst:
        return 0
    best = count = 1
    for i in range(1,len(lst)):
        if lst[i-1] == lst[i]:
            count+=1
        else:
            best=max(count,best)
            count=1
    return max(best,count)

print(longest_run([1, 1, 2, 2, 2, 3]))      # normal
print(longest_run([1, 2, 3]))               # no repeats
print(longest_run([]))                      # empty
print(longest_run([4, 4, 4, 4]))            # all same
print(longest_run([1, 1, 2, 2]))            # tie, first wins
```

```python
import numpy as np

# 1. Swap two rows in a 2D array
def swap_rows(mat, i, j):
    out = mat.copy()
    out[[i,j]] = out[[j,i]]
    return out

print(swap_rows(np.array([[1, 2], [3, 4]]), 0, 1))                  # normal
print(swap_rows(np.array([[1, 2], [3, 4], [5, 6]]), 0, 2))         # 3 rows
print(swap_rows(np.array([[1, 2], [3, 4]]), 0, 0))                 # same index
print(swap_rows(np.array([[-1, -2], [3, 4]]), 0, 1))              # negatives
print(swap_rows(np.array([[1, 1], [1, 1]]), 0, 1))               # identical rows
```

```python
# 2. Return the n largest values in an array (sorted descending)
def n_largest(arr, n):
    n = min(n, len(arr))
    sorted_arr = np.sort(arr)
    return sorted_arr[-n:][::-1]
print(n_largest(np.array([5, 1, 9, 3, 7]), 3))      # normal
print(n_largest(np.array([1, 2, 3]), 3))            # n equals length
print(n_largest(np.array([4, 4, 4]), 2))            # all same
print(n_largest(np.array([10]), 5))                 # n bigger than length
print(n_largest(np.array([-1, -5, -2]), 2))         # negatives
```

```python
import numpy as np

# 1. Min-max scale each column of a 2D feature matrix independently
def scale_columns(mat):
    cmat = np.asarray(mat, dtype=float)
    col_min = mat.min(axis=0, keepdims=True)
    col_max = mat.max(axis=0, keepdims=True)
    ptp = col_max - col_min
    return np.where(ptp == 0, 0.0, (mat - col_min) / ptp)

print(scale_columns(np.array([[1, 10], [2, 20], [3, 30]])))     # normal
print(scale_columns(np.array([[5, 5], [5, 5]])))               # constant column (watch div by zero)
print(scale_columns(np.array([[1, 2, 3]])))                    # single row
print(scale_columns(np.array([[1.0], [2.0], [3.0]])))          # single column
print(scale_columns(np.array([[-1, 0], [0, 10], [1, 20]])))    # negatives
```
