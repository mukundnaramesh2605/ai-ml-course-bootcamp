```python
# 1. FizzBuzz List  [Service / screening]
# Return a list of length n (1..n): "Fizz" if divisible by 3,
# "Buzz" if by 5, "FizzBuzz" if both, else the number as a string.
# n=5 -> ["1","2","Fizz","4","Buzz"]
def fizzbuzz(n):
    result =[]
    if n%3 ==0 and n%5 ==0:
        return "FizzBuzz"
    else:
        for i in range(1,n+1):
            if i%3==0:
                result.append("Fizz")
            elif i%5==0:
                result.append("Buzz")
            else:
                result.append(str(i))
    return result
print(fizzbuzz(5))# -> ["1","2","Fizz","4","Buzz"] ; 
print(fizzbuzz(15))#[-1] -> "FizzBuzz"
```

```python
# 2. Two-Sum Exists  [AI screen / logic]
# Return True if any two DISTINCT elements sum to target, else False.
# [1,2,3], 5 -> True ; [1,2,3], 7 -> False
def has_pair(nums, target):
    s=set()
    for num in nums:
        complement = target-num
        if complement in s:
            return True
        s.add(num)
    return False
print(has_pair([1,2,3], 5))# -> True ; 
print(has_pair([1,2,3], 7))# -> False ; 
print(has_pair([3], 6))# -> False
```

```python
# 3. Compress String  [AI screen / string]
# Run-length encode: "aaabbc" -> "a3b2c1". Return the compressed string.
def compress(s):
    if s =="":
        return ""
    result =[]
    count = 1
    current = s[0]
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current=s[i]
    result.append(current+str(count))
    return "".join(result)
print(compress("aaabbc"))# -> "a3b2c1" ; 
print(compress(""))# -> "" ; 
print(compress("x"))# -> "x1"
```

```python
# 4. Balanced Brackets  [AI screen / stack]
# Return True if (), [], {} are all correctly matched and nested.
# "{[()]}" -> True ; "{[(])}" -> False
def balanced(s):
    brackets = {"}":"{","]":"[",")":"("}
    seen = []
    for i in s:
        if i in brackets.values():
            seen.append(i)
        elif i in brackets:
            if not seen or seen[-1]!=brackets[i]:
                return False
            seen.pop()
    return len(seen) == 0
print(balanced("{[()]}"))# -> True ; 
print(balanced("{[(])}"))# -> False ; 
print(balanced(""))# -> True

```
