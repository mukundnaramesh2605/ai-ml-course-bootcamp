```python
# 1. FizzBuzz List  [Service / screening]
# Return a list of length n (1..n): "Fizz" if divisible by 3,
# "Buzz" if by 5, "FizzBuzz" if both, else the number as a string.
# n=5 -> ["1","2","Fizz","4","Buzz"]
def fizzbuzz(n):
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:          # both, must come first
            out.append("FizzBuzz")
        elif i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(str(i))
    return out
print(fizzbuzz(5))# -> ["1","2","Fizz","4","Buzz"] ; 
print(fizzbuzz(15)[-1])#-> "FizzBuzz"
```

```python
# 2. Two-Sum Exists  [AI screen / logic]
# Return True if any two DISTINCT elements sum to target, else False.
# [1,2,3], 5 -> True ; [1,2,3], 7 -> False
def has_pair(nums, target):
    s=set()
    for num in nums:
        complement = target-num
        if complement in s:
            return True
        s.add(num)
    return False
print(has_pair([1,2,3], 5))# -> True ; 
print(has_pair([1,2,3], 7))# -> False ; 
print(has_pair([3], 6))# -> False
```

```python
# 3. Compress String  [AI screen / string]
# Run-length encode: "aaabbc" -> "a3b2c1". Return the compressed string.
def compress(s):
    if s =="":
        return ""
    result =[]
    count = 1
    current = s[0]
    for i in range(1,len(s)):
        if s[i] == s[i-1]:
            count+=1
        else:
            result.append(current+str(count))
            count=1
            current=s[i]
    result.append(current+str(count))
    return "".join(result)
print(compress("aaabbc"))# -> "a3b2c1" ; 
print(compress(""))# -> "" ; 
print(compress("x"))# -> "x1"
```

```python
# 4. Balanced Brackets  [AI screen / stack]
# Return True if (), [], {} are all correctly matched and nested.
# "{[()]}" -> True ; "{[(])}" -> False
def balanced(s):
    brackets = {"}":"{","]":"[",")":"("}
    seen = []
    for i in s:
        if i in brackets.values():
            seen.append(i)
        elif i in brackets:
            if not seen or seen[-1]!=brackets[i]:
                return False
            seen.pop()
    return len(seen) == 0
print(balanced("{[()]}"))# -> True ; 
print(balanced("{[(])}"))# -> False ; 
print(balanced(""))# -> True

```

```python
# 5. Second Largest  [Service / logic]
# Return the second largest DISTINCT value, or None if there isn't one.
# [4,1,7,7,3] -> 4
def second_largest(nums):
    if len(set(nums)) == 1:
        return None
    first = second = float('-inf')
    for i in nums:
        if i > first:
            second=first
            first = i
        elif i > second and i!= first:
            second=i
    return second
        
print(second_largest([4,1,7,7,3]))# -> 4 ; 
print(second_largest([5,5]))# -> None
```

```python
# 6. Longest Word  [Service / string]
# Return the longest word. On a tie, return the one appearing first.
# "the quick brown fox" -> "quick"
def longest_word(s):
    counts = {word:len(word) for word in s.split()}
    sorted_counts = dict(sorted(counts.items(),key=lambda x: x[1],reverse=True))
    return([x for x in sorted_counts][0])
print(longest_word("the quick brown fox"))# -> "quick" ; 
print(longest_word("a bb cc"))# -> "bb"
```

```python
# 7. Clean and Dedupe  [AI screen / data handling]
# Given messy strings, return a SORTED list of distinct values
# after stripping whitespace and lowercasing.
# ["  Apple", "apple", "BANANA "] -> ["apple", "banana"]
import re
def clean_dedupe(items):
    s = re.findall(r'[a-zA-Z]+'," ".join(items).lower())
    return list(sorted((x.lower() for x in set(s))))
print(clean_dedupe(["  Apple","apple","BANANA "]))# -> ["apple","banana"]
```

```python
# 9. Group Sum, Filter, Sort  [AI screen / pandas]
# Return total amount per region, keep only regions with total >= 100,
# sorted descending, as a DataFrame (reset index), columns "region" and "total".
import pandas as pd
sales = pd.DataFrame({
    "region": ["N","S","N","E","S","N"],
    "amount": [60, 40, 50, 200, 70, 30],
})
grouped = sales.groupby('region').agg(total=('amount','sum'))
grouped_100 = grouped[grouped['total']>=100].sort_values('total',ascending=False).reset_index()
grouped_100
```

```python
# 10. Fill Missing by Group  [AI screen / pandas]
# Fill NaN in "score" with the MEAN score of that row's own "team".
# Return the full df.
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "team": ["A","A","B","B"],
    "score": [10.0, np.nan, 20.0, np.nan],
})
df['score'] = df['score'].fillna(df.groupby('team')['score'].transform('mean'))
df
```

```python
# 1. Capitalize Words  [Service / string]
# Capitalize the first letter of each word, leave the rest as-is.
# "hello world" -> "Hello World"
def cap_words(s):
    return " ".join([x.title() for x in s.split()])
print(cap_words("hello world"))# -> "Hello World" ; 
print(cap_words(""))# -> ""
```

```python
# 2. Missing Number  [AI screen / logic]
# A list holds 0..n with exactly one missing. Return the missing number.
# [0,1,3,4] -> 2
def missing(nums):
    min_,max_ = min(nums),max(nums)
    return list(set(range(0,max_+1)) ^ set(nums))[0]
print(missing([0,1,3,4]))# -> 2 ; 
print(missing([1]))# -> 0
```

```python
# 3. Count Substrings  [Service / string]
# Return how many times substring t appears in s (overlaps count).
# "aaa", "aa" -> 2
def count_occurrences(s, t):
    count = 0
    m=len(t)
    for i in range(len(s)-m+1):
        if s[i:i+m] == t:
            count+=1
    return count
print(count_occurrences("aaa","aa"))# -> 2 ; 
print(count_occurrences("abc","x"))# -> 0
```

```python
# 4. Move Zeros  [AI screen / logic]
# Move all 0s to the end, keep the order of the non-zero values.
# [0,1,0,3,2] -> [1,3,2,0,0]
def move_zeros(nums):
    return [x for x in nums if x>0] + [0]*nums.count(0)
print(move_zeros([0,1,0,3,2]))# -> [1,3,2,0,0]
```

```python
# 5. Anagram Check  [AI screen / string]
# Return True if two strings are anagrams, case-insensitive, ignoring spaces.
# "Listen", "Silent" -> True
def is_anagram(a, b):
    return sorted(a.lower()) == sorted(b.lower())
print(is_anagram("Listen","Silent"))# -> True ; 
print(is_anagram("abc","abd"))# -> False
```
