```python
# 1. Count Elements Appearing Once  [set]
# Return how many values appear EXACTLY once in the list.
def count_singletons(nums):
    return sum([1 for i in nums if nums.count(i) == 1])
print(count_singletons([1, 2, 2, 3, 4, 4]))   # -> 2   (1 and 3)
print(count_singletons([5, 5]))               # -> 0
print(count_singletons([]))    
```

```python
# 2. Move Zeros to End  [two-pointer]
# Move all 0s to the end, keep order of the rest. Negatives stay in order.
def move_zeros(nums):
    return [i for i in nums if i!=0]+[0]*nums.count(0)
print(move_zeros([0, -1, 0, 3, 12]))   # -> [-1, 3, 12, 0, 0]
print(move_zeros([0, 0, 1]))  
```

```python
# 3. Title Case Words  [string]
# Capitalize first letter of each word, lowercase the rest. Collapse extra spaces.
def title_case(s):
    return " ".join([word[0].upper()+word[1:] for word in s.lower().split()])
print(title_case("hELLO   wORLD"))   # -> "Hello World"
print(title_case(""))       
```

```python
# 4. Group Anagrams  [dict]
# Group words that are anagrams. Return a list of groups, each group a list,
# groups in first-appearance order.
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for word in words:
        grouped["".join(sorted(word))].append(word)
    return list(grouped.values())
print(group_anagrams(["eat", "tea", "tan", "ate", "nat"]))
# -> [['eat', 'tea', 'ate'], ['tan', 'nat']]
print(group_anagrams([]))   # -> []
```
