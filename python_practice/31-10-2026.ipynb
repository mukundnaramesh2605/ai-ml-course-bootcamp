```python
# 1. Count Elements Appearing Once  [set]
# Return how many values appear EXACTLY once in the list.
def count_singletons(nums):
    return sum([1 for i in nums if nums.count(i) == 1])
print(count_singletons([1, 2, 2, 3, 4, 4]))   # -> 2   (1 and 3)
print(count_singletons([5, 5]))               # -> 0
print(count_singletons([]))    
```

```python
# 2. Move Zeros to End  [two-pointer]
# Move all 0s to the end, keep order of the rest. Negatives stay in order.
def move_zeros(nums):
    return [i for i in nums if i!=0]+[0]*nums.count(0)
print(move_zeros([0, -1, 0, 3, 12]))   # -> [-1, 3, 12, 0, 0]
print(move_zeros([0, 0, 1]))  
```

```python
# 3. Title Case Words  [string]
# Capitalize first letter of each word, lowercase the rest. Collapse extra spaces.
def title_case(s):
    return " ".join([word[0].upper()+word[1:] for word in s.lower().split()])
print(title_case("hELLO   wORLD"))   # -> "Hello World"
print(title_case(""))       
```

```python
# 4. Group Anagrams  [dict]
# Group words that are anagrams. Return a list of groups, each group a list,
# groups in first-appearance order.
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for word in words:
        grouped["".join(sorted(word))].append(word)
    return list(grouped.values())
print(group_anagrams(["eat", "tea", "tan", "ate", "nat"]))
# -> [['eat', 'tea', 'ate'], ['tan', 'nat']]
print(group_anagrams([]))   # -> []
```

```python
# 5. Two Sum Indices  [hashing]
# Return indices of the two numbers that add to target. Exactly one solution.
# Must work with negatives.
def two_sum(nums, target):
    seen = {}
    for i,num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return [seen[comp],i]
        seen[num] = i
    return []
print(two_sum([2, 7, 11, 15], 9))   # -> [0, 1]
print(two_sum([-3, 4, 1, 90], -2))  # -> [0, 2]   (-3 + 1)
```

```python
# 6. Compress Consecutive Duplicates  [string]
# Collapse consecutive duplicate chars to a single char. "aaabbbc" -> "abc".
import re
def dedupe_consecutive(s):
    return re.sub(r'(.)\1+', r'\1', s)
print(dedupe_consecutive("aaabbbc"))   # -> "abc"
print(dedupe_consecutive("aabaa"))     # -> "aba"   <- non-adjacent repeats stay
print(dedupe_consecutive(""))  
```

```python
# 7. Max Subarray Sum  [logic]
# Return the largest sum of any contiguous non-empty subarray (Kadane).
def max_subarray(nums):
    max_so_far = current_max = nums[0]
    for num in nums[1:]:
        current_max = max(num, current_max + num)
        max_so_far = max(max_so_far, current_max)
    return max_so_far
print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))   # -> 6
print(max_subarray([-1, -2, -3]))   
```
