```python
# 1. Count Elements Appearing Once  [set]
# Return how many values appear EXACTLY once in the list.
def count_singletons(nums):
    return sum([1 for i in nums if nums.count(i) == 1])
print(count_singletons([1, 2, 2, 3, 4, 4]))   # -> 2   (1 and 3)
print(count_singletons([5, 5]))               # -> 0
print(count_singletons([]))    
```

```python
# 2. Move Zeros to End  [two-pointer]
# Move all 0s to the end, keep order of the rest. Negatives stay in order.
def move_zeros(nums):
    return [i for i in nums if i!=0]+[0]*nums.count(0)
print(move_zeros([0, -1, 0, 3, 12]))   # -> [-1, 3, 12, 0, 0]
print(move_zeros([0, 0, 1]))  
```

```python
# 3. Title Case Words  [string]
# Capitalize first letter of each word, lowercase the rest. Collapse extra spaces.
def title_case(s):
    return " ".join([word[0].upper()+word[1:] for word in s.lower().split()])
print(title_case("hELLO   wORLD"))   # -> "Hello World"
print(title_case(""))       
```

```python
# 4. Group Anagrams  [dict]
# Group words that are anagrams. Return a list of groups, each group a list,
# groups in first-appearance order.
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for word in words:
        grouped["".join(sorted(word))].append(word)
    return list(grouped.values())
print(group_anagrams(["eat", "tea", "tan", "ate", "nat"]))
# -> [['eat', 'tea', 'ate'], ['tan', 'nat']]
print(group_anagrams([]))   # -> []
```

```python
# 5. Two Sum Indices  [hashing]
# Return indices of the two numbers that add to target. Exactly one solution.
# Must work with negatives.
def two_sum(nums, target):
    seen = {}
    for i,num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return [seen[comp],i]
        seen[num] = i
    return []
print(two_sum([2, 7, 11, 15], 9))   # -> [0, 1]
print(two_sum([-3, 4, 1, 90], -2))  # -> [0, 2]   (-3 + 1)
```

```python
# 6. Compress Consecutive Duplicates  [string]
# Collapse consecutive duplicate chars to a single char. "aaabbbc" -> "abc".
import re
def dedupe_consecutive(s):
    return re.sub(r'(.)\1+', r'\1', s)
print(dedupe_consecutive("aaabbbc"))   # -> "abc"
print(dedupe_consecutive("aabaa"))     # -> "aba"   <- non-adjacent repeats stay
print(dedupe_consecutive(""))  
```

```python
# 7. Max Subarray Sum  [logic]
# Return the largest sum of any contiguous non-empty subarray (Kadane).
def max_subarray(nums):
    max_so_far = current_max = nums[0]
    for num in nums[1:]:
        current_max = max(num, current_max + num)
        max_so_far = max(max_so_far, current_max)
    return max_so_far
print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))   # -> 6
print(max_subarray([-1, -2, -3]))   
```

```python
# 8. Count Substring Occurrences  [string]
# Return how many times sub appears in s, including OVERLAPPING matches.
import re
def count_overlapping(s, sub):
    return len(re.findall(f"(?={re.escape(sub)})",s))
print(count_overlapping("aaaa", "aa"))   # -> 3   (positions 0,1,2 overlap)
print(count_overlapping("abcabc", "bc")) # -> 2
```

```python
# 9. Valid Parentheses  [stack]
# Return True if (), [], {} are correctly matched and nested.
def is_valid(s):
    brackets = {"}":"{","]":"[",")":"("}
    stack=[]
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0
print(is_valid("([{}])"))   # -> True
print(is_valid("([)]"))     # -> False
print(is_valid("("))        # -> False
```

```python
# 10. Second Largest Distinct  [logic]
# Return the second largest DISTINCT value, or None if there isn't one.
def second_largest(nums):
    first = second = float('-inf')
    for i in nums:
        if i > first:
            second = first
            first = i
        elif i > second and i != first:
            second = i 
    return second if second != float('-inf') else None
print(second_largest([4, 1, 4, 3]))     # -> 3
print(second_largest([7, 7]))           # -> None
print(second_largest([9, 2, 5, 5, 9]))  # -> 5
```

```python
import numpy as np

# 11. Row Sums  [axis]
# Return a 1-D array of the sum of each ROW.
def row_sums(mat):
    return np.sum(mat,axis=1)
print(row_sums(np.array([[1, 2, 3], [4, 5, 6]])))   # -> [ 6 15]
```

```python
# 12. Count Values Above Mean  [masking]
# Return HOW MANY values are strictly greater than the array's mean.
def above_mean(a):
    return np.sum(a > a.mean())
print(above_mean(np.array([1, 2, 3, 10])))   # -> 1
```

```python
# 13. Min-Max Normalize  [normalize]
# Scale a 1-D array to 0..1 using (x - min) / (max - min). Float.
def min_max(a):
    return (a - a.min())/(a.max()-a.min())
print(min_max(np.array([10., 20., 30.])))   # -> [0.  0.5 1. ]
```

```python
# 14. Replace Negatives With Zero  [where]
# Return the array with every negative value set to 0. Do not mutate the input.
def clip_negatives(a):
    return np.where(a < 0,0,a)
orig = np.array([-1, 5, -3, 0, 2])
print(clip_negatives(orig))   # -> [0 5 0 0 2]
print(orig)                   # -> [-1  5 -3  0  2]   <- unchanged
```

```python
# 15. Column With Largest Sum  [argmax + axis]
# Return the INDEX of the column whose values sum to the most.
def max_col_index(mat):
    return np.argmax(np.sum(mat,axis=0))
print(max_col_index(np.array([[1, 5], [2, 6], [3, 7]])))   # -> 1
```

```python
# 16. Count in Closed Range  [masking]
# Return HOW MANY values fall in [low, high] inclusive. Handle negatives.
def count_in_range(a, low, high):
    return np.sum((a>=low)&(a<=high))
print(count_in_range(np.array([-5, 0, 3, 10, 10, -20]), 0, 10))   # -> 4
```

```python
# 17. Standardize 1D  [normalize]
# Return (x - mean) / std for a 1-D array. Population std (ddof=0).
def zscore(a):
    return (a - a.mean())/a.std()
print(zscore(np.array([2., 4., 6., 8.])))   # -> [-1.3416 -0.4472 0.4472 1.3416]

```

```python
# 18. Row-wise Max  [axis]
# Return a 1-D array of the max of each ROW.
def row_max(mat):
    return np.max(mat,axis=1)
print(row_max(np.array([[1, 9, 3], [8, 2, 5]])))   # -> [9 8]
```

```python
# 19. Count Nonzero Per Column  [axis + masking]
# Return a 1-D array: for each COLUMN, how many values are nonzero.
def nonzero_per_col(mat):
    return np.count_nonzero(mat,axis=0)
print(nonzero_per_col(np.array([[0, 1], [2, 0], [3, 4]])))   # -> [2 2]
```

```python
# 20. Clip and Count Changed  [clip]
# Clip to [0, 100], return HOW MANY values were actually changed.
def count_clipped(a):
    return np.sum(a != np.clip(a,0,100))
print(count_clipped(np.array([-5, 50, 150, 0, 100])))   # -> 2
```

```python
import pandas as pd

# 21. Filter Rows by Condition  [masking]
# Return rows where "age" >= 30. Keep all columns and the original index.
def adults(df):
    return df[df['age']>=30]
df21 = pd.DataFrame({"name":["A","B","C","D"], "age":[25, 30, 45, 18]})
print(adults(df21))   # B and C
```

```python
# 22. Value Counts as DataFrame  [groupby]
# Return a DataFrame with columns "city" and "count": rows per city, highest first.
# Reset index.
def city_counts(df):
    return df['city'].value_counts()
df22 = pd.DataFrame({"name":["A","B","C","D","E"], "city":["CHN","MUM","CHN","CHN","MUM"]})
print(city_counts(df22))   # CHN 3, MUM 2
```
