```python
# ============ PYTHON (1-10) ============

# 1. Sum of Digits  [logic]
# Return the sum of the digits of a non-negative integer.
def digit_sum(n):
    return sum([int(i) for i in str(n)])
print(digit_sum(1234))   # -> 10
print(digit_sum(0))      # -> 0
```

```python
# 2. Count Vowels  [string]
# Return how many vowels (a,e,i,o,u) are in the string, case-insensitive.
def count_vowels(s):
    return sum([1 for i in s.lower() if i in 'aeiou'])
print(count_vowels("Hello World"))   # -> 3
print(count_vowels("xyz"))           # -> 0
print(count_vowels(""))              # -> 0
```

```python
# 3. Dedupe Keep Order  [logic]
# Remove duplicates, keep first-occurrence order.
def dedupe(nums):
    seen = []
    for i in nums:
        if i not in seen:
            seen.append(i)
    return seen
print(dedupe([3, 1, 3, 2, 1]))   # -> [3, 1, 2]
print(dedupe([]))                # -> []
```

```python
# 4. Group by First Letter  [dict]
# Return a dict: first letter -> list of words starting with it, original order.
from collections import defaultdict
def group_by_first(words):
    grouped = defaultdict(list)
    for word in words:
        grouped[word[0]].append(word)
    return dict(grouped)
print(group_by_first(["apple", "banana", "avocado"]))   # -> {'a': ['apple', 'avocado'], 'b': ['banana']}
print(group_by_first([]))           
```

```python
# 5. Count Occurrences of Target  [logic]
# Return how many times target appears in the list.
def count_target(nums, target):
    return nums.count(target)
print(count_target([1, 2, 2, 3, 2], 2))   # -> 3
print(count_target([1, 2, 3], 5))         # -> 0
```

```python
# 6. Sum of Squares of Evens  [logic]
# Return the sum of squares of only the EVEN numbers. Negatives count.
def sum_even_squares(nums):
    return sum(i**2 for i in nums if i%2 == 0)
print(sum_even_squares([1, 2, 3, 4]))   # -> 20   (4 + 16)
print(sum_even_squares([-2, 1]))        # -> 4    ((-2)^2)
print(sum_even_squares([1, 3]))         # -> 0
```
