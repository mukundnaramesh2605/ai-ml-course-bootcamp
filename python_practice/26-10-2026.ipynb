```python
# 1. Reverse Integer  [Service / screening]
# Return the integer with its digits reversed. Preserve the sign.
# 1234 -> 4321 ; -520 -> -25
def reverse_int(n):
    sign = -1 if n < 0 else 1
    reversed_num = int(str(abs(n))[::-1])
    return sign * reversed_num
    
print(reverse_int(1234))# -> 4321 ; 
print(reverse_int(-520))# -> -25 ; 
print(reverse_int(0))# -> 0
```

```python
# 2. Palindrome Number  [Service / screening]
# Return True if the integer reads the same forwards and backwards.
# 121 -> True ; 123 -> False ; negatives are never palindromes
def is_palindrome_num(n):
    return str(n) == str(n)[::-1]
print(is_palindrome_num(121))# -> True ; 
print(is_palindrome_num(123))# -> False ; 
print(is_palindrome_num(-121))# -> False
```
