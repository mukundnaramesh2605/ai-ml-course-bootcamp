```python
# 1. Reverse Integer  [Service / screening]
# Return the integer with its digits reversed. Preserve the sign.
# 1234 -> 4321 ; -520 -> -25
def reverse_int(n):
    sign = -1 if n < 0 else 1
    reversed_num = int(str(abs(n))[::-1])
    return sign * reversed_num
    
print(reverse_int(1234))# -> 4321 ; 
print(reverse_int(-520))# -> -25 ; 
print(reverse_int(0))# -> 0
```

```python
# 2. Palindrome Number  [Service / screening]
# Return True if the integer reads the same forwards and backwards.
# 121 -> True ; 123 -> False ; negatives are never palindromes
def is_palindrome_num(n):
    return str(n) == str(n)[::-1]
print(is_palindrome_num(121))# -> True ; 
print(is_palindrome_num(123))# -> False ; 
print(is_palindrome_num(-121))# -> False
```

```python
# 3. Prime Check  [Service / screening]
# Return True if n is prime (n >= 2). 1 and below are not prime.
# 7 -> True ; 9 -> False
def is_prime(n):
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True
print(is_prime(7))# -> True ; 
print(is_prime(9))# -> False ; 
print(is_prime(1))# -> False ; 
print(is_prime(2))# -> True

```

```python
# 4. Fibonacci Series  [Service / screening]
# Return a list of the first n Fibonacci numbers starting 0, 1.
# 5 -> [0,1,1,2,3]
def fib(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]
    res = [0, 1]
    while len(res) < n:
        res.append(res[-1] + res[-2])
    return res
    
print(fib(5))# -> [0,1,1,2,3] ; 
print(fib(1))# -> [0] ; 
print(fib(0))# -> []
```

```python
# 5. Armstrong Number  [Service / screening]
# A 3-digit number equal to the sum of the cubes of its digits.
# Return True/False. 153 -> True (1+125+27) ; 154 -> False
def is_armstrong(n):
    return n == sum([int(i)**3 for i in str(n)])
    
print(is_armstrong(153))# -> True ; 
print(is_armstrong(154))# -> False ; 
print(is_armstrong(370))# -> True

```

```python
# 6. GCD  [Service / screening]
# Return the greatest common divisor of a and b (Euclid's algorithm).
# 12, 18 -> 6
def gcd(a, b):
    while b != 0:
        a,b = b, a % b
    return a
print(gcd(12,18))# -> 6 ; 
print(gcd(7,13))# -> 1 ; 
print(gcd(0,5))# -> 5
```

```python
# 7. Count Vowels and Consonants  [Service / screening]
# Return a tuple (vowels, consonants). Count letters only, case-insensitive.
# "Hello World" -> (3, 7)
def count_vc(s):
    vowels = sum([1 for i in s if i.lower() in 'aeiou' and i.isalpha()])
    cons = sum([1 for i in s if i.lower() not in 'aeiou' and i.isalpha()])
    return (vowels,cons)
print(count_vc("Hello World"))# -> (3, 7) ; 
print(count_vc("123"))# -> (0, 0)
```

```python
# 8. Anagram Check  [Service / screening]
# Return True if a and b are anagrams, case-insensitive.
# "Listen", "Silent" -> True
def is_anagram(a, b):
    return sorted(a.lower()) == sorted(b.lower())
    
print(is_anagram("Listen","Silent"))# -> True ; 
print(is_anagram("abc","abd"))# -> False
```

```python
# 9. Factorial (iterative, no library)  [Service / screening]
# Return n! using a loop. 0! is 1.
# 5 -> 120
def factorial(n):
    if n == 0:
        return 1
    count=1
    for i in range(1,n+1):
        count*=i
    return count
print(factorial(5))# -> 120 ; 
print(factorial(0))# -> 1
```

```python
# 10. Second Largest  [Service / screening]
# Return the second largest DISTINCT value, or None if there isn't one.
# [4,1,7,7,3] -> 4
def second_largest(nums):
    first = second = float('-inf')
    for i in set(nums):
        if i > first:
            second = first
            first = i
        elif i > second and i != first:
            second = i
    return second if second != float('-inf') else None
print(second_largest([4,1,7,7,3]))# -> 4 ; 
print(second_largest([5,5]))# -> None
```
