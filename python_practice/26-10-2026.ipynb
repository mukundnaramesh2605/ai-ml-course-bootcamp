```python
# 1. Reverse Integer  [Service / screening]
# Return the integer with its digits reversed. Preserve the sign.
# 1234 -> 4321 ; -520 -> -25
def reverse_int(n):
    sign = -1 if n < 0 else 1
    reversed_num = int(str(abs(n))[::-1])
    return sign * reversed_num
    
print(reverse_int(1234))# -> 4321 ; 
print(reverse_int(-520))# -> -25 ; 
print(reverse_int(0))# -> 0
```

```python
# 2. Palindrome Number  [Service / screening]
# Return True if the integer reads the same forwards and backwards.
# 121 -> True ; 123 -> False ; negatives are never palindromes
def is_palindrome_num(n):
    return str(n) == str(n)[::-1]
print(is_palindrome_num(121))# -> True ; 
print(is_palindrome_num(123))# -> False ; 
print(is_palindrome_num(-121))# -> False
```

```python
# 3. Prime Check  [Service / screening]
# Return True if n is prime (n >= 2). 1 and below are not prime.
# 7 -> True ; 9 -> False
def is_prime(n):
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True
print(is_prime(7))# -> True ; 
print(is_prime(9))# -> False ; 
print(is_prime(1))# -> False ; 
print(is_prime(2))# -> True

```

```python
# 4. Fibonacci Series  [Service / screening]
# Return a list of the first n Fibonacci numbers starting 0, 1.
# 5 -> [0,1,1,2,3]
def fib(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]
    res = [0, 1]
    while len(res) < n:
        res.append(res[-1] + res[-2])
    return res
    
print(fib(5))# -> [0,1,1,2,3] ; 
print(fib(1))# -> [0] ; 
print(fib(0))# -> []
```

```python
# 5. Armstrong Number  [Service / screening]
# A 3-digit number equal to the sum of the cubes of its digits.
# Return True/False. 153 -> True (1+125+27) ; 154 -> False
def is_armstrong(n):
    return n == sum([int(i)**3 for i in str(n)])
    
print(is_armstrong(153))# -> True ; 
print(is_armstrong(154))# -> False ; 
print(is_armstrong(370))# -> True

```

```python
# 6. GCD  [Service / screening]
# Return the greatest common divisor of a and b (Euclid's algorithm).
# 12, 18 -> 6
def gcd(a, b):
    while b != 0:
        a,b = b, a % b
    return a
print(gcd(12,18))# -> 6 ; 
print(gcd(7,13))# -> 1 ; 
print(gcd(0,5))# -> 5
```

```python
# 7. Count Vowels and Consonants  [Service / screening]
# Return a tuple (vowels, consonants). Count letters only, case-insensitive.
# "Hello World" -> (3, 7)
def count_vc(s):
    vowels = sum([1 for i in s if i.lower() in 'aeiou' and i.isalpha()])
    cons = sum([1 for i in s if i.lower() not in 'aeiou' and i.isalpha()])
    return (vowels,cons)
print(count_vc("Hello World"))# -> (3, 7) ; 
print(count_vc("123"))# -> (0, 0)
```

```python
# 8. Anagram Check  [Service / screening]
# Return True if a and b are anagrams, case-insensitive.
# "Listen", "Silent" -> True
def is_anagram(a, b):
    return sorted(a.lower()) == sorted(b.lower())
    
print(is_anagram("Listen","Silent"))# -> True ; 
print(is_anagram("abc","abd"))# -> False
```

```python
# 9. Factorial (iterative, no library)  [Service / screening]
# Return n! using a loop. 0! is 1.
# 5 -> 120
def factorial(n):
    if n == 0:
        return 1
    count=1
    for i in range(1,n+1):
        count*=i
    return count
print(factorial(5))# -> 120 ; 
print(factorial(0))# -> 1
```

```python
# 10. Second Largest  [Service / screening]
# Return the second largest DISTINCT value, or None if there isn't one.
# [4,1,7,7,3] -> 4
def second_largest(nums):
    first = second = float('-inf')
    for i in set(nums):
        if i > first:
            second = first
            first = i
        elif i > second and i != first:
            second = i
    return second if second != float('-inf') else None
print(second_largest([4,1,7,7,3]))# -> 4 ; 
print(second_largest([5,5]))# -> None
```

```python
# 1. Roman to Integer  [AI screen / string]
# Convert a Roman numeral to an integer. Values: I=1,V=5,X=10,L=50,C=100,D=500,M=1000.
# If a smaller value sits before a larger one, subtract it (IV=4, IX=9).
# "LVIII" -> 58 ; "MCMXCIV" -> 1994
# Idea: walk left to right; if the current value is less than the NEXT one, subtract it, else add it.
def roman_to_int(s):
    romans = {'I':1,'V':5,'X':10,'L':50,'C':100,'D':500,'M':1000}
    int_val = 0
    for i in range(len(s)):
        if i+1 < len(s) and romans[s[i]] < romans[s[i+1]]:
            int_val-=romans[s[i]]
        else:
            int_val+=romans[s[i]]
    return int_val
print(roman_to_int("LVIII"))# -> 58 ; 
print(roman_to_int("MCMXCIV"))# -> 1994 ; 
print(roman_to_int("III"))# -> 3
```

```python
# 2. Best Time to Buy and Sell Stock  [AI screen / logic]
# prices[i] is the price on day i. Return the max profit from one buy then
# one later sell. If no profit is possible, return 0.
# [7,1,5,3,6,4] -> 5   (buy at 1, sell at 6)
# Idea: track the lowest price seen so far; at each day, profit = price - min_so_far, keep the best.
def max_profit(prices):
    min_price = float('inf')
    max_p = 0
    for i in prices:
        if i < min_price:
            min_price = i
        elif i-min_price > max_p:
            max_p = i - min_price
    return max_p
print(max_profit([7,1,5,3,6,4]))# -> 5 ; 
print(max_profit([7,6,4,3,1]))# -> 0 ; 
print(max_profit([]))# -> 0
```

```python
# 3. Single Number  [AI screen / logic]
# Every element appears twice except one. Return the one that appears once.
# [4,1,2,1,2] -> 4
# Idea: XOR everything (a ^ a == 0), or count and return the odd one out.
def single_number(nums):
    for i in nums:
        if nums.count(i) == 1:
            return i
    return None
print(single_number([4,1,2,1,2]))# -> 4 ; 
print(single_number([1]))# -> 1
```

```python
# 4. First Unique Character Index  [AI screen / string]
# Return the INDEX of the first non-repeating character, or -1 if none.
# "leetcode" -> 0 ; "aabb" -> -1
# Idea: count all chars, then scan left to right for the first with count 1.
def first_uniq_index(s):
    for idx,char in enumerate(s):
        if s.count(char) == 1:
            return idx
    return -1
print(first_uniq_index("leetcode"))# -> 0 ; 
print(first_uniq_index("loveleetcode"))# -> 2 ; 
print(first_uniq_index("aabb"))# -> -1
```

```python
# 5. Plus One  [AI screen / logic]
# digits is a number as a list of digits. Add 1 and return the new digit list.
# [1,2,3] -> [1,2,4] ; [1,9,9] -> [2,0,0] ; [9,9] -> [1,0,0]
# Idea: go right to left adding the carry; if a digit was 9 it becomes 0 and carries.
def plus_one(digits):
    num = str(int("".join([str(x) for x in digits]))+1)
    return [int(x) for x in num]
print(plus_one([1,2,3]))# -> [1,2,4] ; 
print(plus_one([1,9,9]))# -> [2,0,0] ; 
print(plus_one([9,9]))# -> [1,0,0]
```

```python
# 1. Climbing Stairs  [AI screen / logic]
# You climb 1 or 2 steps at a time. Return how many distinct ways to reach
# the top of n steps. n=3 -> 3  (1+1+1, 1+2, 2+1)
# Idea: ways(n) = ways(n-1) + ways(n-2). It's Fibonacci. Build it up iteratively.
def climb(n):
    if n <= 1:
        return 1
    a, b = 1, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b
print(climb(3))# -> 3 ; 
print(climb(5))# -> 8 ; 
print(climb(1))# -> 1 ; climb(0) -> 1
```
