```python
# 1. Count Vowels
# Write a function that counts how many vowels (a, e, i, o, u) are in a string.
# Example: "machine learning" -> 5

def count_vowels(s):
    count = sum([1 for i in list(s) if i.lower() in 'aeiou'])
    return count
count_vowels('Machine Learning')
```

```python
# 2. Sum of a List
# Write a function that returns the sum of all numbers in a list without using sum().
# Example: [3, 7, 2, 8] -> 20

def sum_list(nums):
    sum_val = 0
    if not nums:
        return sum_val

    for i in nums:
        if (isinstance(i,int) or isinstance(i,float)):
            sum_val+=i

    return sum_val

display(sum_list([3,7,2,8]))
display(sum_list([]))
```
