```python
# 12. Fill Missing With Group Mean  [AI screen / pandas]
# Fill missing scores with the mean score of that student's class.
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "class": ["X", "X", "Y", "Y", "X", "Y"],
    "score": [80, np.nan, 70, 90, 60, np.nan],
})
df['score']=df['score'].fillna(df.groupby('class')['score'].transform('mean'))
df
```

```python
# 1. Group Consecutive Numbers  [AI screen / logic]
# Split a sorted list into groups of consecutive runs.
# Example: [1, 2, 3, 5, 6, 8] -> [[1, 2, 3], [5, 6], [8]]
nums = [1, 2, 3, 5, 6, 8, 9, 10]
result = []
current_group = [nums[0]]
    
for i in range(1, len(nums)):
    if nums[i] == nums[i - 1] + 1:
        current_group.append(nums[i])
    else:
        result.append(current_group)
        current_group = [nums[i]]
        
result.append(current_group)
print(result)
```

```python
# 2. Count Word Frequency and Return Top 2  [AI screen / text]
# Return the two most frequent words with their counts.
# Example: "a b a c a b" -> [('a', 3), ('b', 2)]
from collections import Counter 
text = "the sun the moon the sun the star"
counts = Counter(text.lower().split())
sorted_dict = dict(sorted(counts.items(),key=lambda x : x[1],reverse=True))
[(k,sorted_dict[k]) for k in list(sorted_dict)[:2]]
```

```python
# 3. Check Balanced Brackets  [Service / logic]
# Return True if every opening bracket has a matching closing one, in order.
# Example: "(a[b]c)" -> True, "(a[b)c]" -> False
expr = "(a[b]{c})"
dict_brack = {")":"(","}":"{","]":"["}
stack = []

def brackets(expr):
    for char in expr:
        if char in dict_brack.values():
            stack.append(char)
        elif char in dict_brack:

            if not stack or stack.pop() != dict_brack[char]:
                return False
    return len(stack)==0

print(brackets(expr))
print(brackets(']a]'))

```

```python
# 4. Flatten a Nested List (any depth)  [AI screen / logic]
# Flatten a list that may contain lists inside lists, to any depth.
# Example: [1, [2, [3, 4], 5], 6] -> [1, 2, 3, 4, 5, 6]
data = [1, [2, [3, 4], 5], [6, [7, [8]]]]
def flatten_list(nested_list):
    result = []
    for item in nested_list:
        # Check if the item is a list or tuple
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result
flatten_list(data)
```

```python
# 5. Find the Missing and Duplicate  [AI screen / logic]
# A list should hold 1 to n once each. One number is missing, one is duplicated.
# Return (duplicate, missing).
# Example: [1, 2, 2, 4], n=4 -> (2, 3)
arr = [1, 2, 3, 3, 5, 6]
n = 6
for i in range(1,n+1):
    if arr[i-1] != i:
        print((arr[i-1],i))
```
