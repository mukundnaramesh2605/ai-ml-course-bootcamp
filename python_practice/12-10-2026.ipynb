```python
# 3. Check Balanced Brackets  [Service / logic]
# Return True if every opening bracket has a matching closing one, in order.
# Example: "(a[b]c)" -> True, "(a[b)c]" -> False
expr = "(a[b]{c})"
dict_brack = {")":"(","}":"{","]":"["}
stack = []

def brackets(expr):
    for char in expr:
        if char in dict_brack.values():
            stack.append(char)
        elif char in dict_brack:

            if not stack or stack.pop() != dict_brack[char]:
                return False
    return len(stack)==0

print(brackets(expr))
print(brackets(']a]'))

```

```python
# 4. Flatten a Nested List (any depth)  [AI screen / logic]
# Flatten a list that may contain lists inside lists, to any depth.
# Example: [1, [2, [3, 4], 5], 6] -> [1, 2, 3, 4, 5, 6]
data = [1, [2, [3, 4], 5], [6, [7, [8]]]]
def flatten_list(nested_list):
    result = []
    for item in nested_list:
        # Check if the item is a list or tuple
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result
flatten_list(data)
```

```python
# Can You Attend All Meetings?  [AI screen / logic]
# Given meeting time intervals, return True if a person can attend all of them,
# meaning none of them overlap.
# Example: [[0, 30], [35, 50], [60, 75]] -> True   (no clashes)
#          [[0, 30], [25, 50], [60, 75]] -> False  (first two overlap)
meetings = [[9, 10], [11, 12], [10, 11], [13, 15], [14, 16]]
# expected for this input: False  (13-15 and 14-16 overlap)
can_attend = True
sorted_meetings = sorted(meetings,key=lambda x : x[0])
for i in range(1,len(sorted_meetings)):
    prev_end = sorted_meetings[i-1][1]
    current_start = sorted_meetings[i][0]
    if current_start<prev_end:
        can_attend = False
        break
print(can_attend)
```

```python
# Minimum Meeting Rooms  [AI screen / logic]
# Given meeting intervals, return how many rooms you need so that no two
# overlapping meetings share a room.
# Example: [[0, 30], [5, 10], [15, 20]] -> 2
#          (0-30 overlaps both 5-10 and 15-20, so you need a 2nd room)
meetings = [[1, 10], [2, 7], [3, 19], [8, 12], [10, 20], [11, 30]]

starts = sorted(m[0] for m in meetings)   # all start times
ends = sorted(m[1] for m in meetings)     # all end times

rooms = 0
max_rooms = 0
s = e = 0                                  # pointers into starts and ends

while s < len(starts):
    if starts[s] < ends[e]:                # a meeting starts before the next one ends
        rooms += 1                         # take a room
        s += 1
        max_rooms = max(max_rooms, rooms)
    else:                                   # a meeting ends first
        rooms -= 1                          # free a room
        e += 1

print(max_rooms)                            # 4
```

```python
# Find Free Time  [AI screen / logic]
# Given busy meeting intervals in a day, return the gaps when the person is free.
# Assume the day runs from the end of one meeting to the start of the next.
# Example: [[1, 3], [5, 8], [10, 12]] -> [[3, 5], [8, 10]]
#          (free between 3 and 5, and between 8 and 10)
meetings = [[9, 11], [13, 14], [16, 18], [12, 13]]
# expected for this input: [[11, 12], [14, 16]]
meetings.sort(key=lambda x: x[0])
free_time = []
for i in range(1,len(meetings)):
    prev_end_time = meetings[i-1][1]
    current_start = meetings[i][0]
    if current_start > prev_end_time:
        free_time.append([prev_end_time,current_start])
print(free_time)
```

```python
# Total Busy Time  [AI screen / logic]
# Given meeting intervals that may overlap, return the total amount of time
# the person is busy. Overlapping time counts once, not twice.
# Example: [[1, 4], [2, 5], [7, 9]] -> 5
#          (1 to 5 is one merged busy block = 4, plus 7 to 9 = 2, total 6)
# wait, recheck: 1-5 is 4 units, 7-9 is 2 units, total = 6
meetings = [[1, 3], [2, 6], [8, 10], [15, 18]]
# expected for this input: 9  (1-6 = 5 units, 8-10 = 2, 15-18 = 3, total 10)
meetings.sort(key=lambda x: x[0])
merged=[meetings[0]]
for start,end in meetings[1:]:
    last = merged[-1]
    if start <= last[1]:
        last[1] = max(last[1],end)
    else:
        merged.append([start,end])
print(sum([abs(x[0]-x[1]) for x in merged]))
```

```python
# Insert and Merge  [AI screen / logic]
# You have a sorted, non-overlapping list of intervals. Insert a new interval
# and merge if it overlaps anything.
# Example: existing [[1, 3], [6, 9]], new [2, 5] -> [[1, 5], [6, 9]]
#          (the new [2,5] overlaps [1,3], merges into [1,5])
intervals = [[1, 2], [3, 5], [6, 7], [8, 10], [12, 16]]
new = [4, 9]
# expected: [[1, 2], [3, 10], [12, 16]]
#          ([4,9] overlaps [3,5], [6,7], [8,10], all merge into [3,10])
intervals.append(new)
intervals.sort(key=lambda x: x[0])

merged = [intervals[0]]

for s,e in intervals[1:]:
    last = merged[-1]
    if s<=last[1]:
        last[1]=max(last[1],e)
    else:
        merged.append([s,e])
print(merged)
```

```python
# Q2. Reverse Only Letters  [Service / string]
# Reverse the letters in a string but keep every non-letter in its place.
# Example: "a1b2c" -> "c1b2a"  (letters a,b,c reversed; digits stay put)
def reverse_letters(s):
    only_letter = [char for char in s if char.isalpha()]
    return "".join([only_letter.pop() if char.isalpha() else char for char in s])
        
    

# test: reverse_letters("a1b2c") -> "c1b2a"
reverse_letters("a1b2c")
```

```python
# Q3. Group Scores by Pass/Fail  [AI screen / data handling]
# Given a dict of name -> score, return a dict with two lists: names who
# passed (score >= 40) and names who failed. Keep input order.
# Example: {"A": 80, "B": 30, "C": 55} -> {"pass": ["A","C"], "fail": ["B"]}
def group_by_result(scores):
    grouped = {'pass':[],'fail':[]}
    for name in scores:
        if scores[name]>=40:
            grouped['pass'].append(name)
        else:
            grouped['fail'].append(name)
    return grouped

# test: group_by_result({"A": 80, "B": 30, "C": 55}) -> {"pass": ["A","C"], "fail": ["B"]}
group_by_result({"A": 80, "B": 30, "C": 55})
```
