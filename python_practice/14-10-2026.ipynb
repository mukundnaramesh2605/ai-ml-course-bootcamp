```python
# 1. Count Words Longer Than N  [Service / string]
# Return how many words in a sentence are longer than n characters.
# Example: ("the quick brown fox", 3) -> 2  ("quick","brown")
def count_long_words(sentence, n):
    return sum(1 for x in sentence.split() if len(x)>n)
count_long_words("the quick brown fox", 3)
```

```python
# 2. Second Smallest  [Service / logic]
# Return the second smallest DISTINCT value, or None if there isn't one.
# Example: [4, 2, 2, 7, 1] -> 2
def second_smallest(nums):
    first= second=float('inf')
    for i in nums:
        if i < first:
            second = first
            first = i
        elif i> first and i < second:
            second = i
    return second if second!= float('inf') else None

second_smallest([4, 2, 2, 7, 1])
```

```python
# 3. Title Case Without .title()  [Service / string]
# Capitalize the first letter of each word.
# Example: "hello world" -> "Hello World"
def title_case(s):
    return " ".join([i[0].upper()+i[1:] for i in s.split()])
title_case('hello world')
```

```python
# 4. Sum of Even-Indexed Elements  [Service / logic]
# Return the sum of values at index 0, 2, 4, ...
# Example: [10, 20, 30, 40, 50] -> 90
def sum_even_index(nums):
    return sum([nums[x] for x in range(0,len(nums),2)])

sum_even_index([10, 20, 30, 40, 50])
```

```python
# 5. Count Character Types  [Service / string]
# Return a dict: how many letters, digits, and spaces in the string.
# Example: "ab 12 c" -> {"letters": 3, "digits": 2, "spaces": 2}
from collections import Counter
def char_types(s):
    return Counter(["letter" if i.isalpha() else "digits" if i.isalnum() else "spaces" for i in s])
char_types("ab 12 c")
```

```python
# 6. Find the Odd One Out  [AI screen / logic]
# Every value appears twice except one. Return the one that appears once.
# Example: [2, 3, 2, 4, 4] -> 3
from collections import Counter
def find_single(nums):
    counts = Counter(nums)
    return [x for x,v in counts.items() if v==1][0]
find_single([2, 3, 2, 4, 4])
```

```python
# 7. Group by Even / Odd  [AI screen / logic]
# Return a dict {"even": [...], "odd": [...]}, order preserved.
# Example: [1, 2, 3, 4] -> {"even": [2, 4], "odd": [1, 3]}
from collections import defaultdict
def split_even_odd(nums):
    grouped = defaultdict(list)
    for i in nums:
       grouped['even'].append(i) if i%2 == 0 else grouped['odd'].append(i)
            
    return dict(grouped)
split_even_odd([1,2,3,4,5,6])
            
```

```python
# 8. Most Frequent Word  [AI screen / text]
# Return the most frequent word, case-insensitive. On a tie, the first seen.
# Example: "the cat the dog" -> "the"
from collections import Counter
def most_frequent_word(text):
    counts = Counter(text.split())
    return [x for x,v in counts.items() if v==max(counts.values())][0]
most_frequent_word("dog the cat the dog")
```

```python
# 9. Running Difference  [AI screen / logic]
# Return a list of differences between each element and the previous one.
# First element has no previous, so start from the second.
# Example: [10, 7, 12, 5] -> [-3, 5, -7]
def running_diff(nums):
    return [nums[i]-nums[i-1] for i in range(1,len(nums))]
running_diff([10, 7, 12, 5])
```

```python
# 10. Merge Two Sorted Lists  [AI screen / logic]
# Merge two already-sorted lists into one sorted list, WITHOUT sorted().
# Example: [1, 3, 5], [2, 4, 6] -> [1, 2, 3, 4, 5, 6]
from heapq import merge
def merge_sorted(a, b):
    return list(merge(a,b))
merge_sorted([1, 3, 5], [2, 4, 6])
```

```python
# 1. Count Digits in a String  [Service / string]  (easy)
# Return how many characters in the string are digits.
# Example: "a1b2c3" -> 3
import re
def count_digits(s):
    digits = re.findall(r'[0-9]',s)
    return len(digits)
count_digits("a1b2c3")
    
```

```python
# 4. Group Anagrams  [AI screen / text]  (medium)
# Group words that are anagrams of each other. Return a list of groups
# (each group a list of words). Order within/between groups doesn't matter.
# Example: ["eat","tea","tan","ate","nat"] -> [["eat","tea","ate"],["tan","nat"]]
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for i in words:
        grouped["".join(sorted(i))].append(i)
    values = list(grouped.values())
    return values
group_anagrams(["eat","tea","tan","ate","nat"])
```
