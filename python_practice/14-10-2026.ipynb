```python
# 1. Count Words Longer Than N  [Service / string]
# Return how many words in a sentence are longer than n characters.
# Example: ("the quick brown fox", 3) -> 2  ("quick","brown")
def count_long_words(sentence, n):
    return sum(1 for x in sentence.split() if len(x)>n)
count_long_words("the quick brown fox", 3)
```

```python
# 2. Second Smallest  [Service / logic]
# Return the second smallest DISTINCT value, or None if there isn't one.
# Example: [4, 2, 2, 7, 1] -> 2
def second_smallest(nums):
    first= second=float('inf')
    for i in nums:
        if i < first:
            second = first
            first = i
        elif i> first and i < second:
            second = i
    return second if second!= float('inf') else None

second_smallest([4, 2, 2, 7, 1])
```

```python
# 5. Count Character Types  [Service / string]
# Return a dict: how many letters, digits, and spaces in the string.
# Example: "ab 12 c" -> {"letters": 3, "digits": 2, "spaces": 2}
from collections import Counter
def char_types(s):
    return Counter(["letter" if i.isalpha() else "digits" if i.isalnum() else "spaces" for i in s])
char_types("ab 12 c")
```

```python
# 6. Find the Odd One Out  [AI screen / logic]
# Every value appears twice except one. Return the one that appears once.
# Example: [2, 3, 2, 4, 4] -> 3
from collections import Counter
def find_single(nums):
    counts = Counter(nums)
    return [x for x,v in counts.items() if v==1][0]
find_single([2, 3, 2, 4, 4])
```

```python
# 4. Group Anagrams  [AI screen / text]  (medium)
# Group words that are anagrams of each other. Return a list of groups
# (each group a list of words). Order within/between groups doesn't matter.
# Example: ["eat","tea","tan","ate","nat"] -> [["eat","tea","ate"],["tan","nat"]]
from collections import defaultdict
def group_anagrams(words):
    grouped = defaultdict(list)
    for i in words:
        grouped["".join(sorted(i))].append(i)
    values = list(grouped.values())
    return values
group_anagrams(["eat","tea","tan","ate","nat"])
```

```python
# 5. Longest Substring Without Repeating Characters  [AI screen / logic]  (hard)
# Return the LENGTH of the longest substring with no repeating characters.
# Example: "abcabcbb" -> 3  ("abc")  ;  "bbbb" -> 1  ;  "pwwkew" -> 3  ("wke")
def longest_unique_substring(s):
    best = 1
    current = 1
    for i in range(len(s)-1):
        if s[i]!=s[i+1]:
            current+=1
            best = max(best,current)
        else:
            current=1
    return best
longest_unique_substring('abcabcbb')
```

```python
# Q2. Compress a String  [Service / string]
# Collapse runs of the same char into char+count. Keep order.
# Example: "aaabbc" -> "a3b2c1"
from collections import Counter
def compress(s):
    if not s:
        return ""
        
    counts = Counter(s)
    return "".join([str(k)+str(v) for k,v in counts.items()])

print(compress("aaabbc"))# -> "a3b2c1"
print(compress("")) # -> ""
print(compress("abc")) # -> "a1b1c1"
```

```python
# Q3. Merge Two Dicts, Summing Shared Keys  [AI screen / data handling]
# Combine two dicts. If a key is in both, add the values. Keep all keys.
# Must work with zero and negative values.
# Example: {"a":2,"b":3}, {"b":4,"c":1} -> {"a":2,"b":7,"c":1}
def merge_sum(d1, d2):
    dict_merged = d1.copy()
    for i in d2:
        dict_merged[i] = dict_merged.get(i,0)+d2[i]

    return dict_merged
    
print(merge_sum({"a":2,"b":3}, {"b":4,"c":1})) # -> {"a":2,"b":7,"c":1}
print(merge_sum({"a":5}, {"a":-5})) #-> {"a":0}
```

```python
# Q3. Valid Parentheses  [AI screen / logic]
# Return True if every bracket is closed correctly, in order.
# Brackets: ( ) [ ] { }
# Example: "(a[b]{c})" -> True ;  "(a[b)c]" -> False
def is_valid(s):
    brackets = {"}":"{","]":"[",")":"("}
    stack=[]
    for i in s:
        if i in brackets.values():
            stack.append(i)
        elif i in brackets:
            if not stack or stack[-1] != brackets[i]:
                return False
            stack.pop()
    return len(stack) == 0
        
print(is_valid("(a[b]{c})")) # -> True
# test: is_valid("(]") -> False
# test: is_valid("(") -> False
# test: is_valid("") -> True
```

```python
# Q4. Group by Length  [AI screen / data handling]
# Given words, return a dict mapping each length to the list of words of
# that length, preserving input order.
# Example: ["a","bb","cc","d"] -> {1: ["a","d"], 2: ["bb","cc"]}
from collections import defaultdict
def group_by_length(words):
    grouped = defaultdict(list)
    for i in words:
        grouped[len(i)].append(i)
    return dict(grouped)
print(group_by_length(["a","bb","cc","d"]))# -> {1:["a","d"], 2:["bb","cc"]}
print(group_by_length([]))# -> {}
```

```python
# Minimum Add to Make Parentheses Valid  [AI screen / logic]
# Given a string of only "(" and ")", return the MINIMUM number of
# brackets you must add to make it valid (every "(" closed, every ")" opened).
# Example: "())"  -> 1   (add one "(" at the start)
#          "(((" -> 3   (need three ")")
#          "()"  -> 0
#          "))((" -> 4
def min_add_to_make_valid(s):
    open_count = 0
    add = 0
    for ch in s:
        if ch == "(":
            open_count+=1         # blank 1: a waiter appears
        else:                        # ch == ")"
            if open_count>0:       # blank 2: is there a waiter to close?
                open_count-=1       # blank 3: close one
            else:
                add += 1             # stranded, must add
    return add+open_count            # blank 4: strays already counted, plus leftover waiters

print(min_add_to_make_valid("())"))#  -> 1
# test: min_add_to_make_valid("(((")  -> 3
# test: min_add_to_make_valid("()")   -> 0
# test: min_add_to_make_valid("))((") -> 4
```

```python
import time

def time_calculator(func):
    def wrapper(*args,**kwargs):
        start_time = time.perf_counter()
        result = func(*args,**kwargs)
        end_time = time.perf_counter()
        execution_time = end_time - start_time
        print(f"Function {func.__name__} excuted in {execution_time:.6f} seconds")
        return result
    return wrapper

@time_calculator
def compute_heavy_math(n):
    return sum(i * i for i in range(n))

# Run the function
total = compute_heavy_math(10_000_000)
print(total)
```

```python
# 2. Reverse Vowels Only  [Service / string]
# Reverse only the vowels in the string, leave other chars in place.
# Example: "hello" -> "holle"  (e and o swap)
def reverse_vowels(s):
    vowels = [i for i in s if i.lower() in 'aeiou']
    return "".join([i if i.lower() not in 'aeiou' else vowels.pop() for i in s])
print(reverse_vowels("hello"))# -> "holle" 
reverse_vowels("xyz")# -> "xyz"

```

```python
# 3. Count Streaks  [Service / logic]   (consecutive)
# Return how many groups of consecutive equal values there are.
# Example: [1,1,2,3,3,3,1] -> 4
def count_streaks(nums):
    if not nums:
        return 0
    count=1
    for i in range(1,len(nums)):
        if nums[i] != nums[i-1]:
            count+=1
    return count
        
print(count_streaks([1,1,2,3,3,3,1]))# -> 4 ; 
print(count_streaks([5,5,5])) #-> 1 ; 
count_streaks([])# -> 0

```

```python
# 2. Safe Parse With else + finally  [Service / exceptions]
# Convert s to int. On failure return None. Use else to return the value on
# success, and finally to print "parse attempted" every time.
def safe_parse(s):
    try:
        value = int(s)
        return value
    except ValueError:
        return None
    finally:
        print("parse attempted")
# safe_parse("42") -> prints "parse attempted", returns 42
# safe_parse("ab") -> prints "parse attempted", returns None
print(safe_parse("42"))
print(safe_parse("ab"))
```
