```python
# Q1. Second Largest Unique Number
# Return the second largest UNIQUE number in the list.
# Return None if it does not exist.

def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in nums:
        if num > first:
            sceond ,first = first,num
        elif num > second and num != first:
            second = num

    return second if second != neg_inf else None

print(second_largest([10, 20, 4, 45, 99]))   # Expected: 45
print(second_largest([5, 5, 5]))             # Expected: None
print(second_largest([-3, -1, -7]))          # Expected: -3
print(second_largest([7]))                   # Expected: None
print(second_largest([]))                    # Expected: None
```

```python
# Q2. Anagram Check
# Return True if s1 and s2 are anagrams.
# Ignore case and spaces.

def is_anagram(s1, s2):
    s1 = sorted([i for i in s1.lower() if i !=" "])
    s2 = sorted([i for i in s2.lower() if i !=" "])
    return s1==s2
print(is_anagram("Listen", "Silent"))          # Expected: True
print(is_anagram("Dormitory", "Dirty room"))   # Expected: True
print(is_anagram("hello", "world"))            # Expected: False
print(is_anagram("aab", "abb"))                # Expected: False
print(is_anagram("", ""))                      # Expected: True
```

```python
# Q3. First Non-Repeating Character
# Return the first character that appears exactly once (case-sensitive).
# Return None if there is no such character.

def first_non_repeating_char(s):
    for char in s:
        if s.count(char) == 1 : return char 
    return None
print(first_non_repeating_char("swiss"))    # Expected: w
print(first_non_repeating_char("aabbcc"))   # Expected: None
print(first_non_repeating_char("a"))        # Expected: a
print(first_non_repeating_char("Aa"))       # Expected: A
print(first_non_repeating_char(""))         # Expected: None
```

```python
# Q4. Flatten a Nested List (any depth)
# Return a single flat list. Do not use any library.

def flatten(lst):
    result = []
    for i in lst:
        if isinstance(i,list):
            result.extend(flatten(i))
        else:
            result.append(i)
    return result

print(flatten([1, [2, [3, [4]]], 5]))   # Expected: [1, 2, 3, 4, 5]
print(flatten([1, 2, 3]))               # Expected: [1, 2, 3]
print(flatten([[], [[]]]))              # Expected: []
print(flatten([[1, 'a'], [None]]))      # Expected: [1, 'a', None]
print(flatten([]))                      # Expected: []
```

```python
# Q5. Top K Frequent Words (NLP style)
# Lowercase the text, remove punctuation, count words.
# Return a list of (word, count) sorted by count desc, ties broken alphabetically.
# If k <= 0 or text is empty, return [].
import re
def top_k_words(text, k):
    result = []
    if k <=0 or not text:
        return result
    only_text = re.findall(r'[a-z]+',text.lower())
    sorted_text = sorted(set((word,only_text.count(word)) for word in only_text),key=lambda x: [-x[1],x[0]])
    return sorted_text[:k]

print(top_k_words("The cat and the hat. The END!", 2))   # Expected: [('the', 3), ('and', 1)]
print(top_k_words("Hello hello HELLO", 1))               # Expected: [('hello', 3)]
print(top_k_words("a b c", 5))                           # Expected: [('a', 1), ('b', 1), ('c', 1)]
print(top_k_words("Data, data; science!", 0))            # Expected: []
print(top_k_words("", 3))                                # Expected: []
```

```python
# N1. Min-Max Normalization
# Scale values to the range [0, 1].
# If all values are equal, return an array of zeros. Empty input returns empty array.
# Use vectorized NumPy only (no loops).

def min_max_normalize(arr):
    if arr.size == 0:
        return np.array([])
    if arr.min() == arr.max():
        return np.zeros_like(arr, dtype=float)
        
    return (arr - arr.min()) / (arr.max() - arr.min()) if arr.any() else []

print(min_max_normalize(np.array([1, 2, 3, 4, 5])))   # Expected: [0.   0.25 0.5  0.75 1.  ]
print(min_max_normalize(np.array([-10, 0, 10])))      # Expected: [0.  0.5 1. ]
print(min_max_normalize(np.array([5, 5, 5])))         # Expected: [0. 0. 0.]
print(min_max_normalize(np.array([42])))              # Expected: [0.]
print(min_max_normalize(np.array([])))                # Expected: []
```

```python
# N2. Fill NaN with Column Mean
# In a 2D array, replace each NaN with the mean of its column (ignoring NaNs).
# If an entire column is NaN, fill it with 0.

def fill_nan_with_col_mean(X):
    col_mean = np.nanmean(X,axis=0)
    col_mean = np.nan_to_num(col_mean, nan=0.0)
    nan_indices = np.isnan(X)
    X[nan_indices] = np.take(col_mean, np.where(nan_indices)[1])
    return X

print(fill_nan_with_col_mean(np.array([[1, np.nan], [3, 4]])))
# Expected: [[1. 4.] [3. 4.]]
print(fill_nan_with_col_mean(np.array([[1, 2, np.nan], [np.nan, 5, 6], [7, np.nan, 9]])))
# Expected: [[1. 2. 7.5] [4. 5. 6.] [7. 3.5 9.]]
print(fill_nan_with_col_mean(np.array([[np.nan, 2], [np.nan, 4]])))
# Expected: [[0. 2.] [0. 4.]]
print(fill_nan_with_col_mean(np.array([[1.0, 2.0], [3.0, 4.0]])))
# Expected: [[1. 2.] [3. 4.]]
print(fill_nan_with_col_mean(np.array([[np.nan]])))
# Expected: [[0.]]
```

```python
# N3. One-Hot Encoding
# Convert integer labels to a one-hot 2D array.
# If num_classes is None, use max(label) + 1.
# Empty labels should return shape (0, num_classes).

def one_hot(labels, num_classes=None):
    labels = np.asarray(labels, dtype=int)

    # Handle empty input
    if num_classes is None:
        if labels.size == 0:
            num_classes = 0
        else:
            num_classes = labels.max() + 1

    # Create a zero matrix
    result = np.zeros((len(labels), num_classes), dtype=int)

    # Set the corresponding class positions to 1
    result[np.arange(len(labels)), labels] = 1

    return result

print(one_hot(np.array([0, 2, 1])))
# Expected: [[1 0 0] [0 0 1] [0 1 0]]
print(one_hot(np.array([1, 1, 1]), num_classes=3))
# Expected: [[0 1 0] [0 1 0] [0 1 0]]
print(one_hot(np.array([0, 1]), num_classes=4))
# Expected: [[1 0 0 0] [0 1 0 0]]
print(one_hot(np.array([3])))
# Expected: [[0 0 0 1]]
print(one_hot(np.array([], dtype=int), num_classes=2).shape)
# Expected: (0, 2)
```

```python
# N4. Cosine Similarity
# Return cosine similarity between two vectors, rounded to 4 decimals.
# If either vector is all zeros, return 0.0.

def cosine_similarity(a, b):
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    if norm_a == 0 or norm_b == 0:
        return 0.0
    similarity = np.dot(a, b) / (norm_a * norm_b)

    return round(float(similarity), 4)
print(cosine_similarity(np.array([1, 2, 3]), np.array([2, 4, 6])))   # Expected: 1.0
print(cosine_similarity(np.array([1, 0]), np.array([0, 1])))         # Expected: 0.0
print(cosine_similarity(np.array([1, 1]), np.array([-1, -1])))       # Expected: -1.0
print(cosine_similarity(np.array([3, 4]), np.array([4, 3])))         # Expected: 0.96
print(cosine_similarity(np.array([0, 0]), np.array([1, 2])))         # Expected: 0.0
```

```python
# N5. Moving Average
# Return the moving average with the given window size (only full windows).
# If window <= 0 or window > len(arr), return an empty array.

def moving_average(arr, window):
    if window <= 0 or window > len(arr):
        return np.array([], dtype=float)

    # Calculate cumulative sum
    cumulative_sum = np.cumsum(arr, dtype=float)

    # Calculate window sums
    cumulative_sum[window:] = (
        cumulative_sum[window:] - cumulative_sum[:-window]
    )

    # Calculate averages
    return cumulative_sum[window - 1:] / window

print(moving_average(np.array([1, 2, 3, 4, 5]), 3))          # Expected: [2. 3. 4.]
print(moving_average(np.array([10, 20]), 1))                 # Expected: [10. 20.]
print(moving_average(np.array([4, 8, 15, 16, 23, 42]), 6))   # Expected: [18.]
print(moving_average(np.array([1, 2]), 3))                   # Expected: []
print(moving_average(np.array([5, 5, 5]), 0))                # Expected: []
```

```python
# D1. Fill Missing Values
# Numeric columns: fill NaN with median.
# Text (object) columns: fill NaN with mode (first value of mode).
# If a column is entirely NaN, leave it as is. Return a new DataFrame.

def fill_missing(df):
    result = df.copy()

    for col in result.columns:
        if not result[col].isna().any():
            continue
    
        if result[col].isna().all():
            continue
    
        if pd.api.types.is_numeric_dtype(result[col]):
            result[col] = result[col].fillna(result[col].median())
    
        elif result[col].dtype == 'object':
            mode_value = result[col].mode().iloc[0]
            result[col] = result[col].fillna(mode_value)

    return result

df1 = pd.DataFrame({'age': [25, np.nan, 35, 40], 'city': ['Chennai', None, 'Pune', 'Chennai']})
df2 = pd.DataFrame({'age': [20, 30], 'city': ['Delhi', 'Goa']})
df3 = pd.DataFrame({'score': [np.nan, np.nan], 'name': ['A', 'B']})
df4 = pd.DataFrame({'x': [1, np.nan, 3, np.nan]})
df5 = pd.DataFrame({'a': []})

print(fill_missing(df1))   # Expected: age row 1 -> 35.0, city row 1 -> Chennai
print(fill_missing(df2))   # Expected: unchanged
print(fill_missing(df3))   # Expected: score stays NaN, name unchanged
print(fill_missing(df4))   # Expected: x -> [1.0, 2.0, 3.0, 2.0]
print(fill_missing(df5))   # Expected: Empty DataFrame
```
