```python
# Q1. Second Largest Unique Number
# Return the second largest UNIQUE number in the list.
# Return None if it does not exist.

def second_largest(nums):
    neg_inf = float('-inf')
    first = second = neg_inf
    for num in nums:
        if num > first:
            sceond ,first = first,num
        elif num > second and num != first:
            second = num

    return second if second != neg_inf else None

print(second_largest([10, 20, 4, 45, 99]))   # Expected: 45
print(second_largest([5, 5, 5]))             # Expected: None
print(second_largest([-3, -1, -7]))          # Expected: -3
print(second_largest([7]))                   # Expected: None
print(second_largest([]))                    # Expected: None
```
