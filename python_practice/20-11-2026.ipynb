```python
# 1. Reverse Each Word  [easy / string]
# Reverse the letters of each word but keep the word order. Words are separated by single spaces.
# Keep EVERY space and punctuation mark exactly where it is.
# "hi there" -> "ih ereht"
def reverse_each_word(s):
    return " ".join([i[::-1] for i in s.split()])

print(reverse_each_word("hi there"))        # -> 'ih ereht'
print(reverse_each_word(""))                # -> ''
print(reverse_each_word("a  b"))            # -> 'a  b'   <- the double space must survive (split() drops it)
print(reverse_each_word("Hello, World!"))   # -> ',olleH !dlroW'   <- punctuation stays inside its word
print(reverse_each_word("x"))               # -> 'x'
```
