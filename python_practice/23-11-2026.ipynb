```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
# 1. Shrink Long Words  [easy / string]
# Replace every word LONGER than n (strictly) with its length as text.
# Keep ALL spaces exactly as they are (split on a single space).
# ("hello big", 3) -> "5 big"
def shrink_words(s, n):
    return " ".join([str(len(word)) if len(word)>n else word for word in s.split(" ")])

print(shrink_words('hello  big world', 3))   # -> '5  big 5'   <- double space must survive
print(shrink_words('', 3))   # -> ''   <- empty string
print(shrink_words('  ab  ', 1))   # -> '  2  '   <- leading and trailing spaces kept
print(shrink_words('tiny', 4))   # -> 'tiny'   <- length == n is NOT replaced
print(shrink_words('a bb ccc', 0))   # -> '1 2 3'   <- n = 0, every word shrinks
```

```python
# 2. Smallest Even Numbers  [easy / list]
# Return the n smallest DISTINCT even numbers, ascending. Zero counts as even.
# If fewer than n exist, return all of them.
# ([8, 2, 2, 4, 7], 2) -> [2, 4]
def smallest_evens(nums, n):
    return sorted([i for i in set(nums) if i%2==0])[:n]

print(smallest_evens([8, 2, 2, 4, 7, 6, 4], 3))   # -> [2, 4, 6]   <- duplicates
print(smallest_evens([], 2))   # -> []   <- empty
print(smallest_evens([1, 3, 5], 2))   # -> []   <- no evens
print(smallest_evens([-4, 0, -4, 6], 5))   # -> [-4, 0, 6]   <- negatives, zero, n too big
print(smallest_evens([10], 1))   # -> [10]   <- single
```

```python
# 3. Pass and Fail Lists  [easy / dict]
# records is a list of (name, marks). Return a dict with EXACT keys "passed" and "failed".
# Pass means marks >= cutoff (INCLUSIVE). Keep input order inside each list.
# ([("A", 50), ("B", 30)], 40) -> {"passed": ["A"], "failed": ["B"]}
def split_results(records, cutoff):
    result = {'passed': [], 'failed': []}
    if not result :
        return records
    for k,v in records:
        if v>= cutoff:
            result['passed'].append(k)
        else:
            result['failed'].append(k)
    return result

print(split_results([('Ravi', 72), ('Anu', 40), ('Kiran', 39)], 40))   # -> {'passed': ['Ravi', 'Anu'], 'failed': ['Kiran']}   <- 40 exactly passes
print(split_results([], 40))   # -> {'passed': [], 'failed': []}   <- empty, both keys still present
print(split_results([('Zed', 0)], 40))   # -> {'passed': [], 'failed': ['Zed']}   <- zero marks
print(split_results([('A', 50), ('B', 60)], 40))   # -> {'passed': ['A', 'B'], 'failed': []}   <- nobody fails
print(split_results([('A', 40), ('B', 40)], 41))   # -> {'passed': [], 'failed': ['A', 'B']}   <- nobody passes
```

```python
# 4. Words In Both Sentences  [medium / string + set]
# Return the words that appear in BOTH sentences, CASE-INSENSITIVE, DISTINCT,
# lowercase, sorted alphabetically.
# ("The cat", "a CAT") -> ["cat"]
def common_words(s1, s2):
    return sorted(list(set(s1.lower().split()) & set(s2.lower().split())))

print(common_words('The cat and the Dog', 'the DOG and a bird'))   # -> ['and', 'dog', 'the']   <- mixed case
print(common_words('', 'a b'))   # -> []   <- empty sentence
print(common_words('a a a', 'A'))   # -> ['a']   <- duplicates collapse
print(common_words('x y', 'z'))   # -> []   <- no match
print(common_words('Hello hello', 'HELLO'))   # -> ['hello']   <- all same word
```

```python
# 5. Merge Two Sorted Lists  [medium / two pointers]
# Merge two ASCENDING lists into one ascending list, KEEPING duplicates.
# Do NOT use sorted() or .sort(). Do NOT mutate the inputs.
# ([1, 3], [2, 3]) -> [1, 2, 3, 3]
def merge_sorted(a, b):
    result =[]
    i,j = 0,0

    while i<len(a) and j < len(b):
        if a[i]<b[j]:
            result.append(a[i])
            i+=1
        else:
            result.append(b[j])
            j+=1
    result.extend(a[i:])
    result.extend(b[j:])
    return result

print(merge_sorted([1, 3, 5], [2, 3, 6]))   # -> [1, 2, 3, 3, 5, 6]   <- duplicate 3 kept twice
print(merge_sorted([], []))   # -> []   <- both empty
print(merge_sorted([], [4, 5]))   # -> [4, 5]   <- one empty
print(merge_sorted([-3, 0], [-5, -3, 2]))   # -> [-5, -3, -3, 0, 2]   <- negatives
print(merge_sorted([7], [7]))   # -> [7, 7]   <- single each, equal
```

```python
# 6. Positive Column Sums  [easy / mask + axis]
# Return a 1D array: for each column, the sum of its POSITIVE values only.
# A column with no positives gives 0. Do not change the input.
def positive_col_sums(a):
    mask = np.where(a<0,0,a)
    return np.sum(mask,axis=0)

a26 = np.array([[ 3, -2, -1],
                [-1,  5, -4],
                [ 2,  0, -7],
                [-6,  1, -2]])   # <- last column has no positives
print(positive_col_sums(a26))   # -> expected below
# [5 6 0]
```

```python
# 7. Column Means Ignoring NaN  [easy / nan]
# Return a 1D array of per-column means that IGNORE NaN, rounded to 2 decimals.
def nan_col_means(a):
    return np.nanmean(a,axis=0).round(2)

a27 = np.array([[1, np.nan, 3],
                [2, 4, np.nan],
                [np.nan, 6, 9],
                [5, np.nan, 6]])
print(nan_col_means(a27))   # -> expected below
# [2.67 5.   6.  ]
```

```python
# 8. Stable Sort By First Column  [easy / argsort]
# Return the rows of a sorted by column 0 ascending.
# Rows with equal first values must KEEP their original order (stable). Return a new array.
def sort_rows(a):
    idx = np.argsort(a[:, 0], kind='stable')
    return a[idx]

a28 = np.array([[2, 10],
                [1, 20],
                [2,  5],
                [1, 30],
                [0,  1]])   # <- ties on column 0
print(sort_rows(a28))   # -> expected below
# [[ 0  1]
#  [ 1 20]
#  [ 1 30]
#  [ 2 10]
#  [ 2  5]]
```

```python
# 9. Mean Of Positives Per Row  [medium / mask + zero guard]
# Return a 1D float array: per row, the mean of its POSITIVE values only.
# A row with no positives gives 0.0 (no warnings, no NaN).
def mean_positive_per_row(a):
    pos = a > 0
    sums = np.where(pos, a, 0).sum(axis=1)
    counts = pos.sum(axis=1)
    return np.divide(sums, counts, out=np.zeros_like(sums, dtype=float), where=counts > 0)

a29 = np.array([[ 3, -1,  5],
                [-2, -4,  0],
                [ 0,  0,  0],
                [ 4,  4, -9]])   # <- rows 1 and 2 have no positives
print(mean_positive_per_row(a29))   # -> expected below
# [4. 0. 0. 4.]
```

```python
# 10. Add Totals Row And Column  [medium / stacking]
# Return a new (rows+1, cols+1) array: extra LAST column = row sums,
# extra LAST row = column sums, bottom-right corner = grand total.
def add_totals(a):
    row_sums = a.sum(axis=1)
    with_row_totals = np.column_stack([a, row_sums])
    col_sums = with_row_totals.sum(axis=0)
    return np.vstack([with_row_totals, col_sums])

a30 = np.array([[1, -2],
                [3,  4],
                [5,  6]])
print(add_totals(a30))   # -> expected below
# [[ 1 -2 -1]
#  [ 3  4  7]
#  [ 5  6 11]
#  [ 9  8 17]]
```

```python
# 11. Overwrite Status  [easy / overwrite column]
# Overwrite the EXISTING column 'status': "Pass" if score >= cutoff (INCLUSIVE) else "Fail".
# Do NOT add a new column. Do not mutate the input.
# Return columns ['name', 'score', 'status'] in that order, index reset.
def mark_status(df, cutoff):
    return (
        df[['name', 'score', 'status']]
        .assign(status=np.where(df['score'] >= cutoff, 'Pass', 'Fail'))
        .reset_index(drop=True)
    )

df31 = pd.DataFrame({
    'name': ['Asha', 'Bala', 'Chitra', 'Dev', 'Esha'],
    'city': ['Chennai', 'Madurai', 'Chennai', 'Salem', 'Trichy'],
    'score': [72, 40, 39, 0, 85],   # <- 40 is the boundary
    'status': ['pending'] * 5,
})
cutoff31 = 40
print(mark_status(df31,cutoff31))   # -> expected below
#      name  score status
# 0    Asha     72   Pass
# 1    Bala     40   Pass
# 2  Chitra     39   Fail
# 3     Dev      0   Fail
# 4    Esha     85   Pass
```

```python
# 12. Top N Earners  [easy / sort + head]
# Return the n highest salaries: sort by salary DESC, then name ASC for ties.
# Take the first n rows, columns ['name', 'salary'], index reset (0..n-1).
def top_earners(df, n):
    return (
        df.sort_values(by=['salary', 'name'], ascending=[False, True])
        .head(n)[['name', 'salary']]
        .reset_index(drop=True)
    )

df32 = pd.DataFrame({
    'name': ['Meena', 'Zoya', 'Arun', 'Bala', 'Chris'],
    'dept': ['HR', 'IT', 'IT', 'Ops', 'HR'],
    'salary': [50000, 60000, 60000, 55000, 50000],   # <- Meena and Chris tie at 50000
})
n32 = 4
print(top_earners(df32,n32))   # -> expected below
#     name  salary
# 0   Arun   60000
# 1   Zoya   60000
# 2   Bala   55000
# 3  Chris   50000
```
