```python
# 1. Count vowels and consonants in a string
def count_vowels_consonants(s):
    return (sum([1 for i in s.lower() if i in 'aeiou']),sum([1 for i in s.lower() if i not in 'aeiou'and i.isalpha()]))

print(count_vowels_consonants("hello world"))   # normal
print(count_vowels_consonants(""))              # empty
print(count_vowels_consonants("12345!"))        # no letters
print(count_vowels_consonants("AEIOU"))         # all vowels, uppercase
print(count_vowels_consonants("rhythm"))        # no standard vowels
```

```python
# 2. Return the factorial of a number without math.factorial
def factorial(n):
    if n==0 or n==1:
        return 1
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact

print(factorial(5))     # normal
print(factorial(0))     # zero
print(factorial(1))     # one
print(factorial(10))    # larger
print(factorial(3))     # small
```

```python
# 3. Check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    
    # Exclude all other even numbers
    if n % 2 == 0:
        return False
    result = []
    for i in range(3,int(n**0.5),2):
        if n % i == 0:
            return False
    return True     

print(is_prime(7))      # prime
print(is_prime(1))      # not prime (edge)
print(is_prime(2))      # smallest prime
print(is_prime(0))      # zero
print(is_prime(100))    # composite
```

```python
# 4. Return the Fibonacci sequence up to n terms
def fibonacci(n):
    if n == 0:
        return 0
    res = [0,1]
    while len(res) < n:
        res.append(res[-1]+res[-2])
    return res

print(fibonacci(5))     # normal
print(fibonacci(0))     # zero terms
print(fibonacci(1))     # single term
print(fibonacci(2))     # two terms
print(fibonacci(10))    # longer
```

```python
# 5. Find the most frequent element in a list
from collections import Counter
def most_frequent(lst):
    if not lst:
        return []
    counts = dict(sorted(Counter(lst).items(),key=lambda x:x[1],reverse=True))
    max_occ = max(counts.values())
    return [k for k,v in counts.items() if v== max_occ]

print(most_frequent([1, 3, 3, 2, 1, 3]))    # clear winner
print(most_frequent([5]))                   # single element
print(most_frequent([]))                    # empty
print(most_frequent([1, 1, 2, 2]))          # tie
print(most_frequent(["a", "b", "a"]))       # strings
```

```python
# 6. Merge two dictionaries, summing values for shared keys
def merge_sum(d1, d2):
    result = d1.copy()
    for k,v in d2.items():
        result[k] = result.get(k,0)+v
    return result
print(merge_sum({"a": 1, "b": 2}, {"b": 3, "c": 4}))    # overlap
print(merge_sum({}, {}))                                # both empty
print(merge_sum({"a": 1}, {}))                          # one empty
print(merge_sum({"x": 5}, {"x": 5}))                    # same key
print(merge_sum({"a": 1}, {"b": 2}))                    # no overlap
```

```python
# 7. Check if two lists have any common element
def has_common(a, b):
    return list(set(a) & set(b))

print(has_common([1, 2, 3], [3, 4, 5]))     # common exists
print(has_common([1, 2], [3, 4]))           # none
print(has_common([], [1, 2]))               # one empty
print(has_common([1, 1], [1]))              # duplicates
print(has_common([], []))                   # both empty
```

```python
# 8. Capitalize the first letter of each word without .title()
def capitalize_words(s):
    return " ".join([i[0].upper()+i[1:] for i in s.lower().split()])

print(capitalize_words("hello world"))      # normal
print(capitalize_words(""))                 # empty
print(capitalize_words("python"))           # single word
print(capitalize_words("a b c"))            # single-char words
print(capitalize_words("  spaced  out "))   # extra spaces
```
