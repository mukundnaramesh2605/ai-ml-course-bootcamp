```python
# 1. Count vowels and consonants in a string
def count_vowels_consonants(s):
    return (sum([1 for i in s.lower() if i in 'aeiou']),sum([1 for i in s.lower() if i not in 'aeiou'and i.isalpha()]))

print(count_vowels_consonants("hello world"))   # normal
print(count_vowels_consonants(""))              # empty
print(count_vowels_consonants("12345!"))        # no letters
print(count_vowels_consonants("AEIOU"))         # all vowels, uppercase
print(count_vowels_consonants("rhythm"))        # no standard vowels
```

```python
# 2. Return the factorial of a number without math.factorial
def factorial(n):
    if n==0 or n==1:
        return 1
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact

print(factorial(5))     # normal
print(factorial(0))     # zero
print(factorial(1))     # one
print(factorial(10))    # larger
print(factorial(3))     # small
```

```python
# 3. Check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    
    # Exclude all other even numbers
    if n % 2 == 0:
        return False
    result = []
    for i in range(3,int(n**0.5),2):
        if n % i == 0:
            return False
    return True     

print(is_prime(7))      # prime
print(is_prime(1))      # not prime (edge)
print(is_prime(2))      # smallest prime
print(is_prime(0))      # zero
print(is_prime(100))    # composite
```

```python
# 4. Return the Fibonacci sequence up to n terms
def fibonacci(n):
    if n == 0:
        return 0
    res = [0,1]
    while len(res) < n:
        res.append(res[-1]+res[-2])
    return res

print(fibonacci(5))     # normal
print(fibonacci(0))     # zero terms
print(fibonacci(1))     # single term
print(fibonacci(2))     # two terms
print(fibonacci(10))    # longer
```

```python
# 5. Find the most frequent element in a list
from collections import Counter
def most_frequent(lst):
    if not lst:
        return []
    counts = dict(sorted(Counter(lst).items(),key=lambda x:x[1],reverse=True))
    max_occ = max(counts.values())
    return [k for k,v in counts.items() if v== max_occ]

print(most_frequent([1, 3, 3, 2, 1, 3]))    # clear winner
print(most_frequent([5]))                   # single element
print(most_frequent([]))                    # empty
print(most_frequent([1, 1, 2, 2]))          # tie
print(most_frequent(["a", "b", "a"]))       # strings
```

```python
# 6. Merge two dictionaries, summing values for shared keys
def merge_sum(d1, d2):
    result = d1.copy()
    for k,v in d2.items():
        result[k] = result.get(k,0)+v
    return result
print(merge_sum({"a": 1, "b": 2}, {"b": 3, "c": 4}))    # overlap
print(merge_sum({}, {}))                                # both empty
print(merge_sum({"a": 1}, {}))                          # one empty
print(merge_sum({"x": 5}, {"x": 5}))                    # same key
print(merge_sum({"a": 1}, {"b": 2}))                    # no overlap
```

```python
# 7. Check if two lists have any common element
def has_common(a, b):
    return list(set(a) & set(b))

print(has_common([1, 2, 3], [3, 4, 5]))     # common exists
print(has_common([1, 2], [3, 4]))           # none
print(has_common([], [1, 2]))               # one empty
print(has_common([1, 1], [1]))              # duplicates
print(has_common([], []))                   # both empty
```

```python
# 8. Capitalize the first letter of each word without .title()
def capitalize_words(s):
    return " ".join([i[0].upper()+i[1:] for i in s.lower().split()])

print(capitalize_words("hello world"))      # normal
print(capitalize_words(""))                 # empty
print(capitalize_words("python"))           # single word
print(capitalize_words("a b c"))            # single-char words
print(capitalize_words("  spaced  out "))   # extra spaces
```

```python
# 9. Find the longest word in a sentence
def longest_word(sentence):
    return list(sorted(sentence.split(),key=len,reverse=True))[0] if sentence != "" else ""

print(longest_word("the quick brown fox"))      # normal
print(longest_word("hi"))                       # single word
print(longest_word(""))                         # empty
print(longest_word("cat dog bird"))             # tie on length
print(longest_word("a bb ccc dddd"))            # increasing lengths
```

```python
# 10. Count occurrences of a substring (overlapping allowed)
import re
def count_substring(text, sub):
    pattern = rf"(?={re.escape(sub)})"
    return len(re.findall(pattern,text))

print(count_substring("ababab", "ab"))      # normal
print(count_substring("aaaa", "aa"))        # overlapping
print(count_substring("hello", "z"))        # not found
print(count_substring("", "a"))             # empty text
print(count_substring("abc", ""))           # empty substring
```

```python
import numpy as np

# 1. Create an identity matrix of size n and multiply by a scalar
def scaled_identity(n, k):
    return np.eye(n)*k

print(scaled_identity(3, 5))    # normal
print(scaled_identity(1, 9))    # 1x1
print(scaled_identity(2, 0))    # scalar zero
print(scaled_identity(4, -2))   # negative scalar
print(scaled_identity(3, 1))    # plain identity
```

```python
# 2. Find the cumulative sum of a 1D array
def cumulative_sum(arr):
    return np.cumsum(arr)

print(cumulative_sum(np.array([1, 2, 3, 4])))   # normal
print(cumulative_sum(np.array([5])))            # single element
print(cumulative_sum(np.array([])))             # empty
print(cumulative_sum(np.array([-1, -2, -3])))   # negatives
print(cumulative_sum(np.array([0, 0, 0])))      # zeros
```

```python
# 3. Replace all even numbers in an array with -1
def replace_evens(arr):
    return np.where(arr%2 == 0 , -1,arr)

print(replace_evens(np.array([1, 2, 3, 4, 5])))     # normal
print(replace_evens(np.array([2, 4, 6])))           # all even
print(replace_evens(np.array([1, 3, 5])))           # all odd
print(replace_evens(np.array([0])))                 # zero (even)
print(replace_evens(np.array([-2, -3, -4])))        # negatives
```

```python
# 4. Compute row-wise and column-wise sums of a 2D array
def axis_sums(mat):
    row_sums = np.sum(mat, axis=1)
    col_sums = np.sum(mat, axis=0)
    return row_sums, col_sums 

print(axis_sums(np.array([[1, 2], [3, 4]])))        # normal
print(axis_sums(np.array([[5]])))                   # single element
print(axis_sums(np.array([[1, 1, 1]])))             # single row
print(axis_sums(np.array([[0, 0], [0, 0]])))        # zeros
print(axis_sums(np.array([[-1, 2], [3, -4]])))      # mixed signs
```

```python
# 5. Find the unique values and their counts in an array
def value_counts(arr):
    unique_vals, counts = np.unique(arr, return_counts=True)
    return unique_vals, counts

print(value_counts(np.array([1, 2, 2, 3, 3, 3])))   # normal
print(value_counts(np.array([5, 5, 5])))            # all same
print(value_counts(np.array([1, 2, 3])))            # all unique
print(value_counts(np.array([])))                   # empty
print(value_counts(np.array([-1, -1, 0])))          # negatives + zero
```

```python
# 6. Clip array values to a given min and max range
def clip_values(arr, low, high):
    # your code here
    return np.clip(arr,low,high)

print(clip_values(np.array([1, 5, 10, 15]), 3, 12))     # normal
print(clip_values(np.array([0, 0, 0]), 1, 5))           # all below min
print(clip_values(np.array([20, 30]), 1, 10))           # all above max
print(clip_values(np.array([5, 6, 7]), 5, 7))           # already in range
print(clip_values(np.array([-5, 0, 5]), -2, 2))         # negatives
```

```python
# 7. Compute the dot product of two 1D arrays
def dot_product(a, b):
    return np.dot(a,b)

print(dot_product(np.array([1, 2, 3]), np.array([4, 5, 6])))    # normal
print(dot_product(np.array([0, 0]), np.array([5, 9])))          # zeros
print(dot_product(np.array([1]), np.array([7])))                # single element
print(dot_product(np.array([-1, 2]), np.array([3, -4])))        # negatives
print(dot_product(np.array([1, 1, 1]), np.array([1, 1, 1])))    # all ones
```

```python
# 8. Flatten a 2D array and sort it in descending order
def flatten_sort_desc(mat):
    return np.sort(mat.flatten())[::-1]

print(flatten_sort_desc(np.array([[3, 1], [4, 2]])))    # normal
print(flatten_sort_desc(np.array([[5]])))               # single element
print(flatten_sort_desc(np.array([[1, 1], [1, 1]])))    # all same
print(flatten_sort_desc(np.array([[-1, -5], [-2, -9]]))) # negatives
print(flatten_sort_desc(np.array([[0, 0], [0, 1]])))    # mostly zeros
```

```python
# 9. Find the index of the value closest to a given target
def closest_index(arr, target):
    # your code here
    return np.argmin(np.abs(arr - target))

print(closest_index(np.array([1, 5, 10, 15]), 7))       # between values
print(closest_index(np.array([3]), 100))                # single element
print(closest_index(np.array([2, 4, 6]), 4))            # exact match
print(closest_index(np.array([-5, 0, 5]), -3))          # negatives
print(closest_index(np.array([10, 20]), 15))            # equidistant
```

```python
# 10. Create a boolean mask for values divisible by 3
def divisible_by_three(arr):
    mask = (arr%3==0)
    return mask
print(divisible_by_three(np.array([1, 3, 6, 7, 9])))    # normal
print(divisible_by_three(np.array([3, 6, 9])))          # all divisible
print(divisible_by_three(np.array([1, 2, 4])))          # none divisible
print(divisible_by_three(np.array([0])))                # zero (divisible)
print(divisible_by_three(np.array([-3, -6, -5])))       # negatives
```
