```python
# 1. Sum Positives  [Service / logic]
# Return the sum of only the positive numbers.
# [3, -1, 4, -5, 2] -> 9
def sum_positives(nums):
    return sum([x for x in nums if x>0])
print(sum_positives([3,-1,4,-5,2]))# -> 9 ; 
print(sum_positives([-1,-2]))# -> 0 ; 
print(sum_positives([]))# -> 0
```

```python
# 2. Count Characters  [Service / dict]
# Return a dict of character -> count, case-insensitive, skipping spaces.
# "A a b" -> {"a":2, "b":1}
from collections import Counter
def char_count(s):
    return dict(Counter(s.lower().split()))
print(char_count("A a b"))# -> {"a":2,"b":1}
```

```python
# 3. Remove Duplicates Keep Order  [Service / logic]
# Return the list with duplicates removed, keeping first-occurrence order.
# [3,1,3,2,1] -> [3,1,2]
def dedupe(nums):
    result = []
    for i in nums:
        if i not in result:
            result.append(i)
    return result
print(dedupe([3,1,3,2,1]))# -> [3,1,2] ; 
print(dedupe([]))# -> []
```

```python
# 4. Title Case Sentence  [Service / string]
# Capitalize the first letter of each word, lowercasing the rest.
# "hELLO wORLD" -> "Hello World"
def title_case(s):
    return " ".join([word[0].upper()+word[1:] for word in s.lower().split()])
print(title_case("hELLO wORLD"))# -> "Hello World"
```

```python
# 5. Find Pairs  [AI screen / logic]
# Return the COUNT of index pairs (i < j) where nums[i] + nums[j] == target.
# [1,2,3,2], 4 -> 2
def count_pairs(nums, target):
    result = []
    for i in range(len(nums)):
        for j in range(i,len(nums)):
            if nums[i] + nums[j] == target and i !=j:
                result.append([i,j])
    return len(result)
print(count_pairs([1,2,3,2], 4))# -> 2 ; 
print(count_pairs([1,1,1], 2))# -> 3
```

```python
# 6. Second Smallest  [Service / logic]
# Return the second smallest DISTINCT value, or None if there isn't one.
# [4,1,1,3] -> 3
def second_smallest(nums):
    if len(set(nums)) == 1:
        return None
    first = second = float('inf')
    for i in nums:
        if i<first:
            second=first
            first=i
        elif i<second and i!=first:
            second = i
    return second if second!= float('inf') else None
print(second_smallest([4,1,1,3]))# -> 3 ; 
print(second_smallest([2,2]))# -> None
```

```python
# 1. Character Frequency  [Service / dict]
# Return a dict char -> count, case-insensitive, skipping spaces.
# "ab ba" -> {"a":2, "b":2}   (note: multi-letter tokens, so no .split shortcut)
from collections import Counter
def char_count(s):
    joined = "".join([x.lower() for x in s if x.isalpha()])
    return dict(Counter(joined))
print(char_count("ab ba"))# -> {"a":2,"b":2} ; 
print(char_count("Hi hi"))# -> {"h":2,"i":2}
```

```python
# 2. Second Smallest  [Service / logic]
# Return the second smallest DISTINCT value, or None if there isn't one.
# [1,2,3,4] -> 2   (four distinct values, so a second-largest shortcut fails)
def second_smallest(nums):
    if len(set(nums)) == 1:
        return None
    first = second = float('inf')
    for i in nums:
        if i<first:
            second=first
            first=i
        elif i<second and i!=first:
            second = i
    return second if second!= float('inf') else None
print(second_smallest([1,2,3,4]))# -> 2 ; 
print(second_smallest([5,5]))# -> None ; 
print(second_smallest([3,1,2]))# -> 2
print(second_smallest([9,10,8]))# -> 9
```

```python
# 3. Move Zeros  [AI screen / logic]
# Move all 0s to the end, keep the order of non-zero values (negatives included).
# [0, -1, 0, 3, -2] -> [-1, 3, -2, 0, 0]
def move_zeros(nums):
    return [x for x in nums if x!=0]+[0]*nums.count(0)
print(move_zeros([0,-1,0,3,-2]))# -> [-1,3,-2,0,0]
```

```python
# 4. Anagram Check  [AI screen / string]
# True if a and b are anagrams, case-insensitive, ignoring spaces.
# "Dormitory", "Dirty Room" -> True   (spaces must be ignored)
def is_anagram(a, b):
    a = [x.lower() for x in a if x.isalnum()]
    b = [x.lower() for x in b if x.isalnum()]
    return sorted(a) == sorted(b)
print(is_anagram("Dormitory","Dirty Room"))# -> True ; 
print(is_anagram("abc","abd"))# -> False
```

```python
# 5. Running Totals  [AI screen / logic]
# Return a list of the running cumulative sum.
# [1,2,3,4] -> [1,3,6,10]
def running_totals(nums):
    if not nums:
        return []
    result = [nums[0]]
    sum_ = nums[0]
    for i in nums[1:]:
        sum_+=i
        result.append(sum_)
    return result
print(running_totals([1,2,3,4]))# -> [1,3,6,10] ; 
print(running_totals([]))# -> []
```

```python
# 6. Most Frequent  [AI screen / dict]
# Return the most frequent element; on a tie, the one appearing first.
# [1,2,2,3,3] -> 2   (2 and 3 tie at 2, but 2 appears first)
def most_frequent(nums):
    sorted_ = dict(sorted({k:nums.count(k) for k in set(nums)}.items(),key=lambda x: x[1],reverse=True))
    return [x for x in sorted_][0]
    
print(most_frequent([1,2,2,3,3]))# -> 2 ; 
print(most_frequent([4]))# -> 4
```

```python
import functools

# 1. Call Counter  [AI screen / decorators]
# Write a decorator that counts how many times the function has been called,
# storing the count on the wrapper as an attribute `.calls`.
# The wrapped function should still return its normal result.
def count_calls(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)

    wrapper.calls = 0
    return wrapper

@count_calls
def do_nothing():
    print('None')
do_nothing()
do_nothing()
do_nothing()
print(do_nothing.calls)
do_nothing()
print(do_nothing.calls)
```

```python
# 2. Repeat N Times (decorator WITH an argument)  [AI screen / decorators]
# Write a decorator repeat(n) that calls the function n times and returns
# a LIST of the n results. Preserve the function's name with functools.wraps.
def repeat(n):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args,**kwargs):
            return [func(*args, **kwargs) for _ in range(n)]
        return wrapper
    return decorator

@repeat(3)
def roll():
    return 7

print(roll())
print(roll.__name__)
# test:
# @repeat(3)
# def roll(): return 7
# roll() -> [7, 7, 7]
# roll.__name__ -> "roll"
```

```python
# 3. Memoize (cache results)  [AI screen / decorators]
# Write a decorator that caches results by the function's positional args,
# so a repeated call with the same args returns the stored result without
# recomputing. Use functools.wraps. (Assume args are hashable.)
def memoize(func):
    cache = {}
    @functools.wraps(func)
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    wrapper.cache = cache
    return wrapper

# test:
calls = []
@memoize
def square(x):
     calls.append(x)       # records each real computation
     return x * x
print(square(4))
print(square(4))
print(square(5))
# calls -> [4, 5]          # 4 computed once, not twice
square(4)# -> 16
```
