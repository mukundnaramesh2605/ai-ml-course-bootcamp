```python
# 1. Sum Positives  [Service / logic]
# Return the sum of only the positive numbers.
# [3, -1, 4, -5, 2] -> 9
def sum_positives(nums):
    return sum([x for x in nums if x>0])
print(sum_positives([3,-1,4,-5,2]))# -> 9 ; 
print(sum_positives([-1,-2]))# -> 0 ; 
print(sum_positives([]))# -> 0
```

```python
# 2. Count Characters  [Service / dict]
# Return a dict of character -> count, case-insensitive, skipping spaces.
# "A a b" -> {"a":2, "b":1}
from collections import Counter
def char_count(s):
    return dict(Counter(s.lower().split()))
# test: char_count("A a b") -> {"a":2,"b":1}
```
