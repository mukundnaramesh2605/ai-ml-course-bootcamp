```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
# 1. Abbreviate Long Words  [easy / string]
# Replace each word of length >= n with: first letter + (length - 2) + last letter.
# n is at least 3. Keep ALL spaces exactly as they are (split on a single space).
# ("internet is big", 5) -> "i6t is big"
def abbreviate(s, n):
    return " ".join([i[0]+str(len(i)-2)+i[-1] if len(i)>=n else i for i in s.split(" ")])

print(abbreviate('internet  is big', 5))   # -> 'i6t  is big'   <- double space must survive
print(abbreviate('', 4))   # -> ''   <- empty string
print(abbreviate('  cat  ', 3))   # -> '  c1t  '   <- length == n counts; outer spaces kept
print(abbreviate('go to', 3))   # -> 'go to'   <- nothing long enough
print(abbreviate('abc abcd', 3))   # -> 'a1c a2d'   <- 3-letter word becomes a1c
```

```python
# 2. Letters Only In First  [easy / set]
# Return the letters that appear in a but NOT in b, CASE-INSENSITIVE, DISTINCT,
# lowercase, sorted. Ignore anything that is not a letter.
# ("abc", "b") -> ["a", "c"]
import re
def letters_only_in_first(a, b):
    a_chars = re.findall(r'[a-z]',a.lower())
    b_chars = re.findall(r'[a-z]',b.lower())
    return sorted(list(set(a_chars) - set(b_chars)))

print(letters_only_in_first('Hello World', 'low'))   # -> ['d', 'e', 'h', 'r']   <- mixed case
print(letters_only_in_first('', ''))   # -> []   <- both empty
print(letters_only_in_first('a1 b2!', 'B'))   # -> ['a']   <- digits and symbols ignored
print(letters_only_in_first('aaa', 'a'))   # -> []   <- nothing left
print(letters_only_in_first('AaBb', 'c'))   # -> ['a', 'b']   <- duplicates collapse
```

```python
# 3. Number Summary Dict  [easy / dict]
# Return a dict with EXACT keys "min", "max", "total", "count".
# For an empty list: min and max are None, total and count are 0.
# ([3, 1]) -> {"min": 1, "max": 3, "total": 4, "count": 2}
def summarize(nums):
    result = {'min': None, 'max': None, 'total': 0, 'count': 0}
    if nums:
        result['min'] = min(nums)
        result['max'] = max(nums)
        result['t0tal'] = sum(nums)
        result['count'] = len(nums)
    return result

print(summarize([4, -2, 9]))   # -> {'min': -2, 'max': 9, 'total': 11, 'count': 3}   <- negative
print(summarize([]))   # -> {'min': None, 'max': None, 'total': 0, 'count': 0}   <- empty, keys still present
print(summarize([5]))   # -> {'min': 5, 'max': 5, 'total': 5, 'count': 1}   <- single
print(summarize([0, 0]))   # -> {'min': 0, 'max': 0, 'total': 0, 'count': 2}   <- all zero
print(summarize([-3, -3, -1]))   # -> {'min': -3, 'max': -1, 'total': -7, 'count': 3}   <- all negative
```

```python
# 4. Best Average Students  [medium / dict + average]
# records is a list of (name, mark); a name can repeat. Return a sorted list of the
# name(s) with the HIGHEST average mark. If several tie, return ALL of them.
# ([("A", 50), ("B", 60), ("A", 80)]) -> ["A"]
def best_average(records):
    totals= {}
    counts = {}
    if not records:
        return []
    for i,j in records:
        totals[i]=totals.get(i,0)+j
        counts[i]=counts.get(i,0)+1
        highest = {i:(totals[i]/counts[i]) for i in totals.keys()}
    return sorted([i for i,j in highest.items() if j==max(highest.values())])

print(best_average([('A', 50), ('B', 60), ('A', 80)]))   # -> ['A']   <- A averages 65
print(best_average([]))   # -> []   <- empty
print(best_average([('Zed', 70)]))   # -> ['Zed']   <- single
print(best_average([('Ravi', 60), ('Anu', 40), ('Ravi', 60), ('Anu', 80)]))   # -> ['Anu', 'Ravi']   <- tie at 60, both returned
print(best_average([('A', 0), ('B', -5)]))   # -> ['A']   <- zero beats negative
```

```python
# 5. Longest Unique Substring  [medium / sliding window]
# Return the LENGTH of the longest substring with no repeated character.
# Linear time: one pass with a dict of last-seen positions.
# ("abcabc") -> 3
def longest_unique(s):
    left = 0
    best = 0
    last_seen = {}

    for right, ch in enumerate(s):
        if ch in last_seen and last_seen[ch] >= left:
            left = last_seen[ch] + 1

        last_seen[ch] = right
        best = max(best, right - left + 1)

    return best

print(longest_unique('abcabcbb'))   # -> 3
print(longest_unique(''))   # -> 0   <- empty
print(longest_unique('bbbb'))   # -> 1   <- all same
print(longest_unique('pwwkew'))   # -> 3   <- substring, not subsequence
print(longest_unique('abba'))   # -> 2   <- old index must not move the window back
```

```python
# 6. NaN Count Per Column  [easy / nan + axis]
# Return a 1D int array: the number of NaN values in each column.
def nan_per_col(a):
    return np.sum(np.isnan(a),axis=0)

a36 = np.array([[1, np.nan, 3],
                [np.nan, np.nan, 6],
                [7, 8, 9],
                [np.nan, np.nan, 3]])   # <- last column has no NaN
print(nan_per_col(a36))   # -> expected below
# [2 3 0]
```

```python
# 7. Cells Above Threshold  [easy / argwhere]
# Return a 2D array of [row, col] pairs for every value STRICTLY above t,
# in row-major order (np.argwhere order). Shape (k, 2).
def cells_above(a, t):
    return np.argwhere(a > t)

a37 = np.array([[5, -1, 9],
                [6,  7, 2],
                [8,  8, -3]])
t37 = 6   # <- the 6 is NOT above 6
print(cells_above(a37,t37))   # -> expected below
# [[0 2]
#  [1 1]
#  [2 0]
#  [2 1]]
```

```python
# 8. Step Changes  [easy / diff]
# Return a 1D array, same length as a: each value minus the one before it.
# The FIRST element is 0. Do not loop.
def step_changes(a):
    return np.diff(a, prepend=a[0])

a38 = np.array([4, 4, 9, -2, -2, 0])
print(step_changes(a38))   # -> expected below
# [  0   0   5 -11   0   2]
```

```python
# 9. Fill NaN With Row Mean  [medium / where + keepdims]
# Return a new float array where every NaN is replaced by the mean of the
# NON-NaN values in ITS ROW. Every row has at least one real value.
def fill_row_mean(a):
    row_means = np.nanmean(a, axis=1, keepdims=True)
    return np.where(np.isnan(a), row_means, a)

a39 = np.array([[1, np.nan, 3],
                [np.nan, 4, 8],
                [np.nan, np.nan, 5],
                [2, 2, np.nan]])
print(fill_row_mean(a39))   # -> expected below
# [[1. 2. 3.]
#  [6. 4. 8.]
#  [5. 5. 5.]
#  [2. 2. 2.]]
```

```python
# 10. Rank Within Row  [medium / argsort]
# Return an int array, same shape: the rank of each value inside its ROW,
# 1 = largest. TIES: the value in the LOWER column index gets the better rank.
def rank_in_row(a):
    order = np.argsort(-a, axis=1, kind='stable')
    return np.argsort(order, axis=1) + 1

a40 = np.array([[5, 9, 5],
                [1, 1, 3],
                [7, 2, 7]])   # <- ties in rows 0, 1 and 2
print(rank_in_row(a40))   # -> expected below
# [[2 1 3]
#  [2 3 1]
#  [1 3 2]]
```

```python
# 11. Add Grade Column  [easy / np.select]
# Add a NEW column 'grade': "A" if score >= 75, "B" if score >= 50, else "C".
# Return columns ['name', 'score', 'grade'] in that order, index unchanged. Do not mutate the input.
def add_grade(df):
    conditions = [df['score'] >= 75, df['score'] >= 50]
    choices = ['A', 'B']
    return df.assign(grade=np.select(conditions, choices, default='C'))[
      ['name', 'score', 'grade']]

df41 = pd.DataFrame({
    'name': ['Asha', 'Bala', 'Chitra', 'Dev', 'Esha'],
    'city': ['Chennai', 'Madurai', 'Salem', 'Trichy', 'Erode'],
    'score': [90, 75, 74, 50, 10],   # <- 75 and 50 are boundaries
})
print(add_grade(df41))   # -> expected below
#      name  score grade
# 0    Asha     90     A
# 1    Bala     75     A
# 2  Chitra     74     B
# 3     Dev     50     B
# 4    Esha     10     C
```

```python
# 12. Name Contains  [easy / str.contains]
# Keep rows where 'name' contains "an", CASE-INSENSITIVE. Missing names are NOT kept.
# Return columns ['name', 'city'], index reset.
def name_has_an(df):
  mask = df['name'].str.contains('an', case=False, na=False)
  return df.loc[mask, ['name', 'city']].reset_index(drop=True)

df42 = pd.DataFrame({
    'name': ['Ananya', 'Ravi', 'Hanan', None, 'DANIEL', 'Kiran'],   # <- None, and upper case
    'city': ['Chennai', 'Madurai', 'Salem', 'Trichy', 'Erode', 'Vellore'],
})
print(name_has_an(df42))   # -> expected below
#      name     city
# 0  Ananya  Chennai
# 1   Hanan    Salem
# 2  DANIEL    Erode
# 3   Kiran  Vellore
```

```python
# 13. Clean City Column  [easy / str overwrite]
# Overwrite the EXISTING column 'city' with its value stripped of outer spaces and lowercase.
# Do NOT add a column. Return all columns in the original order. Do not mutate the input.
def clean_city(df):
  return df.assign(city=df['city'].str.strip().str.lower())

df43 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'city': [' Chennai', 'MADURAI ', 'salem', '  Trichy  '],   # <- spaces and case
})
print(clean_city(df43))   # -> expected below
#    id     city
# 0   1  chennai
# 1   2  madurai
# 2   3    salem
# 3   4   trichy
```

```python
# 14. Running Average Per Customer  [medium / groupby cumulative]
# Sort by cust, then date (ASC), reset index. Add 'running_avg': the average of the
# customer's amounts from their FIRST row up to and INCLUDING this row, rounded to 2.
# Return columns ['cust', 'date', 'amount', 'running_avg']. Do not mutate the input.
def running_avg(df):
  out = df.sort_values(['cust', 'date']).reset_index(drop=True)
  avg = out.groupby('cust')['amount'].expanding().mean().reset_index(level=0, drop=True)
  return out.assign(running_avg=avg.round(2))[['cust', 'date', 'amount', 'running_avg']]

df44 = pd.DataFrame({
    'cust': ['B', 'A', 'A', 'B', 'A'],
    'date': ['2026-04-05', '2026-04-09', '2026-04-01', '2026-04-02', '2026-04-04'],   # <- not in date order
    'amount': [200, 50, 30, 125, 70],
})
print(running_avg(df44))   # -> expected below
#   cust        date  amount  running_avg
# 0    A  2026-04-01      30         30.0
# 1    A  2026-04-04      70         50.0
# 2    A  2026-04-09      50         50.0
# 3    B  2026-04-02     125        125.0
# 4    B  2026-04-05     200        162.5
```

```python
# 15. Combine Two Customer Lists  [medium / concat + dedupe]
# Stack df1 on top of df2. If an 'id' appears in both, KEEP the row from df1.
# Return columns ['id', 'name'], sorted by id ASC, index reset.
def combine(df1, df2):
    return (
      pd.concat([df1[['id', 'name']], df2[['id', 'name']]], ignore_index=True)
      .drop_duplicates(subset='id', keep='first')
      .sort_values('id')
      .reset_index(drop=True)
  )

df45a = pd.DataFrame({'id': [3, 1, 5], 'name': ['Chitra', 'Asha', 'Esha']})
df45b = pd.DataFrame({'id': [2, 3, 4], 'name': ['Bala', 'CHITRA R', 'Dev']})   # <- id 3 is in both
print(combine(df45a,df45b))   # -> expected below
#    id    name
# 0   1    Asha
# 1   2    Bala
# 2   3  Chitra
# 3   4     Dev
# 4   5    Esha
```
