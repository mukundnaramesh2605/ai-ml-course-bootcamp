```python
# 1. Count Divisible  [Service / logic]
# Return how many numbers in the list are divisible by both 2 and 3.
# [6, 4, 12, 9, 18] -> 3   (6, 12, 18)
def count_div(nums):
    return list(filter(lambda x: (x%2==0)&(x%3==0),nums))
print(count_div([6,4,12,9,18]))# -> 3 ; 
print(count_div([]))# -> 0
```

```python
# 2. Initials  [Service / string]
# Return the uppercase initials of a full name as a string.
# "mukundan ramesh" -> "MR"
def initials(name):
    return "".join([i[0].upper() for i in name.split()])
print(initials("mukundan ramesh"))# -> "MR" ; 
print(initials("grace hopper"))# -> "GH"
```
