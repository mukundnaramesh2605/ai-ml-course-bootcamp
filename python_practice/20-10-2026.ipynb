```python
# 1. Count Divisible  [Service / logic]
# Return how many numbers in the list are divisible by both 2 and 3.
# [6, 4, 12, 9, 18] -> 3   (6, 12, 18)
def count_div(nums):
    return list(filter(lambda x: (x%2==0)&(x%3==0),nums))
print(count_div([6,4,12,9,18]))# -> 3 ; 
print(count_div([]))# -> 0
```

```python
# 2. Initials  [Service / string]
# Return the uppercase initials of a full name as a string.
# "mukundan ramesh" -> "MR"
def initials(name):
    return "".join([i[0].upper() for i in name.split()])
print(initials("mukundan ramesh"))# -> "MR" ; 
print(initials("grace hopper"))# -> "GH"
```

```python
# 3. Group Anagrams Count  [AI screen / dict]
# Given a list of words, return how many DISTINCT anagram groups there are.
# ["eat","tea","tan","ate","nat"] -> 2   ({eat,tea,ate}, {tan,nat})
from collections import defaultdict
def anagram_groups(words):
    grouped = defaultdict(list)
    for i in words:
        grouped["".join(sorted(i.lower()))].append(i)
    return len(grouped.values())
print(anagram_groups(["eat","tea","tan","ate","nat"])) #-> 2
```

```python
def anagram_groups(words):
    grouped = set()
    for i in words:
        sorted_word = "".join(sorted(i))
        grouped.add(sorted_word)
    return len(grouped)
print(anagram_groups(["eat","tea","tan","ate","nat"])) #-> 2
```
