```python
# 1. Count Even Digits  [easy / logic]
# Return HOW MANY digits of a non-negative integer are even (0 counts as even).
# 2468 -> 4
def even_digits(n):
    return sum([1 for i in str(n) if int(i)%2==0])

print(even_digits(2468))    # -> 4
print(even_digits(0))       # -> 1   <- 0 is an even digit
print(even_digits(13579))   # -> 0
print(even_digits(102))     # -> 2
print(even_digits(7))       # -> 0
```

```python
# 2. Longest Word  [easy / string]
# Return the longest word in the sentence. On a tie, return the FIRST one. "" if no words.
# "I love python" -> "python"
def longest_word(s):
    if not s:
        return ""
    return sorted(s.split(),key=len,reverse=True)[0]

print(longest_word("I love python"))   # -> 'python'
print(longest_word(""))                # -> ''
print(longest_word("ab cd"))           # -> 'ab'    <- tie, first one wins
print(longest_word("a bb ccc bb"))     # -> 'ccc'
print(longest_word("  hi  "))          # -> "hi"
```

```python
# 3. Char Frequency Sorted  [medium / dict + sort]
# Case-insensitive, ignore spaces. Return a list of (char, count) sorted by count DESC,
# ties broken by char ASC.
# "banana" -> [('a', 3), ('n', 2), ('b', 1)]
def char_freq(s):
    s=s.lower()
    count = list(sorted(set((k,s.count(k)) for k in s if k != " "),key=lambda x:[-x[1],x[0]]))
    return count
print(char_freq("banana"))   # -> [('a', 3), ('n', 2), ('b', 1)]
print(char_freq(""))         # -> []
print(char_freq("Aa b"))     # -> [('a', 2), ('b', 1)]
print(char_freq("abc"))      # -> [('a', 1), ('b', 1), ('c', 1)]
print(char_freq("zzyy"))     # -> [('y', 2), ('z', 2)]   <- same count, so y before z
```

```python
# 4. Nth Largest Distinct  [medium / bounds]
# Return the n-th largest DISTINCT value (1-based). None if there are fewer than n distinct values.
# [4,1,4,3], 2 -> 3
def nth_largest(nums, n):
    distinct = sorted(set(nums), reverse=True)
    if n < 1 or n > len(distinct):
        return None
    return distinct[n - 1]

print(nth_largest([4, 1, 4, 3], 2))      # -> 3   <- the two 4s count once
print(nth_largest([5], 1))               # -> 5
print(nth_largest([5], 2))               # -> None   <- n is bigger than the distinct count
print(nth_largest([], 1))                # -> None
print(nth_largest([-1, -2, -3], 3))      # -> -3
```

```python
# 5. Common Elements  [easy / set]
# Return values present in BOTH lists, no duplicates, in the order they first appear in a.
# [1,2,2,3], [2,3,4] -> [2, 3]
def common(a, b):
    seen = []
    for k in a:
        if k in b and k not in seen:
            seen.append(k)
    return seen

print(common([1, 2, 2, 3], [2, 3, 4]))   # -> [2, 3]
print(common([], [1]))                   # -> []
print(common([1, 1], [1]))               # -> [1]   <- no duplicates in the output
print(common([5, 4], [4, 5]))            # -> [5, 4]   <- order follows a, not b
print(common([1, 2], [3]))               # -> []
```
