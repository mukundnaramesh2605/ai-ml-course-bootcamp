```python
# 1. Average of List  [Service / logic]
# Return the average (mean) of the numbers, or 0 for an empty list.
# [2,4,6] -> 4.0
def average(nums):
    if not nums:
        return 0
    return sum(nums)/len(nums)
print(average([2,4,6]))# -> 4.0 ; 
print(average([]))#-> 0
```

```python
# 2. Count Words  [Service / string]
# Return how many words have length >= 4.
# "the quick brown fox" -> 2   (quick, brown)
def count_long(s):
    return sum([1 for i in s.split() if len(i) >=4])
print(count_long("the quick brown fox"))# -> 2 ; 
print(count_long(""))# -> 0
```
