```python
# 1. Average of List  [Service / logic]
# Return the average (mean) of the numbers, or 0 for an empty list.
# [2,4,6] -> 4.0
def average(nums):
    if not nums:
        return 0
    return sum(nums)/len(nums)
print(average([2,4,6]))# -> 4.0 ; 
print(average([]))#-> 0
```

```python
# 2. Count Words  [Service / string]
# Return how many words have length >= 4.
# "the quick brown fox" -> 2   (quick, brown)
def count_long(s):
    return sum([1 for i in s.split() if len(i) >=4])
print(count_long("the quick brown fox"))# -> 2 ; 
print(count_long(""))# -> 0
```

```python
# 3. Merge Dicts Summing  [AI screen / dict]
# Merge two dicts; if a key is in both, SUM the values.
# {"a":1,"b":2}, {"b":3,"c":4} -> {"a":1,"b":5,"c":4}
def merge_sum(d1, d2):
    result = d1.copy()
    for i in d2:
        result[i] = result.get(i,0)+d2[i]
    return result
print(merge_sum({"a":1,"b":2}, {"b":3,"c":4}))# -> {"a":1,"b":5,"c":4}
```

```python
# 4. Is Sorted  [AI screen / logic]
# Return True if the list is sorted in non-decreasing order.
# [1,2,2,3] -> True ; [1,3,2] -> False
def is_sorted(nums):
    return nums == list(sorted(nums))
print(is_sorted([1,2,2,3]))# -> True ; 
print(is_sorted([1,3,2]))# -> False ; 
print(is_sorted([]))# -> True
```
