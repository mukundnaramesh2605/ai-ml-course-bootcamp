```python
# 1. Average of List  [Service / logic]
# Return the average (mean) of the numbers, or 0 for an empty list.
# [2,4,6] -> 4.0
def average(nums):
    if not nums:
        return 0
    return sum(nums)/len(nums)
print(average([2,4,6]))# -> 4.0 ; 
print(average([]))#-> 0
```

```python
# 2. Count Words  [Service / string]
# Return how many words have length >= 4.
# "the quick brown fox" -> 2   (quick, brown)
def count_long(s):
    return sum([1 for i in s.split() if len(i) >=4])
print(count_long("the quick brown fox"))# -> 2 ; 
print(count_long(""))# -> 0
```

```python
# 3. Merge Dicts Summing  [AI screen / dict]
# Merge two dicts; if a key is in both, SUM the values.
# {"a":1,"b":2}, {"b":3,"c":4} -> {"a":1,"b":5,"c":4}
def merge_sum(d1, d2):
    result = d1.copy()
    for i in d2:
        result[i] = result.get(i,0)+d2[i]
    return result
print(merge_sum({"a":1,"b":2}, {"b":3,"c":4}))# -> {"a":1,"b":5,"c":4}
```

```python
# 4. Is Sorted  [AI screen / logic]
# Return True if the list is sorted in non-decreasing order.
# [1,2,2,3] -> True ; [1,3,2] -> False
def is_sorted(nums):
    return nums == list(sorted(nums))
print(is_sorted([1,2,2,3]))# -> True ; 
print(is_sorted([1,3,2]))# -> False ; 
print(is_sorted([]))# -> True
```

```python
# 5. First Unique Character  [AI screen / string]
# Return the first character appearing exactly once, case-sensitive.
# If none, return None. "aabbc" -> "c"
def first_unique(s):
    for i in s:
        if s.count(i) == 1:
            return i
    return None
print(first_unique("aabbc"))# -> "c" ; 
print(first_unique("aabb"))# -> None
```

```python
# 6. Group by Parity  [AI screen / dict]
# Return a dict {"even":[...], "odd":[...]} preserving original order.
# [1,2,3,4] -> {"even":[2,4], "odd":[1,3]}
def group_parity(nums):
    grouped = {'even':[],'odd':[]}
    for i in nums:
        if i %2 == 0:
            grouped['even'].append(i)
        else:
            grouped['odd'].append(i)
    return grouped
print(group_parity([1,2,3,4]))# -> {"even":[2,4], "odd":[1,3]}
```

```python
# 7. Count Outside Range  [AI screen / data handling]
# Return HOW MANY values are < 20 OR > 80. (a count)
import numpy as np
data = np.array([10, 20, 50, 80, 90, 5])
print(((data > 80) | (data < 20)).sum())
```

```python
# 8. Normalize to Sum 1  [AI screen / numpy]
# Return the array divided by its total, so the result sums to 1.
import numpy as np
a = np.array([10.0, 20.0, 70.0])
a / a.sum()
```

```python
# 9. Top 2 Per Group  [AI screen / pandas]
# For each dept, return the 2 highest-paid rows. Columns name, dept, salary.
import pandas as pd
emp = pd.DataFrame({
    "name": ["A","B","C","D","E"],
    "dept": ["IT","IT","IT","HR","HR"],
    "salary": [70000, 90000, 60000, 50000, 80000],
})

emp.groupby("dept", as_index=False).apply(lambda g: g.nlargest(2, "salary"))
```

```python
# 10. Pivot Count with Fill  [AI screen / pandas]
# Rows = city, columns = status, values = count of rows, missing filled with 0.
import pandas as pd
df = pd.DataFrame({
    "city": ["X","Y","X","X"],
    "status": ["open","open","closed","open"],
})
pd.pivot_table(data=df,index='city',columns='status',aggfunc='size',fill_value=0)
```
