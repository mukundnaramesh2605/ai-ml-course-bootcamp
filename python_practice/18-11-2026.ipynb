```python
# 1. Words Starting With a Vowel  [easy / string]
# Return HOW MANY words start with a vowel (a, e, i, o, u), case-insensitive.
# "Apple is orange" -> 3
def vowel_words(s):
    return sum([1 for i in s.lower().split() if i[0] in 'aeiou'])

print(vowel_words("Apple is orange"))       # -> 3
print(vowel_words(""))                      # -> 0
print(vowel_words("   "))                   # -> 0   <- whitespace only, no words
print(vowel_words("sky fly"))               # -> 0
print(vowel_words("Umbrella under tree"))   # -> 2
```

```python
# 2. Swap Keys and Values  [easy / dict]
# Return a NEW dict with value -> key. If a value repeats, keep the LAST key.
# {"a": 1, "b": 2} -> {1: 'a', 2: 'b'}
def swap_dict(d):
    return {v:k for k,v in d.items()}

print(swap_dict({"a": 1, "b": 2}))    # -> {1: 'a', 2: 'b'}
print(swap_dict({}))                  # -> {}
print(swap_dict({"a": 1, "b": 1}))    # -> {1: 'b'}   <- duplicate value, last key wins
print(swap_dict({"x": "y"}))          # -> {'y': 'x'}
print(swap_dict({"a": 0, "b": -1}))   # -> {0: 'a', -1: 'b'}
```

```python
# 3. Label Signs  [easy / logic]
# Return a list with "neg", "zero" or "pos" for each number. Use EXACTLY these strings.
# [-2, 0, 5] -> ['neg', 'zero', 'pos']
def label_signs(nums):
    return ['pos' if i > 0 else 'neg' if i != 0 else 'zero' for i in nums]

print(label_signs([-2, 0, 5]))     # -> ['neg', 'zero', 'pos']
print(label_signs([]))             # -> []
print(label_signs([0, 0]))         # -> ['zero', 'zero']   <- 0 is neither pos nor neg
print(label_signs([7]))            # -> ['pos']
print(label_signs([-1, -1, 3]))    # -> ['neg', 'neg', 'pos']
```

```python
# 4. Duplicate Values  [medium / Counter]
# Return a SORTED list of values that appear MORE than once. Each value listed once.
# [3, 1, 3, 2, 1] -> [1, 3]
def duplicates(nums):
    return sorted(set(i for i in nums if nums.count(i)>1))

print(duplicates([3, 1, 3, 2, 1]))   # -> [1, 3]
print(duplicates([]))                # -> []
print(duplicates([5]))               # -> []
print(duplicates([2, 2, 2]))         # -> [2]   <- appears 3 times, listed once
print(duplicates([-1, 1, -1, 0]))    # -> [-1]
```
