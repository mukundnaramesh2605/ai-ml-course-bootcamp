```python
nums = [10, 25, 7, 90, 56]
max1 = max2 = float('-inf')
for n in nums:
    if n > max1:
        max2 = max1  
        max1 = n     
    elif n > max2 and n != max1:
        max2 = n  
print(max2)
```

```python
employees = [
    {"name":"Alice","salary":50000},
    {"name":"Bob","salary":75000},
    {"name":"Charlie","salary":60000}
]
max_salary = max([i['salary'] for i in employees])
for i in employees:
    if i['salary'] == max_salary:
        print(i['name'])
```
