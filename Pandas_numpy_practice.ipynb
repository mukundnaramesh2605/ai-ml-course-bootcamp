```python
import pandas as pd
import numpy as np
```

```python
# ============================================================================
# 20 PANDAS Practice Problems: EDA, Filtering, Transformations, Groupby
# Easy to Medium Difficulty - Bootcamp + Interview Ready
# ============================================================================
# These directly apply the functional patterns you just mastered to Pandas
# ============================================================================


# ============================================================================
# BLOCK 1: Selection & Filtering (1-4)
# ============================================================================

# Problem 1: Filter rows by condition
# Create a DataFrame with: name, age, city, salary
# Filter: age > 25
# Expected output: 2 rows (Bob and Charlie)
def problem_1():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [23, 28, 30, 22],
        'city': ['NYC', 'LA', 'NYC', 'Chicago'],
        'salary': [50000, 60000, 75000, 45000]
    })
    # Your code here
    return df[df.age > 25]

# Test:
print("Problem 1:")
print(problem_1())
print()
```

```python
# Problem 2: Filter multiple conditions
# Same DataFrame
# Filter: age > 25 AND city == 'NYC'
# Expected: 1 row (Charlie)
def problem_2():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [23, 28, 30, 22],
        'city': ['NYC', 'LA', 'NYC', 'Chicago'],
        'salary': [50000, 60000, 75000, 45000]
    })
    # Your code here
    return df[(df.age > 25) & (df.city == "NYC")]

# Test:
print("Problem 2:")
print(problem_2())
print()
```

```python
# Problem 3: Select specific columns
# Same DataFrame
# Select: name, salary
# Expected: DataFrame with 4 rows, 2 columns
def problem_3():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [23, 28, 30, 22],
        'city': ['NYC', 'LA', 'NYC', 'Chicago'],
        'salary': [50000, 60000, 75000, 45000]
    })
    # Your code here
    return df[['name','salary']]

# Test:
print("Problem 3:")
print(problem_3())
print()
```

```python
# Problem 4: Filter using .isin()
# Same DataFrame
# Filter: city is either 'NYC' or 'Chicago'
# Expected: 3 rows (Alice, Charlie, David)
def problem_4():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [23, 28, 30, 22],
        'city': ['NYC', 'LA', 'NYC', 'Chicago'],
        'salary': [50000, 60000, 75000, 45000]
    })
    # Your code here
    return df[df.city.isin(['NYC','Chicago'])]

# Test:
print("Problem 4:")
print(problem_4())
print()
```

```python
# ============================================================================
# BLOCK 2: Transformations & Creating New Columns (5-8)
# ============================================================================

# Problem 5: Create new column with .apply()
# DataFrame: name, age, salary
# Create: 'age_group' column
#   - 'young' if age < 25
#   - 'mid' if 25 <= age < 30
#   - 'senior' if age >= 30
# Expected: new column with appropriate labels
def categorize_age(age):
    if age < 25:
        return 'young'
    elif age < 30:
        return 'mid'
    else:
        return 'senior'
def problem_5():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [23, 28, 30, 22],
        'salary': [50000, 60000, 75000, 45000]
    })
    # Your code here
    df["age_group"] = df.age.apply(categorize_age)
    return df

# Test:
print("Problem 5:")
print(problem_5())
print()
```

```python
# Problem 6: Create new column with .map()
# DataFrame: city
# Create: 'region' column mapping city to region
#   - 'NYC' -> 'Northeast'
#   - 'LA' -> 'West'
#   - 'Chicago' -> 'Midwest'
def problem_6():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'city': ['NYC', 'LA', 'NYC', 'Chicago']
    })
    city_to_region = {'NYC': 'Northeast', 'LA': 'West', 'Chicago': 'Midwest'}
    # Your code here
    df['region'] = df['city'].map(city_to_region)
    return df
 
# Test:
print("Problem 6:")
print(problem_6())
print()
```

```python
# Problem 7: Simple arithmetic transformation
# DataFrame: salary
# Create: 'salary_after_raise' = salary * 1.1 (10% raise)
def problem_7():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [50000, 60000, 75000]
    })
    # Your code here
    df['salary_after_raise'] = df['salary'] * 1.1
    return df
    

# Test:
print("Problem 7:")
print(problem_7())
print()
```

```python
# Problem 8: Apply function to multiple columns
# DataFrame: first_name, last_name
# Create: 'full_name' = first_name + ' ' + last_name
def problem_8():
    df = pd.DataFrame({
        'first_name': ['Alice', 'Bob', 'Charlie'],
        'last_name': ['Smith', 'Johnson', 'Brown']
    })
    # Your code here
    df['full_name'] = df['first_name'] + ' ' + df['last_name']
    return df

# Test:
print("Problem 8:")
print(problem_8())
print()
```

```python
# ============================================================================
# BLOCK 3: Groupby & Aggregation (9-12)
# ============================================================================

# Problem 9: Group by and count
# DataFrame: name, department, salary
# Group by department, count employees
# Expected: department with count
def problem_9():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [50000, 60000, 55000, 65000, 48000]
    })
    # Your code here
    return df.groupby('department')['department'].value_counts()

# Test:
print("Problem 9:")
print(problem_9())
print()
```

```python
# Problem 10: Group by and aggregate multiple columns
# DataFrame: name, department, salary
# Group by department
# Aggregate: count, mean salary, max salary
def problem_10():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [50000, 60000, 55000, 65000, 48000]
    })
    # Your code here
    result = df.groupby('department').agg(
        count=('salary', 'count'),
        mean_salary=('salary', 'mean'),
        max_salary=('salary', 'max')
    ).reset_index()
    
    return result

# Test:
print("Problem 10:")
print(problem_10())
print()
```

```python
# Problem 11: Group by multiple columns
# DataFrame: name, city, department, salary
# Group by city AND department
# Aggregate: average salary
def problem_11():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'city': ['NYC', 'NYC', 'LA', 'LA', 'NYC'],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [50000, 60000, 55000, 65000, 48000]
    })
    # Your code here
    return df.groupby(['city','department'])['salary'].mean()

# Test:
print("Problem 11:")
print(problem_11())
print()
```

```python
# Problem 12: Groupby then filter by aggregated value
# DataFrame: name, category, sales
# Group by category, get total sales
# Filter: categories with total sales > 100
def problem_12():
    df = pd.DataFrame({
        'name': ['A', 'B', 'C', 'D', 'E'],
        'category': ['X', 'Y', 'X', 'Y', 'Z'],
        'sales': [30, 40, 50, 60, 10]
    })
    # Your code here
    result = df.groupby('category').filter(lambda x: x['sales'].sum() >= 100)
    return result

# Test:
print("Problem 12:")
print(problem_12())
print()
```

```python
# ============================================================================
# BLOCK 4: Sorting & Ranking (13-15)
# ============================================================================

# Problem 13: Sort by single column
# DataFrame: name, age, salary
# Sort by salary (ascending)
def problem_13():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'age': [23, 28, 30, 22],
        'salary': [50000, 60000, 75000, 45000]
    })
    # Your code here
    result = df.sort_values(by='salary', ascending=True)
    return result

# Test:
print("Problem 13:")
print(problem_13())
print()
```

```python
# Problem 14: Sort by multiple columns
# DataFrame: name, department, salary
# Sort by: department (ascending), then salary (descending)
def problem_14():
    df = pd.DataFrame({
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [50000, 60000, 55000, 65000, 48000]
    })
    # Your code here
    result = df.sort_values(by=['department','salary'], ascending=[True,False])
    return result

# Test:
print("Problem 14:")
print(problem_14())
print()
```
