```python
def is_prime_simple(n: int) -> bool:
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
row,col = map(int,input().split())
lst =[]
arr = [list(map(int, input().split())) for _ in range(row)]
for i in range(row):
    for j in range(col):
        if is_prime_simple((i+j)):
            lst.append(arr[i][j])
print("YES" if is_prime_simple(sum(lst)) else "NO")
```

```python
n = int(input())
cities = [input().strip() for _ in range(n)]

# Initialize matrix: identity (diagonal = 1)
matrix = [[1 if i == j else 0 for j in range(n)] for i in range(n)]

m = int(input())
city_index = {c: i for i, c in enumerate(cities)}

for _ in range(m):
    a, b = input().split()
    i, j = city_index[a], city_index[b]
    matrix[i][j] = 1
    matrix[j][i] = 1   # undirected

# Output
print(*cities)
for i, city in enumerate(cities):
    print(city, *matrix[i])
```

```python
n = int(input())
arr = list(map(int, input().split()))

# Pair each value with its original index, sort by value to find target positions
indexed = sorted(range(n), key=lambda i: arr[i])

visited = [False] * n
swaps = 0

for i in range(n):
    if visited[i] or indexed[i] == i:
        continue
    # walk the cycle starting at i
    cycle_size = 0
    j = i
    while not visited[j]:
        visited[j] = True
        j = indexed[j]
        cycle_size += 1
    swaps += cycle_size - 1

print(swaps)
```

```python
n = int(input())
tokens = input().split()

months_31 = {'JANUARY', 'MARCH', 'MAY', 'JULY', 'AUGUST', 'OCTOBER', 'DECEMBER'}
month_num = {
    'JANUARY': 1, 'FEBRUARY': 2, 'MARCH': 3, 'APRIL': 4,
    'MAY': 5, 'JUNE': 6, 'JULY': 7, 'AUGUST': 8,
    'SEPTEMBER': 9, 'OCTOBER': 10, 'NOVEMBER': 11, 'DECEMBER': 12
}

result = []
for i in range(n):
    day = int(tokens[i*3])
    month = tokens[i*3 + 1].upper()
    year = int(tokens[i*3 + 2])
    dob = (year, month_num[month], day)
    cutoff = (1987, 7, 22)
    
    if dob <= cutoff and month in months_31:
        result.append(i + 1) 

if result:
    print(*result)
else:
    print(-1)
```
