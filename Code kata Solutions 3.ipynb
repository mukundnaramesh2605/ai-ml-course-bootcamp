```python
n = int(input())
arr = list(map(int, input().split()))

# Pair each value with its original index, sort by value to find target positions
indexed = sorted(range(n), key=lambda i: arr[i])

visited = [False] * n
swaps = 0

for i in range(n):
    if visited[i] or indexed[i] == i:
        continue
    # walk the cycle starting at i
    cycle_size = 0
    j = i
    while not visited[j]:
        visited[j] = True
        j = indexed[j]
        cycle_size += 1
    swaps += cycle_size - 1

print(swaps)
```

```python
656. Parenthesis Balance Check

Geekoin10
Easy
Topics
Attempted!
Problem Statement:
you are given a string made up of parenthesis only.Your task is to check whether parenthesis are balanced or not.If they are balanced print 1 else print 0


Input Description:
You are given a string ‘s’


Output Description:
Print 1 for balanced and 0 for imbalanced


Sample Input:
{({})}


Sample Output:
1


Explanation:
check whether the braces are balanced and print 1 or 0

```

```python
s = input().strip()
stack = []
mapping = {')': '(', '}': '{', ']': '['}
balanced = True

for char in s:
    if char in mapping:  
        if not stack or stack.pop() != mapping[char]:
            balanced = False
            break
    else:  
        stack.append(char)


if balanced and not stack:
    print("1")
else:
    print("0")
```

```python
def sum_proper_divisors(n):
    if n <= 1:
        return 0
    total = 1   # 1 is always a proper divisor for n > 1
    i = 2
    while i * i <= n:
        if n % i == 0:
            total += i
            if i != n // i:   # avoid counting sqrt twice for perfect squares
                total += n // i
        i += 1
    return total

n = int(input())
m = sum_proper_divisors(n)
if m != n and sum_proper_divisors(m) == n:
    print(m)
else:
    print(-1)
```
