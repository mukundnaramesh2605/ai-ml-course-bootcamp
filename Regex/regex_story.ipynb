Chapter 1: Decoding coordinates (\d and quantifiers)

The ship intercepts this transmission from a derelict probe:

SIGNAL LOG: probe-042 last position 4471.88 by -122.4194, fuel 7 percent

```python
import re

text = "probe - 042 last position 4471.88 by -122.4194, fuel 7 percent"
numbers = re.findall(r'-?\d+\.?\d*', text)
print(numbers)
# ['042', '4471.88', '-122.4194', '7']
```
