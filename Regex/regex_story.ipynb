```python
import re

text = "ALERT: hull breach on deck 7"
result = re.findall(r'\d', text)
print(result)
# 7
```

```python
import re

text = "ALERT: hull breach on deck 142 on cabin 304"
result = re.search(r'\d', text)
print(result.group())
result = re.search(r'\d+', text)
print(result.group())
```

```python
# matching letters and words, not just digits — \w

text = "Incoming transmission from CALLSIGN NovaHawk7"
result = re.search(r'\w+', text)
print(result.group())
# Incoming
```

```python
text = """
UNIT7 reporting in, fuel at 82
UNIT3 reporting in, fuel at 15
UNIT9 reporting in, fuel at 60"""
re.findall(r'UNIT\d+',text)
```

```python
#Want to level up one notch — pulling out the fuel numbers too, using \d+ again but on the "fuel at" part of each line?
re.findall(r'fuel at \d+', text)
```

```python
re.findall(r'fuel at (\d+)', text)
```

```python
# [('UNIT7', '82'), ('UNIT3', '15'), ('UNIT9', '60')]
text = """
UNIT7 reporting in, fuel at 82
UNIT3 reporting in, fuel at 15
UNIT9 reporting in, fuel at 60"""
re.findall(r'(UNIT\d+).*fuel at (\d+)',text)
```

```python
text = """
CARGO-A weight 1200
CARGO-B weight 340
CARGO-C weight 8990"""

"""Get back pairs like [('CARGO-A', '1200'), ('CARGO-B', '340'), ('CARGO-C', '8990')].

One wrinkle: the ID part isn't just digits this time — it's CARGO-A, CARGO-B, CARGO-C.
So \d+ alone won't grab the letter after the dash. You already have the tool for that — a wildcard for letters, not just digits. Which one was it?"""
re.findall(r'(CARGO-\w).*?(\d+)',text)
```

```python
# [('CARGO-A', '1200', 'Mars-7'), ('CARGO-B', '340', 'Titan-2'), ('CARGO-C', '8990', 'Ceres-1')]
text = """
CARGO-A weight 1200 destination Mars-7
CARGO-B weight 340 destination Titan-2
CARGO-C weight 8990 destination Ceres-1
"""
re.findall(r'(CARGO-\w).*?(\d+).*? destination ([\w-]+)', text)
re.findall(r'(CARGO-\w).*?(\d+).*?([\w]+-\d+)', text)
```
