```python
def show_sequence(n):
    lst = [str(x) for x in range(n+1)]
    return "+".join(lst)+" = "+str(sum(int(x) for x in lst))
show_sequence(6)
```

```python
def to_float_array(arr): 
    return [int(x) if "." not in x else float(x) for x in arr]
```

```python
def vowel_indices(word):
	return [x+1 for x in range(len(word)) if word[x] in "aeiouyAEIOUY"]
```

```python
def partlist(arr):
    lst = []
    for n in range(len(arr)-1):
        lst.append([" ".join(arr[:n+1])," ".join(arr[n+1:])])
    return lst
partlist(["az", "toto", "picaro", "zone", "kiwi"])
```
