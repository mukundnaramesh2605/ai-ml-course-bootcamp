```python
# outlier detection
# IQR
# Z score or the Standard Score
```

```python
var1 = np.random.normal(0,1,1000)
var2 = np.random.normal(5,2,100)
fig,ax = plt.subplots(1,2,figsize=(12,6))
ax[0].hist(var1,bins=10)
ax[1].hist(var2,bins=10)
plt.show()
```

```python
var3 = np.concat([var1,var2])
fig,ax = plt.subplots(1,1,figsize=(12,6))
ax.hist(var3,bins=10)
plt.show()
```

```python
plt.boxplot(var3)
plt.show()
```

```python
# IQR

q1 = np.percentile(var3,25)
q3 = np.percentile(var3,75)

iqr = q3-q1

print(q1,q3,iqr)
```

```python
min_threshold = q1 - 1.5*iqr
max_threshold = q3 + 1.5*iqr

print(min_threshold,max_threshold)
```

```python
data_clean = [i for i in var3 if i>min_threshold and i<max_threshold] 
data_clean2 = list(filter(lambda i: i>min_threshold and i<max_threshold,var3))

outliers = [i for i in var3 if i<min_threshold or i>max_threshold]


plt.hist(data_clean2,bins = 20)
plt.show()
```

```python
# Standard Score method
# std_score = (x - mean) / std dev
# abs(stdscore) < 1.5 -> then its clean data

mean = np.mean(var3)
std = np.std(var3)

print(mean,std)
```

```python
data_clean_std_score = [x for x in var3 if (x-mean)/std < 1.5]
outliers_std_score = [x for x in var3 if (x-mean)/std > 1.5]

plt.hist(data_clean_std_score,bins = 20)
plt.show()
```

```python
# min max normalisation
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


var1 = np.random.normal(0,1,1000)
var2 = np.random.normal(5,2,1000)
var3 = np.concatenate([var1,var2])
plt.hist(var3,bins=20)
plt.show()
```

```python
# min max normalisation
logvar3_minmax = (var3 - var3.min()) / (var3.max()-var3.min())
plt.hist(logvar3_minmax,bins=20)
plt.show()
```

```python
var1 = np.random.uniform(2,3,1000)
var2 = np.random.uniform(4,5,100)

var3 = np.concatenate([var1,var2])

plt.hist(var3,bins = 20)
plt.show()
```

```python
# Log Tranformation

var3_log = np.log(var3)

plt.hist(var3_log,bins = 20)
plt.show()
```

```python
var3_min_max = (var3 - var3.min())/ (var3.max() - var3.min())

plt.hist(var3_min_max,bins = 20)
plt.show()
```

```python
# Standardization : Z Score Transformation

mean = np.mean(var3)
std = np.std(var3)

var3_std = (var3 - mean)/std

plt.hist(var3_std,bins = 20)
plt.show()
```

```python
movies_meta_data = pd.read_csv('movies_metadata.csv',low_memory=False)
movies_meta_data.head()
```

```python
def display_null_entries(df):
  df_null = movies_meta_data.isnull().sum().sort_values(ascending = False)
  print(df_null)


display_null_entries(movies_meta_data)
```

```python
movie_df = movies_meta_data.drop(columns = ["belongs_to_collection","homepage","tagline","runtime","poster_path"])


display_null_entries(movie_df)
```

```python
movie_df["budget"] = pd.to_numeric(movie_df["budget"],errors = "coerce")
movie_df["popularity"] = pd.to_numeric(movie_df["popularity"],errors = "coerce")
movie_df["release_date"] = pd.to_datetime(movie_df["release_date"],errors = "coerce")
movie_df["revenue"] = pd.to_numeric(movie_df["revenue"],errors = "coerce")
movie_df["vote_count"] = pd.to_numeric(movie_df["vote_count"],errors = "coerce")
movie_df["vote_average"] = pd.to_numeric(movie_df["vote_average"],errors = "coerce")
movie_df["id"] = pd.to_numeric(movie_df["id"],errors = "coerce")

movie_df.info()
```

```python
import ast
```

```python
def parse_cell(cell):
    try:
        data = ast.literal_eval(cell)
        if isinstance(data,list):
            return [item['name'] for item in data]
        else:
            return []
    except Exception:
        return []
```

```python
movie_df['genres'] = movie_df['genres'].apply(parse_cell)
movie_df['genres'].head()
```

```python
clean_cols = ["spoken_languages","production_countries","production_companies"]

for col in clean_cols:
  movie_df[col] = movie_df[col].apply(parse_cell)

movie_df.head()
```

```python
import re
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
url_pattern = r"https?://[^\s]+|www\.[^\s]+"
url = re.compile(url_pattern)
email = re.compile(email_pattern)
```

```python
#Removal of stop words
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

#nltk.download('stopwords')
#nltk.download('punkt')
stop_words = set(stopwords.words("english"))
```

```python
def tex_clean(cell):
    if not isinstance(cell,str):
        return str
    cell = email.sub(" ",cell)
    cell = url.sub(" ",cell)
    cell=cell.lower()
    cell = re.sub(r'\d'," ",cell)
    cell = re.sub(r"[^\w\s]","",cell)

    cell = [word for word in cell.split() if word not in stop_words]
    cell = " ".join(cell)
    return cell
movie_df['overview'] = movie_df['overview'].apply(tex_clean)
```
