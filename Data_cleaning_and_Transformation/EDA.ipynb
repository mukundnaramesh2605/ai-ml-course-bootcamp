```python
import pandas as pd
import numpy as np
```

```python
weather_data = {
    "day" : ["1/1/2017","1/2/2017","1/3/2017","1/4/2017"]*3,
    "city" : ["Mysore"]*4 + ["Delhi"]*4 + ["Mumbai"]*4,
    "temperature" : [32,None,28,None,32,None,None,34,40,None,None,25],
    "windspeed" : [6,7,None,7,None,None,None,8,None,None,8,8],
    "event" : ["Rain",None]*2 + [None]*2 + ["Snow"]*2+["Rain","Sunny"]*2
}

df = pd.DataFrame(weather_data)
df
```

```python
df1=df.dropna()
df1
```

```python
#fill a  specific value on all columns
df2 = df.fillna('######')
df2
```

```python
#Fill Column Specific Values:
df3 = df.fillna({'temperature':25,'windspeed':'6.5','event':"No Event"})
df3
```

```python
# forward fill
df4=df.ffill()
df4
```

```python
#backward fill
df5 = df.bfill()
df5
```
