```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

# Load data
X, y = load_iris(return_X_y=True)
feature_names = load_iris().feature_names
class_names = load_iris().target_names

# Train
rf = RandomForestClassifier(n_estimators=100, max_depth=3, random_state=42)
rf.fit(X, y)

# 1. Visualize one tree from the forest
plt.figure(figsize=(16, 8))
plot_tree(rf.estimators_[0],
          feature_names=feature_names,
          class_names=class_names,
          filled=True, rounded=True)
plt.title("Single tree from the Random Forest")
plt.show()

# 2. Visualize a few trees side by side
fig, axes = plt.subplots(1, 3, figsize=(24, 8))
for i in range(3):
    plot_tree(rf.estimators_[i],
              feature_names=feature_names,
              class_names=class_names,
              filled=True, rounded=True, ax=axes[i])
    axes[i].set_title(f"Tree {i}")
plt.show()

# 3. Feature importance
importances = rf.feature_importances_
plt.figure(figsize=(8, 4))
plt.barh(feature_names, importances)
plt.xlabel("Importance")
plt.title("Feature Importance")
plt.tight_layout()
plt.show()
```
