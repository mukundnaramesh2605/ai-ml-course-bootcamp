```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder,MultiLabelBinarizer
```

```python
df = pd.DataFrame({
    'listing_id':  ['L1', 'L2', 'L3', 'L4', 'L5', 'L6', 'L7', 'L8', 'L9', 'L10'],
    'property':    ['Studio', 'Villa', 'Apartment', 'Studio', 'Villa',
                    'Apartment', 'Penthouse', 'Studio', 'Apartment', 'Villa'],
    'furnishing':  ['Unfurnished', 'Semi', 'Full', 'Semi', 'Full',
                    'Unfurnished', 'Full', 'Full', 'Semi', 'Unfurnished'],
    'city':        ['Chennai', 'Chennai', 'Bangalore', 'Mumbai', 'Bangalore',
                    'Chennai', 'Mumbai', 'Chennai', 'Bangalore', 'Mumbai'],
    'verified':    ['Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No'],
    'rent':        [18000, 95000, 32000, 15500, 88000, 27000, 250000, 16000, 30000, 78000],
    'area_sqft':   [450, 3200, 1100, 400, 2900, 950, 5500, 480, 1050, 2700],
    'age_yrs':     [2, 15, 5, 1, 12, 8, 3, 2, 6, 20]
})
```

```python
#Ordinal encode furnishing with order Unfurnished < Semi < Full, using .map().
unique_furnishing = 'Unfurnished < Semi < Full'.split(' < ')
furnishing_mapping = {k:i for i,k in enumerate(unique_furnishing)}
df['furnishing'] = df['furnishing'].map(furnishing_mapping)
df
```

```python
#One-hot city with drop_first=True. Drop the original column after.
df = pd.get_dummies(df,columns=['city'],drop_first=True,dtype=int)
df
```

```python
#Label encode verified. Print the mapping.
lbl = LabelEncoder()
df['verified'] = lbl.fit_transform(df['verified'])
verified_mapping = dict(zip(lbl.classes_, range(len(lbl.classes_))))
verified_mapping
```
