Problem Statement:
Given a number N followed by N elements for every 2 consecutive numbers print the maximum of the 2.


Input Description:
The input consists of an integer N, followed by N elements. N is an integer such that N <= 100000, implying an O(n) time complexity solution is expected.


Output Description:
The output is a space-separated sequence of the maximums of every two consecutive numbers from the input.


Sample Input:

5

1 1 3 0 5


Sample Output:

1 3 3 5

```python
n = int(input())
grid = [list(map(int, input().split())) for _ in range(n)]
count = 0
for r in range(n):
    for c in range(n):
        if grid[r][c] == 1:
            isolated = True
            for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
                nr, nc = r + dr, c + dc
                if 0 <= nr < n and 0 <= nc < n and grid[nr][nc] == 1:
                    isolated = False
                    break
            if isolated:
                count += 1

print(count if count > 0 else -1)
```

```python
from collections import defaultdict, deque

n, m = map(int, input().split())
graph_edges = [tuple(map(int, input().split())) for _ in range(m)]

if m != n - 1:
    print("no")
else:
    adj = defaultdict(list)
    for u, v in graph_edges:
        adj[u].append(v)
        adj[v].append(u)

    visited = set()
    queue = deque([1])
    visited.add(1)
    while queue:
        node = queue.popleft()
        for neighbor in adj[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    print("yes" if len(visited) == n else "no")
```
