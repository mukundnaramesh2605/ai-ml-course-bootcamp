```python
import pandas as pd
import numpy as np
 
```

```python
# ============================================================================
# BLOCK 1: EASY PROBLEMS (1-5)
# ============================================================================
 
# Problem 1: Access single value from DataFrame
# DataFrame: id, name, salary (id as index)
# Get salary of id 102
# Expected: scalar value
def easy_1():
    df = pd.DataFrame({
        'id': [101, 102, 103],
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [50000, 60000, 75000]
    })
    df = df.set_index('id')
    # Your code here
    return df.loc[102]
 
# Test:
print("Easy Problem 1:")
print(easy_1())
print()
 
```

```python
# Problem 2: Add a new column with calculation
# DataFrame: id, cost, tax_rate
# Create 'total_cost' = cost + (cost * tax_rate)
# Expected: DataFrame with new column
def easy_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'cost': [100, 200, 150, 300],
        'tax_rate': [0.1, 0.1, 0.15, 0.15]
    })
    # Your code here
    df['total_cost'] = df.cost * df.tax_rate
    return df
 
# Test:
print("Easy Problem 2:")
print(easy_2())
print()
```

```python
# Problem 3: Basic groupby with single aggregation
# DataFrame: id, department, salary
# Group by department, get average salary
# Expected: Series with avg salary per department
def easy_3():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [50000, 65000, 60000, 70000, 55000]
    })
    # Your code here
    return df.groupby('department')['salary'].mean()
 
# Test:
print("Easy Problem 3:")
print(easy_3())
print()
 
```

```python
# Problem 4: Filter with NOT condition
# DataFrame: id, status, amount
# Filter: status != 'cancelled'
# Expected: all rows except cancelled
def easy_4():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'status': ['active', 'cancelled', 'active', 'pending', 'cancelled'],
        'amount': [100, 200, 150, 300, 250]
    })
    # Your code here
    return df[df.status != 'cancelled']
 
# Test:
print("Easy Problem 4:")
print(easy_4())
print()
```
