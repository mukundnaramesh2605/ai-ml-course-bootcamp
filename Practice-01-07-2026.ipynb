```python
import pandas as pd
import numpy as np
 
```

```python
# ============================================================================
# BLOCK 1: EASY PROBLEMS (1-5)
# ============================================================================
 
# Problem 1: Access single value from DataFrame
# DataFrame: id, name, salary (id as index)
# Get salary of id 102
# Expected: scalar value
def easy_1():
    df = pd.DataFrame({
        'id': [101, 102, 103],
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [50000, 60000, 75000]
    })
    df = df.set_index('id')
    # Your code here
    return df.loc[102]
 
# Test:
print("Easy Problem 1:")
print(easy_1())
print()
 
```

```python
# Problem 2: Add a new column with calculation
# DataFrame: id, cost, tax_rate
# Create 'total_cost' = cost + (cost * tax_rate)
# Expected: DataFrame with new column
def easy_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'cost': [100, 200, 150, 300],
        'tax_rate': [0.1, 0.1, 0.15, 0.15]
    })
    # Your code here
    df['total_cost'] = df.cost * df.tax_rate
    return df
 
# Test:
print("Easy Problem 2:")
print(easy_2())
print()
```

```python
# Problem 3: Basic groupby with single aggregation
# DataFrame: id, department, salary
# Group by department, get average salary
# Expected: Series with avg salary per department
def easy_3():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'department': ['Sales', 'IT', 'Sales', 'IT', 'HR'],
        'salary': [50000, 65000, 60000, 70000, 55000]
    })
    # Your code here
    return df.groupby('department')['salary'].mean()
 
# Test:
print("Easy Problem 3:")
print(easy_3())
print()
 
```

```python
# Problem 4: Filter with NOT condition
# DataFrame: id, status, amount
# Filter: status != 'cancelled'
# Expected: all rows except cancelled
def easy_4():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'status': ['active', 'cancelled', 'active', 'pending', 'cancelled'],
        'amount': [100, 200, 150, 300, 250]
    })
    # Your code here
    return df[df.status != 'cancelled']
 
# Test:
print("Easy Problem 4:")
print(easy_4())
print()
```

```python
# Problem 5: Get row count and column count
# DataFrame: id, name, email, city
# Return tuple: (number of rows, number of columns)
# Expected: (4, 4) or similar
def easy_5():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'email': ['a@test.com', 'b@test.com', 'c@test.com', 'd@test.com'],
        'city': ['NYC', 'LA', 'Chicago', 'Boston']
    })
    # Your code here
    return df.shape
 
# Test:
print("Easy Problem 5:")
print(easy_5())
print()
 
```

```python
# Problem 1: Filter with multiple OR conditions
# DataFrame: id, category, price
# Filter: category == 'A' OR category == 'C' (use .isin())
# Expected: rows with category A or C
def medium_1():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5, 6],
        'category': ['A', 'B', 'C', 'A', 'B', 'C'],
        'price': [100, 150, 200, 120, 180, 160]
    })
    # Your code here
    return df[df.category.isin(['A','C'])]
 
# Test:
print("Medium Problem 1:")
print(medium_1())
print()
```

```python
# Problem 2: Groupby with multiple aggregations using .agg()
# DataFrame: id, team, score, games
# Group by team
# Aggregate: avg score, sum games, count records
# Expected: DataFrame with 3 aggregated columns
def medium_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5, 6],
        'team': ['A', 'B', 'A', 'B', 'A', 'B'],
        'score': [85, 92, 88, 79, 95, 88],
        'games': [5, 4, 5, 3, 4, 5]
    })
    # Your code here
    return df.groupby('team').agg(
        avg_score = ('score','mean'),
        sum_games = ('games','sum'),
        count_records = ('team','count')
    )
 
# Test:
print("Medium Problem 2:")
print(medium_2())
print()
```
