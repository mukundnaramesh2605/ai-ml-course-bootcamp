```python
# ============================================================
# Problem 1: Find Floor of a Number
# Difficulty: Easy
#
# Given a sorted array and a target,
# return the largest element that is
# less than or equal to the target.
#
# If no such element exists,
# return -1.
#
# Example:
# nums = [2,4,6,8,10]
# target = 7
# Output = 6
#
# Example:
# nums = [2,4,6,8,10]
# target = 1
# Output = -1
# ============================================================

def find_floor(nums, target):
    left, right = 0, len(nums) - 1
    ans = -1

    while left <= right:
        mid = left + (right - left) // 2

        if nums[mid] == target:
            return nums[mid]

        elif nums[mid] < target:
            ans = nums[mid]
            left = mid + 1

        else:
            right = mid - 1

    return ans
print(find_floor([2,4,6,8,10],7))
```

```python
# ============================================================
# Problem 2: Find Ceiling of a Number
# Difficulty: Easy
#
# Given a sorted array and a target,
# return the smallest element that is
# greater than or equal to the target.
#
# If no such element exists,
# return -1.
#
# Example:
# nums = [2,4,6,8,10]
# target = 7
# Output = 8
#
# Example:
# nums = [2,4,6,8,10]
# target = 11
# Output = -1
# ============================================================

def find_ceiling(nums, target):
    left, right = 0, len(nums) - 1
    ans = -1

    while left <= right:
        mid = left + (right - left) // 2

        if nums[mid] == target:
            return nums[mid]

        elif nums[mid] < target:
            left = mid + 1

        else:
            ans = nums[mid]
            right = mid - 1

    return ans
print(find_ceiling([2,4,6,8,10],7))   
```
