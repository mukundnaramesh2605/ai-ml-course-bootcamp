```python
# ============================================================
# Problem 1: Find Floor of a Number
# Difficulty: Easy
#
# Given a sorted array and a target,
# return the largest element that is
# less than or equal to the target.
#
# If no such element exists,
# return -1.
#
# Example:
# nums = [2,4,6,8,10]
# target = 7
# Output = 6
#
# Example:
# nums = [2,4,6,8,10]
# target = 1
# Output = -1
# ============================================================

def find_floor(nums, target):
    left, right = 0, len(nums) - 1
    ans = -1

    while left <= right:
        mid = left + (right - left) // 2

        if nums[mid] == target:
            return nums[mid]

        elif nums[mid] < target:
            ans = nums[mid]
            left = mid + 1

        else:
            right = mid - 1

    return ans
print(find_floor([2,4,6,8,10],7))
```

```python
# ============================================================
# Problem 2: Find Ceiling of a Number
# Difficulty: Easy
#
# Given a sorted array and a target,
# return the smallest element that is
# greater than or equal to the target.
#
# If no such element exists,
# return -1.
#
# Example:
# nums = [2,4,6,8,10]
# target = 7
# Output = 8
#
# Example:
# nums = [2,4,6,8,10]
# target = 11
# Output = -1
# ============================================================

def find_ceiling(nums, target):
    left, right = 0, len(nums) - 1
    ans = -1

    while left <= right:
        mid = left + (right - left) // 2

        if nums[mid] == target:
            return nums[mid]

        elif nums[mid] < target:
            left = mid + 1

        else:
            ans = nums[mid]
            right = mid - 1

    return ans
print(find_ceiling([2,4,6,8,10],7))   
```

```python
# ============================================================
# Problem 3: Count Occurrences
# Difficulty: Easy
#
# Given a sorted array and a target,
# return how many times the target appears.
#
# Example:
# nums = [1,2,2,2,3,4]
# target = 2
# Output = 3
#
# Example:
# nums = [1,2,3,4]
# target = 5
# Output = 0
#
# Hint:
# Use two binary searches.
# ============================================================

def count_occurrences(nums, target):
    left, right = 0, len(nums) - 1
    ans = 0
    while left <= right:
        mid = left + (right - left) // 2

        if nums[mid] == target:
            return nums[mid]
            left = mid + 1
        elif nums[mid] < target:
            left = mid + 1

        else:
            ans = nums[mid]
            right = mid - 1

    return ans
```

```python
def count_occurrences(nums, target):

    def first():
        left, right = 0, len(nums)-1
        ans = -1

        while left <= right:
            mid = left + (right-left)//2

            if nums[mid] == target:
                ans = mid
                right = mid - 1

            elif nums[mid] < target:
                left = mid + 1

            else:
                right = mid - 1

        return ans

    def last():
        left, right = 0, len(nums)-1
        ans = -1

        while left <= right:
            mid = left + (right-left)//2

            if nums[mid] == target:
                ans = mid
                left = mid + 1

            elif nums[mid] < target:
                left = mid + 1

            else:
                right = mid - 1

        return ans

    f = first()

    if f == -1:
        return 0

    l = last()

    return l - f + 1
```

```python
"""
Given a sorted array and a target,
return True if the target exists,
otherwise return False.

Example:

nums = [2,4,6,8,10]
target = 6

Output:
True

Example:

nums = [2,4,6,8,10]
target = 5

Output:
False
"""

def contains_(nums, target):
    left, right = 0, len(nums)-1

    while left <= right:
        mid = left+(right-left) //2
        if nums[mid] == target:
            return True
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return False
print(contains_([2,4,6,8,10],5))
```

```python
"""
Return the INDEX of the first element
strictly greater than the target.

If no such element exists,
return -1.

Example 1:
nums = [2,4,6,8,10]
target = 6

Output:
3

because nums[3] = 8


Example 2:
nums = [2,4,6,8,10]
target = 10

Output:
-1
"""

def first_greater(nums, target):
    left, right = 0, len(nums)-1
    ans =-1

    while left <= right:
        mid = left+(right-left) //2
        if nums[mid] < target:
            left = mid+1
        else:
            ans = mid
            right = mid - 1
    return ans
first_greater([2,4,6,8,10],3)
```

```python
"""
Return the INDEX of the first element
greater than or equal to the target.

If no such element exists,
return -1.

Example 1:
nums = [2,4,6,8,10]
target = 5

Output:
2

because nums[2] = 6


Example 2:
nums = [2,4,6,8,10]
target = 6

Output:
2
"""
def lower_bound(nums, target):
    left, right = 0, len(nums)-1
    ans =-1

    while left <= right:
        mid = left+(right-left) //2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid+1
        else:
            ans = mid
            right = mid - 1
    return ans
lower_bound([2,4,6,8,10],11)
```

```python
"""
Return the INDEX of the last element
less than or equal to the target.

If no such element exists,
return -1.

Example 1:
nums = [2,4,6,8,10]
target = 6

Output:
2

Example 2:
nums = [2,4,6,8,10]
target = 7

Output:
2

Example 3:
nums = [2,4,6,8,10]
target = 1

Output:
-1
"""

def upper_bound(nums, target):
    left, right = 0, len(nums)-1
    ans =-1

    while left <= right:
        mid = left+(right-left) //2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            ans = mid
            left = mid+1
        else:
            right = mid - 1
    return ans
upper_bound([2,4,6,8,10],7)
```
