```python
# Bonus: Extract nested values with default handling
# Use: map() + filter() or list comprehension
# Input: [{'name': 'Alice', 'age': 25}, {'name': 'Bob'}, {'name': 'Charlie', 'age': 30}]
# Extract (name, age) for records that have 'age' key AND age > 26
# Expected: [('Charlie', 30)]
def _is_age_available(record):
    if len(record)>1:
        if record["age"] > 26:
            return True
    else:
        return False
def extract_with_key_check(records):
    return list(map(lambda x: (x["name"],x["age"]),filter(lambda y: y.get('age',0)>26,records)))

# Test:
print("Bonus:", extract_with_key_check([{'name': 'Alice', 'age': 25}, {'name': 'Bob'}, {'name': 'Charlie', 'age': 30}]))
```

```python
# Problem 1: Nested transformation with filter
# Input: list of dicts with user data
# Filter: users with age >= 25 AND has_email=True
# Map: extract (name, email, age_group) where age_group = '25-35' or '35+'
# Expected: list of tuples with filtered/transformed data
def problem_1():
    users = [
        {'name': 'Alice', 'age': 28, 'email': 'alice@test.com', 'has_email': True},
        {'name': 'Bob', 'age': 22, 'email': None, 'has_email': False},
        {'name': 'Charlie', 'age': 35, 'email': 'charlie@test.com', 'has_email': True},
        {'name': 'David', 'age': 40, 'email': 'david@test.com', 'has_email': True},
        {'name': 'Eve', 'age': 24, 'email': 'eve@test.com', 'has_email': True}
    ]
    # Your code here
    filtered = list(map(lambda x:(x['name'],x['email'], "35+" if x['age'] > 35 else '25-35'),filter(lambda x : (x['age'] > 25) & (x['has_email']),users)))
    return filtered
 
# Test:
print("Problem 1:")
print(problem_1())
print()
```

```python
 
# Problem 2: Zip + filter + map pattern
# Input: names=['Alice', 'Bob', 'Charlie', 'David']
#        scores=[85, 72, 92, 68]
# Filter: pairs where score >= 75
# Map: create (name, grade) where grade = 'A' if score >= 90, else 'B'
# Expected: [('Alice', 'B'), ('Charlie', 'A'), ('David', 'B')] or similar
def problem_2():
    names = ['Alice', 'Bob', 'Charlie', 'David']
    scores = [85, 72, 92, 68]
    # Your code here
    return list(map(lambda x:(x[0],"A" if x[1]>=90 else "B"),zip(names,scores)))
 
# Test:
print("Problem 2:")
print(problem_2())
print()
 
```
