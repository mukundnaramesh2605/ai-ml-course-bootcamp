```python
# ============================================================================
# 20 ADVANCED Practice Problems: map(), filter(), list comprehensions + lambda
# Medium to Hard Difficulty
# ============================================================================
# These problems combine multiple concepts and real-world patterns.
# ============================================================================

# BLOCK 1: Lambda + map with multiple operations (1-4)
# ============================================================================

# Problem 1: Extract and transform: get word and its reverse as tuple
# Use: map() with lambda returning tuple
# Input: ['hello', 'world', 'python']
# Expected: [('hello', 'olleh'), ('world', 'dlrow'), ('python', 'nohtyp')]
def word_and_reverse(words):
    return list(map(lambda x:(x,x[::-1]),words))

# Test:
print("Problem 1:", word_and_reverse(['hello', 'world', 'python']))
```

```python
# Problem 2: Conditional transformation using lambda (ternary in lambda)
# Use: map() with lambda using if-else
# Input: [1, 2, 3, 4, 5, 6]
# Expected: ['odd', 'even', 'odd', 'even', 'odd', 'even']
def odd_even_label(nums):
    return list(map(lambda x: "odd" if x%2 !=0 else "even",nums))

# Test:
print("Problem 2:", odd_even_label([1, 2, 3, 4, 5, 6]))
```

```python
# Problem 3: Apply discount based on price tier
# Use: map() with lambda and conditional logic
# Input: [50, 150, 200, 75]
# Rule: if price > 100, apply 10% discount; else 5%
# Expected: [47.5, 135.0, 180.0, 71.25]
def apply_discount(prices):
    return list(map(lambda x: x*0.9 if x>100 else x*0.95,prices))

# Test:
print("Problem 3:", apply_discount([50, 150, 200, 75]))
```

```python
# Problem 4: Create key-value pairs with transformed values
# Use: map() with lambda returning dict or tuple
# Input: ['a', 'b', 'c']
# Expected: [('a', 0), ('b', 1), ('c', 2)]  (char with index)
def char_with_index(chars):
    return list(map(lambda x: (x,chars.index(x)),chars))

# Test:
print("Problem 4:", char_with_index(['a', 'b', 'c']))
```

```python
# ============================================================================
# BLOCK 2: Filter with complex conditions (5-8)
# ============================================================================

# Problem 5: Filter numbers divisible by 3 AND greater than 5
# Use: filter() with lambda and multiple conditions
# Input: [3, 6, 9, 12, 15, 2, 4, 8]
# Expected: [6, 9, 12, 15]
def filter_div3_gt5(nums):
    return list(filter(lambda x: (x%3 ==0) & (x>5),nums))

# Test:
print("Problem 5:", filter_div3_gt5([3, 6, 9, 12, 15, 2, 4, 8]))
```

```python
# Problem 6: Filter strings that contain at least one vowel AND are longer than 2 chars
# Use: filter() with lambda and multiple conditions
# Input: ['a', 'bc', 'aei', 'bcd', 'xyz', 'aoe']
# Expected: ['aei', 'aoe']
def filter_vowel_strings(words):
    return list(filter(lambda x: (("a" in x.lower()) | ("e" in x.lower()) | ("i" in x.lower()) | ("o" in x.lower()) | ("u" in x.lower())) & (len(x)>2),words))

# Test:
print("Problem 6:", filter_vowel_strings(['a', 'bc', 'aei', 'bcd', 'xyz', 'aoe']))
```

```python
# Problem 7: Filter tuples where second element > first element
# Use: filter() with lambda on tuples
# Input: [(1, 5), (3, 2), (4, 4), (2, 8)]
# Expected: [(1, 5), (2, 8)]
def filter_tuple_compare(tuples):
    return list(filter(lambda x: x[1]>x[0],tuples))

# Test:
print("Problem 7:", filter_tuple_compare([(1, 5), (3, 2), (4, 4), (2, 8)]))
```

```python
# Problem 8: Filter dictionary items where value is string AND starts with uppercase
# Use: filter() with dict.items() and multiple conditions
# Input: {'name': 'Alice', 'city': 'boston', 'age': 25, 'country': 'India'}
# Expected: [('name', 'Alice'), ('country', 'India')]
def filter_items(items):
    return True if isinstance(items[1],str) and items[1][0].isupper() else False
    
def filter_dict_uppercase_strings(d):
    return list(map(lambda x: (x[0],x[1]),filter(filter_items,[(key,value) for key,value in d.items()])))

# Test:
print("Problem 8:", filter_dict_uppercase_strings({'name': 'Alice', 'city': 'boston', 'age': 25, 'country': 'India'}))
```

```python
# ============================================================================
# BLOCK 3: Nested map/filter (9-12)
# ============================================================================

# Problem 9: Map then filter: square numbers, keep only those between 10 and 50
# Use: map() then filter()
# Input: [2, 3, 4, 5, 6, 7, 8]
# Expected: [16, 25, 36, 49]
def square_between_10_50(nums):
    return list(filter(lambda x: x in range(10,50),map(lambda y: y**2 ,nums)))

# Test:
print("Problem 9:", square_between_10_50([2, 3, 4, 5, 6, 7, 8]))
```
