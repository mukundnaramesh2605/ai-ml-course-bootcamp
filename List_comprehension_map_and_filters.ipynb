```python
# ============================================================================
# 20 ADVANCED Practice Problems: map(), filter(), list comprehensions + lambda
# Medium to Hard Difficulty
# ============================================================================
# These problems combine multiple concepts and real-world patterns.
# ============================================================================

# BLOCK 1: Lambda + map with multiple operations (1-4)
# ============================================================================

# Problem 1: Extract and transform: get word and its reverse as tuple
# Use: map() with lambda returning tuple
# Input: ['hello', 'world', 'python']
# Expected: [('hello', 'olleh'), ('world', 'dlrow'), ('python', 'nohtyp')]
def word_and_reverse(words):
    return list(map(lambda x:(x,x[::-1]),words))

# Test:
print("Problem 1:", word_and_reverse(['hello', 'world', 'python']))
```

```python
# Problem 2: Conditional transformation using lambda (ternary in lambda)
# Use: map() with lambda using if-else
# Input: [1, 2, 3, 4, 5, 6]
# Expected: ['odd', 'even', 'odd', 'even', 'odd', 'even']
def odd_even_label(nums):
    return list(map(lambda x: "odd" if x%2 !=0 else "even",nums))

# Test:
print("Problem 2:", odd_even_label([1, 2, 3, 4, 5, 6]))
```

```python
# Problem 3: Apply discount based on price tier
# Use: map() with lambda and conditional logic
# Input: [50, 150, 200, 75]
# Rule: if price > 100, apply 10% discount; else 5%
# Expected: [47.5, 135.0, 180.0, 71.25]
def apply_discount(prices):
    return list(map(lambda x: x*0.9 if x>100 else x*0.95,prices))

# Test:
print("Problem 3:", apply_discount([50, 150, 200, 75]))
```

```python
# Problem 4: Create key-value pairs with transformed values
# Use: map() with lambda returning dict or tuple
# Input: ['a', 'b', 'c']
# Expected: [('a', 0), ('b', 1), ('c', 2)]  (char with index)
def char_with_index(chars):
    return list(map(lambda x: (x,chars.index(x)),chars))

# Test:
print("Problem 4:", char_with_index(['a', 'b', 'c']))
```
