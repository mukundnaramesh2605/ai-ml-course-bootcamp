```python
import pandas as pd
import numpy as np
```

```python
# ============================================================================
# BLOCK 1: EASY PROBLEMS (1-10)
# ============================================================================

# Problem 1: Convert column to different data type
# DataFrame: id, salary_str (strings like '50000')
# Convert salary_str to integer
# Expected: integer dtype
def easy_1():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'salary_str': ['50000', '60000', '75000', '55000']
    })
    # Your code here
    df['salary_str'] = pd.to_numeric(df.salary_str)
    return df

# Test:
print("Easy Problem 1:")
print(easy_1())
print()
```

```python
# Problem 2: Reset index after filtering
# DataFrame: id, value (after filtering, index is [0, 2, 4])
# Reset index to 0, 1, 2
# Expected: DataFrame with fresh index
def easy_2():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5],
        'value': [10, 20, 30, 40, 50]
    })
    df_filtered = df[df['value'] > 20].reset_index(drop='index')  # index becomes [2, 3, 4]
    return df_filtered

# Test:
print("Easy Problem 2:")
print(easy_2())
print()
```

```python
# Problem 3: Set index to a column
# DataFrame: id, name, salary
# Set 'id' as index
# Expected: DataFrame with id as index
def easy_3():
    df = pd.DataFrame({
        'id': [101, 102, 103],
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [50000, 60000, 75000]
    })
    # Your code here
    df = df.set_index('id')
    return df

# Test:
print("Easy Problem 3:")
print(easy_3())
print()
```

```python
# Problem 4: Get value at specific index
# DataFrame: id, name, salary (with id as index)
# Get salary of id 102
# Expected: scalar value 60000
def easy_4():
    df = pd.DataFrame({
        'id': [101, 102, 103],
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [50000, 60000, 75000]
    })
    df = df.set_index('id')
    # Your code here - get salary for id 102
    return df.salary.loc[102]

# Test:
print("Easy Problem 4:")
print(easy_4())
print()
```

```python
# Problem 5: Apply lambda to create new column
# DataFrame: id, price, quantity
# Create 'total_price' = price * quantity using apply/lambda
# Expected: new column with values [100, 300, 200]
def easy_5():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'price': [10, 15, 20],
        'quantity': [10, 20, 10]
    })
    # Your code here
    df['total_price'] = df.apply(lambda row: row['price'] * row['quantity'], axis=1)
    return df["total_price"]

# Test:
print("Easy Problem 5:")
print(easy_5())
print()
```

```python
# Problem 6: Get rows with max value in a column
# DataFrame: id, name, score
# Get row with maximum score
# Expected: 1 row (the one with score=95)
def easy_6():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4],
        'name': ['Alice', 'Bob', 'Charlie', 'David'],
        'score': [85, 92, 78, 95]
    })
    # Your code here
    max_row = df.loc[[df['score'].idxmax()]]
    return max_row

# Test:
print("Easy Problem 6:")
print(easy_6())
print()
```

```python
# Problem 7: Get value counts sorted
# DataFrame: id, color
# Count occurrences of each color, sorted descending
# Expected: Series with counts
def easy_7():
    df = pd.DataFrame({
        'id': [1, 2, 3, 4, 5, 6, 7],
        'color': ['red', 'blue', 'red', 'green', 'blue', 'red', 'blue']
    })
    # Your code here
    counts = df['color'].value_counts()
    return counts

# Test:
print("Easy Problem 7:")
print(easy_7())
print()
```

```python
# Problem 8: Sample random rows
# DataFrame: id, value (10 rows)
# Get random sample of 3 rows
# Expected: 3 rows (random)
def easy_8():
    df = pd.DataFrame({
        'id': range(1, 11),
        'value': range(10, 110, 10)
    })
    # Your code here (use random_state=42 for reproducibility)
    return df.sample(n=3,random_state=42)

# Test:
print("Easy Problem 8:")
print(easy_8())
print()
```

```python
# Problem 9: Get basic info about DataFrame
# DataFrame: id, name, salary, department
# Return DataFrame info (dtypes, non-null counts)
# Expected: info display
def easy_9():
    df = pd.DataFrame({
        'id': [1, 2, 3],
        'name': ['Alice', 'Bob', 'Charlie'],
        'salary': [50000.5, 60000.75, 75000.0],
        'department': ['Sales', 'IT', 'HR']
    })
    # Your code here (return df.info())
    # Note: df.info() returns None, so you might return a string or df description
    return df.info()

# Test:
print("Easy Problem 9:")
easy_9()
print()
```
