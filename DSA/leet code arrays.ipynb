```python
def numberGame(nums: list[int]) -> list[int]:
    new_lst = []
    for i in range(len(nums)):
        if nums:
            a = nums.pop(nums.index(min(nums)))
            b = nums.pop(nums.index(min(nums)))
            new_lst.append(b)
            new_lst.append(a)
    return new_lst
print(numberGame([5,4,2,3]))
print(numberGame([2,5]))
```
