```python
# ------------------------------------------------------------
# Problem 1: Find the Maximum Element in an Array
#
# Given an integer array nums, return the largest element.
#
# Example:
# Input:  [3, 7, 2, 9, 5]
# Output: 9
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------
def find_maximum(nums: list[int]) -> int:
    """
    Finds the largest element in the given integer array.
    """
    if not nums:
        raise ValueError("Array cannot be empty")
        
    return max(nums)
print(find_maximum([3, 7, 2, 9, 5]))
print(find_maximum([]))
```

```python
# ------------------------------------------------------------
# Problem 2: Find the Minimum Element in an Array
#
# Given an integer array nums, return the smallest element.
#
# Example:
# Input:  [8, 4, 6, 1, 9]
# Output: 1
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------
def find_minimum(nums: list[int]) -> int:
    """
    Finds the smallest element in the given integer array.
    """
    if not nums:
        raise ValueError("Array cannot be empty")
        
    return min(nums)

print(find_minimum([8, 4, 6, 1, 9]))
print(find_minimum([]))
```

```python
# ------------------------------------------------------------
# Problem 3: Check if an Array is Sorted
#
# Given an integer array nums, return True if the array is
# sorted in non-decreasing order, otherwise return False.
#
# Example 1:
# Input:  [1, 2, 2, 4, 7]
# Output: True
#
# Example 2:
# Input:  [1, 5, 3]
# Output: False
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------

def is_sorted_pythonic(nums: list[int]) -> bool:
    return all(nums[i] <= nums[i + 1] for i in range(len(nums) - 1))
print(is_sorted_pythonic([1, 2, 2, 4, 7]))
```
