```python
# ------------------------------------------------------------
# Problem 1: Find the Maximum Element in an Array
#
# Given an integer array nums, return the largest element.
#
# Example:
# Input:  [3, 7, 2, 9, 5]
# Output: 9
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------
def find_maximum(nums: list[int]) -> int:
    """
    Finds the largest element in the given integer array.
    """
    if not nums:
        raise ValueError("Array cannot be empty")
        
    return max(nums)
print(find_maximum([3, 7, 2, 9, 5]))
print(find_maximum([]))
```

```python
# ------------------------------------------------------------
# Problem 2: Find the Minimum Element in an Array
#
# Given an integer array nums, return the smallest element.
#
# Example:
# Input:  [8, 4, 6, 1, 9]
# Output: 1
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------
def find_minimum(nums: list[int]) -> int:
    """
    Finds the smallest element in the given integer array.
    """
    if not nums:
        raise ValueError("Array cannot be empty")
        
    return min(nums)

print(find_minimum([8, 4, 6, 1, 9]))
print(find_minimum([]))
```

```python
# ------------------------------------------------------------
# Problem 3: Check if an Array is Sorted
#
# Given an integer array nums, return True if the array is
# sorted in non-decreasing order, otherwise return False.
#
# Example 1:
# Input:  [1, 2, 2, 4, 7]
# Output: True
#
# Example 2:
# Input:  [1, 5, 3]
# Output: False
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------

def is_sorted_pythonic(nums: list[int]) -> bool:
    return all(nums[i] <= nums[i + 1] for i in range(len(nums) - 1))
print(is_sorted_pythonic([1, 2, 2, 4, 7]))
```

```python
# ------------------------------------------------------------
# Problem 4: Reverse an Array (In-place)
#
# Reverse the given array without using another array.
#
# Example:
# Input:  [1, 2, 3, 4, 5]
# Output: [5, 4, 3, 2, 1]
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------

def reverse_array(nums: list[int]) -> list[int]:
    """
    Reverses the given array in-place.
    """
    left = 0
    right = len(nums) - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
        
    return nums
print(reverse_array([1, 2, 3, 4, 5]))
```

```python
# ------------------------------------------------------------
# Problem 5: Remove Duplicates from a Sorted Array
#
# Given a sorted array, remove duplicates in-place and
# return the number of unique elements.
#
# Example:
# Input:  [1, 1, 2, 2, 3]
#
# Output:
# Length = 3
# Array = [1, 2, 3, ...]
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------

def remove_duplicates(nums: list[int]) -> int:
    """
    Removes duplicates from a sorted array in-place.
    """
    # Edge case: an empty array has 0 unique elements
    if not nums:
        return 0
        
    # 'unique_index' tracks where the last unique element sits
    unique_index = 0
    
    for i in range(1, len(nums)):
        if nums[i] != nums[unique_index]:
            unique_index += 1
            nums[unique_index] = nums[i]
            
    return unique_index + 1 

remove_duplicates([1, 1, 2, 2, 3])
```

```python
# ------------------------------------------------------------
# Problem 6: Move All Zeros to the End
#
# Move all zeros to the end while maintaining the relative
# order of the non-zero elements.
#
# Example:
# Input:  [0, 1, 0, 3, 12]
# Output: [1, 3, 12, 0, 0]
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------

def push_zeroes(nums: list(int)) -> list(int):
    write_ptr = 0
    
    for read_ptr in range(len(nums)):
        if nums[read_ptr] != 0:
            nums[write_ptr], nums[read_ptr] = nums[read_ptr], nums[write_ptr]
            write_ptr += 1
    return nums
print(push_zeroes([0,1,0,3,12]))
print(push_zeroes([0,0,1]))
```

```python
# ------------------------------------------------------------
# Problem 7: Left Rotate an Array by One Place
#
# Rotate the array to the left by one position.
#
# Example:
# Input:  [1, 2, 3, 4, 5]
# Output: [2, 3, 4, 5, 1]
#
# Expected Time Complexity: O(n)
# Expected Space Complexity: O(1)
# ------------------------------------------------------------
def rotate_left(nums: list[int]) -> list[int]:
    n = len(nums)
    k =1
    def reverse(left: int, right: int):
        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1
    reverse(0, k - 1)
    reverse(k, n - 1)
    reverse(0, n - 1)
    
    return nums

# Test Case
print(rotate_left([1, 2, 3, 4, 5,]))
```
