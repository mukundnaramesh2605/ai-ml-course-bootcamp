# Regex - Solutions

For each problem below, open [regex101.com](https://regex101.com), set the flavor to **Python**, and paste the following text into the test string box:

```
FOUND WALLET - turned in March 3, 2026 - call 123-482-1847
Piano lessons - $35.00/hr - starting April 10, 2024
Garage sale - Saturday May 18, 2026 - items from $0.50 to $175.00
For rent: 2BR apt available June 1, 2025 - call 123-903-4621
Tech help - beginner-friendly - $20.00/hr - since January 7, 2026
Babysitter available - from February 14, 2023 - rates from $15.00/hr
```

Then write your pattern in the pattern box at the top.

## Problem 1

Write a regular expression that matches all of the phone numbers in the text.

Each phone number has the format `123-XXX-XXXX` — a literal area code `123`, a hyphen, three digits, a hyphen, and four digits.

### Solution

**Pattern:** `123-[0-9]{3}-[0-9]{4}`

Every phone number starts with the same area code, so we write `123-` as a literal.

Next come three digits. To match a single digit we use the character class `[0-9]`, and to require exactly three of them we follow it with the quantifier `{3}`.

Then we write the hyphen as a literal, followed by `[0-9]{4}` to match the final four digits.

Together, `123-[0-9]{3}-[0-9]{4}` matches both phone numbers in the text.

## Problem 2

Write a regular expression that matches all of the prices in the text.

Each price has the format `$X.XX` — a dollar sign, one or more digits, a dot, and exactly two digits (e.g. `$35.00`, `$0.50`, `$75.00`).

### Solution

**Pattern:** `\$[0-9]+\.[0-9]{2}`

Every price starts with a dollar sign. In regex, `$` is special syntax — it means "end of line" — so to match a literal dollar sign, we have to escape it with a backslash: `\$`.

Next come the digits before the decimal. Since prices can have one or more digits (`$0.50` vs `$75.00`), we use `[0-9]+` — the `+` quantifier means one or more.

Then we need to match the dot. Just like `$`, the dot is special in regex — it's a wildcard that matches any character. To match a literal dot, we escape it: `\.`.

Finally, every price ends with exactly two digits, so we use `[0-9]{2}`.

Together, `\$[0-9]+\.[0-9]{2}` matches all five prices in the text.

## Problem 3

Write a regular expression that matches all of the dates in the text.

Each date has the format `Month D, YYYY` — a month name, a space, one or two digits for the day, a comma and space, and a four-digit year (e.g. `March 3, 2024`, `April 10, 2024`).

> **Heads up:** the word `Saturday` also matches part of your pattern. That's expected — think about why, and whether the pattern is still correct for the task.

### Solution

**Pattern:** `[A-Za-z]+ \d{1,2}, \d{4}`

We'll build this left to right.

Month names are made up of letters — both upper and lowercase. We create a character class `[A-Za-z]` to match any letter, then follow it with `+` to match one or more of them.

After the month name comes a literal space.

The day can be one digit or two — `3` vs. `18`. We use `\d` (shorthand for any digit) with a range quantifier: `\d{1,2}` means "one or two digits."

Then we write `, ` — a comma and a space — as literals.

Finally, the year is always four digits: `\d{4}`.

The full pattern `[A-Za-z]+ \d{1,2}, \d{4}` matches all six dates in the text.

You may have noticed that `Saturday` also partially matched — the letters `Saturday` fit `[A-Za-z]+`. But the match stops there, because `Saturday` isn't followed by a space and then digits. The full pattern only highlights complete dates, so the result is still correct.
